### Страница собрана из кусочков — ячеек

Ячейки бывают двух видов.

**Текстовые** — вроде этой. В них объяснения и картинки.

**Кодовые** — в них команды для компьютера. Слева у такой ячейки
круглая кнопка ▶. Нажал — команда выполнилась, результат появился внизу.

### Как запускать

* нажать ▶ слева от ячейки;
* или клавиши **Shift + Enter** — так быстрее.

Пока ячейка думает, слева крутится кружок. Обычно это доли секунды.

> **Важно.** Ячейки выполняются в том порядке, в котором ты жмёшь кнопку.

> следующая может не сработать. Так что иди сверху вниз.

## Смотрим, как это работает

### Первая команда

Нажми ▶ у ячейки ниже.

````python
print("Привет!")
````

Появилось слово. Разберём, что тут написано:

* `print` — по-английски «печатай». Команда «выведи на экран»;
* скобки — в них кладут то, что печатать;
* кавычки — говорят компьютеру: внутри обычный текст, не команда.

Кавычки на экран не попадают. Они как рамка: показывают, где текст
начинается и где кончается.

Попробуй поменять слово внутри кавычек на своё имя и запустить снова.

### Компьютер считает

Если кавычки убрать, а написать числа — он посчитает.

````python
print(2 + 2)
print(100 * 365)
print("100 * 365")
````

### Ячейка-калькулятор

Эта ячейка пригодится в задачах: меняй в ней пример и запускай.
Сломать нельзя.

````python
# калькулятор: замени пример на свой и нажми ▶
print(3 * 7)
````

Последняя строчка — самая интересная. В кавычках компьютер ничего
не считает, он просто печатает то, что видит.

Запомни: **кавычки выключают вычисление**.

### Красный текст — это не страшно

Ячейка ниже сломана специально. Запусти её.

````python
# СПЕЦИАЛЬНО СЛОМАННАЯ ЯЧЕЙКА  demo-error
print("Забыли закрыть кавычку)
````

Вылезло красное. Ничего не сгорело, компьютер просто не понял команду
и честно об этом сказал.

Главное в этом сообщении — последняя строчка. Там написано
`unterminated string`, то есть «незакрытая строка». Компьютер даже
показал стрелочкой, где споткнулся.

Почини ячейку выше: поставь кавычку перед скобкой и запусти снова.

Программисты ломают код по сто раз на дню. Это не провал, а обычная
работа: написал, сломалось, починил, поехали дальше.
