## Смотрим, как это работает

### Заготовка: викторина

Ответы игрока в настоящем проекте — поля формы Colab (`#@param`),
как в задачах учебника. Здесь они уже вписаны.

```python
вопросы = ["Сколько ног у паука?", "Столица Франции?", "Сколько будет 7 · 8?"]
правильные = ["8", "Париж", "56"]
ответы_игрока = ["8", "Лондон", "56"]

верных = 0
for i in range(len(вопросы)):
    print(вопросы[i])
    if ответы_игрока[i] == правильные[i]:
        print("  Верно!")
        верных = верных + 1
    else:
        print("  Неверно, правильно:", правильные[i])

print("Верных ответов:", верных, "из", len(вопросы))
```

### Заготовка: рисунок кодом

```python
for i in range(1, 6):
    print(" " * (5 - i) + "*" * (2 * i - 1))
print("    |")
```

Ёлка из пяти строк. Поменяй значки и размер — получится своя.

### Заготовка: копилка на мечту

```python
мечта = 3000      # сколько стоит
есть = 500        # уже накоплено
в_неделю = 200    # откладываю каждую неделю

недель = 0
while есть < мечта:
    есть = есть + в_неделю
    недель = недель + 1

print("Накоплю через", недель, "недель")
```

### Заготовка: угадай число 2.0

Очков тем больше, чем быстрее угадал.

```python
import random

random.seed(32)
загаданное = random.randint(1, 50)
попытки = [25, 12, 6, 3, 4, 5]

очки = 10
for попытка in попытки:
    if попытка == загаданное:
        print(попытка, "— угадал! Очков:", очки)
        break
    if попытка < загаданное:
        print(попытка, "— больше")
    else:
        print(попытка, "— меньше")
    очки = очки - 1
```

`break` останавливает цикл сразу, как только угадали: дальше
перебирать попытки незачем.
