### К июню будешь уметь

- понимать, из чего собран компьютер и что происходит, когда ты жмёшь кнопку
- не попадаться мошенникам и держать свои данные при себе
- находить в сети то, что нужно, и отличать правду от выдумки
- рисовать, считать и оформлять — в таблицах, в редакторе и кодом
- собрать и показать собственный проект

Уроки идут по порядку и опираются друг на друга — перепрыгивать через них не стоит.

---

### Как здесь работать

Один и тот же порядок на каждом уроке:

1. Открыть урок по ссылке из таблицы ниже.
2. Если Colab предупредит, что блокнот «создан не в Google», — нажать **«Всё равно выполнить»**. Так он встречает любой чужой файл.
3. В первой ячейке вписать ФИО, класс и адрес журнала (его даёт учитель) и запустить её — кнопкой ▶ или Shift+Enter.
4. Дальше идти по ячейкам сверху вниз, решать задачи и запускать проверки. В конце — ячейка `si.report()` с итогом.

Нужен Google-аккаунт: без него кнопки запуска работать не будут.

Результат каждой проверки уходит учителю сразу, так что сохранять урок не обязательно. Хочешь оставить свои решения себе — «Файл → Сохранить копию на Диске».

Чтобы не вписывать ФИО и адрес журнала каждую неделю, в конце первых уроков есть ячейка `si.запомнить()`: она объясняет, как сохранить их один раз навсегда.
