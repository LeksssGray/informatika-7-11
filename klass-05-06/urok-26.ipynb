## Что делаем

Собираем настоящую игру. Компьютер загадывает число, игрок
угадывает, а программа подсказывает: «больше» или «меньше».

Всё, что для этого нужно, мы уже знаем — кроме двух новых вещей.
С них и начнём.

## Разбираемся

### Новое первое: компьютер умеет выбирать случайно

Чтобы загадать число, нужна случайность. Она живёт в наборе готовых
команд, который называется **модуль**. Модуль подключают одной
строчкой:

```python
import random

загаданное = random.randint(1, 100)
```

`random.randint(1, 100)` даёт случайное число от 1 до 100,
причём **обе границы входят** — не как в `range`.

### Новое второе: компьютер умеет сравнивать и выбирать

Чтобы сказать «больше» или «меньше», программе нужно **сравнить**
и выбрать один из ответов. Для этого есть `if`:

```python
if попытка < загаданное:
    print("Больше!")
else:
    print("Меньше!")
```

Читается: «если попытка меньше загаданного — напечатай „больше“,
иначе — „меньше“».

Строчки с отступом — то, что выполнится. Как в цикле.

### Три ответа вместо двух

У игры три исхода: угадал, больше, меньше. Для среднего случая
есть слово `elif` — «иначе если»:

```python
if попытка == загаданное:
    print("Угадал!")
elif попытка < загаданное:
    print("Больше!")
else:
    print("Меньше!")
```

Обрати внимание: сравнение — **два** знака равно. Один знак равно
означает «положи значение в переменную», а два — «сравни».
Это разные вещи, и путают их все.

### Как играть без ввода

По-настоящему игра спрашивала бы число у человека. Мы пока сделаем
иначе: сложим попытки в список заранее — как будто игрок уже
сходил несколько раз. Так игру можно проверять сколько угодно раз,
и она не будет ждать.

```python
попытки = [50, 25, 37, 42]

for попытка in попытки:
    ...
```

Цикл `for попытка in попытки` идёт **сразу по значениям** списка:
на первом шаге `попытка` равна 50, на втором 25 и так далее.

### Хорошая стратегия

Чтобы угадать число от 1 до 100, не нужно перебирать всё подряд.
Достаточно каждый раз называть **середину** оставшегося отрезка:

```
1…100 → назвал 50 → «больше» → осталось 51…100
51…100 → назвал 75 → «меньше» → осталось 51…74
```

Каждый раз отрезок уменьшается вдвое. Сотня «съедается» за семь
попыток, а тысяча — за десять. Так ищут слово в бумажном словаре:
открывают посередине.

## Смотрим, как это работает

### Случайное число

````python
import random

random.seed(5)          # чтобы у всех вышло одинаково

загаданное = random.randint(1, 100)

print("Компьютер загадал число от 1 до 100.")
print("Подсмотрим (в игре так нельзя):", загаданное)
````

Строчка `random.seed(5)` делает случайность предсказуемой: у всех
в классе выйдет одно и то же число. Без неё каждый запуск давал бы
новое — и проверка не смогла бы ничего сверить.

### Сравниваем и отвечаем

````python
загаданное = 42
попытка = 30

if попытка == загаданное:
    print("Угадал!")
elif попытка < загаданное:
    print("Больше!")
else:
    print("Меньше!")
````

### Пробуем разные попытки

````python
загаданное = 42

for попытка in [30, 60, 42]:
    if попытка == загаданное:
        print(попытка, "→ Угадал!")
    elif попытка < загаданное:
        print(попытка, "→ Больше!")
    else:
        print(попытка, "→ Меньше!")
````

### Игра целиком

````python
загаданное = 42
попытки = [50, 25, 37, 43, 42]

номер = 0

for попытка in попытки:
    номер = номер + 1
    if попытка == загаданное:
        print("Попытка", номер, ":", попытка, "— Угадал!")
    elif попытка < загаданное:
        print("Попытка", номер, ":", попытка, "— Больше!")
    else:
        print("Попытка", номер, ":", попытка, "— Меньше!")

print()
print("Угадал за попыток:", номер)
````

### Почему середина — хорошая стратегия

````python
осталось = 100
попыток = 0

for i in range(10):
    if осталось > 1:
        осталось = осталось // 2
        попыток = попыток + 1
        print("Попытка", попыток, "— осталось вариантов:", осталось)

print()
print("Сотню хватает угадать за", попыток, "попыток.")
````

### Считаем, сколько подсказок

````python
попытки = [50, 25, 37, 43, 42]

print("Попыток сделано:", len(попытки))
print("Из них подсказок:", len(попытки) - 1)
print("Последняя попытка — уже ответ.")
````

### Любопытно

Случайные числа в компьютере не случайны. `random.randint` вычисляет
следующее число из предыдущего по формуле — получается
последовательность, которая **выглядит** беспорядочной, но полностью
предсказуема, если знать начальное значение. Его и задаёт `seed`.

Для игры этого достаточно. А там, где случайность должна быть
настоящей — например, при защите переписки, — её берут из мира:
из шума радиоприёмника, дрожания температуры процессора или даже
из съёмки лавовых лампочек. Такая стена с лампами правда стоит
в одной большой интернет-компании и помогает шифровать
соединения.
