## Разбираемся

### Одна команда, которую ты уже знаешь

`print` — «напечатай». Мы её уже пробовали, теперь разберём как следует.

```python
print("Привет")
```

Внутри скобок кладут то, что надо вывести. В кавычках — текст,
без кавычек — числа и вычисления.

### Несколько кусков сразу

Через запятую можно перечислить что угодно, компьютер сам расставит
пробелы:

```python
print("Мне", 12, "лет")     →  Мне 12 лет
```

Текст в кавычках, число без — и всё это мирно уживается.

### Строки складываются

Звучит странно, но так и есть:

```python
"при" + "вет"     →  "привет"
```

Плюс склеивает два текста в один. Только имей в виду: пробел сам
не появится. `"при" + "вет"` даст `привет`, а если нужен пробел —
его добавляют вручную: `"при" + " " + "вет"`.

### Строки умножаются

Ещё страннее, но очень полезно:

```python
"-" * 20     →  "--------------------"
"ха" * 3     →  "хахаха"
```

Звёздочка повторяет текст заданное число раз. Из плюса и звёздочки
собирается почти любая картинка из символов.

### Перевод строки

Чтобы текст пошёл с новой строки, внутрь кавычек ставят `\n`:

```python
print("первая\nвторая")
```

напечатает две строки. Обратный слеш с буквой n — это не два символа
на экране, а команда «перейди на новую строку».

### Комментарии

Всё, что после знака `#`, компьютер пропускает. Это заметки для людей:

```python
# тут считаем, сколько дней до каникул
дней = 45
```

Пиши в комментариях не «что» (это и так видно), а «зачем».

## Смотрим, как это работает

### Склеиваем и повторяем

````python
print("ха" * 5)
print("=" * 30)
print("при" + "вет")
print("при" + " " + "вет")
````

Обрати внимание на две последние строки: без пробела слова слиплись.
Компьютер не догадался, что там нужен пробел — потому что не догадывается
вообще ни о чём.

### Рамка

````python
print("+" + "-" * 18 + "+")
print("|" + " " * 18 + "|")
print("|" + " Привет, 5 класс! " + "|")
print("|" + " " * 18 + "|")
print("+" + "-" * 18 + "+")
````

Посмотри, как собрана верхняя строка: плюсик, восемнадцать дефисов,
ещё плюсик. Меняй число 18 и смотри, как рамка растягивается.

### Смайлик из символов

````python
смайлик = "^_^"

print(смайлик)
print(смайлик * 3)
print((смайлик + " ") * 5)
````

Скобки нужны, чтобы сначала склеить смайлик с пробелом, а уже потом
повторить всё вместе пять раз. Без скобок получилось бы другое —
попробуй убрать и посмотри.
