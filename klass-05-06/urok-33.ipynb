### Как работать

2. Сделай **одну** часть — и сразу запусти.

### Застрял? Чек-лист

1. **Прочитай последнюю строку ошибки.** Там написано, что случилось.
2. **Посмотри на номер строки** в сообщении — ошибка там или чуть выше.
3. **Напечатай переменные** перед местом, где что-то идёт не так.
4. **Объясни код вслух** строчку за строчкой — хоть игрушке на столе.
   Программисты называют это «методом утёнка»: пока объясняешь,
   сам замечаешь ошибку.
5. **Сравни с заготовкой** из урока 32.
6. Спроси соседа. Потом — учителя.

### Как читать ошибку

Python сообщает об ошибке несколькими строками, но главное —
**последняя**:

```
NameError: name 'сумм' is not defined
```

Это значит «не знаю никакого `сумм`» — скорее всего, опечатка
в имени. Частые ошибки:

| Ошибка | Что значит | Частая причина |
|---|---|---|
| `NameError` | нет переменной с таким именем | опечатка, е вместо ё, переменную не завели |
| `TypeError` | нельзя так делать с этим типом | склеили текст с числом: `"Мне " + 11` |
| `IndentationError` | неправильный отступ | под `for` или `if` забыли отступ |
| `SyntaxError` | Python не понимает запись | забыли двоеточие, скобку, кавычку |
| `ZeroDivisionError` | деление на ноль | делитель оказался нулём |

## Смотрим, как это работает

### Печатаем переменные, чтобы найти баг

Копилка должна набрать 1000, а программа говорит «12 недель»
вместо ожидаемых 10. Добавим `print` внутрь цикла:

````python
есть = 0
недель = 0

while есть < 1000:
    есть = есть + 90
    недель = недель + 1
    print("неделя", недель, "— накоплено", есть)
````

Видно, что откладываем по 90, а не по 100: опечатка в числе.
Без печати её пришлось бы искать глазами.

### Склеиваем текст и число правильно

`print` сам ставит пробелы между частями. А если нужно собрать
строку в переменную, число превращают в текст через `str`:

````python
возраст = 11

print("Мне", возраст, "лет")

фраза = "Мне " + str(возраст) + " лет"
print(фраза)
````
