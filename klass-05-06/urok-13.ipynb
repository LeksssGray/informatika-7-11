## Как это было: феи из Коттингли

Две сестры, Элси и Фрэнсис, в 1917 году сделали первые снимки «фей»
у ручья в деревне Коттингли. Фотографии попали к известным людям,
и в них поверил сам Артур Конан Дойл — автор Шерлока Холмса, самого
внимательного сыщика на свете. Он написал о феях статью и целую
книгу.

Правду сёстры рассказали только в 1983 году, когда уже были
старушками. Фей они вырезали из детской книжки, приклеили на картон
и закрепили булавками в траве. Обман продержался больше шестидесяти лет.

Главный урок этой истории: люди охотнее верят в то, во что **хотят**
верить. Конан Дойл очень хотел, чтобы феи существовали.

## Разбираемся

### Почему фейки так быстро разлетаются

**Фейк** — выдуманная новость, поддельное фото или видео, которое
выдают за правду. Некоторые делают ради шутки, некоторые — чтобы
напугать, разозлить или заработать на просмотрах.

Фейки распространяются быстрее правды, потому что бьют по чувствам:
«Срочно!», «Шок!», «Перешли всем, пока не удалили!». Удивительное
хочется показать друзьям прямо сейчас, а проверять — некогда.

Классический пример: после каждого сильного наводнения в сети появляется
фото акулы, плывущей по затопленной улице. Много лет подряд, в разных
странах, одна и та же картинка — склеенная из двух снимков.

### Пять вопросов к любой новости

1. **Кто автор?** Известное издание, учёный, очевидец — или никто
   не знает?
2. **Откуда взято?** Есть ссылка на первоисточник или «говорят,
   что…»?
3. **Когда?** Может, новости десять лет, а её выдают за вчерашнюю.
4. **Что пишут другие?** Если важная новость есть только в одном
   чате, это подозрительно.
5. **Что я чувствую?** Если хочется срочно разозлиться или испугаться —
   это повод притормозить и проверить.

### Как проверить картинку

Самый быстрый способ — **поиск по картинке**. В Яндексе это кнопка
с фотоаппаратом в строке поиска «Картинок», в Google — Google Объектив.
Загружаешь фото и видишь, где оно появлялось раньше. Часто оказывается,
что «вчерашнее фото» сделано пять лет назад в другой стране.

### Картинки, которые нарисовал компьютер

Сегодня нейросети рисуют фотографии людей, которых не существует,
и видео, где известный человек говорит то, чего никогда не говорил.
Их всё труднее отличить от настоящих, но подсказки есть:

* **пальцы** — лишние, сросшиеся, странно согнутые;
* **надписи** на вывесках и футболках — буквы-каракули без смысла;
* **мелочи** — серьги разные, очки сливаются с кожей, фон «плывёт»;
* **слишком гладко** — кожа как у куклы, свет слишком красивый.

Даже если подсказок не видно, помогают те же пять вопросов: кто
выложил и где это ещё можно найти.
