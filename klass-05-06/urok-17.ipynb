## Разбираемся

### Три лампочки

Возьми лупу и посмотри на экран телефона очень близко. Каждая точка
на самом деле — три крошечные полоски: **красная, зелёная и синяя**.

Меняя их яркость, получают любой цвет. Красная на полную плюс
зелёная на полную — жёлтый. Все три на полную — белый.
Все погасли — чёрный.

Такой способ называют **RGB**: от английских слов Red, Green, Blue.

### Как записывают цвет

Яркость каждой полоски — число **от 0 до 255**. Ноль — не светится
совсем, 255 — во всю силу.

| Цвет | R | G | B |
|---|---|---|---|
| чёрный | 0 | 0 | 0 |
| белый | 255 | 255 | 255 |
| красный | 255 | 0 | 0 |
| зелёный | 0 | 255 | 0 |
| синий | 0 | 0 | 255 |
| жёлтый | 255 | 255 | 0 |
| оранжевый | 255 | 165 | 0 |
| серый | 128 | 128 | 128 |

### Почему жёлтый, а не зелёный

На уроках рисования красный и зелёный дают буро-зелёную кашу.
А тут — яркий жёлтый. Дело в том, что краски и лампочки работают
по-разному.

Краска **забирает** часть света: чем больше красок смешаешь, тем
темнее. Лампочки **добавляют** света: чем больше включишь, тем
светлее. Поэтому на экране три цвета вместе дают белый,
а не чёрный.

### Почему именно 255

Потому что на каждую полоску компьютер отводит один байт — восемь
бит. Разных значений получается 256: от 0 до 255.

Три полоски по 256 оттенков — это 256 × 256 × 256, то есть больше
шестнадцати миллионов цветов. Больше, чем человек способен
различить.

## Смотрим, как это работает

### Записываем цвет числами

```python
красный = 255
зелёный = 165
синий = 0

print("Мой цвет:", красный, зелёный, синий)
print("Сумма яркостей:", красный + зелёный + синий)
```

Это оранжевый. Проверить легко: набери в поиске
«rgb 255 165 0» — и увидишь квадратик такого цвета.

### Собираем запись цвета строкой

```python
r = 128
g = 0
b = 128

цвет = "rgb(" + str(r) + ", " + str(g) + ", " + str(b) + ")"

print(цвет)
print("Это фиолетовый: красная и синяя полоски горят вполсилы.")
```

Обрати внимание на `str(r)`: число нужно превратить в текст,
иначе Python не даст склеить его со строкой. Это и есть
**преобразование типа**.

### Считаем, сколько всего цветов

```python
оттенков = 256

всего = оттенков * оттенков * оттенков

print("Оттенков у одной полоски:", оттенков)
print("Всего цветов:", всего)
```

### Из текста в число

```python
из_формы = "200"

яркость = int(из_формы)

print("Было текстом:", из_формы)
print("Стало числом:", яркость)
print("Половина яркости:", яркость // 2)
```

Зачем это нужно: всё, что человек вписывает в поле, компьютер
получает **текстом**, даже если это цифры. Чтобы считать,
текст превращают в число функцией `int`.

### Делаем цвет темнее

```python
r = 200
g = 100
b = 50

print("Было:  ", r, g, b)

r = r // 2
g = g // 2
b = b // 2

print("Стало: ", r, g, b)
print("Каждую яркость уменьшили вдвое — цвет тот же, но темнее.")
```
