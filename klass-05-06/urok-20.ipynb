## Разбираемся

### Звук — это колебания воздуха

Когда струна дрожит, она толкает воздух: сжатие, разрежение,
сжатие, разрежение. Эти колебания добегают до уха — и мы слышим.

Нарисовать их можно **волной**: выше — воздух сжат сильнее,
ниже — слабее.

### Что делает микрофон

Внутри микрофона тонкая плёнка. Воздух её толкает, она дрожит,
и дрожание превращается в слабый электрический ток — такую же
волну, только не из воздуха.

Дальше начинается работа компьютера: волну нужно превратить
в числа. Компьютер ведь ничего, кроме чисел, не умеет.

### Как волну превращают в числа

Компьютер измеряет высоту волны много раз в секунду и записывает
числа. Это называют **оцифровкой**.

```
волна:   ／‾＼＿／‾＼＿
числа:   3 5 4 1 2 5 4 1
```

Два главных вопроса: **как часто измерять** и **насколько точно
записывать каждое измерение**.

### Как часто измеряют

Обычный музыкальный файл — **44 100 измерений в секунду**.
Сорок четыре тысячи раз за одну секунду!

Почему так много: человек слышит колебания до 20 000 в секунду,
а чтобы волну потом узнать, измерять надо чаще, чем она колеблется.

Чем реже измеряешь, тем хуже узнаётся звук. Поэтому голос
по телефону звучит «плоско»: там измерений всего 8 000 в секунду —
для слов хватает, для музыки нет.

### Насколько точно записывают

Каждое измерение — число. Если на число отвести один байт,
получится 256 разных уровней громкости; если два байта — 65 536.

Чем больше уровней, тем точнее передаются тихие места. Обычный
музыкальный файл использует два байта на измерение.

### Сколько это весит

Считается просто: **измерений в секунду × байт на измерение ×
секунд**. И ещё умножить на два, если звук стерео — то есть
отдельно для левого и правого уха.

Минута музыки без сжатия — около десяти мегабайт. Поэтому файлы
**сжимают**: убирают то, что человек всё равно не услышит —
очень тихое рядом с очень громким, слишком высокое, слишком низкое.

Так получается mp3: файл меньше в десять раз, а разницу большинство
людей не замечает. Но она есть, и на хороших наушниках слышна.

## Смотрим, как это работает

### Рисуем волну значками

````python
print("высоко  " + "█" * 2 + " " * 6 + "█" * 2)
print("средне  " + "█" * 4 + " " * 2 + "█" * 4)
print("низко   " + "█" * 6 + "█" * 4)
````

### Сколько измерений в минуте

````python
в_секунду = 44100
секунд = 60

измерений = в_секунду * секунд

print("Измерений в секунду:", в_секунду)
print("В минуте:", измерений)
````

### Сколько весит минута музыки

````python
в_секунду = 44100
байт_на_измерение = 2
каналов = 2
секунд = 60

байт = в_секунду * байт_на_измерение * каналов * секунд

print("Байт:", байт)
print("Килобайт:", байт // 1024)
print("Мегабайт:", байт // 1024 // 1024)
````

Больше десяти мегабайт за одну минуту! Альбом из десяти песен
занял бы целый диск.

### Что даёт сжатие

````python
без_сжатия = 10
сжатый = 1

print("Минута без сжатия, Мбайт:", без_сжатия)
print("Минута в mp3, Мбайт:     ", сжатый)
print("Меньше в", без_сжатия // сжатый, "раз")
````

### Голос по телефону против музыки
