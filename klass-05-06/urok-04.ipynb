## Разбираемся

### У телефона внутри то же самое

Открой список деталей телефона — увидишь знакомое: процессор,
оперативная память, встроенная память. Те же четыре штуки, что
и в системном блоке, только размером с ноготь.

Телефон — это компьютер. Просто маленький, с сенсорным экраном
и батарейкой.

### Чем он всё-таки отличается

**Всё на одном чипе.** В компьютере процессор, видеокарта и остальное —
отдельные детали. В телефоне их упаковали в один кристалл размером
с ноготь. Компактно, экономно, но заменить ничего нельзя: сломалось —
меняют целиком.

**Он работает от батареи.** Отсюда главное отличие в характере:
телефон постоянно экономит. Приложение, которое ты свернул,
он замораживает. Экран притухает. Процессор сбрасывает скорость,
когда нагрузки нет.

Компьютеру всё равно, он в розетке.

**Память не расширить.** В компьютер можно доставить планку памяти
или второй диск. В телефоне — нет, сколько купил, столько и будет.

### Датчики

А вот чего в обычном компьютере нет — целой кучи датчиков.

| Датчик | Что делает |
|---|---|
| акселерометр | чувствует наклон и поворот |
| гироскоп | понимает вращение |
| освещённости | подстраивает яркость экрана |
| приближения | гасит экран, когда подносишь к уху |
| GPS | определяет, где ты находишься |

Когда картинка поворачивается вслед за телефоном — это акселерометр.
Когда экран гаснет во время звонка — датчик приближения.

### Куда девается память

Знакомая история: купил телефон на 64 гигабайта, а свободно 12.
Куда всё делось?

* **система** съедает 8–15 гигабайт сразу, ещё до первого включения;
* **приложения** растут: игра при установке весила 2 гигабайта,
  через полгода занимает 6;
* **кэш** — временные файлы, картинки, музыка «на всякий случай».
  Мессенджер легко накапливает несколько гигабайт;
* **фото и видео** — минута видео в хорошем качестве это 100–150
  мегабайт.

Первым делом при нехватке места чистят кэш и удаляют старые видео.
Это освобождает больше, чем удаление приложений.

## Смотрим, как это работает

### Куда ушла память

```python
print("Всего гигабайт:    ", 64)
print("Съела система:     ", 12)
print("Приложения:        ", 21)
print("Фото и видео:      ", 18)
print("Кэш:               ", 6)
print("Занято всего:      ", 12 + 21 + 18 + 6)
print("Осталось свободно: ", 64 - (12 + 21 + 18 + 6))
```

Семь гигабайт из шестидесяти четырёх. Знакомо?

Поменяй числа на свои и посмотри, что получится.

### Сколько видео влезет

```python
print("Свободно мегабайт: ", 7 * 1024)
print("Минута видео весит:", 130)
print("Минут видео влезет:", 7 * 1024 // 130)
```

Пятьдесят с небольшим минут. Меньше часа съёмки — и место кончилось.

Именно поэтому облако и внешние диски существуют.

### Сколько держит батарея

```python
print("Ёмкость батареи:        ", 4000)
print("Игра съедает за час:    ", 1200)
print("Часов игры примерно:    ", 4000 // 1200)
print()
print("В покое съедает за час: ", 40)
print("Часов в покое:          ", 4000 // 40)
```

Разница в тридцать раз. Игра сажает батарею за три часа, а лежащий
на столе телефон живёт четыре дня.

Потому что процессор в игре работает на полной скорости, а в покое
почти спит.
