Это вся «начинка» телефона, кроме экрана и батареи: одна плата

видеокарта и ещё десяток помощников. Рядом — память и чипы связи.

Телефон — это компьютер. Просто маленький, с сенсорным экраном
и батарейкой.

### Чем он всё-таки отличается

**Всё на одном чипе.** В компьютере процессор, видеокарта и остальное —
отдельные детали. В телефоне их упаковали в один кристалл размером
с ноготь. Компактно, экономно, но заменить ничего нельзя: сломалось —
меняют плату целиком.

**Он работает от батареи.** Отсюда главное отличие в характере:
телефон постоянно экономит. Приложение, которое ты свернул,
он замораживает. Экран притухает. Процессор сбрасывает скорость,
когда нагрузки нет. Компьютеру всё равно, он в розетке.

**Память не расширить.** В компьютер можно доставить планку памяти
или второй диск. В большинство телефонов — нет: сколько купил,
столько и будет.

### Датчики

А вот чего в обычном компьютере нет — целой кучи датчиков.

| Датчик | Что делает |
|---|---|
| акселерометр | чувствует наклон и тряску |
| гироскоп | понимает, как телефон вращают |
| освещённости | подстраивает яркость экрана |
| приближения | гасит экран, когда подносишь к уху |
| GPS | ловит сигналы спутников и определяет, где ты |
| микрофон и камеры | слышат и видят |
| отпечаток пальца | узнаёт хозяина |

Когда картинка поворачивается вслед за телефоном — это акселерометр.
Когда экран гаснет во время звонка — датчик приближения. Шагомер
тоже работает на акселерометре: он замечает, как телефон
подпрыгивает в кармане при каждом шаге.

Акселерометр в телефоне меньше макового зёрнышка. Внутри у него
крошечный грузик на пружинках: телефон наклонили — грузик сместился,
и датчик это заметил.

### Батарея: как с ней обращаться

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/akkumulyator.jpg" width="360" alt="Литий-ионный аккумулятор телефона">

*Литий-ионный аккумулятор телефона*

<sub>Eugene Varnavsky · общественное достояние · Wikimedia Commons</sub>

В телефонах стоят литий-ионные аккумуляторы. Они запасают много
энергии в маленьком объёме, но характер у них капризный.

* **Не любят жару.** Не оставляй телефон на солнце и на батарее
  отопления.
* **Не любят разряжаться в ноль.** Лучше подзаряжать раньше.
* **Стареют.** Через два-три года батарея держит заметно меньше,
  это нормально.

> **Если батарея вздулась** — корпус приподнялся, экран отходит —
> пользоваться таким телефоном больше нельзя. Не заряжай его, не дави
> и не прокалывай: вздувшийся аккумулятор может загореться. Отнеси
> в мастерскую.

### Куда девается память

Знакомая история: купил телефон на 64 гигабайта, а свободно 12.
Куда всё делось?

* **система** съедает 8–15 гигабайт сразу, ещё до первого включения;
* **приложения** растут: игра при установке весила 2 гигабайта,
  через полгода занимает 6;
* **кэш** — временные файлы, картинки, музыка «на всякий случай».
  Мессенджер легко накапливает несколько гигабайт;
* **фото и видео** — минута видео в хорошем качестве это 100–150
  мегабайт.

Первым делом при нехватке места чистят кэш и удаляют старые видео.
Это освобождает больше, чем удаление приложений.

## Смотрим, как это работает

### Куда ушла память

```python
print("Всего гигабайт:    ", 64)
print("Съела система:     ", 12)
print("Приложения:        ", 21)
print("Фото и видео:      ", 18)
print("Кэш:               ", 6)
print("Осталось свободно: ", 64 - (12 + 21 + 18 + 6))
```

Семь гигабайт из шестидесяти четырёх. Знакомо?
Поменяй числа на свои и посмотри, что получится.

### Сколько держит батарея

```python
print("Часов игры:  ", 4000 // 1200)
print("Часов в покое:", 4000 // 40)
```

Батарея на 4000 единиц. Игра съедает 1200 в час, а лежащий телефон —
40. Разница в тридцать раз: в игре процессор и экран работают
на полную, а в покое почти спят.
