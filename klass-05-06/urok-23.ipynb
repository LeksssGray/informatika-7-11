## Разбираемся

### Зачем диаграмма

Диаграмма отвечает на вопрос быстрее, чем таблица. Где больше?
Что растёт? Из чего состоит? На всё это глаз отвечает за секунду,
если картинка правильная.

Но у диаграммы есть и опасное свойство: она **убеждает**. Ей верят
даже тогда, когда она обманывает. Поэтому уметь их читать важнее,
чем рисовать.

### Какую выбрать

| Вопрос | Диаграмма |
|---|---|
| Что больше, что меньше? | **столбчатая** |
| Как меняется со временем? | **линейная** |
| Из чего состоит целое? | **круговая** |

Два правила, которые почти никто не соблюдает:

1. **Время — всегда линия.** Круговая показывает, из чего состоит
   целое **сейчас**, а не как что-то менялось.
2. **Круговая — только если части складываются в целое.**
   «Кто чем занимается: спорт, музыка, рисование» — не круг:
   кто-то ходит и туда, и туда, и сумма получится больше 100 %.

И третье: **больше пяти-шести кусков круг не читается**. Мелкие
собирают в «прочее».

### Как диаграммой обманывают

**Обрезали ось.** Столбики считают не от нуля, а, скажем, от 95.
Разница в один шаг превращается в столбик вдвое выше. Это самый
частый обман в рекламе.

**Взяли удобный кусок.** Показали три года, где линия растёт,
а двадцать предыдущих, где падала, не показали. Формально всё
правда.

**Сделали объёмные столбики.** Передний кажется больше только
потому, что он ближе.

Проверять чужую диаграмму надо так: **где начинается ось? какой
период? складываются ли части в целое?**

## Смотрим, как это работает

Рисовать будем значками — как на уроке про рисование кодом.
Так виднее механика.

### Данные

```python
классы = ["5А", "5Б", "5В", "5Г"]
человек = [26, 22, 28, 24]

print("Класс  Человек")
print("5А    ", человек[0])
print("5Б    ", человек[1])
print("5В    ", человек[2])
print("5Г    ", человек[3])
```

### Честная столбчатая: считаем от нуля

```python
print("5А |" + "█" * человек[0], человек[0])
print("5Б |" + "█" * человек[1], человек[1])
print("5В |" + "█" * человек[2], человек[2])
print("5Г |" + "█" * человек[3], человек[3])
print("   +" + "-" * 30)
print("    0" + " " * 24 + "30")
```

Разница между классами маленькая — и на картинке она маленькая.
Так и должно быть.

### А теперь обрежем ось

```python
низ = 20

print("5А |" + "█" * ((человек[0] - низ) * 4), человек[0])
print("5Б |" + "█" * ((человек[1] - низ) * 4), человек[1])
print("5В |" + "█" * ((человек[2] - низ) * 4), человек[2])
print("5Г |" + "█" * ((человек[3] - низ) * 4), человек[3])
print()
print("Числа те же! А 5В теперь выглядит вчетверо «больше» 5Б,")
print("хотя разница всего в 6 человек.")
```

Вот и весь фокус: ось начали не с нуля. Когда увидишь такую
картинку в рекламе — посмотри на подписи оси.

### Линейная: как меняется со временем

```python
месяцы = ["сен", "окт", "ноя", "дек", "янв"]
прочитано = [2, 3, 5, 4, 6]

print("Книг прочитано:")
print("сен |" + "▇" * прочитано[0], прочитано[0])
print("окт |" + "▇" * прочитано[1], прочитано[1])
print("ноя |" + "▇" * прочитано[2], прочитано[2])
print("дек |" + "▇" * прочитано[3], прочитано[3])
print("янв |" + "▇" * прочитано[4], прочитано[4])
```

### Круговая: когда можно

```python
да = 12
нет = 5
не_знаю = 3

всего = да + нет + не_знаю

print("Опрос: части складываются в целое — круг подходит")
print("да      ", "●" * да, да)
print("нет     ", "●" * нет, нет)
print("не знаю ", "●" * не_знаю, не_знаю)
print("всего:", всего)
```

### Круговая: когда нельзя

```python
спорт = 15
музыка = 9
рисование = 7
детей = 25

сумма_кружков = спорт + музыка + рисование

print("Детей в классе:", детей)
print("Записей в кружки:", сумма_кружков)
print("Записей больше, чем детей — кто-то ходит в два кружка.")
print("Круговая соврёт: части не складываются в целое.")
print("Нужна столбчатая.")
```

### Считаем доли в процентах

```python
всего = 20
да = 12

доля = да / всего * 100

print("Ответили «да»:", да, "из", всего)
print("Это процентов:", доля)
```
