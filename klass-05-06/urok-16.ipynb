## Что делаем

Вторая четверть была про безопасность: пароли, мошенники, цифровой
след, фейки, файлы и копии. Теперь соберём главное в **памятку** —
одну страницу, которую ты покажешь родителям, бабушке или младшему

Задача непростая. Взрослые обычно не любят, когда их учат. Значит,

## Вспоминаем четверть

**Пароли.** Длина важнее хитрых значков. Пять случайных слов
подбирают сотни лет. Свой пароль для почты. Двухфакторная защита.

**Мошенники.** Давят на спешку, страх, жадность и жалость. Код из SMS —
никому. Перезвони сам по знакомому номеру. Семейное секретное слово.

**Цифровой след.** Удалённое не исчезает. Геометки выдают дом.
Правило школьной доски.

**Правда или выдумка.** Кто автор, откуда, когда, что пишут другие,
что я чувствую. Поиск по картинке.

**Файлы.** Понятные имена, дата в начале, осторожно с `.exe`. Правило
«3-2-1»: три копии, два носителя, одна — в другом месте.

## Какой должна быть памятка

* **Не длиннее одной страницы.** Семь–десять правил, не больше.
* **Каждое правило — одна строчка**, которую можно выполнить. Не «будьте
  осторожны в интернете», а «код из SMS не называйте никому, даже
  „сотруднику банка“».
* **Для конкретного человека.** Бабушке важнее мошенники по телефону,
  младшему брату — бесплатные скины и геометки.
* **С объяснением «почему»** — коротко, в скобках. Правило без причины
  забывают.

чтобы увидеть, как она набрана.

# Как не попасться мошенникам

**Памятка для бабушки от Ани**

1. Звонят «из банка» или «из полиции»? **Положи трубку** и перезвони
   сама по номеру с обратной стороны карты. *(Настоящий банк
   не звонит с просьбами назвать код.)*
2. **Код из SMS не называй никому.** *(По нему снимают деньги
   с карты.)*
3. Пишут «мама, я попал в беду, срочно переведи»? **Спроси наше
   секретное слово.** *(Мошенник его не знает.)*

## Делаем

1. Реши, **для кого** памятка: для мамы, бабушки, младшего брата,
   для всего класса.
2. Выбери из четверти **самое важное для этого человека** —
   семь–десять правил.
3. Наведи мышку под эту ячейку, нажми **+ Текст** и набери памятку.
   Разметку ты помнишь со страницы о себе: `#` — заголовок,
   `1.` — список, `**жирный**`, `*курсив*`.
4. Нажми Shift + Enter и посмотри, что получилось.

### Проверь по списку

* заголовок и для кого памятка;
* от семи до десяти правил, каждое — одна строчка;
* у каждого правила — короткое «почему»;
* нет личных данных: адресов, настоящих паролей, номеров карт;
* можно прочитать за минуту.
