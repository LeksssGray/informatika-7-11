## Начнём с игры: пакеты

Эту игру проводят всем классом, без компьютера.

Учитель пишет короткое сообщение — например, «ЗАВТРА КОНТРОЛЬНАЯ
ОТМЕНЯЕТСЯ» — и режет его на карточки по одному-два слова. На каждой
карточке пишет номер и адрес: кому доставить. Карточки раздают
ученикам в разных концах класса, и те передают их друг другу, пока
карточки не доберутся до адресата. Каждую карточку — своим путём.

Адресат собирает сообщение по номерам. Одна карточка потерялась?
Он просит прислать заново только её.

Ровно так работает интернет. Карточки называются **пакетами**,
ученики, которые их передают, — **маршрутизаторами**.

## Разбираемся

### Что происходит, когда ты жмёшь «отправить»

Сообщение доходит до друга за полсекунды, и кажется, что оно просто
перелетело по воздуху. На самом деле путь у него довольно длинный.

```
  твой телефон
       ↓  по воздуху
  вышка или роутер
       ↓  по проводу
  сервер мессенджера
       ↓  по проводу
  вышка рядом с другом
       ↓  по воздуху
  телефон друга
```

*Вышка сотовой связи с высоты*

Беспроводная связь работает только на первом и последнем участке —
от телефона до ближайшей вышки или роутера. Вышки стоят в городе
на крышах домов, а за городом — на высоких мачтах, как эта. От вышки
сообщение бежит дальше **по проводам**.

### Интернет — это провода

Слово «облако» сбивает с толку. Никакого облака нет: есть здания
с тысячами компьютеров и провода между ними.

Прокладывают кабели специальные корабли. В трюме у них огромные
барабаны с тысячами километров кабеля, и судно неделями идёт через

### Что такое сервер

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/data-centr.jpg" width="560" alt="Ряды серверов в дата-центре">

*Ряды серверов в дата-центре*

<sub>Carl Lender · CC BY 2.0 · Wikimedia Commons</sub>

Сервер — это компьютер, который всё время включён и ждёт запросов.
У него нет монитора и клавиатуры, он стоит в стойке вместе с сотнями
таких же и работает годами без выключения. Здание, где стоят тысячи
серверов, называется **дата-центром**.

Серверы сильно греются, поэтому дата-центры тратят огромное количество
электричества на охлаждение. Некоторые компании строят их в холодных
странах или охлаждают морской водой.

Когда ты отправляешь сообщение, оно сначала попадает на сервер
мессенджера, а уже оттуда — другу. Даже если вы сидите за соседними
партами, сообщение может слетать за тысячу километров и вернуться.

### Почему так быстро

Сигнал по кабелю идёт со скоростью, близкой к скорости света.
Москва — Нью-Йорк это примерно 7500 километров, и свет пробегает
их за сотые доли секунды.

К этому добавляется время на оборудовании по пути: каждый узел должен
принять пакет, посмотреть адрес и отправить дальше.

### Сообщение едет кусочками

Как в нашей игре, большой файл не отправляется целиком. Он режется
на **пакеты** — маленькие кусочки с номерами и адресом.

Пакеты едут разными путями и могут прийти вперемешку. На той стороне
их собирают обратно по номерам. Если какой-то потерялся — запрашивают
заново, только его, а не весь файл.

Поэтому видео при плохой связи не пропадает целиком, а тормозит
и рассыпается на квадратики: часть пакетов дошла, часть нет.

## Смотрим, как это работает

### Сколько времени идёт сигнал

````python
расстояние_км = 7500
скорость_км_в_сек = 200000    # свет в оптоволокне идёт медленнее, чем в вакууме

секунд = расстояние_км / скорость_км_в_сек

print("Расстояние:", расстояние_км, "км")
print("Время в пути:", секунд, "секунды")
print("В миллисекундах:", секунд * 1000)
````

Тридцать семь миллисекунд от Москвы до Нью-Йорка. Моргнуть успеваешь
за сто.

### Сколько качается файл

````python
файл_мегабайт = 100
скорость_мегабит = 50

# в одном байте восемь бит, поэтому скорость в мегабайтах делим на 8
скорость_мегабайт = скорость_мегабит / 8
секунд = файл_мегабайт / скорость_мегабайт

print("Скорость в мегабайтах в секунду:", скорость_мегабайт)
print("Файл скачается за секунд:", секунд)
````

Тут есть подвох: скорость интернета указывают **в мегабитах**,
а размер файла — **в мегабайтах**. Разница в восемь раз.

Поэтому «интернет 100 мегабит» качает около 12 мегабайт в секунду.
Провайдеры не обманывают: скорость связи просто принято мерить в битах.

### Пакеты

````python
файл_мегабайт = 10
пакет_килобайт = 64

файл_килобайт = файл_мегабайт * 1024
пакетов = файл_килобайт // пакет_килобайт

print("Файл разрежется на пакетов:", пакетов)
````

Сто шестьдесят кусочков, каждый со своим номером. Потеряется один —
перезапросят только его.
