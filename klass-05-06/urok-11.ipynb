## Разбираемся

### Что такое цифровой след

**Цифровой след** — всё, что остаётся в интернете после тебя.
Одну часть ты оставляешь сам, другую — даже не замечая.

| Оставляешь сам | Оставляешь незаметно |
|---|---|
| посты, фото, видео | что ищешь в поисковике |
| комментарии и лайки | какие видео досматриваешь |
| ник и аватарка | где находишься, если включена геолокация |
| подписки на группы | с какого устройства заходишь |

По незаметной части тебе подбирают рекламу и рекомендации. Посмотрел
три видео про кроссовки — и неделю тебе показывают кроссовки.

### Удалённое не исчезает

Кажется, что удалил пост — и его больше нет. На самом деле:

* **Скриншоты.** Любой, кто успел увидеть, мог сохранить картинку.
  Её уже не удалить.
* **Пересылки.** Сообщение в чате могли переслать дальше, и копии
  живут своей жизнью.
* **Архивы.** Сайты время от времени сохраняют целиком специальные
  службы.

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/arhiv.jpg" width="520" alt="Серверы интернет-архива">

*Серверы интернет-архива*

Например, **Интернет-архив** с 1996 года сохраняет копии сайтов со всего
мира. На его серверах — сотни миллиардов страниц. Можно открыть,
как выглядел любимый сайт десять лет назад. Там же может оказаться
и старая страница, которую автор давно удалил.

### Фотография рассказывает больше, чем видно

Снимок из телефона хранит не только картинку. Внутри файла записаны
дата, время, модель телефона, а если включена геолокация — ещё
и точное место съёмки. Выложил фото из своей комнаты — и по нему можно
узнать, где ты живёшь.

Выдают и детали на самом снимке: вывеска школы за спиной, номер дома
в окне, форма с эмблемой секции, билет с QR-кодом в руке.

### Что не стоит выкладывать никогда

* домашний адрес и номер школы с классом;
* расписание: когда ты один дома, когда уезжаете на дачу;
* фото документов, билетов, банковских карт — даже частично;
* геометки дома и школы;
* пароли и коды, конечно.

### Правило школьной доски

Перед тем как выложить что-то, представь: это распечатают и повесят
на доску объявлений в школе. Увидят учителя, одноклассники, родители
одноклассников. Не стыдно? Тогда можно.

## Как настроить приватность

В любой соцсети и мессенджере есть раздел **«Приватность»** или
**«Конфиденциальность»**. Что стоит там проверить:

1. Кто видит твою страницу: все или только друзья.
2. Кто может писать тебе сообщения.
3. Кто может найти тебя по номеру телефона.
4. Разрешена ли приложению геолокация — и нужна ли она ему вообще.

Сделай это вместе с родителями. Пять минут настройки защищают лучше,
чем годы осторожности.

## Смотрим, как это работает

Посчитаем, сколько следов оставляет обычный школьник.

```python
print("Фото в день:          ", 3)
print("Фото за год:          ", 3 * 365)
print("Поисковых запросов за год:", 20 * 365)
print("Сообщений за год:     ", 100 * 365)
```

Больше тысячи фотографий и десятки тысяч сообщений в год. Поменяй
числа на свои и посмотри, сколько получится у тебя.
