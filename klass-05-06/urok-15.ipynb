## Разбираемся

### Облако — это чужой компьютер

Когда говорят «файл в облаке», он лежит на дисках серверов в большом
**дата-центре**. Мы видели такие здания, когда разбирали, как
сообщение долетает до друга. Компания хранит там твои файлы и отдаёт
их тебе через интернет на любое устройство.

Облачных хранилищ много: Google Диск, Яндекс Диск, iCloud, облако
Mail.ru. Раз у тебя есть Google-аккаунт для этого учебника, у тебя уже
есть и Google Диск на 15 гигабайт — общих для почты, Диска и Фото.

### Синхронизация

Главное удобство облака — **синхронизация**: файл, изменённый на одном
устройстве, сам обновляется на всех остальных. Начал доклад на школьном
компьютере, дописал дома на ноутбуке, показал на телефоне — везде
одна и та же, самая свежая версия.

Так же работают и копии уроков этого учебника: они сохраняются
на твоём Google Диске, в папке *Colab Notebooks*.

### Плюсы и минусы

| Плюсы | Минусы |
|---|---|
| файлы доступны с любого устройства | без интернета до них не добраться |
| не пропадут, если сломался телефон | бесплатного места мало |
| легко поделиться ссылкой | взломали аккаунт — открыты все файлы |
| не надо носить с собой флешку | файлы хранятся у чужой компании |

### Осторожно со ссылками

В облаке файлом можно поделиться по ссылке. Обычно есть выбор:

* **только конкретным людям** — открыть смогут те, кого ты указал;
* **всем, у кого есть ссылка** — удобно, но такую ссылку могут
  переслать кому угодно, и файл увидят незнакомые люди.

И ещё выбирают права: только **просмотр** или **редактирование**.
Давай редактирование только тем, кто вместе с тобой работает над
файлом, иначе кто-нибудь случайно сотрёт половину доклада.

### А флешка?

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/fleshka.jpg" width="360" alt="Флешка на 1 терабайт рядом с монетой">

*Флешка на 1 терабайт рядом с монетой*

Флешки никуда не делись: эта размером чуть больше монеты, а вмещает
терабайт — тысячу гигабайт. Интернет ей не нужен. Зато её легко потерять,
сломать, постирать вместе с джинсами.

### Правило «3-2-1»

Профессионалы хранят важные файлы по правилу трёх цифр:

* **3** копии файла;
* на **2** разных носителях: например, компьютер и флешка;
* **1** копия — в другом месте: например, в облаке.

Тогда файл переживёт и сломанный диск, и потерянную флешку, и даже
пожар. Для домашних фото достаточно простого варианта: телефон плюс
облако плюс раз в полгода копия на компьютер или флешку.

## Смотрим, как это работает

Сколько фотографий поместится в бесплатные 15 гигабайт?

```python
print("Мегабайт в 15 гигабайтах:", 15 * 1024)
print("Фото по 4 мегабайта:     ", 15 * 1024 // 4)
print("Фото по 3 в день — дней: ", 15 * 1024 // 4 // 3)
```

Три с половиной года, если снимать по три фото в день. А видео
съедят место гораздо быстрее: минута — больше сотни мегабайт.
