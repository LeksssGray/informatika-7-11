## Разбираемся

### Что такое алгоритм

> **Алгоритм** — точная инструкция, которую можно выполнить
> по шагам, не догадываясь, что имелось в виду.

Рецепт блинов — алгоритм. Инструкция к конструктору — алгоритм.
Правило деления в столбик — алгоритм. А «сделай уборку» — нет:
непонятно, что именно и в каком порядке.

### Какими должны быть шаги

| Требование | Что значит | Плохой пример |
|---|---|---|
| **по шагам** | действия по одному | «сделай всё сразу» |
| **однозначно** | понятно без догадок | «иди туда, не знаю куда» |
| **заканчивается** | шагов конечное число | «дели, пока не надоест» |
| **годится для разных данных** | не только для одного случая | «сложи именно 2 и 3» |

Самое коварное требование — **однозначность**. Человек догадается,
а компьютер нет.

### Пример: сделать чай

Плохо:

1. Налей воду.
2. Вскипяти.
3. Сделай чай.

Третий шаг — не шаг, а пересказ всей задачи. Хорошо:

1. Налей в чайник воду до отметки.
2. Включи чайник.
3. Дождись, пока выключится.
4. Положи пакетик в кружку.
5. Налей воду в кружку.
6. Подожди три минуты.
7. Вынь пакетик.

Теперь это выполнит кто угодно — и получится чай, а не кипяток
с пакетиком на дне.

### Порядок шагов важен

Поменяй в рецепте два шага местами — и результат другой. «Налей
воду, положи пакетик» и «положи пакетик, налей воду» — разница
небольшая. А вот «надень носки, надень ботинки» наоборот
уже не работает.

Компьютер выполняет шаги **строго по порядку**, сверху вниз.

### Исполнитель

> **Исполнитель** — тот, кто выполняет алгоритм.

У каждого исполнителя свой набор команд, которые он понимает.
Чайник понимает «включись». Робот-пылесос — «едь вперёд»,
«повернись». Python — `print`, `+`, `*` и другие.

Написать алгоритм — значит выразить задачу **через команды,
которые исполнитель знает**. Больше ничего использовать нельзя.

## Смотрим, как это работает

### Алгоритм по шагам

```python
print("Алгоритм: сделать чай")
print("1. Налей воду до отметки")
print("2. Включи чайник")
print("3. Дождись, пока выключится")
print("4. Положи пакетик в кружку")
print("5. Налей воду")
print("6. Подожди три минуты")
print("7. Вынь пакетик")
print()
print("Шагов:", 7)
```

### Компьютер делает строго по порядку

```python
print("Первое действие")
print("Второе действие")
print("Третье действие")
```

Ровно в том порядке, в котором написано. Никакой самодеятельности.

### Порядок меняет результат

```python
деньги = 100

деньги = деньги + 50
деньги = деньги * 2

print("Сначала прибавили, потом умножили:", деньги)

деньги = 100

деньги = деньги * 2
деньги = деньги + 50

print("Сначала умножили, потом прибавили:", деньги)
```

Действия те же, числа те же — итог разный. Порядок шагов
и есть алгоритм.

### Алгоритм для любых данных

```python
сторона = 5

площадь = сторона * сторона

print("Сторона:", сторона, "— площадь:", площадь)

сторона = 12

площадь = сторона * сторона

print("Сторона:", сторона, "— площадь:", площадь)
```

Один и тот же алгоритм сработал для разных чисел. Если бы
мы написали `площадь = 25`, он годился бы только для пятёрки.

### Алгоритм, который не заканчивается

```python
print("Алгоритм: «Вычитай из числа 2, пока не получится 1»")
print()

число = 10
print("10 →", число - 2, "→", число - 4, "→", число - 6, "→", число - 8, "→ …")
print()
print("Из чётного числа вычитаем чётное — всегда получается чётное.")
print("Единицы не будет никогда: алгоритм не закончится.")
```

### Считаем шаги

```python
шагов_чай = 7
шагов_бутерброд = 4

print("Шагов в алгоритме чая:", шагов_чай)
print("Шагов в алгоритме бутерброда:", шагов_бутерброд)
print("Всего:", шагов_чай + шагов_бутерброд)
```
