## Разбираемся

### Рассказ за три минуты

Три части, по минуте на каждую:

1. **Что получилось.** Название и одно предложение: «Викторина
   про космос на 8 вопросов, которая ставит оценку».
2. **Как это работает.** Запусти и покажи вживую. Покажи одну
   интересную строчку кода и объясни её.
3. **Что было трудно.** Где застрял и как выбрался. Что добавил бы,
   будь ещё неделя.

Честно сказать «вот это я не успел» — нормально. Плохо — делать вид,
что всё работает, когда нет.

### Как отвечать на вопросы

* выслушай вопрос до конца;
* не знаешь — так и скажи: «Не проверял, но думаю, что…»;
* вопрос — не нападение: тебе помогают увидеть, что можно улучшить.

### Как оценивать чужой проект

Таблица та же, что в уроке 32:

| Критерий | 0 | 1 | 2 |
|---|---|---|---|
| работает | не запускается | с ошибками | без ошибок |
| задумка выполнена | меньше половины | почти всё | всё |
| код понятный | имена a, b, x | частично | все имена понятные |
| своё | скопировано | чуть изменено | много своего |
| рассказ | не рассказал | без показа | рассказал и показал |

А устно — **два плюса и одно пожелание**: две вещи, которые
понравились, и одна, которую стоит улучшить. Пожелание — конкретное:
не «сделай лучше», а «добавь подсказку после неверного ответа».

## Смотрим, как это работает

### Считаем баллы проекта

Баллы по пяти критериям — в списке. Сумма и оценка — программой.

```python
критерии = ["работает", "задумка", "понятный код", "своё", "рассказ"]
баллы = [2, 2, 1, 2, 2]

for i in range(len(критерии)):
    print(критерии[i], "—", баллы[i])

всего = sum(баллы)
print("Всего:", всего, "из 10")
```

### Оценка по сумме

9–10 баллов — «5», 7–8 — «4», 5–6 — «3», меньше — «нужно доделать».

```python
if всего >= 9:
    оценка = "5"
elif всего >= 7:
    оценка = "4"
elif всего >= 5:
    оценка = "3"
else:
    оценка = "нужно доделать"

print("Оценка:", оценка)
```

### Оценки от всего класса

Проект оценивали несколько одноклассников. Итог — среднее их сумм.

```python
суммы_от_класса = [9, 8, 10, 9, 7]
среднее = sum(суммы_от_класса) / len(суммы_от_класса)
print("Средний балл от класса:", среднее)
```
