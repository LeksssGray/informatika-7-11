Раньше в каждой библиотеке стоял **каталог** — шкаф с узкими ящиками.

Поисковик в интернете — такой же каталог, только карточек в нём
сотни миллиардов, и заполняет их не библиотекарь, а программа.

## Разбираемся

### Как работает поисковик

**1. Робот обходит сайты.** Программа, которую называют **поисковым
роботом**, без остановки ходит по интернету: открывает страницу,
запоминает текст, переходит по всем ссылкам на ней, открывает
следующие — и так бесконечно.

**2. Составляет указатель.** Для каждого слова поисковик записывает,
на каких страницах оно встречается. Как указатель в конце толстой
книги: «вулкан — страницы 12, 47, 103».

**3. Выбирает лучшее.** Когда ты вводишь запрос, поисковик за доли
секунды находит страницы со всеми словами и решает, какие показать
первыми: где слов больше, на какие страницы чаще ссылаются другие,
насколько сайт надёжный.

### Как это было: сервер на пробковых досках

Этот стеллаж теперь стоит в музее. Российский поисковик «Яндекс»
появился годом раньше, в 1997-м.

### Как задать вопрос, чтобы тебя поняли

Поисковик не читает мысли. Он ищет слова. Поэтому писать надо
не «подскажите пожалуйста как сделать чтобы…», а главные слова:
«вулкан извержение как происходит».

Есть и особые знаки. Они работают и в Яндексе, и в Google.

| Что написать | Что будет | Пример |
|---|---|---|
| слова в кавычках | ищется точная фраза, слово в слово | `"где раки зимуют"` |
| минус перед словом | страницы с этим словом отбросятся | `пирог рецепт -вишня` |
| `site:` и адрес сайта | поиск только на одном сайте | `site:ru.wikipedia.org вулканы` |
| `filetype:pdf` | только файлы этого типа | `таблица умножения filetype:pdf` |

### Первая ссылка — не всегда лучшая

* **Реклама.** Первые строчки часто платные, рядом с ними мелко
  написано «Реклама». Их показывают не потому, что они лучшие, а потому,
  что за них заплатили.
* **Хитрые сайты.** Некоторые сайты специально набивают страницы
  популярными словами, чтобы подняться повыше, а толкового ответа
  на них нет.
* **Старое.** Ответ мог устареть: расписание прошлого года, цены
  позапрошлого.

Хорошая привычка — открыть две-три ссылки и сравнить. Если источники
говорят одно и то же, скорее всего, это правда.
