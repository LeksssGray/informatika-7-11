Соберём страницу о себе. Не в тетради — прямо здесь, и она будет
выглядеть как настоящая веб-страница: с заголовками, списками,
выделенным текстом.

### Разметка — это просто

Текстовые ячейки понимают специальные значки. Поставил значок —
текст оформился.

| Что написать | Что получится |
|---|---|
| `# Заголовок` | большой заголовок |
| `## Поменьше` | заголовок поменьше |
| `**жирный**` | **жирный** |
| `*курсив*` | *курсив* |
| `* пункт` | пункт списка |
| `1. пункт` | нумерованный список |
| `> цитата` | выделенный блок |

Это называется **Markdown** — способ размечать текст, не отвлекаясь
на кнопки. Им пользуются программисты, а ещё он работает
в мессенджерах: звёздочки вокруг слова во многих из них делают
текст жирным.

### Как сделать текстовую ячейку

Наведи мышку между ячейками — появится кнопка **+ Текст**. Нажми,
и появится новая текстовая ячейка.

Пока пишешь, слева видно исходный текст, справа — как он будет
выглядеть. Закончил — нажми Shift + Enter, и текст оформится.

Чтобы вернуться к правке, дважды щёлкни по ячейке.

### Сколько символов получилось

Заодно посчитаем, сколько весит такая страница.

```python
символов = 420          # примерно столько в тексте выше
байт_на_символ = 2      # русские буквы занимают по два байта

вес_байт = символов * байт_на_символ

print("Символов:", символов)
print("Вес в байтах:", вес_байт)
print("Это меньше килобайта:", вес_байт < 1024)
```

Меньше килобайта. Одна фотография весит в несколько тысяч раз больше,
чем целая страница текста. Текст — самое лёгкое, что бывает
в компьютере.
