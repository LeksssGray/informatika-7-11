Соберём страницу о себе. Не в тетради — прямо здесь, и она будет
выглядеть как настоящая веб-страница: с заголовками, списками,
выделенным текстом.

### Разметка — это просто

Текстовые ячейки понимают специальные значки. Поставил значок —
текст оформился.

| Что написать | Что получится |
|---|---|
| `# Заголовок` | большой заголовок |
| `## Поменьше` | заголовок поменьше |
| `**жирный**` | **жирный** |
| `*курсив*` | *курсив* |
| `* пункт` | пункт списка |
| `1. пункт` | нумерованный список |
| `> цитата` | выделенный блок |

Это называется **Markdown** — способ размечать текст, не отвлекаясь
на кнопки. Им пользуются программисты, а ещё он работает
в мессенджерах: звёздочки вокруг слова во многих из них делают
текст жирным.

### Как сделать текстовую ячейку

Наведи мышку между ячейками — появится кнопка **+ Текст**. Нажми,
и появится новая текстовая ячейка.

Пока пишешь, слева видно исходный текст, справа — как он будет
выглядеть. Закончил — нажми Shift + Enter, и текст оформится.

Чтобы вернуться к правке, дважды щёлкни по ячейке.

## Делаем свою страницу

### Что должно быть

Обязательно:

* заголовок с твоим именем;
* подзаголовок «Обо мне» и пара предложений;
* список из трёх и более занятий;
* нумерованный список из трёх фактов;
* хотя бы одно жирное и одно курсивное слово.

По желанию: цитата, любимая фраза, что угодно ещё.

### Как делать

1. Наведи мышку под эту ячейку и нажми **+ Текст**.
2. Пиши свою страницу, подглядывая в образец выше.
3. Нажми Shift + Enter, посмотри, что получилось.
4. Не понравилось — дважды щёлкни и правь дальше.

Не старайся сделать красиво с первого раза. Сделай как получится,
потом поправишь.

### Как сделать страницу, которую приятно читать

* **Заголовки короткие.** «Чем занимаюсь», а не «Здесь я расскажу
  о том, чем я занимаюсь в свободное время».
* **Жирным — одно-два слова.** Если выделить всё, не выделено ничего.
* **Списки — для перечислений.** Три увлечения — список. Рассказ
  о лете — обычный текст.
* **Пустая строка между абзацами.** Иначе всё сольётся в одну кашу.
* **Никаких личных данных.** Адрес, номер школы и телефон на странице
  не нужны. Имя и увлечения — пожалуйста.
