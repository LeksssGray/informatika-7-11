## Разбираемся

### Три шага любого робота

Как бы ни был устроен робот, он повторяет одно и то же:

1. **датчик** — узнать, что происходит вокруг;
2. **решение** — по правилу выбрать, что делать;
3. **действие** — сделать: открыть, поехать, включить.

И снова к первому шагу. Так по кругу, тысячи раз в секунду.

| Робот | Датчик | Решение | Действие |
|---|---|---|---|
| автоматическая дверь | датчик движения | кто-то подошёл? | открыть |
| обогреватель | термометр | холоднее 20 °C? | включить нагрев |
| робот-пылесос | бампер, датчик заряда | упёрся? заряд кончается? | повернуть, ехать на базу |
| светофор | часы | прошло 30 секунд? | сменить цвет |

### Решение — это `if`

Шаг «решение» в программе — это ветвление, которое мы уже знаем:

```python
if температура < 20:
    print("Включаю обогрев")
else:
    print("Обогрев выключен")
```

### Повторение — это цикл

Робот не делает дело один раз и не выключается. Светофор меняет
цвета весь день, пылесос ездит, пока есть заряд. Повторение — это
цикл `for` или `while`.

### Робот не думает

Пылесос «не знает», что такое комната. Он знает правила: упёрся —
повернись, заряд меньше 15 % — езжай на базу. Если правило не
предусмотрело ситуацию (носок на полу, провод), робот сделает
глупость. Поэтому программисты так долго придумывают правила
и проверяют их на странных случаях.

## Смотрим, как это работает

### Обогреватель

````python
температура = 18

if температура < 20:
    print("Холодно:", температура, "градусов. Включаю обогрев")
else:
    print("Тепло:", температура, "градусов. Обогрев выключен")
````

Поменяй температуру на 22 и запусти ещё раз.

### Светофор

Цвета лежат в списке, цикл перебирает их по очереди. Внешний цикл
повторяет весь круг три раза.

````python
цвета = ["зелёный", "жёлтый", "красный"]

for круг in range(3):
    for цвет in цвета:
        print(цвет)
    print("— круг закончился —")
````

### Лифт едет на седьмой этаж

````python
for этаж in range(1, 8):
    print("Этаж", этаж)

print("Двери открываются")
````

### Робот-пылесос

Уборка одной комнаты тратит 30 % заряда. Пылесос убирает, пока заряда
хватает, а потом едет на базу.

````python
заряд = 100
комнат = 0

while заряд >= 30:
    заряд = заряд - 30
    комнат = комнат + 1
    print("Убрал комнату", комнат, "— заряд", заряд)

print("Еду на зарядку")
````

Датчик — переменная `заряд`, решение — условие `while`, действие —
уборка и поездка на базу. Все три шага в шести строчках.
