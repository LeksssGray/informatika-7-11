### Правило «если — то»

Почти любой робот работает по правилам вида «если… то…». Пылесос: **если** на полу мусор,
**то** включить щётку. Светофор: **если** горит зелёный, **то** можно идти.

```python
мусор_на_полу = True
if мусор_на_полу:
    print("Пылесос включается")
else:
    print("Пылесос отдыхает")
```

Поменяй значение `мусор_на_полу` на `False` и запусти ещё раз. Программа выберет другую ветку —
это и есть правило «если — то».
