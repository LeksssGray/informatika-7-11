## Разбираемся

### Что такое файл

**Файл** — это кусок информации на диске, у которого есть имя:
фотография, песня, доклад, игра. Компьютер хранит миллионы файлов
и отличает их друг от друга только по именам и по тому, в какой
папке они лежат.

### Имя и расширение

Имя файла состоит из двух частей, разделённых точкой:

```
  доклад_вулканы.docx
  └──── имя ────┘└расширение┘
```

**Расширение** подсказывает компьютеру, что внутри файла и какой
программой его открыть.

| Расширение | Что внутри | Чем открыть |
|---|---|---|
| `.jpg`, `.png` | картинка | просмотрщик фото |
| `.mp3` | музыка | проигрыватель |
| `.mp4` | видео | видеоплеер |
| `.docx` | текстовый документ | Word, Google Документы |
| `.pdf` | документ «для печати» | браузер, программа для PDF |
| `.txt` | простой текст | Блокнот |
| `.zip` | архив: много файлов в одном | архиватор |
| `.exe` | программа для Windows | запускается сама |

> **Осторожно с `.exe`.** Такой файл — это программа, и она выполняет
> всё, что в неё заложили. Файл `.exe` из письма или из «бесплатной
> игры» с неизвестного сайта — самый частый способ заразить компьютер
> вирусом. Не запускай, не спросив взрослых.

### Папки и путь

Файлы раскладывают по **папкам**, папки — по другим папкам. Получается
дерево:

```
  Документы
  ├── Школа
  │   ├── Математика
  │   │   └── контрольная_вопросы.docx
  │   └── История
  │       └── 2025-10-03_доклад_Пётр_I.pdf
  └── Кружки
      └── расписание.pdf
```

Адрес файла в этом дереве называется **путём**. Папки в нём
разделяют косой чертой: `Документы/Школа/История/2025-10-03_доклад_Пётр_I.pdf`.
По пути компьютер находит файл, как почтальон — дом по адресу.

### Как называть файлы

| Плохо | Хорошо |
|---|---|
| `Новый документ (7).docx` | `2025-09-12_доклад_вулканы.docx` |
| `IMG_4821.jpg` | `2025-06-15_дача_закат.jpg` |
| `финал_финал2_точно_последний.docx` | `сочинение_версия_3.docx` |

Правила простые:

1. **Имя объясняет, что внутри.** Через месяц ты не вспомнишь, что было
   в `документ (3)`.
2. **Дата в начале, в виде год-месяц-день:** `2025-09-12`. Тогда файлы
   сами встанут по порядку при сортировке по имени.
3. **Вместо пробелов — подчёркивание.** Не все программы любят пробелы.
4. **Версии нумеруй**, а не пиши «финал».

Посмотри на кнопку «Сохранить» в любой программе. Там нарисован
квадратик с металлической шторкой. Это **дискета** — так раньше
называли переносной диск.

Первые дискеты 1970-х были размером с большую книгу, 8 дюймов,
то есть 20 сантиметров. Потом появились поменьше, 5,25 дюйма. Самые
известные — квадратные, 3,5 дюйма: на них помещалось 1,44 мегабайта.
Одна фотография с твоего телефона на такую дискету уже не влезет.

На фото — дисководы для всех трёх размеров. Дискетами давно никто
не пользуется, а значок остался, и многие ребята даже не знают, что
на нём нарисовано.

## Смотрим, как это работает

Имя файла — это просто текст. Его можно собрать из кусочков знаком `+`,
как мы склеивали строки.

````python
дата = "2025-09-12"
тема = "доклад_вулканы"
расширение = ".docx"

print(дата + "_" + тема + расширение)
````

Поменяй тему и дату — получится новое аккуратное имя. Путь
собирается так же, только куски разделяет косая черта.

````python
print("Документы" + "/" + "Школа" + "/" + "История")
````

### Любопытно

В Windows есть имена, которые файлу дать нельзя, например `CON`,
`PRN` и `AUX`. Попробуй — система откажется. Это наследство операционных

и другие устройства. Правило оставили, чтобы не сломались старые
программы.
