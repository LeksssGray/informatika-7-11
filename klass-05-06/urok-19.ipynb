## Разбираемся

### Повтор значка

Главный приём урока ты уже видел:

```python
print("█" * 10)
```

Звёздочка означает «повтори». Получится ряд из десяти значков.
Поменяешь число — поменяется длина, и ничего больше переписывать
не нужно.

### Склейка

Ряды можно составлять из кусков разной длины:

```python
print("█" * 4 + "░" * 4)
```

Плюс склеивает. Сначала четыре плотных значка, потом четыре
светлых.

### Из чего рисовать

| Значок | Как выглядит |
|---|---|
| `█` | полностью закрашено |
| `▓` | почти закрашено |
| `▒` | наполовину |
| `░` | чуть-чуть |
| пробел | пусто |

Эти значки называют блочными. Ими рисовали задолго до того,
как экраны научились показывать картинки.

### Переменная делает рисунок гибким

Если размер вписан числом, менять придётся в каждой строчке.
А если он в переменной — только в одном месте:

```python
ширина = 8
print("█" * ширина)
print("░" * ширина)
```

Поменял `ширина = 20` — весь рисунок стал шире. Это главная
причина, по которой числа прячут в переменные.

### Как рисовать рамку

Рамка — это три разных ряда:

```
████████    верх: всё закрашено
█      █    середина: края закрашены, внутри пусто
████████    низ: снова всё
```

Середина собирается склейкой: значок, пробелы, значок.

### Как рисовать градиент

Градиент — плавный переход. Берут значки от плотного к светлому
и печатают подряд:

```python
print("█" * 5 + "▓" * 5 + "▒" * 5 + "░" * 5)
```

## Смотрим, как это работает

### Полоса и её длина

````python
print("█" * 5)
print("█" * 10)
print("█" * 20)
````

### Флаг в три полосы

````python
ширина = 14

print("█" * ширина, "красная")
print("░" * ширина, "белая")
print("▓" * ширина, "синяя")
````

Поменяй `ширина` на 30 — флаг станет шире, а код не изменится.

### Рамка

````python
сторона = 10

print("█" * сторона)
print("█" + " " * (сторона - 2) + "█")
print("█" + " " * (сторона - 2) + "█")
print("█" * сторона)
````

Внутри рамки пробелов на два меньше: два места заняли края.
Отсюда `сторона - 2`.

### Шахматная доска

````python
клетка = "██"
пусто = "  "

чёрный_ряд = (клетка + пусто) * 4
белый_ряд = (пусто + клетка) * 4

print(чёрный_ряд)
print(белый_ряд)
print(чёрный_ряд)
print(белый_ряд)
print(чёрный_ряд)
print(белый_ряд)
print(чёрный_ряд)
print(белый_ряд)
````

Хитрость в том, что повторяется не одна клетка, а **пара**:
«закрашено плюс пусто». Четыре пары — восемь клеток, как
на настоящей доске.

### Градиент

````python
print("█" * 6 + "▓" * 6 + "▒" * 6 + "░" * 6)
print("░" * 6 + "▒" * 6 + "▓" * 6 + "█" * 6)
````

### Лестница

````python
print("█" * 1)
print("█" * 2)
print("█" * 3)
print("█" * 4)
print("█" * 5)
````

Ступенек пять, значков всего 1 + 2 + 3 + 4 + 5 = 15.

Скоро мы научимся делать такое одной строчкой — на уроке
про повторы.

### Считаем, сколько значков в рисунке

````python
сторона = 10

в_рамке = сторона + сторона + (сторона - 2) + (сторона - 2)

print("Рамка", сторона, "на", сторона)
print("Закрашенных значков:", в_рамке)
print("А всего клеток:", сторона * сторона)
````

### Любопытно

Рисовать значками начали не от бедности, а от восторга: первые
принтеры умели печатать только буквы, и программисты выкладывали
из них портреты, календари и даже плакаты в полный рост.

Такие картинки называют ASCII-графикой. Их до сих пор рисуют —
например, в заставках программ, которые работают без окон, одним
текстом. А в некоторых играх ASCII-графика — нарочный стиль:
вся карта подземелья из точек, решёток и букв.
