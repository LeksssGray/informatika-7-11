## Разбираемся

### Кто есть кто

* **Программист** пишет код, который делает сайт или приложение работающим.
* **Дизайнер** продумывает, как всё выглядит и как этим удобно пользоваться.
* **Тестировщик** ищет ошибки: нажимает всё, что можно, и проверяет, что будет.
* **Аналитик** разбирается, что нужно людям, и переводит это в понятные задачи.

Все они работают вместе. Без дизайнера программа может быть неудобной, без тестировщика — полной ошибок.
