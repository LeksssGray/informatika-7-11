## Что делаем

Над каждым приложением работает команда: кто-то пишет код, кто-то
рисует экраны, кто-то ищет ошибки. Сегодня примерим на себя четыре
профессии и в каждой решим настоящую рабочую задачу.

## Разбираемся

### Команда одного приложения

| Профессия | Чем занимается | Главный вопрос |
|---|---|---|
| **программист** | пишет и исправляет код | как сделать, чтобы это работало? |
| **дизайнер** | придумывает экраны, кнопки, цвета | удобно ли этим пользоваться? |
| **тестировщик** | ищет ошибки до того, как их найдут люди | где это сломается? |
| **аналитик** | разбирается в данных и цифрах | что на самом деле происходит? |
| **специалист по безопасности** | защищает данные от взлома | как это могут украсть? |

Ни одна из них не работает одна. Дизайнер рисует экран, программист
его оживляет, тестировщик ломает, аналитик смотрит, понравилось ли
людям, — и всё по кругу.

### Тестировщик проверяет границы

Ошибки чаще всего прячутся **на границе**. Если фильм «12+», тестировщик
не станет проверять только возраст 30. Он проверит 11, 12 и 13 — ровно
там, где программа должна переключиться с «нельзя» на «можно».

### Аналитик не верит на слово

«Приложение всем нравится!» — говорит команда. Аналитик открывает
отзывы и считает: средняя оценка, сколько единиц, на что жалуются.
Цифры иногда говорят совсем другое.

### С чего начинают

Почти все начинали так же, как ты сейчас: с маленьких программ,
игр, рисунков. Важнее всего не знать всё, а уметь **разбираться** —
читать, пробовать, искать ошибку.

## Смотрим, как это работает

### Тестировщик: кинотеатр «12+»

Программа решает, пускать ли в зал. Тестировщик проверяет пограничные
возрасты:

```python
for возраст in [11, 12, 13]:
    if возраст > 12:
        print(возраст, "— можно")
    else:
        print(возраст, "— нельзя")
```

Двенадцатилетнего не пустили, а фильм «12+» — значит, с 12 можно.
Ошибка в одном знаке: нужно `>=`, а не `>`. Если бы проверили только
возраст 30, баг ушёл бы к зрителям.

### Аналитик: оценки приложения

```python
оценки = [5, 5, 4, 1, 5, 1, 5, 4, 1, 5]

средняя = sum(оценки) / len(оценки)
единиц = 0
for оценка in оценки:
    if оценка == 1:
        единиц = единиц + 1

print("Средняя оценка:", средняя)
print("Единиц:", единиц, "из", len(оценки))
```

Средняя 3,6 — вроде неплохо. Но каждый третий поставил единицу!
Значит, у части людей что-то серьёзно сломалось. Аналитик пойдёт
читать именно эти отзывы.

### Специалист по безопасности: длина пароля

```python
пароль = "солнце"

if len(пароль) < 8:
    print("Слабый пароль:", len(пароль), "символов. Нужно хотя бы 8")
else:
    print("Длина пароля нормальная")
```

### 2. Найди границу

В магазине скидка, если покупка **от 1000 рублей**. Программа проверяет
три покупки. Тестировщик заметил, что покупатель с 1000 рублями скидку
не получил. Исправь и посчитай, сколько покупок получат скидку.

```python
покупки = [999, 1000, 1500]
со_скидкой = 0

for сумма in покупки:
    if сумма > 1000:
        со_скидкой = со_скидкой + 1

print(со_скидкой)
```
