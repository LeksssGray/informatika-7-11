## Разбираемся

### Как подбирают пароли

У взломщика три главных способа.

**Перебор.** Компьютер пробует все варианты подряд: `aaaa`, `aaab`,
`aaac`… Хорошая видеокарта проверяет миллиарды вариантов в секунду.

**Словарь.** Большинство людей не придумывают пароли, а берут
привычное: `123456`, `qwerty`, `password`, своё имя с годом рождения.
Взломщики собирают списки таких паролей и начинают с них. Пароль
из этого списка подбирается за долю секунды, какой бы длинный он
ни был.

**Утечки.** Сайты иногда взламывают и крадут базу паролей. Если у тебя
один пароль на всё, то после утечки на каком-нибудь игровом форуме
откроются и почта, и соцсеть.

### Почему сайт тебя не блокирует, а взломщика — да

Если вводить пароль на сайте, после пяти-десяти ошибок он попросит
подождать или заблокирует вход. Перебирать так медленно и бесполезно.

Опасно другое: когда базу украли, взломщик перебирает у себя
на компьютере, и никто его не останавливает. Вот тогда и считают
миллиарды вариантов в секунду.

### Сколько вариантов у пароля

Правило то же, что с замком: число символов, из которых можно
выбирать, умножаем само на себя столько раз, какова длина пароля.

| Пароль | Из чего | Вариантов | Подбор при миллиарде в секунду |
|---|---|---|---|
| 4 цифры | 10 цифр | 10 000 | мгновенно |
| 6 маленьких букв | 26 букв | 309 миллионов | меньше секунды |
| 8 букв и цифр | 62 символа | 218 триллионов | около двух с половиной дней |
| 12 букв и цифр | 62 символа | число с 22 цифрами | сто тысяч лет |
| 5 случайных слов | 7776 слов | число с 20 цифрами | 900 лет |

Главный вывод: **длина важнее хитрых значков**. Каждый лишний символ
умножает число вариантов в десятки раз.

Фраза из пяти **случайных** слов — например, `кит лампа ветер облако
пятница` — и длинная, и легко запоминается. Слова должны быть именно
случайными: «я люблю свою маму» взломщик найдёт в словаре фраз.

## Смотрим, как это работает

Посчитаем, сколько вариантов у разных паролей. Знак `**` — это
возведение в степень: `10 ** 4` значит 10 × 10 × 10 × 10.

```python
print("Замок из 3 цифр:     ", 10 ** 3)
print("ПИН-код из 4 цифр:   ", 10 ** 4)
print("6 маленьких букв:    ", 26 ** 6)
print("8 букв и цифр:       ", 62 ** 8)
print("5 случайных слов:    ", 7776 ** 5)
```

А теперь — сколько секунд уйдёт на перебор, если проверять миллиард
вариантов в секунду.

```python
в_секунду = 1000000000

print("8 букв и цифр, секунд:", 62 ** 8 // в_секунду)
print("то же в днях:         ", 62 ** 8 // в_секунду // 86400)
print("5 слов, лет:          ", 7776 ** 5 // в_секунду // 86400 // 365)
```

Восемь символов держатся два с половиной дня, пять случайных слов —
девятьсот лет. Поменяй длину пароля в ячейке и посмотри, как быстро
растут числа.

## Как защитить свои аккаунты

1. **Длинный пароль**: фраза из пяти случайных слов или от двенадцати
   символов вперемешку.
2. **Свой пароль для каждого важного сайта**. Хотя бы для почты — точно
   отдельный: через почту восстанавливают все остальные пароли.
3. **Двухфакторная защита.** Включи в настройках вход с кодом
   из SMS или приложения. Тогда даже с украденным паролем войти
   не получится: нужен ещё твой телефон.
4. **Менеджер паролей** — программа, которая помнит пароли за тебя.
   Помнить надо только один, главный.
5. **Не записывай пароль** на листочке у компьютера и в заметках
   телефона.

> **Код из SMS не сообщай никому.** Ни «сотруднику банка», ни «службе
> поддержки игры», ни другу. Настоящая поддержка его никогда
> не спрашивает. Если просят — это взлом.
