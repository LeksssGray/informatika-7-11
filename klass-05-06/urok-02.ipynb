## Начнём с игры: робот-бутерброд

Эту игру лучше провести всем классом, до того как открывать компьютер.

Один человек — робот. Он стоит у стола, где лежат хлеб, масло и нож
(можно понарошку). Робот делает **ровно то, что ему говорят**, и ничего
сверх этого. Остальные по очереди дают ему команды, чтобы получился
бутерброд.

Скажут «намажь масло на хлеб» — робот положит пачку масла сверху
на батон. Ведь его не просили открыть пачку, взять нож и отрезать
ломтик. Скажут «возьми нож» — возьмёт за лезвие, если не уточнили,
за какой конец.

Через пару минут становится понятно главное: объяснить задачу тому,
кто совсем ничего не понимает, очень трудно. А компьютер именно такой
робот.

## Разбираемся

### Компьютер очень быстрый и очень тупой

Обычный компьютер выполняет **несколько миллиардов действий в секунду**.
Если бы ты складывал числа со скоростью одно в секунду и не спал,
то работу, которую он делает за секунду, ты бы заканчивал лет через сто.

А теперь вторая половина. Компьютер не понимает **ничего**. Ни одного
слова. Он не знает, что такое «дом», «мама» или «завтра». Он умеет
только выполнять инструкции — точно и буквально.

### Что значит «буквально»

Представь, что ты просишь друга: «Сходи в магазин, купи хлеб.
Если будут яйца — возьми десяток».

Друг возвращается с десятью батонами хлеба.

Смешно, но компьютер поступил бы именно так. Он выполнил инструкцию
**буквально**: «если будут яйца — возьми десяток» — яйца были,
значит десяток. Десяток чего? Того, о чём шла речь. Хлеба.

Человек догадался бы по смыслу. Компьютер смысла не видит.

### Что у него получается, а что нет

| Компьютер делает отлично | Компьютеру не даётся |
|---|---|
| считать быстро и без ошибок | понять, что ты имел в виду |
| помнить огромное количество всего | догадаться, что инструкция неправильная |
| повторять одно и то же миллион раз | придумать, зачем вообще это делать |
| искать нужное среди миллиардов записей | отличить шутку от серьёзной просьбы |

Отсюда главное:

* **он почти никогда не ошибается сам** — ошибается тот, кто написал
  инструкцию;
* **он не догадается**, что ты имел в виду, если написал не то;
* **он не устанет и не заскучает** — миллион раз подряд сделает
  одинаково.

Программист — это человек, который умеет объяснить задачу так, чтобы
её понял тот, кто вообще ничего не понимает. После робота-бутерброда
ты знаешь, каково это.

## Как это было: первые компьютеры

### ENIAC — компьютер размером с зал

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/eniac.jpg" width="620" alt="Программисты ENIAC Глен Бек и Бетти Снайдер за работой">

*Программисты ENIAC Глен Бек и Бетти Снайдер за работой*

<sub>Армия США · общественное достояние · Wikimedia Commons</sub>

Один из первых электронных компьютеров, ENIAC, заработал в США в 1946 году.
Он весил около 27 тонн, занимал целый зал и был собран почти
из 18 тысяч радиоламп — стеклянных колбочек размером с палец.

Считал он около 5000 сложений в секунду. Для того времени — чудо:
расчёт, на который у человека с арифмометром уходили сутки, ENIAC
делал за полминуты.

Клавиатуры и экрана у него не было. Чтобы дать ему новую задачу,
люди часами перетыкали провода и переключали рычажки. Первыми
программистами ENIAC были шесть девушек-математиков: они разбирались
в устройстве машины лучше многих инженеров.

### Самый знаменитый «жучок»

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/bag.jpg" width="520" alt="Страница журнала 1947 года с приклеенным мотыльком">

*Страница журнала 1947 года с приклеенным мотыльком*

<sub>ВМС США · общественное достояние · Wikimedia Commons</sub>

В 1947 году в Гарварде сломался другой компьютер, Mark II. Инженеры
полезли внутрь и нашли застрявшего между контактами мотылька. Его
аккуратно вклеили в рабочий журнал с подписью: «Первый настоящий случай,
когда найден жучок». Эта страница хранится в музее до сих пор.

Слово «жучок» — по-английски *bug* — инженеры говорили про неполадки
и раньше. Но после этой истории оно прилипло к компьютерам навсегда.
Ошибку в программе и сегодня называют **багом**, а её исправление —
**отладкой**.

## Знаменитые ошибки

Компьютер сделает всё точно. Но если человек дал неверную инструкцию,
точность только навредит.

### Марсианский аппарат, 1999 год

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/mars-orbiter.jpg" width="520" alt="Mars Climate Orbiter — так его рисовали художники NASA">

*Mars Climate Orbiter — так его рисовали художники NASA*

<sub>NASA/JPL · общественное достояние · Wikimedia Commons</sub>

Аппарат Mars Climate Orbiter летел к Марсу почти год. Одна команда
инженеров считала силу в метрических единицах, другая — в американских,
в фунтах. Никто не сверил. Аппарат подошёл к Марсу слишком низко
и сгорел в атмосфере.

Компьютер всё посчитал верно — по тем числам, что ему дали.

### Ракета Ariane 5, 1996 год

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/ariane.jpg" width="520" alt="Ракета Ariane 5 на старте">

*Ракета Ariane 5 на старте*

<sub>Bill Ingalls, NASA · общественное достояние · Wikimedia Commons</sub>

Во время первого полёта число оказалось слишком большим для той
ячейки памяти, куда его записывали. Программа запуталась, и через
37 секунд после старта ракета разрушилась. Стоимость ошибки — больше
300 миллионов долларов.

Потом ошибку исправили, и Ariane 5 летала ещё много лет. На снимке —
та самая ракета в 2021 году: она выводит в космос телескоп
«Джеймс Уэбб».

Обе катастрофы устроил не компьютер. Их устроили люди, написавшие
инструкции.

## Смотрим, как это работает

### Он буквальный

Попросим компьютер сложить два числа. Сначала обычно, потом —
те же числа, но в кавычках.

````python
print(2 + 3)
print("2" + "3")
````

В первом случае он сложил числа и получил пять. Во втором — склеил
два текста и получил «23», потому что в кавычках это уже не числа,
а просто значки на экране.

Человек бы догадался, что мы хотели пятёрку. Компьютер не догадался.
Он сделал ровно то, что написано.

Попробуй поменять числа в ячейке и запусти снова.

### Он быстрый

Насколько именно — можно прикинуть. Обычный процессор делает около
трёх миллиардов действий в секунду. А ENIAC делал пять тысяч.

````python
print("Действий в секунду:", 3000000000)
print("Действий за минуту:", 3000000000 * 60)
print("Во сколько раз быстрее ENIAC:", 3000000000 // 5000)
````

Числа такие, что их даже прочитать вслух трудно. Три миллиарда
секунд — это почти сто лет. А компьютер столько действий делает
за одну секунду. Твой телефон быстрее ENIAC в сотни тысяч раз,
а весит в сто с лишним тысяч раз меньше.

И при этом каждое отдельное действие — простейшее: сложить, сравнить,
переложить. Вся сложность игр и фильмов собрана из таких кирпичиков.

## Может ли компьютер думать?

В 1997 году компьютер Deep Blue обыграл в шахматы чемпиона мира Гарри

На самом деле Deep Blue думал не так, как человек. Он **перебирал
варианты**: около 200 миллионов позиций в секунду, и для каждой считал
очки по правилам, которые написали программисты. Человек смотрит

Современные программы, которые рисуют картинки и отвечают на вопросы,
устроены сложнее. Они учатся на миллионах примеров и находят в них
закономерности. Но и они не понимают мир так, как понимаешь его ты,
и иногда уверенно говорят полную ерунду. Поэтому то, что написал
компьютер, всегда стоит проверять.
