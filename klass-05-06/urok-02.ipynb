## Разбираемся

### Компьютер очень быстрый и очень тупой

Обе половины этой фразы правда, и обе важны.

Обычный компьютер выполняет **несколько миллиардов действий в секунду**.
Если бы ты складывал числа со скоростью одно в секунду и не спал,
то работу, которую он делает за секунду, ты бы заканчивал лет через сто.

А теперь вторая половина. Компьютер не понимает **ничего**. Ни одного
слова. Он не знает, что такое «дом», «мама» или «завтра». Он умеет
только выполнять инструкции — точно и буквально.

### Что значит «буквально»

Представь, что ты просишь друга: «Сходи в магазин, купи хлеб.
Если будут яйца — возьми десяток».

Друг возвращается с десятью батонами хлеба.

Смешно, но компьютер поступил бы именно так. Он выполнил инструкцию
**буквально**: «если будут яйца — возьми десяток» — яйца были,
значит десяток. Десяток чего? Того, о чём шла речь. Хлеба.

Человек догадался бы по смыслу. Компьютер смысла не видит.

### Из этого следует всё остальное

Раз он не понимает, а только выполняет, то:

* **он никогда не ошибается сам** — ошибается тот, кто написал инструкцию;
* **он не догадается**, что ты имел в виду, если написал не то;
* **он не устанет и не заскучает** — миллион раз подряд сделает
  одинаково.

Программист — это человек, который умеет объяснить задачу так, чтобы
её понял тот, кто вообще ничего не понимает. Это сложнее, чем кажется.

### Знаменитые случаи

**Марсианский аппарат, 1999 год.** Одна команда инженеров считала
в километрах, другая в милях. Никто не проверил. Аппарат вошёл
в атмосферу не под тем углом и сгорел. Компьютер всё посчитал верно —
по тем данным, что ему дали.

**Ракета Ariane 5, 1996 год.** Число оказалось слишком большим
для той ячейки, куда его записывали. Ракета развалилась через 37 секунд
после старта. Стоимость ошибки — больше 300 миллионов долларов.

Обе катастрофы устроил не компьютер. Их устроили люди, написавшие
инструкции.

## Смотрим, как это работает

### Он быстрый

Пусть компьютер сложит все числа от 1 до миллиона. Засечём время.

````python
import time

начало = time.time()

сумма = 0
число = 1
while число <= 1000000:
    сумма = сумма + число
    число = число + 1

прошло = time.time() - начало

print("Сумма всех чисел от 1 до миллиона:", сумма)
print("Заняло секунд:", round(прошло, 2))
````

Миллион сложений — меньше секунды. Тебе на то же самое понадобилось бы
около двух недель без сна.

### Он буквальный

Попросим его сложить два числа. Только числа дадим в кавычках.

````python
print(2 + 3)
print("2" + "3")
````

В первом случае он сложил числа. Во втором — склеил два текста,
потому что в кавычках это уже не числа, а буквы.

Человек бы догадался, что мы хотели пятёрку. Компьютер не догадался.
Он сделал ровно то, что написано.

### Он не устаёт

````python
счётчик = 0
while счётчик < 5:
    print("Я не устал! Повтор номер", счётчик + 1)
    счётчик = счётчик + 1
````

Поставь вместо пятёрки миллион — он честно напечатает миллион строк
и ни разу не собьётся. Правда, страница после этого будет грузиться
долго, так что лучше не надо.
