## Что делаем

Текст можно набрать так, что его будет приятно читать, — а можно
так, что читать не захочется. Разница не в словах, а в оформлении,
и учат этому за один урок.

## Разбираемся

### Из чего состоит текст

| Часть | Зачем нужна |
|---|---|
| **заголовок** | говорит, о чём дальше |
| **абзац** | одна мысль |
| **список** | перечисление |
| **выделение** | показывает главное слово |

Самое важное правило: **одна мысль — один абзац**. Стена текста
без абзацев не читается, даже если написана хорошо.

### Правила набора

Их немного, и они одинаковы везде: в тетради, в документе,
в сообщении.

1. **Пробел после знака препинания, а не до.**
   Правильно: `Привет, мир!` Неправильно: `Привет ,мир !`
2. **Один пробел, а не пять.** Ряд пробелов вместо отступа —
   первый признак, что человек не умеет работать с документом.
3. **Enter — только между абзацами.** Внутри абзаца текст
   переносится сам.
4. **Дефис и тире — разные.** Дефис короткий, без пробелов:
   `кто-то`. Тире длинное, с пробелами: `знание — сила`.
5. **Не пиши всё заглавными.** ТАКОЙ ТЕКСТ ЧИТАЕТСЯ МЕДЛЕННЕЕ
   И ВОСПРИНИМАЕТСЯ КАК КРИК.

### Почему не выравнивают пробелами

Кажется удобным: нажал пробел десять раз — и слово встало на место.
Но стоит поменять шрифт или размер — и всё разъезжается, потому что
ширина пробела меняется.

Отступ задаётся настройкой абзаца, а не пробелами. Тогда он
не зависит ни от шрифта, ни от размера.

### Зачем нужны стили

Можно сделать заголовок так: выделить строку, поставить размер 16,
нажать «жирный», выбрать цвет. А можно применить **стиль**
«Заголовок 1».

Выглядит одинаково. Но представь, что заголовков двадцать
и все надо сделать синими:

* без стилей — двадцать раз повторить всю возню;
* со стилями — поменять стиль один раз.

Стиль — это как переменная в программе: задал в одном месте,
применилось везде.

### Разметка значками

Все текстовые ячейки этого учебника набраны особым способом —
**Markdown**. Оформление задаётся прямо значками, не отрывая рук
от клавиатуры.

| Что написать | Что получится |
|---|---|
| `# Заголовок` | большой заголовок |
| `## Заголовок` | заголовок поменьше |
| `**жирный**` | **жирный** |
| `*курсив*` | *курсив* |
| `* пункт` | пункт списка |
| `1. пункт` | нумерованный пункт |

Уровней заголовков шесть: от одной решётки до шести. Больше
не нужно — такой текст никто не прочтёт.

Дважды щёлкни по этой ячейке — и увидишь, как она набрана.
Shift+Enter вернёт красивый вид.

## Смотрим, как это работает

### Компьютер видит каждый пробел

```python
хорошо = "Привет, мир!"
плохо = "Привет ,  мир !"

print("Правильно:", хорошо, "— символов:", len(хорошо))
print("Неряшливо:", плохо, "— символов:", len(плохо))
```

Три лишних символа в короткой фразе. В тексте на страницу таких
наберётся сотня — и любое изменение шрифта превратит его в кашу.

### Сколько в тексте пробелов

```python
фраза = "Информатика — интересный предмет"

print(фраза)
print("Символов всего:", len(фраза))
```

`len` считает **все** символы: буквы, пробелы, тире, знаки
препинания. Для компьютера они равноправны.

### Одна мысль — один абзац

```python
стена = ("Я был в музее там были картины и ещё динозавры а потом "
         "мы пошли в кафе и я съел мороженое оно было вкусное")

по_абзацам = ("Я был в музее. Там были картины и скелет динозавра.\n\n"
              "Потом мы пошли в кафе. Я съел мороженое — очень вкусное.")

print("СТЕНОЙ:")
print(стена)
print()
print("ПО АБЗАЦАМ:")
print(по_абзацам)
```

Слова почти те же. Второй вариант читается легко, первый — нет.

`\n` — это переход на новую строку. Два подряд дают пустую строку
между абзацами.

### Заголовок и текст

```python
print("# Мой поход в музей")
print()
print("**Когда:** в субботу")
print()
print("Больше всего понравился скелет динозавра.")
print()
print("* картины")
print("* динозавры")
print("* мороженое")
```

Так выглядит разметка Markdown до превращения в красивый вид.
Скопируй эти строчки в текстовую ячейку — и увидишь результат.
