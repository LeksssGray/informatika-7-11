## Что делаем

Нейросети рисуют картинки, пишут тексты и отвечают на вопросы. Как они
этому учатся и почему иногда уверенно говорят неправду? Сегодня соберём
в Python крошечный «искусственный интеллект» и увидим это своими глазами.

## Разбираемся

### Правила и примеры

Обычную программу пишет человек: «если холоднее 20 — включи обогрев».
Правило придумано заранее.

Нейросети правил не дают. Ей показывают **примеры** — тысячи фотографий
с подписью «кошка» или «собака» — и она сама подбирает, по каким
признакам их различать. Это называется **обучением**.

### Нейросеть угадывает самое вероятное

Чат-бот, который пишет текст, на каждом шаге выбирает **самое вероятное
следующее слово** — такое, которое чаще всего встречалось в похожих
текстах. После «Доброе…» почти всегда идёт «утро», поэтому он
и напишет «утро».

Он не «знает» ответ, а продолжает текст так, как это обычно делают люди.

### Почему нейросеть уверенно врёт

Если в текстах, на которых она училась, ошибка встречалась часто —
нейросеть повторит ошибку. Если правильного ответа в примерах не было —
она всё равно напишет что-нибудь гладкое и убедительное. Сомневаться
она не умеет: тон у правды и выдумки одинаковый.

Такие выдумки называют **галлюцинациями**.

### Правила для человека

* **проверяй факты** — даты, числа, имена — в другом источнике;
* **не рассказывай** чат-боту пароли, адрес, номер телефона;
* **думай сам** — нейросеть помогает, но отвечаешь за работу ты.

## Смотрим, как это работает

### Учимся на примерах: кошка или собака

Компьютер знает только вес животного. Дадим ему примеры, и пусть
сам найдёт границу между кошками и собаками.

```python
кошки = [4, 5, 3, 6]      # вес кошек из примеров, кг
собаки = [20, 30, 25, 15]  # вес собак, кг

средняя_кошка = sum(кошки) / len(кошки)
средняя_собака = sum(собаки) / len(собаки)
порог = (средняя_кошка + средняя_собака) / 2

print("Средняя кошка:", средняя_кошка)
print("Средняя собака:", средняя_собака)
print("Граница:", порог)
```

Границу никто не придумывал — она получилась из примеров. Это и есть
обучение, только очень маленькое.

### Угадываем

```python
вес = 8

if вес < порог:
    print(вес, "кг — наверное, кошка")
else:
    print(вес, "кг — наверное, собака")
```

Попробуй вес 2, 12, 40. А теперь вес 14: маленькая собака весит
столько же, сколько огромный кот. Программа ответит уверенно —
и может ошибиться. Ей просто не хватает признаков.

### Какое слово идёт дальше

Возьмём крошечный «интернет» — список слов — и посмотрим, что в нём
идёт после слова «доброе».

```python
слова = ["доброе", "утро", "друзья", "доброе", "утро", "мама",
         "доброе", "дело", "доброе", "утро", "всем"]

for i in range(len(слова) - 1):
    if слова[i] == "доброе":
        print("после «доброе» —", слова[i + 1])
```

«Утро» встретилось три раза, «дело» — один. Нейросеть выберет «утро».
Она не понимает смысла — она считает, что чаще.

### Почему она уверенно врёт

Многие люди думают, что столица Австралии — Сидней. На самом деле —
Канберра. Если в текстах чаще ошибались, чем писали правильно:

```python
продолжения = ["Сидней", "Канберра", "Сидней", "Сидней", "Мельбурн", "Сидней"]

сидней = 0
канберра = 0
for слово in продолжения:
    if слово == "Сидней":
        сидней = сидней + 1
    if слово == "Канберра":
        канберра = канберра + 1

print("Сидней:", сидней, " Канберра:", канберра)

if сидней > канберра:
    print("Ответ: столица Австралии — Сидней")
else:
    print("Ответ: столица Австралии — Канберра")
```

Ответ неверный, но звучит уверенно. Вот почему факты из ответа
нейросети надо проверять.
