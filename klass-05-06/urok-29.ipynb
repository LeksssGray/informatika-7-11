## Разбираемся

### Что такое нейросеть

**Нейросеть** — программа, которая учится на множестве примеров: картинок, текстов, звуков.
Она не получает готовые правила, а подбирает их сама. Поэтому нейросеть хорошо узнаёт то, с чем
много работала, и плохо справляется с тем, чего не видела.

### Почему она может врать

Нейросеть не знает фактов в человеческом смысле. Она выдаёт наиболее правдоподобный ответ.
Если ответ звучит уверенно, это не значит, что он верный. Поэтому любой ответ нейросети стоит проверить.
