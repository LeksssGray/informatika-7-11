## Разбираемся

### Что делает программист

Многие думают, что программист целый день печатает. На самом деле больше времени уходит
на то, чтобы **прочитать** чужой код, понять задачу и найти, почему что-то работает не так.
Поэтому код пишут так, чтобы его было легко читать: понятные имена, короткие строки, комментарии.

```python
print("Привет, мир!")
print("Меня зовут", "Аня")
```

### Ошибка на миллион

Ошибка в программе, которой пользуются миллионы людей, может стоить очень дорого. Поэтому
программисты проверяют свой код, просят коллег его посмотреть и пишут тесты. Ошибку, которая
мешает программе работать, называют **багом**.
