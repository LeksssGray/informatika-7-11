## Разбираемся

### Чем на самом деле занят программист

Кажется, что программист целый день быстро печатает. На деле больше
времени уходит на другое:

* понять, **что именно** нужно сделать;
* **прочитать** код, который написали раньше — другие люди или ты сам
  месяц назад;
* найти, **почему** программа работает не так, как задумано;
* проверить, что исправление ничего не сломало.

Код читают гораздо чаще, чем пишут. Поэтому хороший код — не самый
короткий, а самый **понятный**.

### Понятные имена

Сравни две программы. Они делают одно и то же:

```python
a = 12                    тетрадей = 12
b = 35                    цена = 35
c = a * b                 стоимость = тетрадей * цена
print(c)                  print(стоимость)
```

Левую придётся разгадывать. Правую читаешь как текст. Имя переменной
должно говорить, **что в ней лежит**.

### Баг и отладка

Ошибку в программе называют **багом** — по-английски «жук». Есть история,
что в 1947 году в огромном компьютере Mark II нашли настоящую моль,
застрявшую между контактами, и вклеили её в журнал с подписью «первый
найденный баг».

Поиск и исправление ошибок называется **отладкой**. Главный приём
отладки очень простой: **напечатать, что лежит в переменных**, и сравнить
с тем, что там должно быть.

### Ошибка на миллион

В 1999 году к Марсу летел аппарат Mars Climate Orbiter. Одна команда
инженеров считала силу в фунтах, другая — в ньютонах, и никто
не проверил, что числа понимают одинаково. Аппарат подошёл к планете
слишком низко и сгорел в атмосфере. Стоил он около 125 миллионов
долларов.

Код при этом «работал» — просто считал не то. Самые опасные ошибки
именно такие: программа не ломается, а тихо выдаёт неверный ответ.

## Смотрим, как это работает

### Непонятно и понятно

````python
a = 12
b = 35
c = a * b
print(c)
````

````python
тетрадей = 12
цена = 35
стоимость = тетрадей * цена
print("Тетради стоят", стоимость, "рублей")
````

Ответ одинаковый, но во второй программе сразу видно, что посчитали.

### Программа работает, но врёт

Хотели сложить числа от 1 до 10. Должно быть 55. Запусти:

````python
сумма = 0
for i in range(1, 10):
    сумма = сумма + i
print(сумма)
````

Получилось 45. Ошибок Python не показал — он сделал ровно то,
что написано. Где расхождение?

### Отладка: печатаем, что происходит

````python
сумма = 0
for i in range(1, 10):
    сумма = сумма + i
    print("прибавили", i, "— стало", сумма)
````

Последним прибавилось 9, а десятка не вошла. Правая граница `range`
в счёт не входит — нужно `range(1, 11)`. Одна строчка `print` внутри
цикла показала ошибку за секунду.

### Ещё один частый баг

Копилку завели **внутри** цикла:

````python
for i in range(1, 6):
    копилка = 0
    копилка = копилка + i
print(копилка)
````

Вместо 15 получилось 5: на каждом повторе копилка снова обнуляется.
Копилку заводят **до** цикла, один раз.
