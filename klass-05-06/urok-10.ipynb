## Разбираемся

### Обманывают не компьютер, а человека

Взламывать пароли долго. Гораздо проще уговорить человека самого
отдать пароль, код или деньги. Такой обман называют **социальной
инженерией**. Мошенники хорошо знают, как люди себя ведут, и давят
на чувства:

| Чем давят | Как это звучит |
|---|---|
| **спешка** | «только сегодня», «в течение часа», «пока не заблокировали» |
| **страх** | «ваш аккаунт взломан», «звонят из полиции», «с карты списывают деньги» |
| **жадность** | «вы выиграли», «бесплатные скины», «удвоим твои монеты» |
| **жалость** | «мама, я попал в беду», «помогите больному котёнку» |

Если сообщение заставляет волноваться и торопиться — это первый
повод остановиться и подумать.

### Самые частые приёмы

**Поддельный сайт.** Ссылка ведёт на страницу, которая выглядит как
настоящая: тот же логотип, те же цвета. Но адрес другой:
`sberbank-bonus.ru` вместо `sberbank.ru`, `vk-login.com` вместо `vk.com`.
Ты вводишь пароль — и он уходит мошеннику. Такой обман называется
**фишингом**, от английского «рыбалка»: жертву ловят на наживку.

**«Это я, но с другого номера».** Пишут от имени друга, мамы, учителя:
телефон якобы сломался, срочно нужны деньги или код из SMS. Иногда
вообще взламывают аккаунт знакомого и пишут от него.

**Звонок «из банка» или «из полиции».** Строгий голос, пугающая
новость, просьба назвать данные карты или код. Настоящий банк
и настоящая полиция так **никогда** не делают.

**Бесплатное в играх.** Скины, робуксы, голоса, редкие предметы
«за вход через аккаунт». В лучшем случае украдут аккаунт, в худшем —
ещё и привязанную карту родителей.

**Выигрыш, в котором ты не участвовал.** Приз бесплатный, но «оплатите
доставку» или «налог». Приза нет, деньги пропадут.

   ведёт на поддельный адрес.

Три признака сразу — это почти всегда обман.

## Как это было: письмо от узника

Мошенники придумали свои приёмы задолго до интернета. Ещё в XIX веке
люди получали письма от «богатого узника» из Испании: он якобы томится
в тюрьме, но спрятал сокровища. Помоги деньгами освободиться —
получишь щедрую награду. Деньги отправляли, узника, конечно, не было.

Сто лет спустя тот же приём пришёл в электронную почту: «наследник
африканского принца» просил помочь вывести миллионы и заплатить
небольшую комиссию. Меняются только декорации.

## Что делать, если сомневаешься

1. **Не торопись.** Настоящее важное дело подождёт пять минут.
2. **Не переходи по ссылке** из подозрительного сообщения. Нужен сайт
   банка или игры — набери адрес сам.
3. **Перезвони сам** по номеру, который знаешь: маме — на мамин,
   другу — на его обычный номер.
4. **Никому не называй** пароли и коды из SMS.
5. **Спроси взрослого.** Показать сообщение родителям не стыдно —
   стыдно попасться.

Хорошая идея — договориться в семье о **секретном слове**. Если кто-то
пишет «мама, срочно переведи», спроси это слово. Мошенник его не знает.
