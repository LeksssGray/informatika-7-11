## Разбираемся

### Компьютер умеет повторять

Для повторов есть особая команда — **цикл**:

```python
for i in range(5):
    print("Привет")
```

Читается так: «пять раз сделай то, что ниже с отступом».
«Привет» напечатается пять раз.

### Из чего состоит цикл

```
for i in range(5):
└─┬┘ │  └──┬──┘ │
  │  │     │    └─ двоеточие обязательно
  │  │     └────── сколько раз повторять
  │  └──────────── переменная-счётчик
  └─────────────── слово «для»
```

Строчки **с отступом** под `for` — это тело цикла, они и повторяются.
Отступ делают четырьмя пробелами (или клавишей Tab).

Как только отступ кончился — цикл кончился.

### Счётчик считает с нуля

Переменная `i` на каждом повторе меняется:

```python
for i in range(5):
    print(i)
```

Напечатает 0, 1, 2, 3, 4. Пятёрки не будет: **счёт идёт с нуля**,
и последнее число в `range` не входит.

Так же, как номера в списке, — помнишь, с урока про пиксель-арт?

### Три вида range

| Запись | Какие числа |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(2, 11, 2)` | 2, 4, 6, 8, 10 |

Первое число — откуда начать, второе — до чего (не включая),
третье — с каким шагом.

Шаг бывает и отрицательным: `range(10, 0, -1)` даст 10, 9, 8, …, 1.
Пригодится для обратного отсчёта.

### Копилка

Самое полезное, что делают в цикле, — **копят** результат:

```python
сумма = 0
for i in range(1, 6):
    сумма = сумма + i
print(сумма)
```

Копилку заводят **до** цикла и кладут в неё ноль. На каждом шаге
добавляют очередное число. После цикла в ней сумма: 15.

Важно: `print` стоит **после** цикла, без отступа. Будь он внутри,
число печаталось бы пять раз.

### Частая ошибка

```python
for i in range(1, 6):
    сумма = 0              # ← заводим копилку заново каждый раз!
    сумма = сумма + i
```

На каждом шаге копилка обнуляется, и в конце в ней останется
только последнее число. **Копилку заводят до цикла.**

## Смотрим, как это работает

### Повторяем пять раз

````python
for i in range(5):
    print("Привет")
````

### Что лежит в счётчике

````python
for i in range(5):
    print("Шаг", i)
````

Счёт с нуля, пятёрка не вошла. Это не ошибка, а правило.

### Считаем от одного

````python
for i in range(1, 6):
    print("Номер", i)
````

### Таблица умножения на 7

````python
for i in range(1, 11):
    print("7 ×", i, "=", 7 * i)
````

Десять строк — и всего две строчки кода. Поменяй семёрку
на своё число.

### Обратный отсчёт

````python
for i in range(10, 0, -1):
    print(i)

print("Пуск!")
````

### Копилка растёт

````python
сумма = 0

for i in range(1, 6):
    сумма = сумма + i
    print("Прибавили", i, "— в копилке", сумма)

print("Итого:", сумма)
````

### Рисуем лестницу циклом

Помнишь лестницу с урока про рисование? Тогда мы писали пять строк.
Теперь — одну.

````python
for i in range(1, 6):
    print("█" * i)
````

### Только чётные

````python
for i in range(2, 11, 2):
    print(i)
````

### Любопытно

Первой машиной, которая умела повторять, был не компьютер,
а ткацкий станок. В 1804 году француз Жозеф Жаккар сделал станок,
которым управляли картонные карточки с дырочками: каждая карточка —
один ряд узора.

Карточки сшивали в длинную ленту, и станок снова и снова повторял
рисунок цветов и листьев, пока не кончится ткань. Это и был цикл —
за полтора века до первого компьютера.
