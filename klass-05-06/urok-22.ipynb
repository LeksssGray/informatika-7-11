## Что делаем

Сегодня — про таблицы, которые считают сами. Карманные деньги,
расходы, накопления: вписал числа один раз, а итог пересчитывается

## Разбираемся

### Как устроена таблица

Таблица — это клетки. Столбцы обозначают буквами, строки — числами.
Клетка на пересечении столбца **B** и строки **2** называется **B2**:
сначала буква, потом номер.

```
     A          B        C
1    Что        Сколько  Цена
2    Булочка    2        35
3    Сок        1        60
```

### Формула начинается со знака равно

Если в клетке написать `=B2*C2`, таблица не покажет эти буквы —
она **посчитает**. Знак равно в начале означает: «дальше считай».

```
=35+60        посчитает 95
=B2*C2        умножит то, что лежит в B2, на то, что в C2
```

### Почему в формулах адреса, а не числа

Это главное, что стоит понять про таблицы.

Напишешь `=2*35` — получишь 70 и всё. Поменял количество булочек?
Переписывай формулу.

Напишешь `=B2*C2` — таблица будет **пересчитывать сама** при любом
изменении. Поменял 2 на 5 — итог обновился мгновенно, ничего
не трогая.

Поэтому правило простое: **в формулах пишут адреса клеток**.

### Готовые функции

Складывать двадцать клеток по одной не нужно — есть функции:

| Функция | Что делает |
|---|---|
| `=СУММ(C2:C20)` | складывает всё от C2 до C20 |
| `=СРЗНАЧ(C2:C20)` | считает среднее |
| `=МАКС(C2:C20)` | находит самое большое |
| `=МИН(C2:C20)` | самое маленькое |
| `=СЧЁТ(C2:C20)` | сколько чисел |

Двоеточие означает «все клетки подряд»: `C2:C20` — это девятнадцать
клеток.

### Растянуть формулу

Формулу пишут в первой строке и растягивают вниз за уголок.
Адреса внутри сдвигаются сами:

```
в D2:  =B2*C2
в D3:  =B3*C3     ← таблица поправила номера сама
в D4:  =B4*C4
```

Это экономит часы работы: тысячу строк считают одной формулой.

### Где такое нужно

Карманные деньги за месяц, расходы на подарки, счёт в игре, план
накоплений на что-то дорогое, оценки за четверть. Всё, где есть
числа, которые меняются.

## Смотрим, как это работает

Настоящей таблицы в ноутбуке нет, зато есть Python — и он считает
точно так же. Заодно видно, что внутри формул ничего волшебного.

### Считаем стоимость покупки

````python
количество = 2
цена = 35

стоимость = количество * цена

print("Булочек:", количество, "по", цена, "рублей")
print("Стоимость:", стоимость)
print("В таблице это была бы формула =B2*C2")
````

### Поменяли одно число

````python
количество = 5
цена = 35

стоимость = количество * цена

print("Теперь булочек:", количество)
print("Стоимость:", стоимость)
print("Формулу не трогали — поменялось только количество.")
````

### Итог за неделю

````python
понедельник = 150
вторник = 200
среда = 100
четверг = 250
пятница = 300

всего = понедельник + вторник + среда + четверг + пятница

print("Потрачено за неделю:", всего)
print("В таблице: =СУММ(B2:B6)")
````

### Среднее за день

````python
дней = 5

среднее = всего / дней

print("Всего:", всего)
print("Дней:", дней)
print("В среднем за день:", среднее)
print("В таблице: =СРЗНАЧ(B2:B6)")
````

### Список расходов и сумма

````python
расходы = [150, 200, 100, 250, 300]

print("Расходы:", расходы)
print("Дней:", len(расходы))
print("Сумма:", sum(расходы))
print("Больше всего за день:", max(расходы))
print("Меньше всего:", min(расходы))
````

Четыре функции — ровно те же, что в таблице: СУММ, СЧЁТ,
МАКС и МИН.

### Накопить на покупку

````python
цель = 4500
в_неделю = 300

недель = цель / в_неделю

print("Цель:", цель, "рублей")
print("Откладываю в неделю:", в_неделю)
print("Понадобится недель:", недель)
print("Это примерно месяцев:", недель / 4)
````

### Сколько клеток в диапазоне

````python
первая = 2
последняя = 20

клеток = последняя - первая + 1

print("Диапазон C2:C20")
print("Клеток в нём:", клеток)
````

Считать нужно **включительно** оба конца — отсюда «плюс один».
На этом ошибаются чаще всего.
