## Разбираемся

### Как устроена таблица

Таблица — это клетки, их называют **ячейками**. Столбцы обозначают
буквами, строки — числами. Ячейка на пересечении столбца **C**
и строки **5** называется **C5**: сначала буква, потом номер.

### Формула начинается со знака равно

Если в ячейке написать `=C2*D2`, таблица не покажет эти буквы —
она **посчитает**. Знак равно в начале означает: «дальше считай».

### Почему в формулах адреса, а не числа

Это главное, что стоит понять про таблицы.

Напишешь `=2*35` — получишь 70 и всё. Поменял количество булочек?
Переписывай формулу.

Напишешь `=C2*D2` — таблица будет **пересчитывать сама** при любом
изменении. Поменял 2 на 5 — итог обновился мгновенно.

Поэтому правило простое: **в формулах пишут адреса ячеек**.

### Готовые функции

| Функция | Что делает |
|---|---|
| `=СУММ(E2:E8)` | складывает всё от E2 до E8 |
| `=СРЗНАЧ(E2:E8)` | считает среднее |
| `=МАКС(E2:E8)` | находит самое большое |
| `=МИН(E2:E8)` | самое маленькое |

Двоеточие означает «все ячейки подряд»: `E2:E8` — это семь ячеек,
от второй строки до восьмой **включительно**.
