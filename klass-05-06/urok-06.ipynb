## Разбираемся

### Считать умеет любой калькулятор

Умеет. Но калькулятор считает один пример за раз, а компьютер — тысячу
примеров подряд, и запоминает промежуточные результаты.

В этом вся разница. Разберёмся, как заставить его считать то,
что вручную считать лень.

### Все действия

| Что делает | Знак | Пример | Ответ |
|---|---|---|---|
| сложение | `+` | `7 + 3` | `10` |
| вычитание | `-` | `7 - 3` | `4` |
| умножение | `*` | `7 * 3` | `21` |
| деление | `/` | `7 / 2` | `3.5` |
| деление без остатка | `//` | `7 // 2` | `3` |
| остаток | `%` | `7 % 2` | `1` |

Первые четыре знакомы по математике. Последние два стоит разобрать
отдельно — они пригодятся много раз.

### Деление нацело и остаток

Представь: 7 конфет надо разделить на двоих поровну.

Каждому достанется по 3 конфеты — это `7 // 2`, деление нацело.
Одна конфета останется лишней — это `7 % 2`, остаток.

```
  7 // 2 = 3      сколько целых поместилось
  7 % 2  = 1      сколько осталось
```

Этой парой решается куча бытовых задач: перевести минуты в часы,
копейки в рубли, разложить предметы по коробкам.

### Переменные — коробки для чисел

Результат можно запомнить, положив его в **переменную**:

```python
конфет = 7
друзей = 2
каждому = конфет // друзей
```

Слева имя, справа значение. Знак `=` тут не «равно», а «положи».

Имя придумываешь сам, но лучше понятное: через неделю `x` не скажет
тебе ничего, а `конфет` скажет.

### Порядок действий

Такой же, как в математике: сначала умножение и деление, потом
сложение и вычитание. Скобки всё меняют:

```python
2 + 3 * 4      →  14
(2 + 3) * 4    →  20
```

Сомневаешься — ставь скобки. Лишняя не помешает.

## Смотрим, как это работает

### Сколько ты уже прожил

````python
возраст = 12

дней = возраст * 365
часов = дней * 24
минут = часов * 60

print("Дней прожито примерно:", дней)
print("Часов:", часов)
print("Минут:", минут)
````

Меняй возраст на свой и запускай снова. Числа получаются такие,
что в уме их не посчитаешь, а компьютеру всё равно.

### Конфеты и остаток

````python
конфет = 17
друзей = 5

каждому = конфет // друзей
останется = конфет % друзей

print("Конфет:", конфет, "друзей:", друзей)
print("Каждому достанется:", каждому)
print("Останется лишних:", останется)
````

Три конфеты каждому, две останутся. Проверь в уме: 5 умножить на 3
это 15, плюс 2 лишние — как раз 17.

### Минуты в часы

````python
всего_минут = 250

часов = всего_минут // 60
минут = всего_минут % 60

print(всего_минут, "минут — это", часов, "ч", минут, "мин")
````

Тот же приём, только вместо конфет минуты, а вместо друзей — 60.
Схема одна и та же, меняются только числа.

### Карманные деньги

````python
в_неделю = 500
недель_в_году = 52

за_год = в_неделю * недель_в_году
велосипед = 15000

print("За год накопится:", за_год)
print("Велосипед стоит: ", велосипед)
print("Хватит ли:       ", за_год >= велосипед)
````

Последняя строка сравнивает два числа и печатает `True` или `False` —
«правда» или «неправда». Про сравнения подробно поговорим позже,
пока просто посмотри, что так тоже можно.
