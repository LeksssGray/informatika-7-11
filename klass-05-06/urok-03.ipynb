## Разбираемся

### Что внутри

Если открыть системный блок, увидишь несколько плат и коробочек.
Выглядит сложно, но главных деталей всего четыре, и у каждой
понятная работа.

### Процессор — тот, кто считает

Маленький квадратик под большим радиатором. Это мозг: он выполняет
все команды. Складывает, сравнивает, перекладывает данные.

Больше он ничего не умеет. Совсем ничего. Всё, что ты видишь на экране —
игры, видео, музыка — собрано из миллиардов таких простейших действий.

Греется он так, что без радиатора и вентилятора сгорел бы за секунды.

### Оперативная память — рабочий стол

Представь письменный стол. На нём лежит то, с чем ты работаешь прямо
сейчас: тетрадь, учебник, ручка. Стол не бесконечный, и когда всё
не помещается, приходится что-то убирать.

Оперативная память работает так же. В ней лежат открытые программы
и файлы. Чем её больше, тем больше программ можно держать открытыми
одновременно.

> **Важное свойство.** Выключил компьютер — оперативная память
> очистилась полностью. Всё, что там было, пропало. Именно поэтому
> несохранённый документ теряется при выключении света.

### Диск — шкаф

А это шкаф, куда всё убирают на хранение. Диск помнит информацию
и без электричества: фотографии, игры, документы лежат там годами.

Диски бывают двух видов. **Жёсткий диск** — внутри крутящиеся пластины
и головка, как в проигрывателе пластинок. Дешёвый, вместительный,
медленный. **SSD** — без движущихся частей, работает в несколько раз
быстрее и не боится тряски.

Если компьютер долго включается — почти наверняка у него жёсткий диск.

### Видеокарта — художник

Рисует всё, что ты видишь на экране. В простом компьютере она встроена
в процессор, в игровом — отдельная плата размером с кирпич
и с собственными вентиляторами.

### Почему компьютер тормозит

Теперь понятно, отчего это бывает.

| Что тормозит | Кто виноват |
|---|---|
| долго включается | медленный диск |
| тормозит, когда открыто много вкладок | мало оперативной памяти |
| тормозит в играх | слабая видеокарта |
| тормозит вообще всегда | слабый процессор или всё сразу |

### Как читать характеристики

Когда в описании компьютера пишут:

```
Процессор Intel Core i5, 8 ГБ ОЗУ, SSD 512 ГБ
```

— это значит: мозг средней силы, рабочий стол на 8 гигабайт,
быстрый шкаф на 512 гигабайт.

## Смотрим, как это работает

### Считаем объём памяти

Гигабайты, мегабайты — в них легко запутаться. Разберёмся раз и навсегда.

````python
гигабайт = 1
мегабайт = гигабайт * 1024
килобайт = мегабайт * 1024

print("В 1 гигабайте мегабайт:", мегабайт)
print("В 1 гигабайте килобайт:", килобайт)
````

Обрати внимание: не тысяча, а **1024**. Так уж исторически сложилось —
компьютер считает двойками, а 1024 это ровно две в десятой степени.
Позже разберём почему, пока просто запомни число.

### Сколько фотографий влезет

````python
диск_гигабайт = 512
фото_мегабайт = 4

всего_мегабайт = диск_гигабайт * 1024
фотографий = всего_мегабайт // фото_мегабайт

print("На диск влезет фотографий:", фотографий)
````

Сто тридцать тысяч. Знак `//` — это деление, при котором остаток
отбрасывается: половина фотографии нам всё равно ни к чему.

### Что съедает оперативку

````python
браузер = 1500
игра = 4000
музыка = 300

занято = браузер + игра + музыка
всего = 8 * 1024

print("Занято мегабайт:", занято)
print("Всего мегабайт: ", всего)
print("Осталось:       ", всего - занято)
````

Три программы съели больше половины. Открой ещё несколько вкладок —
и место кончится, компьютер начнёт тормозить.
