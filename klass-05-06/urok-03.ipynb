Так выглядит игровой компьютер, если снять боковую стенку. Цветная

### Процессор — тот, кто считает

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/processor.jpg" width="460" alt="Процессор сверху и снизу">

*Процессор сверху и снизу*

<sub>Intel · CC BY-SA 2.0 · Wikimedia Commons</sub>

Слева — процессор сверху, справа — снизу. Под металлической крышкой
спрятан кристалл кремния размером с ноготь, а на нём больше миллиарда
крошечных переключателей — транзисторов. Снизу видны контакты: у этого
процессора их 1150, и через каждый бегут сигналы.

Процессор — мозг компьютера: он выполняет все команды. Складывает,
сравнивает, перекладывает данные. Больше он ничего не умеет. Всё, что
ты видишь на экране — игры, видео, музыка, — собрано из миллиардов
таких простейших действий.

### Кулер — чтобы не сгореть

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/kuler.jpg" width="300" alt="Кулер процессора: радиатор из тонких пластин">

*Кулер процессора: радиатор из тонких пластин*

<sub>D-Kuru · CC BY-SA 3.0 at · Wikimedia Commons</sub>

Процессор работает так быстро, что сильно греется. Без охлаждения
он перегрелся бы за секунды. Поэтому сверху на нём стоит **кулер**:
металлический радиатор из тонких пластин и вентилятор. Пластины
забирают тепло, а вентилятор его сдувает.

Если компьютер гудит как пылесос, значит, процессору жарко: он много
работает или радиатор забит пылью.

### Оперативная память — рабочий стол

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/operativka.jpg" width="460" alt="Две планки оперативной памяти">

*Две планки оперативной памяти*

<sub>Bretwa · CC BY-SA 4.0 · Wikimedia Commons</sub>

Представь письменный стол. На нём лежит то, с чем ты работаешь прямо
сейчас: тетрадь, учебник, ручка. Стол не бесконечный, и когда всё
не помещается, приходится что-то убирать.

Оперативная память работает так же. В ней лежат открытые программы
и файлы. Чем её больше, тем больше программ можно держать открытыми
одновременно. Выглядит она как длинные планки с чёрными чипами.

> **Важное свойство.** Выключил компьютер — оперативная память
> очистилась полностью. Всё, что там было, пропало. Поэтому
> несохранённый документ теряется при выключении света.

### Диск — шкаф

А это шкаф, куда всё убирают на хранение. Диск помнит информацию
и без электричества: фотографии, игры, документы лежат там годами.

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/zhestkiy-disk.jpg" width="420" alt="Жёсткий диск со снятой крышкой">

*Жёсткий диск со снятой крышкой*

<sub>Zzubnik · общественное достояние · Wikimedia Commons</sub>

**Жёсткий диск** устроен почти как проигрыватель пластинок. Внутри
крутятся блестящие пластины — 5–7 тысяч оборотов в минуту. Над ними
на тонкой ножке летает головка, которая читает и записывает данные.
Между головкой и пластиной зазор в тысячи раз тоньше волоса. Поэтому
жёсткие диски боятся ударов: если уронить работающий компьютер,
головка может чиркнуть по пластине.

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/ssd.jpg" width="420" alt="Современный накопитель SSD">

*Современный накопитель SSD*

<sub>David290 · CC BY-SA 4.0 · Wikimedia Commons</sub>

**SSD** — диск без движущихся частей, просто плата с чипами памяти.
Он в несколько раз быстрее жёсткого диска, а самые быстрые —
в десятки раз. Он тихий и не боится тряски. Зато места на те же
деньги в нём меньше.

| | Жёсткий диск | SSD |
|---|---|---|
| Как устроен | крутящиеся пластины и головка | чипы памяти |
| Скорость | медленный | быстрый |
| Боится ударов | да | почти нет |
| Место за те же деньги | много | меньше |

Если компьютер включается минуту и дольше, скорее всего, у него
жёсткий диск.

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/videokarta.jpg" width="420" alt="Видеокарта">

*Видеокарта*

## Как это было: диск весом в тонну

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g56/ramac.jpg" width="520" alt="Компьютер IBM 305 RAMAC — первый с жёстким диском">

*Компьютер IBM 305 RAMAC — первый с жёстким диском*

<sub>Армия США, Ballistic Research Laboratory · общественное достояние · Wikimedia Commons</sub>

Первый жёсткий диск появился в 1956 году в компьютере IBM RAMAC.
Внутри шкафа вращались пятьдесят алюминиевых дисков размером
с большую пиццу. Весил накопитель около тонны, а вмещал 3,75 мегабайта —
меньше, чем одна фотография с твоего телефона.

Купить его было нельзя, только взять в аренду: больше трёх тысяч
долларов в месяц. Сейчас карта памяти размером с ноготь хранит
в сотни тысяч раз больше.

### Почему компьютер тормозит

Теперь понятно, отчего это бывает.

| Что тормозит | Кто виноват |
|---|---|
| долго включается | медленный диск |
| тормозит, когда открыто много вкладок | мало оперативной памяти |
| тормозит в играх | слабая видеокарта |
| тормозит вообще всегда | слабый процессор или всё сразу |
| гудит и греется | пыль в радиаторе |

### Как читать характеристики

Когда в описании компьютера пишут:

```
Процессор Intel Core i5, 8 ГБ ОЗУ, SSD 512 ГБ
```

— это значит: мозг средней силы, рабочий стол на 8 гигабайт,
быстрый шкаф на 512 гигабайт.

## Смотрим, как это работает

### Сколько всего помещается

Гигабайты и мегабайты легко перепутать. Посчитаем, сколько одних
в других, и сколько фотографий влезет на диск.

````python
print("В 1 гигабайте мегабайт:", 1024)
print("В 16 гигабайтах мегабайт:", 16 * 1024)
print("Диск 512 гигабайт — это мегабайт:", 512 * 1024)
print("Фотографий по 4 мегабайта влезет:", 512 * 1024 // 4)
````

Обрати внимание: не тысяча, а **1024**. Так уж исторически сложилось —
компьютер считает двойками, а 1024 это ровно две в десятой степени.
Позже разберём почему, пока просто запомни число.

Фотографий влезет сто тридцать тысяч. Знак `//` — это деление,
при котором остаток отбрасывается: половина фотографии нам ни к чему.

### Что съедает оперативку

````python
print("Браузер занял мегабайт:", 1500)
print("Игра заняла:           ", 4000)
print("Музыка заняла:         ", 300)
print("Осталось свободно:     ", 8 * 1024 - (1500 + 4000 + 300))
````

Три программы съели больше половины из 8 гигабайт. Открой ещё
несколько вкладок — и место кончится, компьютер начнёт тормозить.
