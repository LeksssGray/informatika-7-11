## Разбираемся

### 3D-печать

3D-принтер делает предмет **слой за слоем**: выдавливает тонкую нитку
расплавленного пластика, рисует ею один срез модели, поднимается
чуть выше и рисует следующий. Слой тонкий — обычно 0,2 миллиметра,
то есть в одном миллиметре пять слоёв.

Сначала модель рисуют на компьютере в программе для 3D, потом
специальная программа режет её на слои — и принтер печатает.

Что уже работает: игрушки и запчасти, протезы рук для детей, детали
для самолётов, даже небольшие дома из бетона.

### Дополненная реальность

**Дополненная реальность** (AR) — это когда программа добавляет
картинки к тому, что видит камера телефона. Наводишь телефон на пустой
угол комнаты — и на экране там стоит диван, который ты собираешься
купить. Игра Pokémon GO, маски в видеозвонках, подсказки
на лобовом стекле машины — всё это AR.

### Беспилотники

**Дрон** — летательный аппарат без пилота на борту. Им управляет
человек с пульта или программа. Дроны снимают кино, опрыскивают поля,
ищут потерявшихся людей.

Главная беда дрона — **батарея**: её хватает на 20–40 минут. Дрон
должен не только долететь, но и **вернуться**, поэтому половина заряда
всегда «принадлежит» обратной дороге.

**Машины без водителя** уже возят пассажиров — но в отдельных районах
нескольких городов и под присмотром инженеров. Ездить так везде
они пока не умеют.

### Работает или обещают?

Читая новость о технике, ищи слова:

| Значит работает | Значит пока нет |
|---|---|
| продаётся, уже пользуются, работает с такого-то года | показали прототип, испытывают, планируют, к 2030 году |

Обещания не всегда сбываются — и почти всегда сбываются позже,
чем обещали.

## Смотрим, как это работает

### Принтер печатает слои

```python
высота = 3      # миллиметров
слоёв = высота * 5

for слой in range(1, слоёв + 1):
    print("Печатаю слой", слой, "из", слоёв)

print("Готово!")
```

### Сколько ждать

Один слой печатается 40 секунд. А модель высотой 3 сантиметра?

```python
высота = 30          # миллиметров
слоёв = высота * 5
секунд = слоёв * 40

print("Слоёв:", слоёв)
print("Минут:", секунд / 60)
```

Фигурка размером с палец — полтора часа печати. Вот почему 3D-принтер
не заменит магазин: он медленный, зато делает то, чего в магазине нет.

### Дрон летит и возвращается

На каждые 100 метров полёта дрон тратит 5 % заряда. Пусть он летит
вперёд, пока ему хватает заряда и на обратную дорогу.

```python
заряд = 100
от_базы = 0

while заряд >= 10:          # 5 % туда и 5 % обратно
    заряд = заряд - 10
    от_базы = от_базы + 100
    print("Отлетел на", от_базы, "м — в запасе на обратную дорогу")

print("Дальше нельзя: не вернусь. Максимум", от_базы, "м")
```

Заряда хватило бы на 2 километра полёта, но улететь можно только на 1:
вторая половина — дорога домой.
