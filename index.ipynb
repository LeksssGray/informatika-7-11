## Как пользоваться учебником

**Ученику.** Открой урок по ссылке, в первой ячейке впиши ФИО и класс и запусти её. Дальше читай теорию, запускай примеры и решай задачи — проверка встроена в ноутбук и сразу показывает, что не сошлось. Чтобы сохранить свои правки, выбери «Файл → Сохранить копию на Диске».

**Учителю.** Результаты всех проверок собираются в вашу Google-таблицу: кто, когда, какую задачу решил, с какой попытки и каким кодом. Настройка журнала описана в файле `teacher/README.md` репозитория.
