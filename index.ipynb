# Информатика и технология, 5–11 класс

Тридцать четыре урока в год, по одному в неделю. Каждый урок — отдельная страница: сначала разбираемся, потом смотрим примеры (их можно менять и ломать), потом задачи. Проверка встроена — она не ставит оценок, а показывает, где ошибка, и ждёт сколько угодно попыток.

Устанавливать ничего не надо: всё работает в браузере, хоть на школьном компьютере, хоть на телефоне.

## Что нужно знать до первого клика

**Понадобится Google-аккаунт.** Тот же, что для почты или Диска. Без него страницу можно только читать: кнопки запуска работать не будут. Если аккаунта нет, его заводят за пару минут.

**При открытии выскочит предупреждение.** Примерно такое:

> ⚠️ *Внимание: этот блокнот создан не в Google.*
> *Он может запрашивать доступ к вашим данным…*

Это стандартное предупреждение — оно появляется у **любого** файла, который сделал не сам Google. Нажмите **«Всё равно выполнить»** (*Run anyway*). Наши уроки не читают вашу почту и не лезут в файлы: весь их код открыт и лежит в этом репозитории.

**Сразу сохраните копию.** «Файл → Сохранить копию на Диске». Иначе всё, что вы напишете, пропадёт при закрытии вкладки: исходный файл общий для всего класса и правки в нём не сохраняются.

## Как это устроено

**Ученику.** Открываешь урок, в первой ячейке вписываешь имя и класс, запускаешь. Дальше читаешь, пробуешь примеры, решаешь задачи. В первых уроках ответы вписываются в поле, как в анкете; код начинаешь писать позже, когда разберёшься, из чего он состоит.

**Учителю.** Результаты проверок собираются в вашу Google-таблицу: кто, когда, какую задачу решил, с какой попытки и каким кодом. Как это подключить — в `teacher/README.md`.
