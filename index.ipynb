## Как устроен урок

Сначала разбираемся, что к чему. Потом примеры — их можно менять, ломать и запускать заново, на то они и живые. Потом задачи, и в конце домашнее.

Проверка встроена в страницу: показывает, где ошибка, и ждёт сколько угодно попыток. В первых уроках ответы вписываются в поле, как в анкете, — код начинается позже, когда станет понятно, из чего он состоит.

Один урок — одна неделя. В году их тридцать четыре.

## Первый раз здесь?

Начните с первого урока своего класса: там по шагам и с картинками показано, как запускать ячейки, решать задачи и настроить автовход, чтобы не вписывать имя каждую неделю.

**Нужен Google-аккаунт** — тот же, что для почты. Без него страницу можно только читать, кнопки запуска работать не будут. Если аккаунта нет, в первом уроке 5–6 и 7 классов расписано, как его завести.

**При первом запуске Colab предупредит:** «Google не имеет отношения к созданию этого блокнота». Так он встречает любую страницу, сделанную не им самим. Нажмите **«Выполнить»** — код уроков открыт целиком, посмотреть можно всё.
