## Разбираемся

### Откуда берутся данные

До сих пор списки мы писали руками. В настоящих задачах они
появляются тремя способами: из ввода, из генератора случайных
чисел и из вычислений в цикле.

**Из ввода в одну строку** — то, что нужно на олимпиадах:

```python
строка = input()            # "3 8 1 9"
числа = []
for кусок in строка.split():
    числа.append(int(кусок))
```

**Случайными числами** — когда нужно проверить программу
на большом объёме:

```python
import random

числа = []
for i in range(10):
    числа.append(random.randint(1, 100))
```

`random.randint(a, b)` даёт случайное целое от `a` до `b`
включительно — обе границы входят, в отличие от `range`.

**Вычислением в цикле** — самый частый школьный случай:

```python
квадраты = []
for i in range(1, 6):
    квадраты.append(i * i)
```

### Три операции над списком

Почти вся обработка списков — это три действия.

| Действие | Что получается | Заготовка |
|---|---|---|
| **Фильтрация** | список короче исходного | `новый = []` + `if` + `append` |
| **Преобразование** | список той же длины | `новый = []` + `append(f(x))` |
| **Свёртка** | одно число | `итог = 0` + накопление |

Свёртку мы разбирали на уроке про типовые алгоритмы цикла — это
сумма, максимум, счётчик.
Сегодня первые две.

### Фильтрация

```python
чётные = []
for x in числа:
    if x % 2 == 0:
        чётные.append(x)
```

Новый список создаётся **до** цикла и наполняется внутри.
Исходный при этом не меняется — и это хорошо: он ещё может
понадобиться.

### Преобразование

```python
удвоенные = []
for x in числа:
    удвоенные.append(x * 2)
```

### Изменение на месте

Иногда список нужно поправить, а не копировать. Тогда идут
по индексам:

```python
for i in range(len(числа)):
    if числа[i] < 0:
        числа[i] = 0
```

> Запись `for x in числа: x = 0` список **не изменит**: `x` —
> отдельная переменная, копия значения. Менять элементы можно
> только через индекс.

Это ловушка, на которую попадаются все, и понять её стоит сейчас.

### Склейка и повтор

Со списками работают знакомые по строкам операции:

```python
[1, 2] + [3]        # [1, 2, 3]
[0] * 5             # [0, 0, 0, 0, 0]
```

Вторая строка — быстрый способ создать список нужной длины,
заполненный нулями.

## Смотрим, как это работает

### Список из строки ввода

````python
строка = "3 8 1 9 4"

числа = []
for кусок in строка.split():
    числа.append(int(кусок))

print(числа)
print("Сумма:", sum(числа))
````

### Случайные числа

````python
import random

random.seed(8)          # чтобы у всех получилось одинаково

случайные = []
for i in range(10):
    случайные.append(random.randint(1, 100))

print(случайные)
````

Строка `random.seed(8)` делает случайность предсказуемой:
при каждом запуске выйдет один и тот же набор. Без неё числа
были бы разными, и автопроверка не смогла бы ничего сверить.

### Фильтрация

````python
числа = [7, 3, 12, 5, 20, 8]

большие = []
for x in числа:
    if x > 7:
        большие.append(x)

print("Было:", числа)
print("Больше семи:", большие)
````

### Преобразование

````python
числа = [1, 2, 3, 4]

квадраты = []
for x in числа:
    квадраты.append(x * x)

print(квадраты)
````

### Изменение на месте, и как это не работает

````python
числа = [5, -3, 8, -1]

for x in числа:
    if x < 0:
        x = 0

print("После цикла по значениям:", числа, "— ничего не изменилось")

for i in range(len(числа)):
    if числа[i] < 0:
        числа[i] = 0

print("После цикла по индексам:", числа)
````

### Заготовка нужной длины

````python
пусто = [0] * 5
print(пусто)

пусто[2] = 42
print(пусто)
````

### Любопытно

Случайные числа в компьютере не случайны. `random.randint`
вычисляет следующее число из предыдущего по формуле — получается
последовательность, которая **выглядит** беспорядочной,
но полностью предсказуема, если знать начальное значение.
Поэтому её называют псевдослучайной, а начальное значение —
зерном (`seed`).

Для игры этого достаточно, а для шифрования — нет: предсказуемый
«случайный» ключ не защитит ничего. Там зерно берут из настоящего
физического шума — дрожания температуры процессора или
микросекунд между нажатиями клавиш.
