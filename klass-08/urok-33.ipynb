## Разбираемся

### Что такое модель

**Модель** — описание явления правилами и формулами, по которым можно предсказать, что будет дальше.
Модель не обязана быть точной: её ценность в том, что её можно проверить и изменить.

Три знакомых примера:

* **движение** — путь равен скорости, умноженной на время;
* **рост популяции** — за каждый период число особей умножается на один и тот же коэффициент;
* **вклад** — каждый год сумма увеличивается на процент, и проценты начисляются уже на новую сумму.

````python
вклад = 10000
процент = 5
for год in range(1, 4):
    вклад = вклад + вклад * процент / 100
    print("год", год, ":", round(вклад, 2))
````

В таблице та же модель записывается строкой на каждый год: в ячейке следующего года стоит
формула с предыдущим значением, например `=B2+B2*$B$1/100`. Знаки доллара закрепляют
ячейку с процентом, потому что при копировании он не должен сдвигаться.
