### Вложенный цикл

Цикл внутри цикла: внешний делает шаг, а внутренний при каждом шаге проходит весь
свой диапазон. Так заполняют таблицы и перебирают пары элементов.

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end=" ")
    print()
```

### Поиск по порядку

Линейный поиск идёт по списку от начала и останавливается на первом совпадении.
Если совпадения нет, он проходит весь список и ничего не находит. Сколько сравнений
сделано, столько раз сработало условие внутри цикла.
