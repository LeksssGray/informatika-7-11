## Повторяем

### 1. Типы и операции

`int("7")` — число из строки, `str(12)` — строка из числа.
`23 // 5` — целая часть (4), `23 % 5` — остаток (3). Строки
складываются склейкой: `"4" + "5"` = `"45"`.

### 2. Ветвления

`if / elif / else` проверяются сверху вниз, срабатывает первая
подходящая ветка. `and` — оба условия, `or` — хотя бы одно.

### 3. Цикл for: типовые алгоритмы

Сумма — копилка `s += x`, произведение — копилка с единицы,
счётчик — `k += 1` под `if`, делители — перебор `d` и проверка `n % d == 0`.

### 4. Цикл while

Цифры числа: `n % 10` — последняя цифра, `n //= 10` — отбросить её;
повторять, пока `n > 0`.

### 5. Вложенные циклы

Внутренний цикл выполняется целиком на **каждом** шаге внешнего:
2 × 5 = 10 повторов. Пары `i < j` — внутренний цикл с `i + 1`.

### 6. Списки и поиск

Линейный поиск идёт по порядку: в худшем случае столько сравнений,
сколько элементов. Двоичный работает **только на отсортированном**
списке и каждым шагом отбрасывает половину.

### 7. Электронные таблицы

`=СУММ`, `=СРЗНАЧ`, `=СЧЁТЕСЛИ(диапазон; условие)`. Диапазон
`C3:E4` — прямоугольник: 3 столбца × 2 строки = 6 ячеек.

### 8. Базы данных

Строка таблицы — **запись**, столбец — **поле**. **Первичный ключ** —
поле, значения которого не повторяются: номер билета, код товара.
