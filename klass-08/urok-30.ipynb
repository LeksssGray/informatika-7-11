## Разбираемся

### Строка — почти список

Строка устроена как список символов: есть длина, индексы, срезы,
перебор циклом. Отличие одно — **строку нельзя менять на месте**:

```python
слово = "кот"
слово[0] = "р"        # ошибка!
слово = "р" + слово[1:]   # так можно: создали новую строку
```

Поэтому все методы строк не меняют её, а возвращают новую.

### Срезы с шагом

Полная форма среза — `строка[начало:конец:шаг]`:

```python
текст = "информатика"
текст[0:5]        # "инфор"
текст[::2]        # "ифраик" — каждый второй символ
текст[::-1]       # "акитамрофни" — задом наперёд
```

Последняя запись — стандартный способ перевернуть строку
в Python. Шаг −1 означает «иди с конца».

### Поиск подстроки

| Запись | Что даёт |
|---|---|
| `"мама".count("м")` | сколько раз встречается |
| `"мама".find("ам")` | индекс первого вхождения или −1 |
| `"ма" in "мама"` | `True` или `False` |

Оператор `in` читается как в обычном языке и потому нагляднее
всего: `if "ошибка" in строка:`.

### Разбиение и сборка

Два метода, которые в паре решают половину задач на текст.

**`split`** режет строку на список:

```python
"мама мыла раму".split()        # ["мама", "мыла", "раму"]
"5,7,9".split(",")              # ["5", "7", "9"]
```

**`join`** собирает список обратно в строку. Пишется непривычно:
метод вызывается у разделителя.

```python
" ".join(["мама", "мыла"])      # "мама мыла"
"-".join(["2026", "09", "24"])  # "2026-09-24"
```

Читать это надо так: «склеить список, вставляя между элементами
вот эту строку».

### Разбор текста по словам

Связка `split` + цикл — основа всех задач про текст:

```python
for слово in текст.split():
    ...
```

Так считают количество слов, ищут самое длинное, отбирают слова
на нужную букву.

### Полезные методы

| Метод | Что делает |
|---|---|
| `.upper()` / `.lower()` | регистр |
| `.strip()` | убирает пробелы по краям |
| `.replace(a, b)` | замена |
| `.startswith(x)` | начинается ли с подстроки |
| `.isdigit()` | состоит ли только из цифр |

`.strip()` незаменим при обработке ввода: пользователь почти
всегда оставляет лишний пробел, и без него сравнение строк
неожиданно даёт `False`.

## Смотрим, как это работает

### Срезы

````python
текст = "информатика"

print("Целиком:      ", текст)
print("Первые пять:  ", текст[0:5])
print("Каждый второй:", текст[::2])
print("Наоборот:     ", текст[::-1])
print("Последние три:", текст[-3:])
````

### Поиск

````python
фраза = "мама мыла раму"

print("Букв «м»:", фраза.count("м"))
print("Где «мыла»:", фраза.find("мыла"))
print("Где «папа»:", фраза.find("папа"))
print("Есть ли «раму»:", "раму" in фраза)
````

### Разбиение и сборка

````python
фраза = "мама мыла раму"

слова = фраза.split()
print("Слова:", слова)
print("Слов:", len(слова))

обратно = " ".join(слова)
print("Собрали:", обратно)

через_дефис = "-".join(слова)
print("Через дефис:", через_дефис)
````

### Самое длинное слово

````python
текст = "информатика самый интересный школьный предмет"

самое_длинное = ""
for слово in текст.split():
    if len(слово) > len(самое_длинное):
        самое_длинное = слово

print("Самое длинное:", самое_длинное, f"({len(самое_длинное)} букв)")
````

Знакомый шаблон «максимум», только сравниваем не числа,
а длины слов.

### Частота букв

````python
текст = "информатика"

for буква in "аиоэ":
    print(f"{буква}: {текст.count(буква)}")
````

### Шифр Цезаря целиком

В седьмом классе мы сдвигали одну букву. Теперь сдвинем весь текст.

````python
def зашифровать(текст, сдвиг):
    результат = ""
    for символ in текст:
        if "a" <= символ <= "z":
            номер = ord(символ) - ord("a")
            новый = (номер + сдвиг) % 26
            результат += chr(ord("a") + новый)
        else:
            результат += символ
    return результат


шифр = зашифровать("attack at dawn", 3)
print("Зашифровано:", шифр)
print("Расшифровано:", зашифровать(шифр, -3))
````

Остаток `% 26` заворачивает алфавит в кольцо: после «z» снова
идёт «a». Без него буквы в конце алфавита превратились бы
в посторонние символы.
