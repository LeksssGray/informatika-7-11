### Аргументы и значения по умолчанию

Если параметру задано значение по умолчанию, при вызове его можно не передавать.
Обязательные параметры пишут первыми, со значениями по умолчанию — после них.

```python
def поздороваться(имя, приветствие="Привет"):
    return приветствие + ", " + имя + "!"

print(поздороваться("Аня"))
print(поздороваться("Аня", "Здравствуй"))
```

ноль, одна буква) и неожиданный. Именно так устроены проверки в этом курсе: функция
получает набор пар «вход — ожидаемый ответ», и каждая пара должна совпасть.
