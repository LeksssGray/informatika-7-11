## Разбираемся

### Диапазон и его имя

Диапазон — это часть таблицы, например `B2:B101`. Диапазону можно дать имя, например
`Оценки`, и потом писать формулы с этим именем: `=СРЗНАЧ(Оценки)`. Так формула читается
как текст и не ломается, если диапазон сдвинули.

### Сортировка — только всей строкой

Сортируя один столбец, нельзя оставлять остальные на месте: фамилия окажется
рядом с чужой оценкой. Выделяй **всю таблицу целиком** (заголовок не входит — его
отделяют) и задавай порядок: сначала по классу, затем по фамилии. Так называется
сортировка по нескольким столбцам.

```python
ученики = [("Петров", "8Б"), ("Иванова", "8А"), ("Сидоров", "8А")]
for имя, класс in sorted(ученики, key=lambda s: (s[1], s[0])):
    print(класс, имя)
```

### Фильтр и условное форматирование

**Фильтр** прячет строки, которые не подходят под условие, но не удаляет их.
Снять фильтр — и все строки на месте. **Условное форматирование** красит ячейки по
правилу: оценки ниже трёх — красным, пятёрки — зелёным. Таблица сама следит за правилом,
когда данные меняются.
