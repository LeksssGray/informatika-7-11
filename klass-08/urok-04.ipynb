## Разбираемся

### Четыре направления перевода

Переводов между системами счисления кажется много, но на самом деле
приёмов всего три, и все они уже частично знакомы.

```
                  развёрнутая запись
     любая  ─────────────────────────→  десятичная
     система ←─────────────────────────
               деление на основание

     2 ←─────── группировка ────────→ 8, 16
```

Разберём каждый.

### Приём 1. В десятичную — развёрнутая запись

Умножаем каждую цифру на степень основания и складываем.
Это мы делали на прошлых уроках:

$$2F_{16} = 2 \cdot 16^1 + 15 \cdot 16^0 = 32 + 15 = 47$$

**Схема Горнера** — тот же расчёт, но удобнее для устного счёта и для кода.
Идём слева направо: результат умножаем на основание и прибавляем цифру.

```
  2F₁₆ :   0 · 16 + 2  = 2
           2 · 16 + 15 = 47
```

Для `237₈`:

```
  0 · 8 + 2 = 2
  2 · 8 + 3 = 19
 19 · 8 + 7 = 159
```

Заметьте: никаких степеней вычислять не надо, только умножение
и сложение. Именно так переводит компьютер.

### Приём 2. Из десятичной — деление на основание

Делим число на основание с остатком, пока частное не станет нулём.
Остатки читаем **снизу вверх**.

```
  Переведём 159 в восьмеричную:

  159 : 8 = 19,  остаток 7   ↑
   19 : 8 =  2,  остаток 3   │  читаем
    2 : 8 =  0,  остаток 2   │  вверх

  Ответ: 237₈
```

Для шестнадцатеричной остатки больше девяти записываются буквами:

```
  Переведём 47 в шестнадцатеричную:

  47 : 16 = 2,  остаток 15  →  F
   2 : 16 = 0,  остаток 2   →  2

  Ответ: 2F₁₆
```

### Приём 3. Между 2, 8 и 16 — через группировку

Самый быстрый путь, разбирали на прошлом уроке. **Через десятичную здесь
ходить не нужно** — это лишняя работа и лишний риск ошибиться.

| Переход | Как |
|---|---|
| 2 → 8 | группы по 3 разряда справа |
| 2 → 16 | группы по 4 разряда справа |
| 8 → 2 | каждая цифра → 3 разряда |
| 16 → 2 | каждая цифра → 4 разряда |
| 8 → 16 | через двоичную: 8 → 2 → 16 |

Последняя строка важна: прямого перехода между восьмеричной
и шестнадцатеричной нет, но двоичная работает мостом.

### Как не ошибиться

1. **Группировать всегда справа.** Число `1011100` режется на `1 011 100`,
   а не на `101 110 0`. Слева при необходимости дописываем нули.
2. **Остатки читать снизу вверх.** Самая частая ошибка на контрольных.
3. **Проверять обратным переводом.** Занимает двадцать секунд и ловит
   почти любую ошибку.

## Смотрим, как это работает

### Пример 1. Схема Горнера в коде

````python
def в_десятичную(запись, основание):
    цифры = "0123456789ABCDEF"
    результат = 0
    for знак in запись:
        результат = результат * основание + цифры.index(знак)
        print(f"  берём {знак} → результат {результат}")
    return результат


print("Переводим 237 из восьмеричной:")
print("Итого:", в_десятичную("237", 8))
print()
print("Переводим 2F из шестнадцатеричной:")
print("Итого:", в_десятичную("2F", 16))
````

Обратите внимание, насколько короче получился код по сравнению
с вычислением степеней. Вся суть — в одной строке
`результат = результат * основание + цифра`.

### Пример 2. Деление с остатком

````python
def из_десятичной(число, основание):
    цифры = "0123456789ABCDEF"
    if число == 0:
        return "0"
    результат = ""
    while число > 0:
        остаток = число % основание
        print(f"  {число} : {основание} = {число // основание}, остаток {остаток} → {цифры[остаток]}")
        результат = цифры[остаток] + результат
        число //= основание
    return результат


print("159 в восьмеричную:")
print("Ответ:", из_десятичной(159, 8))
print()
print("47 в шестнадцатеричную:")
print("Ответ:", из_десятичной(47, 16))
````

Строка `результат = цифры[остаток] + результат` приписывает новую цифру
**слева** — это и есть чтение остатков снизу вверх, только выполненное
сразу, без записи в столбик.

### Пример 3. Полный цикл: 8 → 2 → 16

Переведём `755₈` в шестнадцатеричную, пройдя через двоичную.

````python
восьмеричное = "755"

# Шаг 1: каждая цифра → три двоичных разряда
двоичное = ""
for знак in восьмеричное:
    двоичное += f"{int(знак):03b}"
print(f"{восьмеричное}₈ → {двоичное}₂")

# Шаг 2: группируем по четыре, дополнив слева
длина = (len(двоичное) + 3) // 4 * 4
дополненное = двоичное.zfill(длина)
print(f"дополнили до {длина} разрядов: {дополненное}")

шестнадцат = ""
for i in range(0, длина, 4):
    шестнадцат += f"{int(дополненное[i:i + 4], 2):X}"
print(f"{дополненное}₂ → {шестнадцат}₁₆")

# Проверка через десятичную
print(f"\nПроверка: {восьмеричное}₈ = {int(восьмеричное, 8)}₁₀ = {int(шестнадцат, 16)}₁₀")
````

Проверка в конце — обязательная привычка. Оба числа дали 493,
значит перевод верен.
