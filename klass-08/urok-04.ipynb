## Разбираемся

### Три системы, с которыми работает компьютер

Внутри процессора всё двоичное. Но человеку читать длинные цепочки нулей
и единиц тяжело: адрес памяти `1010110011110000` не запомнить и не проверить
глазом. Поэтому программисты пользуются ещё двумя системами — они короче,
но переводятся в двоичную мгновенно.

| Система | Основание | Алфавит | Зачем нужна |
|---|---|---|---|
| двоичная | 2 | 0, 1 | язык самой машины |
| восьмеричная | 8 | 0 … 7 | компактная запись, права доступа в файлах |
| шестнадцатеричная | 16 | 0 … 9, A … F | адреса памяти, цвета, коды символов |

### Почему именно 8 и 16

Причина в одной строчке: **8 = 2³, а 16 = 2⁴**.

Из этого следует замечательное свойство. Одна восьмеричная цифра — это
ровно **три** двоичных разряда, а одна шестнадцатеричная — ровно **четыре**.
Никаких вычислений, только группировка.

```
  двоичное:      1 0 1 1 1 1 0 0
  по 4 разряда:  1011      1100
  в 16-ричной:     B         C     →  BC₁₆
```

С восьмеричной так же, только группы по три.

Если бы основание не было степенью двойки — скажем, десятичная —
такой фокус не прошёл бы: пришлось бы делить и умножать.

### Шестнадцатеричные цифры

В шестнадцатеричной системе шестнадцать цифр, а привычных знаков только
десять. Недостающие шесть взяли из латинского алфавита:

| Знак | A | B | C | D | E | F |
|---|---|---|---|---|---|---|
| Значение | 10 | 11 | 12 | 13 | 14 | 15 |

Регистр значения не имеет: `2f` и `2F` — одно и то же.

### Таблица соответствия

Эту таблицу стоит держать в голове — она закрывает почти все задачи.

| 10 | 2 | 8 | 16 |   | 10 | 2 | 8 | 16 |
|---|---|---|---|---|---|---|---|---|
| 0 | 0000 | 0 | 0 | | 8 | 1000 | 10 | 8 |
| 1 | 0001 | 1 | 1 | | 9 | 1001 | 11 | 9 |
| 2 | 0010 | 2 | 2 | | 10 | 1010 | 12 | A |
| 3 | 0011 | 3 | 3 | | 11 | 1011 | 13 | B |
| 4 | 0100 | 4 | 4 | | 12 | 1100 | 14 | C |
| 5 | 0101 | 5 | 5 | | 13 | 1101 | 15 | D |
| 6 | 0110 | 6 | 6 | | 14 | 1110 | 16 | E |
| 7 | 0111 | 7 | 7 | | 15 | 1111 | 17 | F |

Обратите внимание на строку с числом 8: в восьмеричной оно записывается
как `10`. Это не «десять» — это «одна восьмёрка и ноль единиц». Такая же
логика, как с числом 10 в десятичной системе.

### Где вы это уже видели

Шестнадцатеричные числа окружают вас, даже если вы не программист.

* **Цвета в вебе.** `#FF0000` — красный: FF красного, 00 зелёного,
  00 синего. Каждая пара — один байт, от 00 до FF, то есть от 0 до 255.
* **MAC-адрес** сетевой карты: `A4:5E:60:C1:0F:2B`.
* **Коды символов Unicode**: буква «А» — это U+0410.

### Права доступа: восьмеричная в действии

В операционных системах права на файл записывают тремя восьмеричными
цифрами: для владельца, группы и всех остальных. Каждая цифра — три бита:
чтение (4), запись (2), выполнение (1).

`755` означает `111 101 101`: владелец может всё, остальные — читать
и выполнять. Восьмеричная система здесь не случайность, а прямое следствие
того, что прав ровно три.

## Смотрим, как это работает

### Пример 1. Таблица соответствия

Построим таблицу сами — это лучший способ её запомнить.

````python
print(f"{'10':>4} {'2':>10} {'8':>6} {'16':>4}")
print("-" * 28)
for число in range(17):
    print(f"{число:>4} {число:>10b} {число:>6o} {число:>4X}")
````

Здесь использована возможность f-строк, о которой мы ещё не говорили:
буква после двоеточия задаёт систему счисления. `b` — двоичная,
`o` — восьмеричная, `X` — шестнадцатеричная заглавными буквами.
Это короче, чем `bin(число)[2:]`, и сразу выравнивает столбцы.

### Пример 2. Группировка по три и по четыре

Главный практический приём урока: перевод между 2, 8 и 16
**без вычислений**.

````python
двоичное = "10111100"

# Дополняем нулями слева до кратности 4 и режем на группы
дополненное = двоичное.zfill(8)
группы4 = []
for i in range(0, len(дополненное), 4):
    группы4.append(дополненное[i:i + 4])

шестнадцатеричные = ""
for группа in группы4:
    шестнадцатеричные += f"{int(группа, 2):X} "

print("Двоичное:      ", двоичное)
print("Группы по 4:   ", " ".join(группы4))
print("Шестнадцат.:   ", шестнадцатеричные)
````

Метод `zfill` дополняет строку нулями слева до нужной длины —
без этого последняя группа получилась бы неполной и результат был бы
неверным. Дополнять всегда нужно **слева**: ведущие нули значение
не меняют, а вот приписанные справа умножили бы число.

### Пример 3. Цвет как три байта

Разберём, как устроен цвет в вебе, и соберём свой.

````python
def цвет(красный, зелёный, синий):
    return f"#{красный:02X}{зелёный:02X}{синий:02X}"


print("Красный:      ", цвет(255, 0, 0))
print("Белый:        ", цвет(255, 255, 255))
print("Тёмно-синий:  ", цвет(0, 0, 139))
print("Серый:        ", цвет(128, 128, 128))
````

Формат `:02X` означает: шестнадцатеричная запись заглавными буквами,
дополненная нулями до двух знаков. Без `02` число 0 записалось бы
как `0`, а не `00`, и код цвета сломался бы.
