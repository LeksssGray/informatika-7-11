## Разбираемся

### Шаблоны, из которых собрано всё

Задач на циклы бесконечно много, а приёмов — пять. Выучив их
как таблицу умножения, почти любую задачу ОГЭ можно собрать
из готовых кусков.

| Что нужно | Заготовка до цикла | Что делать внутри |
|---|---|---|
| Сумма | `s = 0` | `s += x` |
| Произведение | `p = 1` | `p *= x` |
| Счётчик | `k = 0` | `if условие: k += 1` |
| Максимум | `m = первый` | `if x > m: m = x` |
| Минимум | `m = первый` | `if x < m: m = x` |

Обратите внимание на первую колонку: **с чего начинается копилка**.
Это самое частое место ошибки. Сумма — с нуля, произведение —
с единицы, максимум — с первого элемента, а не с нуля.

### Почему максимум нельзя начинать с нуля

```python
m = 0
for x in [-5, -3, -9]:
    if x > m:
        m = x
```

Ни одно отрицательное число не больше нуля, поэтому в `m` так
и останется ноль — значение, которого в данных не было вовсе.

Правильных вариантов два: начать с первого элемента
или с заведомо маленького числа вроде `-10 ** 9`. Первый способ
честнее.

### Сумма и количество вместе

Если задача спрашивает «сколько чисел больше 10 и какова
их сумма» — не нужно два цикла. Один проход, два накопителя:

```python
k = 0
s = 0
for x in числа:
    if x > 10:
        k += 1
        s += x
```

### Среднее только подходящих

Ловушка экзамена: «найдите среднее арифметическое чётных чисел».
Делить нужно не на общее количество, а на количество **чётных** —
то есть на счётчик, а не на длину списка.

И ещё: если подходящих чисел не нашлось, деление на ноль обрушит
программу. Поэтому счётчик проверяют перед делением.

### Обход по индексу

Когда нужен не только элемент, но и его номер, идут по `range(len(...))`:

```python
for i in range(len(числа)):
    if числа[i] > числа[лучший]:
        лучший = i
```

Так находят **позицию** максимума, а не само значение.

### Сколько шагов делает цикл

Вопрос из каждого варианта ОГЭ. В `range(a, b, шаг)` количество
повторов считается так: берём разницу `b - a`, делим на шаг
и округляем вверх.

`range(2, 20, 3)` → числа 2, 5, 8, 11, 14, 17 — шесть повторов.
Двадцатка не входит: правая граница никогда не включается.

## Смотрим, как это работает

### Пять шаблонов на одном списке

````python
числа = [7, 3, 12, 5, 20, 8]

сумма = 0
произведение = 1
счётчик = 0
максимум = числа[0]
минимум = числа[0]

for x in числа:
    сумма += x
    произведение *= x
    if x > 10:
        счётчик += 1
    if x > максимум:
        максимум = x
    if x < минимум:
        минимум = x

print("Сумма:", сумма)
print("Произведение:", произведение)
print("Больше десяти:", счётчик)
print("Максимум:", максимум)
print("Минимум:", минимум)
````

Один проход по списку — пять ответов. Компьютер посмотрел на каждое
число ровно один раз.

### Максимум, начатый с нуля

````python
числа = [-5, -3, -9]

неверно = 0
for x in числа:
    if x > неверно:
        неверно = x

верно = числа[0]
for x in числа:
    if x > верно:
        верно = x

print("Начали с нуля:", неверно, "— такого числа в списке нет")
print("Начали с первого элемента:", верно)
````

### Среднее только чётных

````python
числа = [7, 3, 12, 5, 20, 8]

сумма = 0
сколько = 0

for x in числа:
    if x % 2 == 0:
        сумма += x
        сколько += 1

if сколько > 0:
    print(f"Чётных: {сколько}, их среднее: {сумма / сколько}")
else:
    print("Чётных чисел нет")
````

### Позиция максимума

````python
числа = [7, 3, 12, 5, 20, 8]

лучший = 0
for i in range(len(числа)):
    if числа[i] > числа[лучший]:
        лучший = i

print(f"Максимум {числа[лучший]} стоит на позиции {лучший}")
````

### Сколько повторов

````python
for диапазон in [(1, 10, 1), (2, 20, 3), (10, 0, -2)]:
    a, b, шаг = диапазон
    повторов = 0
    for i in range(a, b, шаг):
        повторов += 1
    print(f"range({a}, {b}, {шаг}) → {повторов} повторов")
````
