## Разбираемся

### Зачем упрощать

Длинное условие в программе трудно читать и легко испортить. А ещё
каждая лишняя операция в схеме процессора — это лишние транзисторы,
лишнее тепло и лишние деньги. Поэтому логические выражения упрощают:
заменяют на **равносильные**, то есть дающие ту же таблицу истинности,
но короче.

Делают это по **законам алгебры логики**. Многие из них похожи
на законы обычной алгебры, но есть и неожиданные.

### Законы

| Закон | Для И | Для ИЛИ |
|---|---|---|
| переместительный | A ∧ B = B ∧ A | A ∨ B = B ∨ A |
| сочетательный | (A ∧ B) ∧ C = A ∧ (B ∧ C) | (A ∨ B) ∨ C = A ∨ (B ∨ C) |
| распределительный | A ∧ (B ∨ C) = (A ∧ B) ∨ (A ∧ C) | A ∨ (B ∧ C) = (A ∨ B) ∧ (A ∨ C) |
| идемпотентности | A ∧ A = A | A ∨ A = A |
| поглощения | A ∧ (A ∨ B) = A | A ∨ (A ∧ B) = A |
| с константами | A ∧ 1 = A, A ∧ 0 = 0 | A ∨ 0 = A, A ∨ 1 = 1 |
| противоречия и исключённого третьего | A ∧ ¬A = 0 | A ∨ ¬A = 1 |

И два закона без пары:

* **двойного отрицания**: ¬¬A = A;
* **де Моргана**: ¬(A ∧ B) = ¬A ∨ ¬B и ¬(A ∨ B) = ¬A ∧ ¬B.

Обратите внимание на второй распределительный закон: в обычной алгебре
`a + b·c` не равно `(a + b)·(a + c)`, а в логике — равно.

### Законы де Моргана словами

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g08/demorgan.jpg" width="200" alt="Огастес де Морган">

*Огастес де Морган*

<sub>из книги С. де Морган, 1882 · общественное достояние · Wikimedia Commons</sub>

Английский математик Огастес де Морган, современник и друг Буля,
сформулировал правило, которым пользуются чаще всех остальных.
Отрицание «раскрывают»: каждое высказывание отрицается, а И меняется
на ИЛИ, и наоборот.

«Неверно, что на улице тепло **и** солнечно» — это то же самое, что
«на улице **не** тепло **или** **не** солнечно».

В коде это встречается постоянно: `not (a and b)` можно заменить
на `not a or not b`.

### Как упрощают

Упростим **(A ∧ B) ∨ (A ∧ ¬B)**.

```
  (A ∧ B) ∨ (A ∧ ¬B)
= A ∧ (B ∨ ¬B)        распределительный закон, выносим A
= A ∧ 1               исключённое третье
= A                   закон с константой
```

Длинное выражение оказалось просто A. Если сомневаетесь в результате —
сравните таблицы истинности до и после.

Ещё пример: **¬(¬A ∨ ¬B)**.

```
  ¬(¬A ∨ ¬B)
= ¬¬A ∧ ¬¬B           де Морган
= A ∧ B               двойное отрицание
```

## Смотрим, как это работает

### Пример 1. Проверяем закон перебором

Доказать тождество — значит показать, что обе части совпадают
**на всех наборах**. Для двух переменных их всего четыре,
и компьютер переберёт их мгновенно.

````python
def левая(a, b):
    return not (a and b)


def правая(a, b):
    return (not a) or (not b)


for a in (False, True):
    for b in (False, True):
        print(a, b, "→", левая(a, b), правая(a, b))
````

Столбцы совпали на всех четырёх строках — закон де Моргана верен.

### Пример 2. Функция, которая доказывает сама

Функции в Python можно передавать в другие функции как обычные
значения. Напишем проверку, которая принимает две функции-выражения
и отвечает, равносильны ли они.

````python
def равносильны(f, g):
    for a in (False, True):
        for b in (False, True):
            if f(a, b) != g(a, b):
                return False
    return True


def длинное(a, b):
    return (a and b) or (a and not b)


def короткое(a, b):
    return a


print("Упрощение верно:  ", равносильны(длинное, короткое))
print("Де Морган верен:  ", равносильны(левая, правая))
````

Если хоть на одном наборе значения разошлись, функция сразу возвращает
`False`. Дошла до конца — значит, выражения равносильны.
