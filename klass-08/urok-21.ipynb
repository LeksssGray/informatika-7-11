## Разбираемся

### Когда повторов не сосчитать заранее

`for` берут, когда известно **сколько раз**. `while` — когда
известно только **до каких пор**.

Признак задачи для `while` простой: в условии есть слово «пока».
«Пока число не станет однозначным», «пока остаток не обнулится»,
«пока пользователь не введёт верный пароль».

```python
while условие:
    тело
```

И три обязательные части, о которых мы говорили: начальное
значение до цикла, условие в заголовке, изменение внутри.
Нет третьей — цикл не кончится никогда.

### Работа с цифрами числа

Главное применение `while` в школьной информатике. Два действия
разбирают число по цифрам:

| Действие | Что даёт | Пример для 4821 |
|---|---|---|
| `n % 10` | последнюю цифру | 1 |
| `n // 10` | число без последней цифры | 482 |

Шаблон выглядит так и почти не меняется от задачи к задаче:

```python
while n > 0:
    цифра = n % 10
    # что-то делаем с цифрой
    n = n // 10
```

Меняется только середина: складываем цифры — получится сумма цифр,
считаем шаги — количество цифр, приписываем в конец — переворот
числа.

> Осторожно с нулём: `while n > 0` для `n = 0` не выполнится
> ни разу. Если по условию «количество цифр числа 0» должно быть
> единицей, этот случай обрабатывают отдельно.

### Переворот числа

```python
перевёрнутое = 0
while n > 0:
    перевёрнутое = перевёрнутое * 10 + n % 10
    n = n // 10
```

Каждый шаг сдвигает накопленное на разряд влево (умножение на 10)
и дописывает очередную цифру. Для 123: 3, затем 32, затем 321.

### Алгоритм Евклида

Самый древний алгоритм, дошедший до нас, — ему больше двух тысяч
лет. Он находит наибольший общий делитель двух чисел.

Идея: НОД не меняется, если большее число заменить остатком
от деления на меньшее. Повторяем, пока одно из чисел не станет
нулём — второе и будет ответом.

```python
while b != 0:
    a, b = b, a % b
return a
```

Запись `a, b = b, a % b` меняет обе переменные сразу, причём
справа используются **старые** значения. Без такой записи
понадобилась бы третья переменная.

Для 48 и 18: (48, 18) → (18, 12) → (12, 6) → (6, 0). Ответ 6.
Три шага вместо перебора всех делителей.

### Защита от зацикливания

Пока алгоритм не проверен, к условию полезно добавить ограничитель:

```python
шагов = 0
while условие and шагов < 1000:
    ...
    шагов += 1
```

Если программа упрётся в ограничитель — значит, в логике ошибка,
и это гораздо лучше, чем зависший ноутбук посреди урока.

## Смотрим, как это работает

### Разбираем число по цифрам

````python
n = 4821

while n > 0:
    цифра = n % 10
    n = n // 10
    print(f"цифра {цифра}, осталось {n}")
````

Цифры выходят **с конца** — это особенность приёма, и её надо
помнить, когда порядок важен.

### Сумма и количество цифр сразу

````python
n = 90210
сумма = 0
количество = 0

while n > 0:
    сумма += n % 10
    количество += 1
    n = n // 10

print("Сумма цифр:", сумма)
print("Количество цифр:", количество)
````

### Переворот

````python
n = 12345
перевёрнутое = 0

while n > 0:
    перевёрнутое = перевёрнутое * 10 + n % 10
    n = n // 10
    print("накопили:", перевёрнутое)
````

### Алгоритм Евклида

````python
a = 48
b = 18

while b != 0:
    print(f"({a}, {b})")
    a, b = b, a % b

print("НОД:", a)
````

### Сколько шагов до единицы

````python
x = 100
шагов = 0

while x > 1:
    x = x // 3
    шагов += 1
    print(f"шаг {шагов}: {x}")
````

### Ограничитель в деле

Алгоритм неверный — он никогда не дойдёт до единицы. Ограничитель
не даёт ячейке зависнуть.

````python
x = 10
шагов = 0

while x != 1 and шагов < 8:
    x -= 2
    шагов += 1

print("Остановились на", x, "после", шагов, "шагов")
print("Единицы не будет: из чётного вычитаем чётное")
````
