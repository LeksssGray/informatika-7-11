## Разбираемся

### Три способа записать алгоритм

Один и тот же алгоритм можно записать:

* **словами** — как рецепт; понятно людям, но легко написать
  двусмысленно;
* **блок-схемой** — картинкой из фигур и стрелок; видно всю логику сразу;
* **программой** — на языке, который понимает компьютер.

Блок-схему удобно рисовать, когда алгоритм ещё не написан: на ней
хорошо видно, где развилка и где повтор, а ошибку в стрелке найти
проще, чем в коде.

*Блок-схема «Светильник не работает»*

Это знаменитая шуточная блок-схема: что делать, если не горит лампа.
Ромбы — вопросы, прямоугольники — действия, стрелки показывают, куда
идти дальше. Такой алгоритм выполнит кто угодно, даже не разбираясь
в электрике.

### Фигуры блок-схем

| Фигура | Что означает | Пример |
|---|---|---|
| овал (скруглённый прямоугольник) | начало и конец | «начало», «конец» |
| прямоугольник | действие | `s = s + i` |
| ромб | условие, развилка | `a > b?` — выходы «да» и «нет» |
| параллелограмм | ввод или вывод данных | «ввести n», «вывести s» |

Внутри ромба — вопрос, на который можно ответить только «да» или «нет».

### Три базовые конструкции

Любой алгоритм собирается из трёх кирпичиков.

**Следование** — действия идут друг за другом:

```
     ┌──────────┐
     │ действие1│
     └────┬─────┘
          ▼
     ┌──────────┐
     │ действие2│
     └──────────┘
```

**Ветвление** — в зависимости от условия выполняется одна из веток,
потом ветки снова сходятся:

```
            ╱ ╲
      да   ╱усл╲   нет
    ┌─────╲ ?  ╱─────┐
    ▼      ╲ ╱       ▼
 ┌──────┐        ┌──────┐
 │ветка1│        │ветка2│
 └──┬───┘        └───┬──┘
    └───────┬────────┘
            ▼
```

**Цикл** — пока условие верно, тело повторяется, и стрелка
возвращается к условию:

```
        ┌────────────────┐
        ▼                │
       ╱ ╲               │
      ╱усл╲  да   ┌──────┴─┐
      ╲ ? ╱──────►│  тело  │
       ╲ ╱        └────────┘
        │ нет
        ▼
```

### От блок-схемы к программе

Каждая конструкция переводится в Python напрямую:

| Блок-схема | Python |
|---|---|
| прямоугольник | присваивание или вызов функции |
| ромб с двумя ветками | `if … else` |
| ромб со стрелкой назад | `while` |
| параллелограмм | `input()` или `print()`, в функциях — параметры и `return` |

## Смотрим, как это работает

### Пример 1. Ветвление: большее из двух

```
  начало → ввести a, b → a > b? ─ да ─→ вывести a → конец
                             └─ нет ─→ вывести b → конец
```

````python
def большее(a, b):
    if a > b:
        return a
    else:
        return b


print(большее(7, 3), большее(2, 9))
````

### Пример 2. Цикл: сумма чисел от 1 до n

```
  начало → ввести n → s = 0, i = 1 → i ≤ n? ─ да ─→ s = s + i, i = i + 1 ─┐
                                       ▲                                  │
                                       └──────────────────────────────────┘
                                       └─ нет ─→ вывести s → конец
```

````python
def сумма_до(n):
    s = 0
    i = 1
    while i <= n:
        s = s + i
        i = i + 1
    return s


print(сумма_до(10))
````

Стрелка назад на схеме превратилась в `while`, прямоугольник
в теле — в две строки внутри цикла.

### Пример 3. Прокрутка вручную

В экзаменационных задачах часто просят выполнить блок-схему
«на бумаге». Удобно записывать значения переменных после каждого шага.

```
  x = 5 → x < 50? ─ да ─→ x = x * 2 ─┐
           ▲                          │
           └──────────────────────────┘
           └─ нет ─→ вывести x
```

| Проверка | x < 50? | x после шага |
|:-:|:-:|:-:|
| 1 | 5 < 50 — да | 10 |
| 2 | 10 < 50 — да | 20 |
| 3 | 20 < 50 — да | 40 |
| 4 | 40 < 50 — да | 80 |
| 5 | 80 < 50 — нет | выводим 80 |

````python
x = 5
шагов = 0
while x < 50:
    x = x * 2
    шагов += 1

print("x =", x, "| повторов:", шагов)
````
