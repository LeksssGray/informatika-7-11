## Разбираемся

### Функции с условием

* `=ЕСЛИ(условие; что_если_да; что_если_нет)` — выбирает значение по условию.
* `=СУММЕСЛИ(диапазон; условие; что_суммировать)` — складывает только подходящие значения.
  Если третий аргумент не указан, складываются сами значения диапазона.
* `=СЧЁТЕСЛИ(диапазон; условие)` — считает подходящие ячейки.

Условие записывается в кавычках, если это сравнение: `">20"`, `"=3"`.

### Вложенные условия

Если вариантов больше двух, ЕСЛИ вкладывают внутрь ЕСЛИ. Оценку по баллам удобно
записать так: `=ЕСЛИ(A1>=90; "5"; ЕСЛИ(A1>=70; "4"; "3"))`.

````python
баллы = [95, 72, 60, 88]
итоги = []
for балл in баллы:
    if балл >= 90:
        итоги.append("5")
    elif балл >= 70:
        итоги.append("4")
    else:
        итоги.append("3")
print(итоги)
````

или в ссылке.
