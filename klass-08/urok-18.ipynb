### Типы данных: три главных

| Тип | Что хранит | Пример |
|---|---|---|
| `int` | целое число | `42`, `-7`, `0` |
| `float` | дробное число | `3.14`, `-0.5`, `2.0` |
| `str` | строку | `"привет"`, `"42"` |

Тип видно функцией `type`, и это первое, что стоит проверить,
когда программа ведёт себя странно:

```python
type(42)        # int
type(42.0)      # float
type("42")      # str
```

Обрати внимание на третью строку: `"42"` — это **строка**, а не
число, хотя выглядит как число. Складывать её с числом нельзя.

### Деление всегда даёт дробь

```python
10 / 2      # 5.0 — float, даже если делится нацело
10 // 3     # 3   — целая часть
10 % 3      # 1   — остаток
```

Это ловушка: `10 / 2` возвращает `5.0`, а не `5`. Если по условию
нужно целое — берите `//`.

### Ввод данных

`input()` всегда возвращает **строку**, даже если ввели число:

```python
n = input("Введите число: ")
n = int(n)                      # обязательное преобразование
```

Обычно это пишут в одну строку:

```python
n = int(input("Введите число: "))
x = float(input("Введите дробное: "))
```

### Несколько значений в одной строке

На олимпиадах и в задачниках данные часто вводят так: `3 5 7` —
три числа через пробел. Разбирает такую строку метод `.split()`:

```python
строка = "3 5 7"
части = строка.split()          # ["3", "5", "7"]
```

`split` режет строку по пробелам и возвращает **список строк**.
Чтобы получить числа, каждую часть надо преобразовать:

```python
a = int(части[0])
b = int(части[1])
```

У `split` есть и другая форма: `"5,7,9".split(",")` разрежет
по запятой. Это пригодится, когда данные придут из таблицы.

### Форматированный вывод

f-строка подставляет значения прямо в текст, и внутри неё можно
считать:

```python
a = 7
b = 3
print(f"{a} + {b} = {a + b}")
```

Для дробей указывают, сколько знаков оставить:

```python
x = 3.14159
print(f"{x:.2f}")      # 3.14
```

Запись `:.2f` читается как «вывести с двумя знаками после точки».
Округление до двух знаков делает и `round(x, 2)` — разница в том,
что `round` возвращает число, а f-строка сразу готовый текст.

## Смотрим, как это работает

### Типы и преобразования

````python
print(type(42))
print(type(42.0))
print(type("42"))

print()
print("Строка в число:", int("42") + 8)
print("Число в строку:", "ответ: " + str(42))
````

### Три вида деления

````python
a = 17
b = 5

print(f"{a} / {b}  = {a / b}")
print(f"{a} // {b} = {a // b}")
print(f"{a} % {b}  = {a % b}")

print()
print("Проверка:", (a // b) * b + (a % b), "— снова", a)
````

Последняя строка — полезное тождество: целая часть, умноженная
на делитель, плюс остаток даёт исходное число. На нём построены
все задачи «сколько коробок и сколько останется».

### Разбор строки на числа

````python
данные = "3 5 7"
части = данные.split()

print("Части:", части)
print("Тип части:", type(части[0]))

a = int(части[0])
b = int(части[1])
c = int(части[2])

print("Сумма:", a + b + c)
````

### Другой разделитель

````python
из_таблицы = "12,8,30,5"
числа = из_таблицы.split(",")

print(числа)

сумма = 0
for кусок in числа:
    сумма += int(кусок)

print("Сумма:", сумма)
````

### Красивый вывод

````python
имя = "Аня"
рост = 1.6234
оценка = 4.6666

print(f"{имя}: рост {рост:.2f} м, средний балл {оценка:.1f}")
print(f"Без форматирования: {рост} и {оценка}")
````

### Любопытно

Дробные числа компьютер хранит приближённо, и это видно
невооружённым глазом:

```python
print(0.1 + 0.2)      # 0.30000000000000004
```

Ошибки нет: 0,1 в двоичной системе — бесконечная дробь, как 1/3
в десятичной. Её приходится обрезать, и крошечная погрешность
всплывает в шестнадцатом знаке.

Поэтому дробные числа никогда не сравнивают на точное равенство,
а деньги в банковских программах считают **в копейках целыми
числами** — чтобы ни одна копейка не потерялась в шестнадцатом знаке.
