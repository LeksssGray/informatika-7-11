### Способ 1. Рассуждение «от противного»

> Разбили окно. Аня говорит: «Это Боря». Боря: «Это не я». Вова: «Это
> не я». Известно, что правду сказал только один. Кто разбил?

Предполагаем по очереди каждого виновником и проверяем, сходится ли
условие «правду сказал только один».

| Если разбил… | Аня «это Боря» | Боря «не я» | Вова «не я» | Правдивых |
|---|:-:|:-:|:-:|:-:|
| Аня | ложь | правда | правда | 2 |
| Боря | правда | ложь | правда | 2 |
| Вова | ложь | правда | ложь | **1** |

Условию подходит только третья строка: разбил Вова.

### Способ 2. Таблица

> Маша, Катя и Лена держат кошку, собаку и попугая — у каждой свой
> питомец. У Маши не кошка и не попугай. Катя не держит птиц.
> У кого попугай?

Рисуем таблицу «человек × питомец» и ставим минусы по условию.
Где в строке остался один пустой квадратик — там плюс, и этот питомец
вычёркивается у остальных.

| | кошка | собака | попугай |
|---|:-:|:-:|:-:|
| Маша | − | **+** | − |
| Катя | **+** | − | − |
| Лена | − | − | **+** |

У Маши остаётся только собака. У Кати — не попугай и не собака, значит
кошка. Лене достаётся попугай.

### Рыцари и лжецы

Классика логических задач: на острове живут **рыцари**, которые всегда
говорят правду, и **лжецы**, которые всегда лгут. Встречаете двоих,
A и B.

> A говорит: «Мы оба лжецы». Кто есть кто?

Если A рыцарь, он сказал правду — значит, он лжец. Противоречие.
Значит, A лжец, и его слова ложны: неверно, что они **оба** лжецы.
Раз A лжец, то B — рыцарь.

## Смотрим, как это работает

### Пример 1. Кто разбил окно — перебором

```python
for виновник in ("Аня", "Боря", "Вова"):
    аня = виновник == "Боря"          # Аня: «Это Боря»
    боря = виновник != "Боря"         # Боря: «Это не я»
    вова = виновник != "Вова"         # Вова: «Это не я»
    правдивых = int(аня) + int(боря) + int(вова)
    print(виновник, "— правдивых:", правдивых)
```

Каждое утверждение — логическое выражение, зависящее от того, кто
виноват. `int(True)` равно 1, поэтому сумма и есть число правдивых.

### Пример 2. Питомцы — перебором

Переберём, какой питомец у Маши и у Кати; Лене достанется оставшийся.

```python
for маша in ("кошка", "собака", "попугай"):
    for катя in ("кошка", "собака", "попугай"):
        if катя == маша:
            continue                          # у всех разные питомцы
        if маша == "кошка" or маша == "попугай":
            continue                          # у Маши не кошка и не попугай
        if катя == "попугай":
            continue                          # Катя не держит птиц
        print("Маша:", маша, "| Катя:", катя)
```

Подошёл один вариант: Маша — собака, Катя — кошка, значит Лена — попугай.
`continue` пропускает остаток тела цикла и переходит к следующему варианту.

### Пример 3. Рыцари и лжецы — перебором

```python
for a_рыцарь in (True, False):
    for b_рыцарь in (True, False):
        слова_a = (not a_рыцарь) and (not b_рыцарь)   # «Мы оба лжецы»
        # рыцарь говорит правду, лжец — ложь
        if слова_a == a_рыцарь:
            print("A рыцарь:", a_рыцарь, "| B рыцарь:", b_рыцарь)
```

Главная строчка — `слова_a == a_рыцарь`: утверждение истинно ровно тогда,
когда говорящий рыцарь. Это та самая эквиваленция с прошлого урока.
