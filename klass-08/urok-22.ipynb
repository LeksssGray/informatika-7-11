## Разбираемся

### Второй инструмент

Пять уроков подряд мы писали код. Сегодня — пауза и другой
инструмент: электронные таблицы.

Это не «упрощённое программирование для тех, кто не осилил Python».
Это другой способ считать, и у него своя сильная сторона: **данные
и вычисления видно одновременно**. В программе данные спрятаны
в переменных, а результат печатается один раз. В таблице всё
на виду, и любое исправление мгновенно пересчитывает остальное.

Поэтому бухгалтерия, школьный журнал, смета ремонта и бюджет
поездки живут в таблицах, а не в коде.

### Адреса и диапазоны

Столбцы называются буквами, строки — числами. Ячейка на пересечении
столбца C и строки 4 называется **C4** — сначала столбец, потом строка.

Несколько ячеек подряд — **диапазон**, записывается через двоеточие:

| Запись | Что означает |
|---|---|
| `A1` | одна ячейка |
| `A1:A10` | столбец, десять ячеек |
| `A1:C1` | строка, три ячейки |
| `A1:C10` | прямоугольник, тридцать ячеек |

### Формула

Всё, что начинается со знака **=**, таблица считает формулой,
а не текстом:

```
=A1+B1
=(A1+A2)/2
=A1*0.13
```

В формуле пишут не числа, а **адреса**. Это главное правило: если
вписать `=5+7`, таблица посчитает один раз и навсегда, а если
`=A1+B1` — пересчитает при любом изменении данных.

### Функции

У таблиц есть готовые функции — ровно те алгоритмы, которые мы
писали циклами.

| Функция | Что делает | Наш цикл |
|---|---|---|
| `=СУММ(A1:A10)` | сумма диапазона | копилка `+= x` |
| `=СРЗНАЧ(A1:A10)` | среднее | сумма делить на количество |
| `=МИН(...)`, `=МАКС(...)` | минимум и максимум | сравнение с лучшим |
| `=СЧЁТ(A1:A10)` | сколько чисел | счётчик |
| `=СЧЁТЕСЛИ(A1:A10; ">3")` | сколько подходят под условие | счётчик с `if` |
| `=ЕСЛИ(A1>=3; "зачёт"; "нет")` | ветвление | `if ... else` |

Обратите внимание на последнюю строку: `ЕСЛИ` — это тот же `if`,
только записанный в одну строку. Сначала условие, потом что вернуть
при истине, потом при лжи.

> В русской версии аргументы разделяются **точкой с запятой**,
> в английской — запятой. Если формула ругается, первым делом
> проверьте разделитель.

### Копирование формулы

Самая сильная вещь в таблицах и самая частая ошибка на экзамене.

Формулу пишут один раз и растягивают на весь столбец. При этом
адреса внутри **сдвигаются вместе с формулой**:

```
в C1:  =A1+B1
копируем в C2 →  =A2+B2
копируем в C3 →  =A3+B3
```

Такие ссылки называются **относительными**: таблица запоминает
не «ячейку A1», а «ячейку на два столбца левее».

Иногда это мешает. Если в E1 лежит курс доллара и все формулы
должны смотреть именно туда, ссылку **закрепляют** знаком доллара:

| Запись | Что происходит при копировании |
|---|---|
| `A1` | меняются и столбец, и строка |
| `$A1` | столбец закреплён, строка меняется |
| `A$1` | строка закреплена, столбец меняется |
| `$A$1` | не меняется ничего |

Правило простое: **доллар держит то, что стоит за ним**.

## Смотрим, как это работает

### Таблица глазами программиста

Таблицу можно представить списком списков — как картинку
в седьмом классе. Тогда видно, что именно делают её функции.

````python
# Журнал: фамилия и три оценки
журнал = [
    ["Иванов", 5, 4, 5],
    ["Петрова", 4, 4, 3],
    ["Сидоров", 3, 3, 4],
    ["Кузнецова", 5, 5, 5],
]

for строка in журнал:
    оценки = строка[1:]
    среднее = sum(оценки) / len(оценки)
    print(f"{строка[0]:<12} {оценки}  среднее {среднее:.2f}")
````

В таблице то же самое пишется одной формулой `=СРЗНАЧ(B2:D2)`
и растягивается на весь столбец. Цикл делает таблица.

### СЧЁТЕСЛИ своими руками

````python
оценки = [5, 4, 3, 5, 4, 5, 2, 4]

пятёрок = 0
for оценка in оценки:
    if оценка == 5:
        пятёрок += 1

print("Пятёрок:", пятёрок)
print("В таблице это было бы: =СЧЁТЕСЛИ(B2:B9; 5)")
````

### ЕСЛИ своими руками

````python
for балл in [2, 3, 5]:
    if балл >= 3:
        итог = "зачёт"
    else:
        итог = "незачёт"
    print(f"{балл} → {итог}")

print()
print('В таблице: =ЕСЛИ(A1>=3; "зачёт"; "незачёт")')
````

### Что происходит при копировании

Покажем на бумаге, во что превращается формула `=A1*$E$1`
при растягивании вниз.

````python
for строка in range(1, 5):
    print(f"строка {строка}:  =A{строка}*$E$1")
````

Первая ссылка ползёт за формулой, вторая стоит на месте —
потому что закреплена долларами.

### Одна задача, два инструмента

Посчитаем стоимость покупок с наценкой 15 %.

````python
товары = [
    ["тетрадь", 40, 10],
    ["ручка", 25, 5],
    ["линейка", 60, 2],
]
наценка = 1.15

итого = 0
for название, цена, количество in товары:
    стоимость = цена * количество * наценка
    итого += стоимость
    print(f"{название:<10} {стоимость:>8.2f}")

print(f"{'ИТОГО':<10} {итого:>8.2f}")
````

В таблице то же самое: в столбце D формула `=B2*C2*$F$1`,
растянутая вниз, и `=СУММ(D2:D4)` в самом низу. Наценка лежит
в F1 — поменяли её один раз, и пересчиталось всё.

Вот ответ на вопрос «что выбрать»: для расчёта, который смотрят
и правят люди, — таблица. Для алгоритма, который нужно повторить
на тысяче файлов, — код.
