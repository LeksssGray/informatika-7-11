### Как проверять свою функцию

Так же, как проверяют задачи в учебнике: заранее выписать случаи
с известным ответом, включая крайние, и сравнить.

```python
def в_двоичную(n):
    if n == 0:
        return "0"
    цифры = ""
    while n > 0:
        цифры = str(n % 2) + цифры
        n //= 2
    return цифры


случаи = [(0, "0"), (1, "1"), (5, "101"), (8, "1000"), (255, "11111111")]
for число, ожидаем in случаи:
    получили = в_двоичную(число)
    отметка = "верно" if получили == ожидаем else "ОШИБКА"
    print(число, "→", получили, "—", отметка)
```

Ноль и 255 — крайние случаи: именно на них функции чаще ломаются.
