## Разбираемся

### Зачем сортировать

На уроке про поиск выяснилось: в отсортированном списке поиск
идёт в тысячи раз быстрее. Отсюда и ответ на вопрос «зачем»:
сортируют один раз, а ищут потом миллион раз.

Плюс сортировка нужна сама по себе: рейтинг игроков, таблица
результатов, список по алфавиту, цены по возрастанию.

Сегодня разберём два классических алгоритма. Оба медленные,
и в реальной программе вы возьмёте встроенный `sorted`, —
но понимать, что внутри, нужно: на этих двух алгоритмах
объясняют саму идею «как машина наводит порядок».

### Сортировка выбором

Идея как при уборке: находим самый маленький элемент и ставим
его в начало. Потом самый маленький из оставшихся — на второе
место. И так далее.

```python
for i in range(len(числа)):
    мин = i
    for j in range(i + 1, len(числа)):
        if числа[j] < числа[мин]:
            мин = j
    числа[i], числа[мин] = числа[мин], числа[i]
```

Внутренний цикл ищет минимум среди оставшихся, последняя строка
меняет два элемента местами.

Запись `a, b = b, a` меняет значения местами без третьей
переменной — Python сначала вычисляет правую часть целиком.
В большинстве языков так нельзя, и там заводят `temp`.

### Сортировка пузырьком

Идея другая: сравниваем **соседей** и меняем их местами, если
они стоят не по порядку. За один проход самый большой элемент
«всплывает» в конец, как пузырёк воздуха.

```python
for i in range(len(числа)):
    for j in range(len(числа) - 1):
        if числа[j] > числа[j + 1]:
            числа[j], числа[j + 1] = числа[j + 1], числа[j]
```

Пузырёк проще запоминается и потому попал во все учебники,
хотя на деле он самый медленный из разумных алгоритмов.

### Сколько это стоит

Оба алгоритма — два вложенных цикла, то есть примерно $n^2$
сравнений. Точнее, $n(n-1)/2$.

| Элементов | Сравнений |
|---|---|
| 5 | 10 |
| 100 | 4 950 |
| 1 000 | ≈ 500 000 |
| 100 000 | ≈ 5 000 000 000 |

Последняя строка — часы работы. Поэтому большие данные так
не сортируют.

### Встроенная сортировка

```python
sorted(числа)                    # новый отсортированный список
sorted(числа, reverse=True)      # по убыванию
числа.sort()                     # отсортировать на месте
```

`sorted` внутри устроен куда хитрее: он делит список пополам,
сортирует части и сливает их обратно. Это даёт примерно
$n \cdot \log n$ сравнений вместо $n^2$ — для миллиона элементов
разница в пятьдесят тысяч раз.

Разбирать, как он устроен, будут в старших классах. Пока
достаточно правила: **свои сортировки пишут на уроке,
в работе берут `sorted`**.

## Смотрим, как это работает

### Обмен значений

````python
a = 1
b = 2

a, b = b, a
print("a =", a, " b =", b)
````

### Сортировка выбором по шагам

````python
числа = [7, 3, 12, 5, 20, 8]
print("Начало:  ", числа)

for i in range(len(числа)):
    мин = i
    for j in range(i + 1, len(числа)):
        if числа[j] < числа[мин]:
            мин = j
    числа[i], числа[мин] = числа[мин], числа[i]
    print(f"шаг {i + 1}:  ", числа)
````

Видно, как отсортированная часть растёт слева: после первого шага
на месте стоит самый маленький элемент, после второго — два
самых маленьких.

### Сортировка пузырьком по проходам

````python
числа = [7, 3, 12, 5, 20, 8]
print("Начало:   ", числа)

for i in range(len(числа)):
    for j in range(len(числа) - 1):
        if числа[j] > числа[j + 1]:
            числа[j], числа[j + 1] = числа[j + 1], числа[j]
    print(f"проход {i + 1}:", числа)
````

После первого прохода двадцатка уже в конце — всплыла.
Заметьте, что дальше список не меняется: он отсортирован,
а алгоритм всё равно делает все проходы. Это можно исправить
флажком «были ли обмены» — попробуйте в домашнем задании.

### Сколько сравнений

````python
for n in [5, 10, 100, 1000]:
    сравнений = n * (n - 1) // 2
    print(f"{n:>5} элементов → {сравнений:>9} сравнений")
````

### Встроенная сортировка

````python
числа = [7, 3, 12, 5, 20, 8]

print("По возрастанию:", sorted(числа))
print("По убыванию:   ", sorted(числа, reverse=True))
print("Исходный цел:  ", числа)

слова = ["груша", "яблоко", "банан"]
print("Строки тоже:   ", sorted(слова))
````

Строки сортируются по алфавиту — точнее, по кодам символов,
которые мы проходили в кодировках. Поэтому все заглавные буквы
окажутся раньше строчных.

### Любопытно

Сортировок придумано несколько десятков, и у них есть свой
чемпион по бесполезности — «сортировка Бого». Алгоритм такой:
перемешать список случайным образом и проверить, не отсортирован
ли он. Не отсортирован — перемешать снова.

Для десяти элементов ей понадобится в среднем три с половиной
миллиона перемешиваний. Для двадцати — больше, чем атомов
в Солнечной системе.

Её придумали не для работы, а для объяснения: на таком фоне
становится понятно, почему «медленный» пузырёк на самом деле
вполне приличный алгоритм.
