## Где обычно ошибаются

Первая четверть 8 класса позади. Разберём ошибки, которые стабильно
встречаются в работах по системам счисления.

### Ошибка 1. Остатки читают сверху вниз

При переводе делением остатки записываются **снизу вверх**. Если прочитать
наоборот, получится перевёрнутое число.

**Профилактика:** сразу после перевода сделайте обратный — двадцать секунд
работы, и ошибка исключена.

### Ошибка 2. Группировка не с той стороны

При переводе 2 → 8 и 2 → 16 группы отсчитываются **справа**.
Число `1011100` режется по три так: `1 011 100`, а не `101 110 0`.
Слева при нехватке дописываются нули: `001 011 100`.

### Ошибка 3. Цифра, равная основанию

В системе с основанием q цифры идут **от 0 до q−1**. Записи `1082₈`
не существует: цифры 8 в восьмеричном алфавите нет. Если такая цифра
появилась в вашем ответе — где-то ошибка в переносе.

### Ошибка 4. Забыт перенос при сложении

Порог переноса равен основанию: в двоичной перенос возникает уже при 2,
в восьмеричной — при 8. Привычка к десятичной подводит: `7 + 5` в
восьмеричной это не 12, а `14₈`.

### Ошибка 5. Путаница диапазонов

| Что спрашивают | Формула |
|---|---|
| сколько чисел помещается в n разрядов | $2^n$ |
| наибольшее беззнаковое | $2^n - 1$ |
| диапазон со знаком | от $-2^{n-1}$ до $2^{n-1} - 1$ |

Помните про минус единицу: комбинаций 256, но наибольшее число 255,
потому что счёт идёт с нуля.

### Ошибка 6. Сравнение вещественных через ==

`0.1 + 0.2 == 0.3` даёт `False`. Всегда сравнивайте через модуль разности.

## Разбор сложных случаев

### Пример 1. Задача на определение основания

> В некоторой системе счисления число 20 записывается как 110.
> Найдите основание системы.

Запишем условие развёрнутой записью. Пусть основание q:

$$110_q = 1 \cdot q^2 + 1 \cdot q^1 + 0 = q^2 + q = 20$$

Получилось квадратное уравнение $q^2 + q - 20 = 0$. Решать его
необязательно — достаточно перебрать основания.

````python
for q in range(2, 17):
    значение = 1 * q ** 2 + 1 * q + 0
    отметка = "  ← подходит!" if значение == 20 else ""
    print(f"  основание {q:>2}: 110 = {значение}{отметка}")
````

Ответ: основание 4. Перебор здесь — не «глупый» метод, а совершенно
законный: вариантов мало, и проверить их быстрее, чем решать уравнение.
На экзамене этот приём экономит много времени.

### Пример 2. Задача на сравнение

> Расположите в порядке возрастания: `1101₂`, `15₁₀`, `E₁₆`, `16₈`.

Приводим всё к десятичной и сортируем.

````python
числа = [("1101", 2), ("15", 10), ("E", 16), ("16", 8)]

значения = []
for запись, основание in числа:
    значение = int(запись, основание)
    значения.append((значение, запись, основание))
    print(f"  {запись}(осн. {основание}) = {значение}")

значения.sort()
print()
строка = ""
for _, з, о in значения:
    строка += f"{з}({о}), "
print("По возрастанию:", строка)
````

Обратите внимание: `13, 14, 14, 15` — два числа оказались равны.
`E₁₆` и `16₈` — это одно и то же значение 14, записанное по-разному.

### Пример 3. Хитрая задача на количество

> Сколько единиц в двоичной записи числа $4^{20} + 2^{15} - 8$?

Такие задачи решаются приведением всех слагаемых к степеням двойки.

$4^{20} = (2^2)^{20} = 2^{40}$, а $8 = 2^3$.

Значит выражение равно $2^{40} + 2^{15} - 2^3$. Проверим программой.

````python
число = 4 ** 20 + 2 ** 15 - 8
запись = bin(число)[2:]

print(f"Число: {число}")
print(f"Разрядов в записи: {len(запись)}")
print(f"Единиц: {запись.count('1')}")
````

Проверять такие задачи кодом — законно и полезно: вы сначала рассуждаете
на бумаге, а потом убеждаетесь, что рассуждение верное. На экзамене
компьютера не будет, но привычка проверять себя останется.
