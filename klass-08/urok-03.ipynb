### Непозиционные системы

В **непозиционной** системе значение знака не зависит от его места.
Самый известный пример — римские числа:

| Знак | I | V | X | L | C | D | M |
|---|---|---|---|---|---|---|---|
| Значение | 1 | 5 | 10 | 50 | 100 | 500 | 1000 |

`XXX` — это 10 + 10 + 10 = 30: каждый `X` означает десять, где бы он ни стоял.

Правило вычитания добавляет исключение: если меньший знак стоит **перед**
большим, он вычитается. `IX` = 10 − 1 = 9, `XL` = 50 − 10 = 40.

Римская система неудобна: попробуйте перемножить `XLVII` на `XIX`
столбиком. Именно поэтому её вытеснила позиционная запись.

### Позиционные системы

В **позиционной** системе значение цифры зависит от её позиции.
В числе 555 три одинаковые цифры означают пятьсот, пятьдесят и пять.

Систему задают два параметра:

* **основание** — сколько разных цифр используется;
* **алфавит** — сами цифры.

| Система | Основание | Алфавит |
|---|---|---|
| двоичная | 2 | 0, 1 |
| восьмеричная | 8 | 0 … 7 |
| десятичная | 10 | 0 … 9 |
| шестнадцатеричная | 16 | 0 … 9, A, B, C, D, E, F |

Обратите внимание: **самой цифры «основание» в алфавите нет**.
В десятичной нет знака для десяти, в двоичной нет двойки. Как только
счёт доходит до основания, разряд обнуляется и появляется единица слева.

### Развёрнутая запись числа

Главная формула позиционной системы. Число записывается как сумма цифр,
умноженных на степени основания:

$$A_q = a_n \cdot q^n + a_{n-1} \cdot q^{n-1} + \ldots + a_1 \cdot q^1 + a_0 \cdot q^0$$

где q — основание, a — цифры. Степени растут справа налево, начиная с нуля.

**Примеры.**

$$275_{10} = 2 \cdot 10^2 + 7 \cdot 10^1 + 5 \cdot 10^0$$

$$1011_2 = 1 \cdot 2^3 + 0 \cdot 2^2 + 1 \cdot 2^1 + 1 \cdot 2^0 = 11_{10}$$

$$237_8 = 2 \cdot 8^2 + 3 \cdot 8^1 + 7 \cdot 8^0 = 128 + 24 + 7 = 159_{10}$$

Развёрнутая запись — универсальный инструмент: она переводит число
**из любой системы в десятичную**. Обратный перевод разберём на уроке 4.

### Как записывают основание

Основание указывают нижним индексом: $1011_2$, $237_8$, $2F_{16}$.
В программах пишут приставки: `0b1011`, `0o237`, `0x2F`.
В обоих случаях смысл один — сообщить, по каким правилам читать запись.

## Смотрим, как это работает

### Пример 1. Развёрнутая запись в действии

Разложим число по разрядам и увидим, как формула превращается в результат.

````python
def развернуть(запись, основание):
    """Показать развёрнутую запись и вычислить значение."""
    цифры = "0123456789ABCDEF"
    результат = 0
    части = []
    степень = len(запись) - 1

    for знак in запись:
        цифра = цифры.index(знак)
        части.append(f"{цифра}·{основание}^{степень}")
        результат += цифра * основание ** степень
        степень -= 1

    print(f"{запись}(осн. {основание}) = " + " + ".join(части) + f" = {результат}")
    return результат


развернуть("275", 10)
развернуть("1011", 2)
развернуть("237", 8)
развернуть("2F", 16)
````

Ключевая строка — `цифра = цифры.index(знак)`. Метод `index` находит
позицию символа в строке `"0123456789ABCDEF"`, и эта позиция как раз
равна числовому значению цифры: у `"7"` она 7, у `"F"` — 15.
Приём удобный, запомните его.

### Пример 2. Одно число в четырёх системах

Посмотрим, как выглядит одно и то же значение в разных системах.

````python
print(f"{'десятичная':>18} {'двоичная':>12} {'восьмеричная':>14} {'шестнадцат.':>12}")
for число in [5, 8, 15, 16, 100, 255]:
    print(f"{число:>18} {bin(число)[2:]:>12} {oct(число)[2:]:>14} {hex(число)[2:].upper():>12}")
````

Обратите внимание на строку с числом 16: в двоичной это `10000`,
в шестнадцатеричной — короткое `10`. Чем больше основание, тем короче
запись. Именно поэтому программисты любят шестнадцатеричную: она вчетверо
короче двоичной и при этом легко в неё переводится.

Функции `bin`, `oct`, `hex` возвращают строку с приставкой, поэтому
мы отрезаем два первых символа срезом `[2:]`.

### Пример 3. Римские числа

Переведём римскую запись в обычную — заодно увидим, чем непозиционная
система неудобна.

````python
# таблицу знаков держим в двух строках: знак и его значение
ЗНАКИ = "IVXLCDM"
ЦЕНЫ = [1, 5, 10, 50, 100, 500, 1000]


def цена(знак):
    return ЦЕНЫ[ЗНАКИ.index(знак)]


def из_римского(запись):
    итог = 0
    for i in range(len(запись)):
        текущее = цена(запись[i])
        # если следующий знак крупнее — текущий вычитается
        следующее = 0
        if i + 1 < len(запись):
            следующее = цена(запись[i + 1])
        if текущее < следующее:
            итог -= текущее
        else:
            итог += текущее
    return итог


for число in ["XXX", "IX", "XLVII", "MCMXCIV"]:
    print(f"{число:>8} = {из_римского(число)}")
````

Сравните с позиционной системой: там значение цифры считается по формуле,
а здесь приходится заглядывать вперёд и разбирать особый случай.
Один этот факт объясняет, почему римские числа остались только
на циферблатах и в номерах глав.
