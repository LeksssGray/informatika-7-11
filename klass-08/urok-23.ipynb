## Разбираемся

### Цикл внутри цикла

Пока каждый наш цикл проходил по одному ряду значений. Но задачи
бывают двумерные: таблица умножения, шахматная доска, все пары
учеников в классе. Там нужен **вложенный цикл** — цикл внутри цикла.

```python
for i in range(3):
    for j in range(4):
        print(i, j)
```

Работает это так: внешний цикл берёт первое значение `i` и передаёт
управление внутреннему. Внутренний прокручивается **целиком** —
все четыре значения `j`. Только потом `i` меняется на следующее.

Порядок вывода получится такой:

```
0 0   0 1   0 2   0 3
1 0   1 1   1 2   1 3
2 0   2 1   2 2   2 3
```

### Сколько раз выполнится тело

Ровно **произведение** длин: 3 × 4 = 12 повторов.

Это главное, что нужно запомнить про вложенные циклы, — и вот
почему. Если каждый цикл идёт по списку из 1000 элементов,
тело выполнится миллион раз. Три вложенных цикла по 1000 —
миллиард, и программа будет считать минутами.

| Вложенность | n = 100 | n = 1000 |
|---|---|---|
| один цикл | 100 | 1 000 |
| два вложенных | 10 000 | 1 000 000 |
| три вложенных | 1 000 000 | 1 000 000 000 |

Поэтому, увидев вложенный цикл, всегда спрашивают: а сколько
это шагов? В девятом классе это назовут сложностью алгоритма,
а пока достаточно правила «перемножь длины».

### Переменные внешнего цикла видны внутри

Внутренний цикл спокойно пользуется `i`:

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end=" ")
    print()
```

Получится таблица умножения 3 × 3. Пустой `print()` стоит
**во внешнем цикле** — он переводит строку после каждого ряда.
Поставите его внутрь — каждое число окажется на своей строке.

### Треугольные циклы

Если внутренний цикл зависит от внешнего, получается треугольник:

```python
for i in range(1, 5):
    for j in range(i):
        print("*", end="")
    print()
```

Первая строка — одна звёздочка, вторая — две, и так далее.
Всего звёздочек 1 + 2 + 3 + 4 = 10.

Такой же приём перебирает **все пары без повторов**: если
внутренний цикл начинается с `i + 1`, пара (2, 5) встретится,
а (5, 2) — нет.

### Обход таблицы

Список списков обходят двумя циклами: внешний берёт строку,
внутренний — элемент в ней.

```python
for строка in таблица:
    for x in строка:
        ...
```

## Смотрим, как это работает

### Порядок обхода

````python
for i in range(3):
    for j in range(4):
        print(f"({i},{j})", end=" ")
    print()
````

### Таблица умножения

````python
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i * j:>4}", end="")
    print()
````

Запись `:>4` выравнивает числа по правому краю в четыре знака —
без неё столбцы разъедутся.

### Треугольник

````python
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
````

### Сколько шагов на самом деле

````python
шагов = 0

for i in range(100):
    for j in range(100):
        шагов += 1

print("Два вложенных цикла по 100:", шагов, "шагов")
````

### Все пары без повторов

````python
числа = [1, 2, 3, 4]

for i in range(len(числа)):
    for j in range(i + 1, len(числа)):
        print(f"({числа[i]}, {числа[j]})", end=" ")

print()
print("Пар:", len(числа) * (len(числа) - 1) // 2)
````

Внутренний цикл стартует с `i + 1`, поэтому каждая пара
встречается один раз и элемент не образует пару сам с собой.

### Сумма таблицы

````python
таблица = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

сумма = 0
for строка in таблица:
    for x in строка:
        сумма += x

print("Сумма всех элементов:", сумма)
````

### Любопытно

Разница между одним циклом и двумя вложенными — это разница
между «мгновенно» и «сходи пообедай».

Пусть компьютер делает сто миллионов простых шагов в секунду.
Тогда для миллиона элементов один цикл отработает за сотую долю
секунды, два вложенных — за три часа, а три вложенных —
за триста лет.

Именно поэтому программисты так носятся с лишним вложенным циклом:
на маленьких данных разницы не видно, а на больших программа
просто перестаёт работать.
