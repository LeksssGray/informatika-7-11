## Разбираемся

### Ветвление посложнее

`if` и `else` мы знаем с седьмого класса. Сегодня разберём случаи,
на которых спотыкаются даже те, кто уверенно пишет простые условия.

### Многовариантный выбор: elif

Когда вариантов больше двух, `if` и `else` выстраивают в цепочку:

```python
if оценка == 5:
    print("отлично")
elif оценка == 4:
    print("хорошо")
elif оценка == 3:
    print("удовлетворительно")
else:
    print("плохо")
```

Главное свойство цепочки: **выполнится ровно одна ветка** — первая
подошедшая. Остальные Python даже проверять не станет.

Из-за этого порядок условий решает всё:

```python
if x > 10:
    размер = "большой"
elif x > 100:          # сюда не попасть никогда!
    размер = "огромный"
```

Число 500 больше десяти, значит, сработает первая ветка, и «огромным»
оно не станет. Условия в цепочке всегда идут **от частного
к общему**.

### Проверка диапазона

Принадлежность отрезку записывают двойным сравнением:

```python
if 10 <= x <= 20:
    print("в диапазоне")
```

Так пишут только в Python — и это как раз тот случай, когда
на его синтаксис стоит перейти с математического: `10 <= x and x <= 20`
означает ровно то же, но длиннее.

> Частая ошибка: `if x == 5 or 6:`. Такая запись не проверяет,
> равен ли `x` шестёрке. Правильно: `if x == 5 or x == 6:`.

### Вложенные условия и логические операторы

Одно и то же можно записать двумя способами:

```python
if возраст >= 14:              # вложенное
    if есть_паспорт:
        print("можно")

if возраст >= 14 and есть_паспорт:   # через and
    print("можно")
```

Второй короче и читается легче, поэтому вложенность держат
на минимуме. Три уровня отступов — верный признак, что условие
пора переписать через `and` и `or`.

### Точка и область

Классическая задача экзаменов: попадает ли точка `(x, y)` в фигуру.
Фигуру разбивают на простые условия и соединяют операторами:

| Область | Условие |
|---|---|
| правая полуплоскость | `x > 0` |
| первая четверть | `x > 0 and y > 0` |
| круг радиуса 5 с центром в нуле | `x * x + y * y <= 25` |
| полоса от −2 до 2 по вертикали | `-2 <= y <= 2` |

Уравнение окружности — `x² + y² = r²`, поэтому «внутри круга» —
это «меньше либо равно `r²`». Корень извлекать не нужно: сравнивать
квадраты и проще, и точнее.

### Треугольник по сторонам

Ещё одна классика. Сначала проверяем, существует ли треугольник:
**каждая сторона меньше суммы двух других**. Только потом
определяем вид.

## Смотрим, как это работает

### Цепочка elif

````python
def словами(оценка):
    if оценка == 5:
        return "отлично"
    elif оценка == 4:
        return "хорошо"
    elif оценка == 3:
        return "удовлетворительно"
    else:
        return "плохо"


for оценка in [5, 4, 3, 2]:
    print(оценка, "—", словами(оценка))
````

### Порядок условий решает

````python
def неправильно(x):
    if x > 10:
        return "большой"
    elif x > 100:
        return "огромный"
    return "маленький"


def правильно(x):
    if x > 100:
        return "огромный"
    elif x > 10:
        return "большой"
    return "маленький"


for x in [5, 50, 500]:
    print(f"{x:>4}: неправильно — {неправильно(x)}, правильно — {правильно(x)}")
````

Число 500 в первой функции навсегда осталось «большим»: до второго
условия дело не дошло.

### Диапазон

````python
for x in [5, 10, 15, 20, 25]:
    if 10 <= x <= 20:
        print(x, "— попадает")
    else:
        print(x, "— мимо")
````

### Точка в круге

````python
def в_круге(x, y):
    return x * x + y * y <= 25


for точка in [(0, 0), (3, 4), (5, 0), (4, 4)]:
    x, y = точка
    print(f"({x}, {y}) → {в_круге(x, y)}")
````

Точка (3, 4) лежит ровно на окружности — 9 + 16 = 25, — и знак
«меньше либо равно» засчитывает её как попадание. В задачах всегда
смотрите, входит ли граница.

### Тарифы: ступеньки

Первые 100 минут — 300 рублей. Каждая следующая минута — 2 рубля.

````python
def стоимость(минуты):
    if минуты <= 100:
        return 300
    return 300 + (минуты - 100) * 2


for м in [50, 100, 150, 200]:
    print(f"{м:>4} минут → {стоимость(м)} руб.")
````
