### Что такое алгоритм

> **Алгоритм** — точное и понятное описание последовательности действий,
> которая за конечное число шагов приводит к нужному результату.

Рецепт пирога, инструкция к шкафу, правила сложения столбиком,
программа на Python — всё это алгоритмы. Но не всякое описание действий
годится: у алгоритма есть обязательные свойства.

### Свойства алгоритма

| Свойство | Что значит | Пример нарушения |
|---|---|---|
| **дискретность** | действия разбиты на отдельные шаги | «приготовь обед» одним шагом |
| **понятность** | каждый шаг понятен исполнителю | инструкция на незнакомом языке |
| **определённость** | каждый шаг однозначен, без вариантов толкования | «посолить по вкусу», «немного подождать» |
| **результативность** | алгоритм заканчивается и даёт результат | «повторяй, пока не надоест» |
| **массовость** | годится для целого класса задач, а не для одной | способ сложить только 23 и 45 |

Компьютер строго требует всех свойств. «Посолить по вкусу» он не поймёт —
ему нужно «добавить 5 граммов соли».

### Исполнитель

**Исполнитель** — тот, кто выполняет алгоритм: человек, робот,
компьютер, станок. У каждого исполнителя есть:

* **среда** — где он действует: клетчатое поле, числа, лист бумаги;
* **система команд исполнителя** (СКИ) — список команд, которые он
  умеет выполнять. Больше он не умеет ничего;
* **отказы** — ситуации, когда команду выполнить нельзя: робот
  упёрся в стену.

Исполнитель выполняет команды **формально**: не задумываясь о смысле,
как наш робот-бутерброд.

### Задачи на исполнителей

В ОГЭ и ЕГЭ часто встречаются числовые исполнители. Например:

> У исполнителя **Вычислитель** две команды:
> **1. прибавь 3**
> **2. умножь на 2**
> Программа — последовательность номеров команд. Какое число получится
> из 2 по программе **12112**?

Выполняем по шагам:

| Шаг | Команда | Было | Стало |
|:-:|---|:-:|:-:|
| 1 | прибавь 3 | 2 | 5 |
| 2 | умножь на 2 | 5 | 10 |
| 3 | прибавь 3 | 10 | 13 |
| 4 | прибавь 3 | 13 | 16 |
| 5 | умножь на 2 | 16 | 32 |

Ответ: 32.

Обратные задачи — «найдите программу, которая из 1 получает 20» —
решают рассуждением с конца или перебором всех программ.

## Смотрим, как это работает

### Пример 1. Симулятор исполнителя

Напишем исполнителя: он получает число и программу-строку и выполняет
команды по одной.

```python
def вычислитель(число, программа):
    for команда in программа:
        if команда == "1":
            число = число + 3
        elif команда == "2":
            число = число * 2
    return число


print(вычислитель(2, "12112"))
```

Функция не думает о смысле программы — просто выполняет команды.
Это и есть формальный исполнитель.

### Пример 2. Перебор программ

Найдём все программы из четырёх команд, которые из 1 получают 20.
Четыре вложенных цикла перебирают все 16 вариантов.

```python
for к1 in "12":
    for к2 in "12":
        for к3 in "12":
            for к4 in "12":
                программа = к1 + к2 + к3 + к4
                if вычислитель(1, программа) == 20:
                    print(программа)
```

Нашлось две программы. Проверьте любую вручную по таблице.
