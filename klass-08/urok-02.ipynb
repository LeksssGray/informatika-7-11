## Справочник

### Единицы информации

| Единица | Сколько это |
|---|---|
| 1 байт | 8 бит |
| 1 килобайт | 1024 байта |
| 1 мегабайт | 1024 килобайта |
| 1 гигабайт | 1024 мегабайта |

**Вниз по таблице умножаем, вверх — делим.** И везде 1024, не тысяча.

### Формула N = 2ⁱ

N — сколько разных вариантов, i — сколько бит на один вариант.

| N | 2 | 4 | 8 | 16 | 32 | 64 | 128 | 256 |
|---|---|---|---|---|---|---|---|---|
| i | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 |

Вариантов всегда больше, чем бит. Если получилось наоборот —
перепутал местами.

**Объём текста** = (бит на символ) × (количество символов).

### Двоичная система

Веса разрядов справа налево: 1, 2, 4, 8, 16, 32, 64, 128.

```
  веса:   16   8   4   2   1
  число:   1   1   0   1   0   →  16 + 8 + 2 = 26
```

Обратно — делением на 2 с остатком, **остатки читаются снизу вверх**.

### Переменные и типы

```python
возраст = 13           # int, целое
рост = 1.62            # float, дробное
имя = "Аня"            # str, строка
отличница = True       # bool, правда или неправда
```

Знак `=` означает «положи», а не «равно». Строка `счёт = счёт + 1`
вполне законна.

`input` всегда возвращает **строку**. Нужно число — оберни в `int()`.

### Операции

| Знак | Что делает | Пример |
|---|---|---|
| `/` | обычное деление | `7 / 2` = 3.5 |
| `//` | деление нацело | `7 // 2` = 3 |
| `%` | остаток | `7 % 2` = 1 |
| `**` | степень | `2 ** 8` = 256 |

### Ветвления

```python
if баллы >= 85:
    оценка = 5
elif баллы >= 70:
    оценка = 4
else:
    оценка = 3
```

Отступы обязательны — они показывают, что относится к условию.
Условия проверяются сверху вниз, срабатывает первое подошедшее.

### Циклы

```python
for i in range(1, 6):      # 1, 2, 3, 4, 5 — правая граница не входит
    print(i)

while число > 0:           # пока условие верно
    число = число // 10
```

`for` — когда число повторений известно, `while` — когда нет.

### Списки

```python
оценки = [5, 4, 5, 3]
оценки[0]        # первый элемент, нумерация с нуля
оценки[-1]       # последний
len(оценки)      # сколько элементов
оценки.append(4) # добавить в конец
```

### Типовые алгоритмы

| Задача | Что завести до цикла | Что делать внутри |
|---|---|---|
| сумма | `сумма = 0` | `сумма += x` |
| количество | `к = 0` | `if условие: к += 1` |
| максимум | `м = список[0]` | `if x > м: м = x` |

Максимум **нельзя** начинать с нуля: на отрицательных числах ответ
будет неверным.

## Смотрим, как это работает

### Единицы и объём текста

````python
мощность = 32          # символов в алфавите
символов = 100

бит_на_символ = 5      # потому что 32 = 2⁵
всего_бит = бит_на_символ * символов

print("Бит:", всего_бит)
print("Байт:", всего_бит / 8)
print("В килобайтах:", всего_бит / 8 / 1024)
````

Три строчки — три единицы. Обрати внимание на порядок: сначала биты,
потом байты, потом килобайты. Прыгать через ступеньку — верный способ
ошибиться.

### Двоичная система

````python
число = 26

print("Двоичная запись 26:", bin(число)[2:])
print("Обратно:", int("11010", 2))
print("Сколько единиц в записи:", bin(число).count("1"))
````

`bin` даёт запись с приставкой `0b`, поэтому первые два символа
отрезают срезом `[2:]`.

### Ветвление и цикл вместе

Классика: пройти по списку, отобрать нужное, посчитать.

````python
оценки = [5, 4, 3, 5, 4, 5, 2]

пятёрок = 0
сумма = 0
лучшая = оценки[0]

for оценка in оценки:
    сумма += оценка
    if оценка == 5:
        пятёрок += 1
    if оценка > лучшая:
        лучшая = оценка

print("Оценок всего:", len(оценки))
print("Пятёрок:", пятёрок)
print("Средний балл:", round(сумма / len(оценки), 2))
print("Лучшая:", лучшая)
````

Один проход по списку, четыре характеристики. Это и есть тот самый
набор приёмов из таблицы: накопитель, счётчик с условием, поиск
максимума.

### Цикл while и цифры числа

````python
число = 4726
сумма_цифр = 0
копия = число

while копия > 0:
    сумма_цифр += копия % 10     # последняя цифра
    копия = копия // 10          # отбрасываем её

print(f"Сумма цифр числа {число}: {сумма_цифр}")
````

Пара `%` и `//` разбирает число на цифры: остаток даёт последнюю цифру,
деление нацело её убирает. Приём понадобится вам в этом году
десятки раз.
