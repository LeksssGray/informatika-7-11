### Пример 1. Всё, что нужно помнить об операциях

````python
a, b = 17, 5

print(f"{a} + {b}  = {a + b}")
print(f"{a} / {b}  = {a / b}      ← обычное деление, всегда дробное")
print(f"{a} // {b} = {a // b}        ← деление нацело")
print(f"{a} % {b}  = {a % b}        ← остаток")
print(f"{a} ** 2 = {a ** 2}      ← степень")
````

Пару `//` и `%` вы будете встречать постоянно. Деление нацело отвечает
«сколько целых раз поместилось», остаток — «сколько не поместилось».

### Пример 2. Цикл, условие и накопление в одной задаче

Найдём сумму всех чётных чисел от 1 до 20.

````python
сумма = 0
for число in range(1, 21):
    if число % 2 == 0:
        сумма += число

print("Сумма чётных от 1 до 20:", сумма)
````

Три элемента, которые встретятся вам сотни раз:

* **накопитель** `сумма = 0` — заводится до цикла;
* **перебор** `for число in range(1, 21)` — обратите внимание, что 21
  не входит в диапазон, поэтому пишем на единицу больше;
* **фильтр** `if число % 2 == 0` — остаток от деления на 2 равен нулю
  только у чётных.

### Пример 3. Строки и списки

Соберём частотную сводку по строке — задача, которая покажет, что вы
помните о работе с текстом.

````python
слово = "информатика"

гласные = "аеёиоуыэюя"
счётчик = 0
согласные_список = []

for буква in слово:
    if буква in гласные:
        счётчик += 1
    else:
        согласные_список.append(буква)

print(f"В слове «{слово}» {len(слово)} букв")
print(f"Гласных: {счётчик}")
print(f"Согласные: {согласные_список}")
print(f"Слово наоборот: {слово[::-1]}")
````

Здесь стоит вспомнить три вещи: оператор `in` проверяет вхождение,
метод `append` добавляет элемент в конец списка, а срез `[::-1]`
переворачивает строку.
