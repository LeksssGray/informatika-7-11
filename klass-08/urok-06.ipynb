## Разбираемся

### Арифметика работает в любой системе

Столбик, которому вас учили в начальной школе, не привязан к десятке.
Правило одно и то же: складываем разряд за разрядом справа налево,
и **как только сумма достигает основания — возникает перенос**.

В десятичной перенос появляется при сумме 10 и больше. В двоичной —
уже при 2. В восьмеричной — при 8. Меняется только порог.

### Сложение в двоичной системе

Таблица сложения самая короткая в математике:

| | 0 | 1 |
|---|---|---|
| **0** | 0 | 1 |
| **1** | 1 | 10 |

Единственный интересный случай — `1 + 1 = 10₂`: ноль пишем, единицу
переносим в старший разряд.

```
    ¹ ¹ ¹
    1 0 1 1        (11)
  +   1 1 0         (6)
  ─────────
  1 0 0 0 1        (17)
```

Разберём по шагам справа налево:

| Разряд | Считаем | Пишем | Переносим |
|---|---|---|---|
| 1-й | 1 + 0 = 1 | 1 | — |
| 2-й | 1 + 1 = 10₂ | 0 | 1 |
| 3-й | 0 + 1 + 1 = 10₂ | 0 | 1 |
| 4-й | 1 + 1 = 10₂ | 0 | 1 |
| 5-й | перенос | 1 | — |

Может встретиться и сумма трёх единиц: `1 + 1 + 1 = 11₂` — пишем 1,
переносим 1.

### Вычитание в двоичной системе

Так же, как в десятичной, но занимаем не десятку, а двойку:

```
    1 0 0 1        (9)
  -   1 1 0        (6)
  ─────────
      0 1 1        (3)
```

Во втором разряде из нуля вычесть единицу нельзя — занимаем у старшего.
Занятая единица старшего разряда даёт две единицы младшего.

### Сложение в восьмеричной и шестнадцатеричной

Точно так же, только порог переноса другой.

```
  Восьмеричная:      Шестнадцатеричная:
     2 3 7               2 F
   +   4 5             + 1 3
   ───────             ─────
     3 0 4               4 2
```

Разбор восьмеричного примера: `7 + 5 = 12` в десятичной, но `12 = 8 + 4`,
значит пишем 4, переносим 1. Дальше `3 + 4 + 1 = 8 = 8 + 0`, пишем 0,
переносим 1. И наконец `2 + 1 = 3`.

Шестнадцатеричный: `F + 3 = 15 + 3 = 18 = 16 + 2`, пишем 2, переносим 1.
Затем `2 + 1 + 1 = 4`.

### Умножение на степень основания

Самая приятная операция. Чтобы умножить число на основание системы,
достаточно **приписать ноль справа** — точно так же, как умножение
на 10 в десятичной.

| Действие | Десятичная | Двоичная |
|---|---|---|
| × основание | 25 → 250 | 101 → 1010 |
| × основание² | 25 → 2500 | 101 → 10100 |
| ÷ основание нацело | 250 → 25 | 1010 → 101 |

Для двоичной это означает: **умножение на 2 — приписать ноль,
деление на 2 — отбросить последний разряд**. Компьютер именно так
и умножает на степени двойки — это быстрее обычного умножения,
и операция называется «сдвиг».

### Сравнение чисел из разных систем

Сравнивать записи напрямую нельзя: `11₂` меньше `9₁₀`, хотя выглядит
«длиннее». Всегда приводите к одной системе — обычно к десятичной.

## Смотрим, как это работает

### Пример 1. Сложение столбиком, как на бумаге

Напишем сложение двоичных чисел, честно повторяя работу с переносами —
без перевода в десятичную.

````python
def сложить_двоичные(а, б):
    длина = max(len(а), len(б))
    а = а.zfill(длина)
    б = б.zfill(длина)

    результат = ""
    перенос = 0

    for i in range(длина - 1, -1, -1):        # идём справа налево
        сумма = int(а[i]) + int(б[i]) + перенос
        цифра = сумма % 2
        новый_перенос = сумма // 2
        print(f"  разряд {длина - i}: {а[i]} + {б[i]} + пришло {перенос}"
              f" → пишем {цифра}, переносим {новый_перенос}")
        результат = str(цифра) + результат
        перенос = новый_перенос

    if перенос:
        результат = "1" + результат
    return результат


print("1011 + 110:")
итог = сложить_двоичные("1011", "110")
print("Ответ:", итог, "=", int(итог, 2), "в десятичной")
````

Обратите внимание на две строки, в которых вся суть:

* `результат = str(сумма % 2) + результат` — остаток от деления на 2
  это цифра, которую пишем;
* `перенос = сумма // 2` — целая часть это то, что уходит в старший разряд.

Заметьте, что формула не зависит от двойки: замените её на основание —
и получите сложение в любой системе.

### Пример 2. Сдвиг вместо умножения

````python
число = "101"          # это 5
print(f"{число}₂ = {int(число, 2)}")

for сдвиг in range(1, 4):
    сдвинутое = число + "0" * сдвиг
    print(f"  приписали {сдвиг} нуля: {сдвинутое}₂ = {int(сдвинутое, 2)}  (это ×{2 ** сдвиг})")
````

В Python сдвиг записывается операторами `<<` и `>>`:
`5 << 3` даёт 40, `40 >> 3` даёт 5. Это те самые «приписать нули»
и «отбросить разряды», выполненные процессором за один такт.

### Пример 3. Восьмеричное сложение

Тот же алгоритм, другое основание — убедимся, что формула универсальна.

````python
def сложить(а, б, основание):
    цифры = "0123456789ABCDEF"
    длина = max(len(а), len(б))
    а, б = а.zfill(длина), б.zfill(длина)

    результат = ""
    перенос = 0
    for i in range(длина - 1, -1, -1):
        сумма = цифры.index(а[i]) + цифры.index(б[i]) + перенос
        результат = цифры[сумма % основание] + результат
        перенос = сумма // основание

    return цифры[перенос] + результат if перенос else результат


print("237 + 45 в восьмеричной =", сложить("237", "45", 8))
print("2F  + 13 в шестнадцатеричной =", сложить("2F", "13", 16))
print("1011 + 110 в двоичной =", сложить("1011", "110", 2))
````

Одна функция обслуживает все системы сразу. Это хороший признак:
если алгоритм записан правильно, частные случаи получаются
подстановкой параметра, а не переписыванием кода.
