### Высказывание

> **Высказывание** — повествовательное предложение, про которое можно
> однозначно сказать, истинно оно или ложно.

| Предложение | Высказывание? |
|---|---|
| «Москва — столица России» | да, истинное |
| «2 + 2 = 5» | да, ложное |
| «Который час?» | нет: вопрос |
| «Закрой дверь» | нет: просьба |
| «x > 3» | нет, пока неизвестен x |

Истину обозначают **1**, ложь — **0**. В Python это `True` и `False`.

### Три базовые операции

| Операция | Как читается | Обозначения | Python | Истинна, когда |
|---|---|---|---|---|
| конъюнкция | «И» | A ∧ B, A & B, A · B | `A and B` | **оба** истинны |
| дизъюнкция | «ИЛИ» | A ∨ B, A + B | `A or B` | **хотя бы одно** истинно |
| инверсия | «НЕ» | ¬A, Ā | `not A` | A ложно |

Конъюнкцию ещё называют **логическим умножением**: результат равен
произведению нулей и единиц. Дизъюнкцию — **логическим сложением**.

### Картинкой: диаграммы Венна

правый — где истинно B.

«И» — только общая часть кругов. «ИЛИ» — всё, что попало хотя бы
в один круг. «НЕ A» — всё вне левого круга.

### Приоритет операций

Как в арифметике умножение раньше сложения:

1. сначала **НЕ**;
2. потом **И**;
3. потом **ИЛИ**.

Скобки меняют порядок. Выражение `A ∨ B ∧ ¬C` вычисляется
как `A ∨ (B ∧ (¬C))`.

### Как переводить слова в формулы

| Словами | Формулой |
|---|---|
| число больше 10 и меньше 20 | `(x > 10) and (x < 20)` |
| число меньше 5 или больше 50 | `(x < 5) or (x > 50)` |
| неверно, что число чётное | `not (x % 2 == 0)` |
| число не меньше 18 | `not (x < 18)` — то же, что `x >= 18` |

Частая ловушка: «число не больше 10» — это `x <= 10`, а не `x < 10`.

## Смотрим, как это работает

### Пример 1. Все сочетания двух высказываний

```python
print(" A  B | A∧B A∨B ¬A")
for A in (False, True):
    for B in (False, True):
        print(f" {int(A)}  {int(B)} |  {int(A and B)}   {int(A or B)}   {int(not A)}")
```

`int(True)` превращает истину в 1, ложь — в 0. Так таблица выглядит
как в учебнике.

### Пример 2. Задача формата ОГЭ

> Для какого из чисел истинно высказывание:
> **НЕ (X < 5) И (X < 8)**?
> Варианты: 3, 5, 8, 10.

Подставим каждое число:

```python
def высказывание(x):
    return not (x < 5) and (x < 8)


for x in (3, 5, 8, 10):
    print(x, "→", высказывание(x))
```

Истинно только для 5. Рассуждение без компьютера: `НЕ (X < 5)` значит
`X ≥ 5`, а вместе с `X < 8` получаем промежуток от 5 до 7.

### Пример 3. Наименьшее подходящее число

Такие вопросы решаются перебором: идём по числам по порядку
и останавливаемся на первом подходящем.

```python
for x in range(1, 100):
    if not (x <= 7) and (x % 2 == 0):
        print("Наименьшее:", x)
        break
```

`break` прекращает цикл сразу, как только ответ найден.
