### Путают приоритет операций

В логике порядок такой же строгий, как в арифметике:

1. инверсия (`не`, ¬),
2. конъюнкция (`и`, ∧),
3. дизъюнкция (`или`, ∨),
4. импликация (→),
5. эквиваленция (≡).

Значит, `A ∨ B ∧ C` — это `A ∨ (B ∧ C)`, а вовсе не слева направо.
Сомневаешься — ставь скобки, за них не снижают.

### Импликацию считают «как удобно»

`A → B` ложна **ровно в одном случае**: когда A истинно, а B ложно.
Во всех остальных — истинна, в том числе когда A ложно.

«Если пойдёт дождь, я возьму зонт». Дождя не было — обещание
не нарушено, что бы ты ни сделал с зонтом.

Полезная замена: `A → B` — это то же самое, что `¬A ∨ B`.
Через неё импликация считается в уме быстрее всего.

### Теряют строки в таблице истинности

Строк всегда $2^n$, где $n$ — число переменных. Две переменные —
4 строки, три — 8, четыре — 16. Написал 6 строк для трёх
переменных — значит, два набора потерял.

Заполняй левую часть по правилу: первый столбец — половина нулей,
половина единиц; второй — четвертями; третий — через один.
Тогда ни один набор не пропадёт.

### Законы де Моргана применяют наполовину

```
¬(A ∧ B) = ¬A ∨ ¬B
¬(A ∨ B) = ¬A ∧ ¬B
```

Забыть поменять «и» на «или» — самая частая ошибка упрощения.
Инверсия переворачивает не только переменные, но и саму операцию.

### Свойства алгоритма называют наугад

| Свойство | Что означает | Что нарушено в примере |
|---|---|---|
| Дискретность | шаги идут по одному | «сделай всё сразу» |
| Определённость | шаг понимается однозначно | «иди туда, не знаю куда» |
| Результативность | заканчивается ответом | «дели, пока не надоест» |
| Массовость | годится для разных данных | «сложи именно 2 и 3» |
| Конечность | шагов конечное число | бесконечный цикл |

### Что проверить перед сдачей

1. Скобки в логическом выражении расставлены?
2. Строк в таблице ровно $2^n$?
3. При упрощении менялась не только переменная, но и операция?
4. У алгоритма есть выход из цикла?

### Таблица, в которой не хватает строк

Ученик строил таблицу истинности для выражения с тремя переменными
и выписал шесть наборов: 000, 001, 010, 100, 101, 111.

Проверим программой, сколько их должно быть.

````python
переменных = 3
print("Должно быть строк:", 2 ** переменных)

for a in range(2):
    for b in range(2):
        for c in range(2):
            print(a, b, c)
````

Восемь наборов, а выписано шесть: потеряны 011 и 110. Три вложенных
цикла не дадут ошибиться — они переберут всё.

### Упрощение, которое сломалось

Нужно упростить `¬(A ∧ B)`. Ученик написал `¬A ∧ ¬B` — забыл
поменять операцию. Сравним обе формулы на всех наборах.

````python
print(" A  B | ¬(A∧B) | ¬A∧¬B | ¬A∨¬B")

for a in range(2):
    for b in range(2):
        правильно = int(not (a and b))
        ошибка = int((not a) and (not b))
        де_морган = int((not a) or (not b))
        print(f" {a}  {b} |   {правильно}    |   {ошибка}   |   {де_морган}")
````

Видно сразу: столбец «ошибка» совпадает с правильным только
в двух строках из четырёх, а столбец де Моргана — во всех.

Такую проверку можно делать на любой контрольной, где разрешён
компьютер: четыре строки кода вместо долгих рассуждений.

### Алгоритм без выхода

Алгоритм: «пока число не станет равно 1, вычитай 2». Запустим
его мысленно для числа 10: 8, 6, 4, 2, 0, −2, −4…

Единицы среди чётных чисел не будет никогда. Нарушено свойство
**результативности**: алгоритм не заканчивается.

````python
число = 10
шагов = 0

while число != 1 and шагов < 10:
    число -= 2
    шагов += 1
    print("шаг", шагов, "→", число)

print("Через 10 шагов единицы так и нет:", число)
````

Ограничитель `шагов < 10` спас нас от зависания. Это обычный приём:
если не уверен в алгоритме, добавь счётчик шагов.
