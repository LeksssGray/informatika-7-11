## Разбираемся

### Таблица в программе

Двумерный список — это таблица: список строк, каждая из которых
сама список. Так хранят матрицы, игровые поля, изображения,
выгрузки из таблиц.

```python
матрица = [
    [1, 2, 3],
    [4, 5, 6],
]
```

Обращение — **два индекса, сначала строка, потом столбец**:
`матрица[1][2]` — это 6. Порядок не путать: первым идёт номер
строки, как в морском бое.

### Размеры

```python
строк = len(матрица)
столбцов = len(матрица[0])
```

Второе работает только если таблица прямоугольная. В Python
строки могут быть разной длины — язык этому не мешает,
но в задачах ЕГЭ таблицы всегда прямоугольные.

### Создание

```python
таблица = [[0] * столбцов for _ in range(строк)]
```

И ещё раз про ловушку с прошлого урока: `[[0] * 3] * 3`
создаёт **три ссылки на одну строку**. Меняете одну — меняются все.
Это самая частая ошибка при работе с матрицами.

### Обход

```python
for i in range(len(матрица)):                # по строкам
    for j in range(len(матрица[i])):         # по столбцам
        ... матрица[i][j] ...
```

Если индексы не нужны, короче:

```python
for строка in матрица:
    for элемент in строка:
        ...
```

### Суммы по строкам и столбцам

Сумма строки — это `sum(матрица[i])`, одна операция.

Сумма столбца требует пройти по всем строкам:

```python
сумма = 0
for строка in матрица:
    сумма += строка[j]
```

Несимметрия неслучайна: строка лежит в памяти единым списком,
а столбец «размазан» по разным строкам.

### Диагонали

Для квадратной матрицы размера n:

| Что | Условие | Элементы |
|---|---|---|
| главная диагональ | `i == j` | `[0][0]`, `[1][1]`, … |
| побочная диагональ | `i + j == n - 1` | `[0][n-1]`, `[1][n-2]`, … |
| выше главной | `i < j` | правый верхний угол |
| ниже главной | `i > j` | левый нижний угол |

Эти четыре условия закрывают почти все школьные задачи
на матрицы — их стоит запомнить.

### Транспонирование

Поменять строки и столбцы местами:

```python
новая = [[матрица[i][j] for i in range(строк)] for j in range(столбцов)]
```

Или встроенным способом: `list(zip(*матрица))` — но результат будет
из кортежей, а не списков.

### Где это встречается

Игровое поле (крестики-нолики, шахматы, сапёр), изображение
(пиксель = элемент матрицы), таблица данных, граф в виде матрицы
смежности — всё это двумерные списки. В одиннадцатом классе
на матрицах смежности будут графы.

## Смотрим, как это работает

### Пример 1. Создание и обход

````python
матрица = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

строк = len(матрица)
столбцов = len(матрица[0])

print(f"Размер: {строк} × {столбцов}")
print("Элемент [1][2]:", матрица[1][2])
print()
for строка in матрица:
    print("  ".join(f"{x:>2}" for x in строка))
````

### Пример 2. Ловушка умножения ещё раз

````python
плохо = [[0] * 3] * 3
хорошо = [[0] * 3 for _ in range(3)]

плохо[1][1] = 5
хорошо[1][1] = 5

print("Через умножение:", плохо)
print("Через генератор:", хорошо)
````

### Пример 3. Суммы по строкам и столбцам

````python
print("Суммы строк:")
for i in range(строк):
    print(f"  строка {i}: {sum(матрица[i])}")

print("Суммы столбцов:")
for j in range(столбцов):
    сумма = 0
    for i in range(строк):
        сумма += матрица[i][j]
    print(f"  столбец {j}: {сумма}")
````

### Пример 4. Диагонали

````python
главная = [матрица[i][i] for i in range(строк)]
побочная = [матрица[i][строк - 1 - i] for i in range(строк)]
выше = [матрица[i][j] for i in range(строк) for j in range(столбцов) if i < j]

print("Главная диагональ: ", главная, "сумма", sum(главная))
print("Побочная диагональ:", побочная, "сумма", sum(побочная))
print("Выше главной:      ", выше)
````

### Пример 5. Транспонирование

````python
транспонированная = [[матрица[i][j] for i in range(строк)] for j in range(столбцов)]

print("Было:")
for строка in матрица:
    print("  ", строка)
print("Стало:")
for строка in транспонированная:
    print("  ", строка)
````

### Пример 6. Игровое поле

````python
поле = [
    ["X", "O", "X"],
    ["O", "X", "O"],
    ["O", "O", "X"],
]

for строка in поле:
    print(" | ".join(строка))

победа_по_диагонали = all(поле[i][i] == "X" for i in range(3))
print()
print("Крестики выиграли по главной диагонали:", победа_по_диагонали)
````

`all(...)` истинно, когда истинны все условия. Его пара —
`any(...)`, истинное, если верно хотя бы одно.

### Пример 7. Поиск максимума и его координат

````python
данные = [
    [3, 17, 8],
    [21, 5, 13],
    [9, 2, 19],
]

лучший_i, лучший_j = 0, 0
for i in range(len(данные)):
    for j in range(len(данные[i])):
        if данные[i][j] > данные[лучший_i][лучший_j]:
            лучший_i, лучший_j = i, j

print(f"Максимум {данные[лучший_i][лучший_j]} в строке {лучший_i}, столбце {лучший_j}")
````

### Любопытно

Почему сумма столбца считается медленнее суммы строки, хотя
операций поровну? Потому что список в памяти лежит подряд,
и процессор, читая один элемент, заодно затягивает в быструю память
соседние — «на всякий случай». Для строки эти соседи и есть
следующие нужные элементы, а для столбца — мусор.

На матрице 1000 × 1000 разница между обходом по строкам и по
столбцам достигает нескольких раз. Код при этом отличается
порядком двух циклов.
