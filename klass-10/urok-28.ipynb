## Разбираемся

### Текстовые файлы и CSV

Текстовый файл — это строки, разделённые переводами строк. **CSV** — таблица в текстовом виде:
значения в строке разделены запятыми (или точками с запятой). Первая строка часто содержит заголовки.
Чтобы разобрать такую строку, её делят по разделителю.

```python
import csv
import io

текст = "имя,балл\nАня,8\nВаня,9\n"
for строка in csv.DictReader(io.StringIO(текст)):
    print(строка["имя"], int(строка["балл"]))
```
