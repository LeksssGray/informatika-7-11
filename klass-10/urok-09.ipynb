### Ошибка 1. Путаница «бит на символ» и «мощность алфавита»

В формуле $N = 2^i$ величина N — это **количество** символов
в алфавите, а i — количество **бит** на один символ. Вариантов всегда
больше, чем бит: 256 символов кодируются 8 битами, не наоборот.

### Ошибка 2. Не округлили вверх

Если мощность алфавита не является степенью двойки, бит нужно взять
**с запасом**. Для 100 символов нужно 7 бит (128 вариантов), а не 6,6.
Дробного числа бит не бывает.

### Ошибка 3. Забыли про количество каналов в аудио

Формула $V = f \cdot i \cdot t \cdot k$ содержит множитель k.
Стерео вдвое больше моно — этот множитель теряют чаще всего.

### Ошибка 4. Перепутали килобайт и килобит

Байтовые единицы идут по 1024, битовые (кбит/с в битрейте) — по 1000.
И не забывайте делить на 8 при переходе от бит к байтам.

### Ошибка 5. Считали большие степени напрямую

Задачу «сколько единиц в записи $2^{40} - 2^{10}$» решают
рассуждением о структуре, а не вычислением триллионного числа.

### Ошибка 6. Не проверили допустимость цифр

Найдя основание системы, убедитесь, что все цифры записи в нём
существуют. Ответ q = 3 для записи `14` неверен: цифры 4
в троичной системе нет.

### Ошибка 7. Сравнили вещественные через ==

Всегда через модуль разности.

## Разбираем сложное

### Пример 1. Составная задача на кодирование

> Автоматическая камера делает снимки 640 × 480 с глубиной цвета
> 24 бита и передаёт их по каналу со скоростью 512 Кбит/с.
> Сколько снимков в минуту сможет передать камера?

````python
ширина, высота, глубина = 640, 480, 24
скорость_кбит = 512
секунд = 60

объём_снимка_бит = ширина * высота * глубина
передано_бит = скорость_кбит * 1024 * секунд

print(f"Объём одного снимка: {объём_снимка_бит:,} бит "
      f"= {объём_снимка_бит / 8 / 1024:.1f} Кбайт")
print(f"За минуту передаётся: {передано_бит:,} бит")
print(f"Снимков в минуту: {передано_бит // объём_снимка_бит}")
````

Обратите внимание на два момента. Килобиты в скорости канала
переводятся через 1024, потому что в условии написано «Кбит»
с большой буквы — это двоичная приставка. И результат округляется
**вниз**: неполный снимок передать нельзя.

### Пример 2. Задача на структуру двоичной записи

> Сколько единиц в двоичной записи числа $8^{25} + 4^{15} - 2^{10}$?

Приводим всё к степеням двойки: $8^{25} = 2^{75}$, $4^{15} = 2^{30}$.

$$2^{75} + 2^{30} - 2^{10}$$

Разность $2^{30} - 2^{10}$ даёт 20 единиц (приём из урока 5),
а $2^{75}$ добавляет ещё одну в старшем разряде — разряды не пересекаются.

````python
число = 8 ** 25 + 4 ** 15 - 2 ** 10
запись = bin(число)[2:]

print(f"Всего разрядов: {len(запись)}")
print(f"Единиц: {запись.count('1')}")
print(f"Рассуждение: 1 (от 2^75) + 20 (от 2^30 - 2^10) = {1 + 20}")
````

Рассуждение и вычисление сошлись. На экзамене вычислить такое число
невозможно, а рассуждение занимает полминуты.
