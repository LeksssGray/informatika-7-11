### Сколько операций

Нас интересует не конкретное время, а рост числа операций, когда данных становится больше.
Цикл по всем элементам списка делает `n` операций. Два вложенных цикла по всем парам — примерно `n²`.

Классы сложности, от быстрых к медленным: **O(1)** — не зависит от размера, **O(log n)** — деление пополам,
**O(n)** — один проход, **O(n log n)** — хорошая сортировка, **O(n²)** — два вложенных прохода.

```python
for n in [10, 100, 1000]:
    операций = 0
    for i in range(n):
        for j in range(n):
            операций += 1
    print(n, операций)
```

Видно, что при увеличении `n` в десять раз число операций растёт в сто раз. Для миллиона элементов
такой алгоритм уже не работает за разумное время, а `O(n log n)` справляется.
