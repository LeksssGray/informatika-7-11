## Разбираемся

### Двоичный поиск

Если список отсортирован, искать можно быстрее. Средний элемент сравнивают с искомым и отбрасывают
половину. Каждый шаг уменьшает поиск вдвое, поэтому тысячу элементов хватит десятью шагами.

### Сортировка вставками

Элементы берут по одному и вставляют в уже отсортированную часть слева на нужное место. Для почти
отсортированных данных это быстро.

Вставками — **устойчивая** сортировка: равные элементы остаются в том
порядке, в каком были. Это важно, когда сортируют записи: отсортировали
учеников по классу — внутри класса сохранился алфавит.

```python
числа = [5, 2, 9, 1]
for i in range(1, len(числа)):
    текущее = числа[i]
    j = i - 1
    while j >= 0 and числа[j] > текущее:
        числа[j + 1] = числа[j]
        j -= 1
    числа[j + 1] = текущее
print(числа)
```
