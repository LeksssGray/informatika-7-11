## Где обычно ошибаются

Четверть была о двух вещах: логике как математике и компьютере как
устройстве. Разберём ошибки, из-за которых теряют баллы, и напишем
контрольную.

### Ошибка 1. Забыли приоритет импликации

Импликация и эквиваленция выполняются **последними**, позже ∧ и ∨.
Выражение `A ∨ B → C` — это `(A ∨ B) → C`. Ошибка в приоритете
рушит всю задачу, даже если дальше всё посчитано верно.

### Ошибка 2. Неверно раскрыли отрицание импликации

`¬(A → B) = A ∧ ¬B`, а вовсе не `¬A ∨ B` и не `A → ¬B`.
Проверяйте себя одной строкой: импликация ложна ровно в одном случае,
значит, у её отрицания ровно одно решение.

### Ошибка 3. Потеряли решения при подсчёте

Считая решения по случаям (A = 0 и A = 1), не забывайте, что
оставшиеся переменные могут быть **свободными**: каждая свободная
переменная удваивает количество наборов.

### Ошибка 4. Цепочка импликаций посчитана как 2ⁿ

У цепочки из n переменных ровно n + 1 решение, а не 2ⁿ и не n.
Проверьте на трёх переменных: должно получиться 4.

### Ошибка 5. Путают биты адреса и объём памяти

Разрядность шины адреса — это **показатель степени**: 32 бита дают
2³² адресов, то есть 4 ГБ, а не 32 ГБ.

### Ошибка 6. Округление размера файла вниз

Файл занимает **целое** число кластеров, всегда с округлением вверх.
4097 байт при кластере 4096 — это два кластера.

### Ошибка 7. Знак `?` в маске считают необязательным

`?` — ровно один символ, ни нуля, ни двух. Необязательную часть
задаёт только `*`.

## Разбираем сложное

### Пример 1. Уравнение с ограничением

> Сколько решений у уравнения
> **(x₁ → x₂) ∧ (x₂ → x₃) ∧ (x₃ → x₄) = 1**, если известно,
> что x₂ = 1?

Рассуждение: решения цепочки — наборы вида «нули, потом единицы».
Если x₂ = 1, то x₃ и x₄ тоже обязаны быть единицами (иначе где-то
будет переход 1 → 0), а x₁ свободен. Значит, два решения.

```python
from itertools import product


def цепочка(*x):
    for i in range(len(x) - 1):
        if x[i] and not x[i + 1]:
            return False
    return True


все = [набор for набор in product([0, 1], repeat=4) if цепочка(*набор)]
с_ограничением = [набор for набор in все if набор[1] == 1]

print("Всего решений цепочки:", len(все))
print("Из них с x₂ = 1:      ", len(с_ограничением))
for набор in с_ограничением:
    print("  ", набор)
```

Приём общий: сначала находим структуру всех решений, потом
накладываем ограничение. Так задача не разрастается, даже если
переменных сорок.

### Пример 2. Составная задача про память

> Компьютер с 32-разрядной шиной адреса адресует память побайтно.
> Сколько **страниц** памяти по 4 КБ он способен адресовать?

```python
всего_байт = 2 ** 32
страница = 4 * 1024
страниц = всего_байт // страница

print(f"Всего памяти: {всего_байт / 1024 ** 3:.0f} ГБ")
print(f"Страниц:      {страниц:,}".replace(",", " "))
print(f"Проверка степенями двойки: 2^32 / 2^12 = 2^{32 - 12} = {2 ** 20:,}".replace(",", " "))
```

Считать такие задачи удобнее в степенях двойки: деление превращается
в вычитание показателей, и калькулятор не нужен.
