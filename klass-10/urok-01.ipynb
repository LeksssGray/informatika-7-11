### Что такое информация: три подхода

В 7 классе мы обошлись рабочим описанием. Теперь стоит знать,
что единого определения нет, и разные науки понимают информацию по-своему.

**Обыденный подход.** Информация — сведения, знания, сообщения.
Годится для разговора, но не для расчётов: как измерить «сведения»?

**Кибернетический подход** (Норберт Винер). Информация — то, что
уменьшает неопределённость. Именно он даёт формулу $N = 2^i$
и позволяет измерять информацию в битах.

**Вероятностный подход** (Клод Шеннон, 1948). Обобщение
кибернетического на случай **неравновероятных** событий.

Если у события вероятность p, то количество информации в сообщении
о нём равно

$$i = \log_2 \frac{1}{p} = -\log_2 p$$

Смысл прозрачен: **чем неожиданнее событие, тем больше информации
в сообщении о нём**. Сообщение «завтра в Сахаре не будет дождя» почти
не несёт информации — это и так известно. Сообщение «завтра в Сахаре
снег» несёт огромную информацию.

Проверим согласованность с прежней формулой. Если исходов N
и они равновероятны, то $p = 1/N$, и

$$i = \log_2 N$$

что в точности равносильно $N = 2^i$. Новая формула не отменяет старую,
а включает её как частный случай.

### Свойства информации

| Свойство | Что означает | Пример нарушения |
|---|---|---|
| **достоверность** | соответствует реальности | слух, фейк |
| **полнота** | достаточна для решения | прогноз без указания города |
| **актуальность** | своевременна | вчерашнее расписание |
| **понятность** | доступна получателю | текст на незнакомом языке |
| **ценность** | полезна для задачи | точное время на необитаемом острове |

Обратите внимание: почти все свойства **относительны**. Информация
ценна или бесполезна не сама по себе, а для конкретного получателя
и конкретной задачи. Единственное безусловное свойство — достоверность.

## Смотрим, как это работает

### Пример 1. Формула Шеннона в действии

Посчитаем количество информации для событий разной вероятности.

```python
import math

события = [
    ("монета — орёл", 0.5),
    ("кубик — шестёрка", 1 / 6),
    ("карта — туз пик", 1 / 36),
    ("выигрыш в лотерею", 1 / 1000000),
    ("завтра будет день", 0.999999),
]

print(f"{'событие':<24} {'вероятность':>12} {'информация, бит':>18}")
for название, p in события:
    i = -math.log2(p)
    print(f"{название:<24} {p:>12.6f} {i:>18.2f}")
```

Крайние строки показывают суть формулы. Почти достоверное событие
несёт около нуля бит информации — мы и так это знали. Невероятное
событие несёт двадцать бит: узнав о выигрыше, мы получили много нового.

Функция `math.log2` вычисляет логарифм по основанию 2 — то самое,
что отвечает на вопрос «в какую степень возвести двойку».

### Пример 2. Согласованность с формулой N = 2ⁱ

Убедимся, что вероятностный подход не противоречит алфавитному.

```python
print(f"{'исходов N':>10} {'через N = 2^i':>15} {'через Шеннона':>15}")
for N in [2, 4, 8, 16, 32, 256]:
    через_степень = math.log2(N)
    через_шеннона = -math.log2(1 / N)
    print(f"{N:>10} {через_степень:>15.1f} {через_шеннона:>15.1f}")
```

Столбцы совпали полностью: для равновероятных исходов обе формулы
дают один и тот же ответ. Это и означает, что подход Шеннона
обобщает, а не заменяет уже известное.

### Пример 3. Информация в реальном тексте

В русском языке буквы встречаются с разной частотой. Посчитаем,
сколько информации несёт каждая, и сравним с наивной оценкой.

```python
# Приблизительные частоты некоторых букв русского языка
частоты = {"о": 0.110, "е": 0.085, "а": 0.080, "и": 0.074,
           "н": 0.067, "ъ": 0.0004, "ф": 0.002, "э": 0.003}

print("Наивная оценка (33 равновероятные буквы):",
      f"{math.log2(33):.2f} бит на букву")
print()
print(f"{'буква':>6} {'частота':>10} {'информация':>12}")
for буква, p in частоты.items():
    print(f"{буква:>6} {p:>10.4f} {-math.log2(p):>12.2f}")
```

Частая буква «о» несёт около 3,2 бита, а редчайший «ъ» — почти 11.
Средняя информация на букву в русском тексте получается примерно
4,35 бита вместо наивных 5,04.

Разница в 0,7 бита на букву кажется мелочью, но именно на ней
работают **архиваторы**: чем неравномернее частоты, тем сильнее
можно сжать текст. К алгоритмам сжатия мы вернёмся позже.
