### Развёрнутая запись: общий случай

$$A_q = \sum_{k} a_k \cdot q^{k}$$

где q — основание, $a_k$ — цифра в разряде k. Разряды нумеруются
справа налево начиная с нуля.

Основание может быть любым: 3, 5, 7, 13. В задачах ЕГЭ встречаются
и такие.

### Приём 1. Числа вида q^n

В системе с основанием q число $q^n$ записывается как единица
и n нулей — точно как 10ⁿ в десятичной.

| Число | В системе q | Запись |
|---|---|---|
| $q$ | q | 10 |
| $q^2$ | q | 100 |
| $q^n$ | q | 1 и n нулей |

### Приём 2. Числа вида q^n − 1

Очень важный случай. Число на единицу меньше степени основания
записывается как n цифр, каждая из которых максимальна:

| Число | Запись в системе q | Пример при q = 2 | Пример при q = 10 |
|---|---|---|---|
| $q - 1$ | одна цифра (q−1) | 1 | 9 |
| $q^2 - 1$ | две цифры | 11 | 99 |
| $q^n - 1$ | n цифр (q−1) | 111…1 | 999…9 |

Отсюда мгновенный ответ на типовой вопрос: «сколько единиц в двоичной
записи числа $2^{15} - 1$?» — пятнадцать.

### Приём 3. Сумма степеней

Задача вида «сколько единиц в записи $2^{20} + 2^{10} + 2^3$?»
решается устно: каждое слагаемое даёт ровно одну единицу
в своём разряде, разряды разные — значит, единиц три.

А вот если слагаемые повторяются, возникает перенос:
$2^5 + 2^5 = 2^6$, и единица остаётся одна.

### Приём 4. Разность

Задача «сколько единиц в записи $4^{20} - 2^{10}$?» сложнее.
Приводим к одному основанию: $4^{20} = 2^{40}$.

$$2^{40} - 2^{10} = 2^{10}(2^{30} - 1)$$

Второй множитель — тридцать единиц (приём 2), первый добавляет
десять нулей справа. Итого **тридцать единиц**.

Общее правило: $2^a - 2^b$ при $a > b$ даёт $(a-b)$ единиц,
за которыми идут b нулей.

### Приём 5. Сравнение записей

Число в системе с меньшим основанием записывается **длиннее**.
Если две записи одного числа имеют разную длину, короче та,
где основание больше.

Обратное сравнение самих чисел требует приведения к общей системе —
на глаз сравнивать записи в разных системах нельзя.

### Определение основания по записи

Задача: «в системе с некоторым основанием число X записывается как Y.
Найти основание». Решается двумя способами.

**Через уравнение.** Записываем развёрнутую форму с неизвестным q
и решаем. Для `21` в неизвестной системе: $2q + 1 = 15$, откуда $q = 7$.

**Перебором.** Проверяем основания от минимально допустимого
(на единицу больше максимальной цифры) до 16. Вариантов мало,
и это надёжнее.

### Ограничение на цифры

В системе с основанием q цифры принимают значения от 0 до q−1.
Значит, минимально возможное основание для записи — на единицу больше
её наибольшей цифры. Запись `178` не может быть восьмеричной,
зато может быть девятеричной и любой большей.

## Смотрим, как это работает

### Пример 1. Единицы в записи без вычислений

```python
случаи = [
    ("2^15 - 1", 2 ** 15 - 1),
    ("2^20 + 2^10 + 2^3", 2 ** 20 + 2 ** 10 + 2 ** 3),
    ("4^20 - 2^10", 4 ** 20 - 2 ** 10),
    ("2^30 - 2^10", 2 ** 30 - 2 ** 10),
    ("8^5 - 1", 8 ** 5 - 1),
]

print(f"{'выражение':<22} {'единиц':>8} {'разрядов':>10}")
for название, значение in случаи:
    запись = bin(значение)[2:]
    print(f"{название:<22} {запись.count('1'):>8} {len(запись):>10}")
```

Сверимся с приёмами из теории.

$2^{15}-1$ — пятнадцать единиц, приём 2 сработал.
Сумма трёх различных степеней — три единицы, приём 3.
$4^{20} - 2^{10} = 2^{40} - 2^{10}$ — тридцать единиц, приём 4.
$8^5 - 1 = 2^{15} - 1$ — снова пятнадцать.

Все ответы получены устно, программа лишь подтвердила рассуждение.

### Пример 2. Определение основания

Два способа для задачи «число 15 записывается как 21».

```python
# Способ 1: уравнение 2q + 1 = 15
q_уравнение = (15 - 1) / 2
print(f"Через уравнение: q = {q_уравнение:.0f}")

# Способ 2: перебор
print("\nПеребором:")
for q in range(3, 17):          # цифра 2 требует основания не меньше 3
    значение = 2 * q + 1
    отметка = "  ← подходит" if значение == 15 else ""
    print(f"  q = {q:>2}: 21 означает {значение:>3}{отметка}")
```

Уравнение быстрее, перебор надёжнее. На экзамене имеет смысл решить
уравнение и проверить ответ подстановкой — это займёт секунды
и исключит арифметическую ошибку.

### Пример 3. Задача с двумя неизвестными

> В некоторой системе счисления записи чисел 20 и 30 оканчиваются
> на цифру 2. Найдите все возможные основания.

Запись оканчивается на 2 означает, что остаток от деления на основание
равен 2.

```python
подходящие = []
for q in range(3, 21):
    if 20 % q == 2 and 30 % q == 2:
        подходящие.append(q)
        print(f"  q = {q}: 20 = ...{20 % q}, 30 = ...{30 % q}  ✓")

print(f"\nОтвет: {подходящие}")
```

Условие «оканчивается на 2» переводится в остаток от деления —
это ключевое наблюдение. Разность 30 − 20 = 10 должна делиться
на основание, что сразу сужает перебор до делителей десятки,
больших двойки: 5 и 10.
