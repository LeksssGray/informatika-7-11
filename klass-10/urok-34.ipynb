## Проект

### Требования

1. **структуры данных:** списки, словари или множества — по делу,
   а не для галочки;
2. **хотя бы один алгоритм года:** поиск, сортировка, рекурсия,
   перебор, обработка файла;
3. **вход и выход** описаны заранее: что программа получает и что
   выдаёт;
4. **проверка:** набор случаев с известным ответом, включая крайние;
5. **оценка сложности:** сколько примерно шагов делает главный
   алгоритм при n входных данных.

Темы: анализатор текста (частоты слов из файла), расписание без
пересечений, перебор паролей заданного вида, генератор лабиринта
и поиск выхода, рекомендации фильмов по оценкам друзей, статистика
по CSV-файлу с графиком.

### Критерии

| Критерий | 0 | 1 | 2 |
|---|---|---|---|
| работает | не запускается | с ошибками | без ошибок |
| алгоритм и структуры | нет | есть, но не по делу | выбраны осмысленно |
| проверка | нет | несколько случаев | набор с крайними случаями |
| сложность | не может сказать | оценка с ошибкой | верная оценка и объяснение |
| защита | не рассказал | без показа | рассказал, показал, ответил |

### Защита — три минуты

Задача → вход и выход → алгоритм и его сложность → проверка →
что получилось. Будь готов к вопросу «а что будет на миллионе
элементов?».
