Логическое выражение можно записать десятком разных способов —
и все они будут давать одну и ту же таблицу истинности. Такие
выражения называют **равносильными** и пишут между ними знак `=`.

Упрощение нужно не ради красоты:

* короткая схема содержит меньше элементов и стоит дешевле;
* упрощённое выражение позволяет решить логическое уравнение
  рассуждением, а не перебором 2⁶⁴ наборов;
* в заданиях ЕГЭ ответ часто находится в две строки после раскрытия
  импликации — и в двадцать строк без него.

### Законы алгебры логики

| Название | Для ∧ | Для ∨ |
|---|---|---|
| переместительный | A ∧ B = B ∧ A | A ∨ B = B ∨ A |
| сочетательный | (A ∧ B) ∧ C = A ∧ (B ∧ C) | (A ∨ B) ∨ C = A ∨ (B ∨ C) |
| распределительный | A ∧ (B ∨ C) = (A∧B) ∨ (A∧C) | A ∨ (B ∧ C) = (A∨B) ∧ (A∨C) |
| идемпотентности | A ∧ A = A | A ∨ A = A |
| с константами | A ∧ 1 = A, A ∧ 0 = 0 | A ∨ 0 = A, A ∨ 1 = 1 |
| противоречия / исключённого третьего | A ∧ ¬A = 0 | A ∨ ¬A = 1 |
| поглощения | A ∧ (A ∨ B) = A | A ∨ (A ∧ B) = A |
| де Моргана | ¬(A ∧ B) = ¬A ∨ ¬B | ¬(A ∨ B) = ¬A ∧ ¬B |
| двойного отрицания | ¬¬A = A | |

Второй распределительный закон — тот, которого нет в обычной алгебре:
в числах `a + b·c` не равно `(a+b)·(a+c)`. В логике равно.

### Законы де Моргана словами

Огастес де Морган — современник Буля, он же придумал термин
«математическая индукция». Его правила читаются так:

> Отрицание конъюнкции — это дизъюнкция отрицаний, и наоборот.

Практическое правило: **отрицание «проваливается» внутрь скобок
и переворачивает операцию**. «Неверно, что я и устал, и голоден» =
«я не устал или не голоден».

### Раскрытие импликации и эквиваленции

```
A → B  =  ¬A ∨ B
A ≡ B  =  (A ∧ B) ∨ (¬A ∧ ¬B)
A ≡ B  =  (A → B) ∧ (B → A)
A ⊕ B  =  (A ∧ ¬B) ∨ (¬A ∧ B)
```

Порядок работы почти всегда один и тот же:

1. избавиться от импликаций и эквиваленций;
2. загнать отрицания внутрь по де Моргану;
3. раскрыть скобки и применить законы поглощения и склеивания.

### Закон склеивания

```
(A ∧ B) ∨ (A ∧ ¬B) = A
```

Смысл: если при истинном A результат истинен и при B, и при ¬B,
то B вообще ни на что не влияет.

### Как проверять себя

Любое преобразование проверяется таблицей истинности: если два
выражения совпали во всех строках, они равносильны. Раньше это делали
руками, теперь — за секунду программой. Именно так мы и будем
проверять каждое упрощение на уроке.

## Смотрим, как это работает

### Пример 1. Проверка равносильности

````python
from itertools import product


def равносильны(первое, второе, переменных):
    for набор in product([0, 1], repeat=переменных):
        if bool(первое(*набор)) != bool(второе(*набор)):
            print("Различие на наборе", набор)
            return False
    return True


# ¬(A ∧ B)  =  ¬A ∨ ¬B
print(равносильны(lambda a, b: not (a and b),
                  lambda a, b: (not a) or (not b), 2))

# ¬(A ∧ B)  =  ¬A ∧ ¬B   — намеренная ошибка
print(равносильны(lambda a, b: not (a and b),
                  lambda a, b: (not a) and (not b), 2))
````

Функция сравнивает две формулы на всех наборах и показывает первый
набор, где они разошлись. Дальше проверяем ею каждое преобразование.

### Пример 2. Упрощаем по шагам

Упростим **F = ¬(A → B) ∨ B**.

```
¬(A → B) ∨ B          исходное
¬(¬A ∨ B) ∨ B         раскрыли импликацию
(A ∧ ¬B) ∨ B          де Морган, двойное отрицание
(A ∨ B) ∧ (¬B ∨ B)    распределительный закон
(A ∨ B) ∧ 1           исключённое третье
A ∨ B                 константа
```

````python
исходное = lambda a, b: (not ((not a) or b)) or b
итог = lambda a, b: a or b

print("Равносильны:", равносильны(исходное, итог, 2))

print(" A B | исходное | итог")
for a in (0, 1):
    for b in (0, 1):
        print(f" {a} {b} |    {int(bool(исходное(a, b)))}     |  {int(bool(итог(a, b)))}")
````

Шесть строк преобразований сократили выражение до двух символов.

### Пример 3. Поглощение и склеивание

````python
проверки = [
    ("A ∨ (A ∧ B) = A",
     lambda a, b: a or (a and b), lambda a, b: a),
    ("A ∧ (A ∨ B) = A",
     lambda a, b: a and (a or b), lambda a, b: a),
    ("(A ∧ B) ∨ (A ∧ ¬B) = A",
     lambda a, b: (a and b) or (a and not b), lambda a, b: a),
    ("A ∨ (¬A ∧ B) = A ∨ B",
     lambda a, b: a or ((not a) and b), lambda a, b: a or b),
]

for название, слева, справа in проверки:
    print(f"{название:<28} {равносильны(слева, справа, 2)}")
````

Последняя строка — тоже поглощение, только с отрицанием. Её часто
не замечают и тащат лишнюю скобку через всё решение.

### Пример 4. Доказательство тождества

Докажем закон контрапозиции: **A → B = ¬B → ¬A**.

Это то самое выражение из домашнего задания прошлого урока, где
во всех строках получалась единица.

````python
прямое = lambda a, b: (not a) or b
обратное = lambda a, b: (not (not b)) or (not a)

print("Контрапозиция верна:", равносильны(прямое, обратное, 2))

print(" A B | A→B | ¬B→¬A")
for a in (0, 1):
    for b in (0, 1):
        print(f" {a} {b} |  {int(bool(прямое(a, b)))}  |   {int(bool(обратное(a, b)))}")
````

«Если идёт дождь, асфальт мокрый» равносильно «если асфальт сухой,
дождя нет». А вот «если асфальт мокрый, идёт дождь» — уже другое
утверждение, и оно неверно: асфальт могла полить машина.

### Пример 5. Сколько единиц осталось

````python
def единиц(функция, переменных):
    return sum(1 for набор in product([0, 1], repeat=переменных) if функция(*набор))


выражения = [
    ("¬A ∨ B", lambda a, b: (not a) or b),
    ("A ∧ ¬B", lambda a, b: a and not b),
    ("A ⊕ B", lambda a, b: bool(a) != bool(b)),
    ("(A → B) ∧ (B → A)", lambda a, b: ((not a) or b) and ((not b) or a)),
]

for название, функция in выражения:
    print(f"{название:<20} единиц в таблице: {единиц(функция, 2)}")
````

Количество единиц — быстрый способ убедиться, что упрощение не сломало
выражение: у равносильных формул оно обязано совпадать. Обратное
неверно: совпадение количества ещё не означает равносильности,
поэтому полную проверку всё равно делаем по всем наборам.
