## Разбираемся

### Когда индекса мало

Список хорош, когда элементы нумеруются по порядку. Но чаще данные
устроены иначе: «Иванову — 92 балла», «коду 101 — цена 45».
Здесь нужен не номер, а **ключ**.

> **Словарь** — набор пар «ключ → значение». Ключи не повторяются.

```python
баллы = {"Иванов": 92, "Петрова": 78}
баллы["Иванов"]          # 92
баллы["Сидоров"] = 85    # добавили
```

### Безопасный доступ

```python
баллы["Кузнецов"]        # KeyError — ключа нет
баллы.get("Кузнецов")    # None
баллы.get("Кузнецов", 0) # 0 — значение по умолчанию
```

`.get(ключ, 0)` — главная рабочая строчка при подсчётах:
«сколько было, или ноль».

### Обход словаря

```python
for ключ in баллы:                  # только ключи
for значение in баллы.values():     # только значения
for ключ, значение in баллы.items():  # пары
```

Начиная с Python 3.7 словарь помнит **порядок добавления** —
это гарантия языка, а не случайность.

### Подсчёт по категориям

Тот самый приём, которым мы делали сводную таблицу:

```python
счёт = {}
for слово in слова:
    счёт[слово] = счёт.get(слово, 0) + 1
```

Четыре строки заменяют вложенный цикл с `count` — и работают
в сотни раз быстрее на больших данных.

### Множество

> **Множество** — набор **уникальных** элементов без порядка.

```python
s = {1, 2, 3}
s.add(2)          # ничего не изменится: 2 уже есть
len({1, 2, 2, 3}) # 3
```

Пустое множество — только `set()`. Запись `{}` создаёт **пустой
словарь**, это классическая путаница.

### Операции над множествами

| Операция | Запись | Что даёт |
|---|---|---|
| объединение | `a \| b` | элементы обоих |
| пересечение | `a & b` | общие элементы |
| разность | `a - b` | те, что есть в a, но не в b |
| симметрическая разность | `a ^ b` | те, что только в одном |

Это буквально операции из теории множеств, которые вы рисовали
кругами Эйлера. Теперь их можно посчитать.

### Чем множество лучше списка

Проверка `x in ...` для списка — перебор, для множества —
практически мгновенная операция независимо от размера. Разница
на десятках тысяч элементов — сотни раз.

Цена — два ограничения: элементы не хранят порядок и должны быть
**неизменяемыми** (число, строка, кортеж; список — нельзя).

### Что выбрать

| Задача | Структура |
|---|---|
| элементы по порядку, нужны индексы | список |
| «есть ли такой» на больших данных | множество |
| убрать повторы | множество |
| «ключ → значение», подсчёт по категориям | словарь |
| пара значений, которую нельзя менять | кортеж |

## Смотрим, как это работает

### Пример 1. Словарь вместо двух списков

````python
баллы = {"Иванов": 92, "Петрова": 78, "Сидоров": 85}

print("Баллы Иванова:", баллы["Иванов"])
print("Кузнецов:", баллы.get("Кузнецов", "нет в списке"))

баллы["Кузнецов"] = 88
print("После добавления:", баллы)
print("Всего учеников:", len(баллы))
````

### Пример 2. Обход

````python
for фамилия, балл in баллы.items():
    оценка = 5 if балл >= 85 else 4
    print(f"{фамилия:<10} {балл:>3} → {оценка}")

print()
print("Средний балл:", sum(баллы.values()) / len(баллы))
print("Лучший:", max(баллы, key=баллы.get))
````

`max(баллы, key=баллы.get)` читается так: «максимум среди ключей,
сравнивая их по значению».

### Пример 3. Подсчёт по категориям

````python
текст = "мама мыла раму а рама мыла маму мама"

счёт = {}
for слово in текст.split():
    счёт[слово] = счёт.get(слово, 0) + 1

for слово in sorted(счёт, key=счёт.get, reverse=True):
    print(f"{слово:<8} {счёт[слово]}")
````

### Пример 4. Множества и повторы

````python
данные = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]

уникальные = set(данные)

print("Было:", данные, "—", len(данные), "элементов")
print("Стало:", sorted(уникальные), "—", len(уникальные), "элементов")
print("Повторов:", len(данные) - len(уникальные))
````

### Пример 5. Операции над множествами

````python
информатика = {"Иванов", "Петрова", "Сидоров"}
математика = {"Петрова", "Сидоров", "Кузнецов"}

print("Ходят хоть куда-то:", sorted(информатика | математика))
print("Ходят на оба:      ", sorted(информатика & математика))
print("Только информатика:", sorted(информатика - математика))
print("Только один кружок:", sorted(информатика ^ математика))
````

### Пример 6. Скорость проверки

````python
import time

размер = 50_000
список = list(range(размер))
множество = set(список)
запросы = [размер + i for i in range(2000)]      # заведомо отсутствующие

начало = time.perf_counter()
for x in запросы:
    x in список
по_списку = time.perf_counter() - начало

начало = time.perf_counter()
for x in запросы:
    x in множество
по_множеству = time.perf_counter() - начало

print(f"Проверка в списке:    {по_списку * 1000:>8.2f} мс")
print(f"Проверка в множестве: {по_множеству * 1000:>8.2f} мс")
print(f"Разница в {по_списку / по_множеству:,.0f} раз")
````

### Пример 7. Словарь как замена длинной цепочке if

````python
def оценка_словами_плохо(оценка):
    if оценка == 5:
        return "отлично"
    elif оценка == 4:
        return "хорошо"
    elif оценка == 3:
        return "удовлетворительно"
    return "плохо"


НАЗВАНИЯ = {5: "отлично", 4: "хорошо", 3: "удовлетворительно"}


def оценка_словами(оценка):
    return НАЗВАНИЯ.get(оценка, "плохо")


for о in [5, 4, 3, 2]:
    print(о, "→", оценка_словами(о), "|", оценка_словами_плохо(о))
````

Вторая версия короче, а главное — таблицу соответствий можно
менять, не трогая код.

### Любопытно

Почему проверка в множестве почти мгновенная? Потому что элемент
не ищут — его **вычисляют**. Хеш-функция превращает значение
в число, число указывает на ячейку памяти, и остаётся только
заглянуть туда.

Отсюда и требование неизменяемости: если элемент поменяется, его
хеш станет другим, и он «потеряется» — будет лежать в ячейке,
которую больше никто не проверит. Поэтому список в множество
положить нельзя, а кортеж — можно.
