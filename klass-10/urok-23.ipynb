## Разбираемся

### Дизайн — это не «покрасивее»

Слово «дизайн» в школе обычно означает «украсить». В работе оно
означает другое: **сделать так, чтобы человек справился с задачей
и не ошибся**.

Кнопка «Удалить» рядом с «Сохранить», форма, где непонятно, какое
поле обязательное, приложение, где не видно, что происходит после
нажатия, — это не «некрасиво». Это сломано.

Отсюда критерий, по которому оценивают макет: не «нравится —
не нравится», а **справился человек или нет**.

### Сначала структура, потом красота

Порядок работы над интерфейсом всегда один:

1. **Задача.** Что человек хочет сделать на этом экране?
2. **Содержание.** Какие данные и действия нужны для этого.
3. **Структура.** Что главное, что второстепенное, что рядом с чем.
4. **Вёрстка.** Сетка, отступы, размеры.
5. **Оформление.** Цвет, шрифт, иконки.

Сразу начать с пятого пункта — типичная ошибка новичка. Получается
красивый экран, на котором непонятно, что делать.

### Сетка и ритм

Всё в интерфейсе выравнивают по **сетке**: отступы, размеры
и положение элементов кратны одному числу. Обычно берут 8 —
отсюда «система восьмёрки»: 8, 16, 24, 32, 40 пикселей.

Зачем: глаз мгновенно замечает несовпадение в пару пикселей,
даже если не понимает, что именно не так. Единый шаг делает экран
спокойным, а работу — быстрой: не нужно каждый раз выбирать отступ.

### Близость и группировка

Элементы, которые связаны по смыслу, ставят **ближе друг к другу**,
чем к остальным. Подпись поля должна быть ближе к своему полю,
чем к соседнему, иначе человек прочтёт её не к тому.

Это самый мощный приём вёрстки, и он бесплатный: достаточно
правильно расставить отступы.

### Иерархия

На экране всегда есть главное. Его выделяют — но не всем сразу,
а одним-двумя способами: размером, жирностью, цветом, положением.

Если выделено всё — не выделено ничто. Три «очень важные» красные
кнопки на экране равны нулю важных кнопок.

### Типографика

| Правило | Почему |
|---|---|
| Не больше двух шрифтов | иначе экран разваливается |
| Размеры из короткого набора (12, 14, 16, 20, 24, 32) | тот же ритм, что у отступов |
| Строка 45–75 символов | длиннее — глаз теряет строку |
| Межстрочный интервал ≈ 1,4 от размера | текст «дышит» |
| Контраст текста к фону не ниже 4,5 : 1 | иначе не прочтут при солнце |

Последнее — не вкусовщина, а требование стандарта доступности
WCAG. Светло-серый текст на белом красив ровно до первого человека
со слабым зрением или до улицы в солнечный день.

### Состояния элемента

Кнопка — это не одна картинка, а несколько состояний:

| Состояние | Когда |
|---|---|
| обычное | по умолчанию |
| наведение | курсор над кнопкой |
| нажатие | в момент клика |
| отключено | действие сейчас недоступно |
| загрузка | запрос ушёл, ответа ещё нет |

Про последние два забывают чаще всего — и получают интерфейс,
где человек жмёт кнопку пять раз, потому что ничего не происходит.

### Figma: что нужно знать

**Figma** — редактор интерфейсов, который работает в браузере
и бесплатен для учебных задач. Четыре понятия закрывают
почти всё:

| Понятие | Что это |
|---|---|
| **фрейм** | экран или блок: рамка со своими размерами |
| **автолейаут** | фрейм сам расставляет содержимое с заданным отступом |
| **компонент** | образец элемента; копии обновляются вслед за ним |
| **прототип** | связи между экранами: нажал здесь — попал туда |

Автолейаут и компоненты — это ровно те же идеи, что цикл и функция
в программировании: не повторять руками то, что можно описать
один раз.

### Зачем это программисту

Макет — это техническое задание на вёрстку. Если в нём указаны
размеры, отступы, цвета и состояния, вёрстка занимает час.
Если макета нет, программист придумывает всё сам — и почти всегда
получается хуже и дольше.

## Смотрим, как это работает

Figma живёт в браузере, поэтому здесь мы посчитаем то, что в ней
задают глазами: сетку, контраст, размеры.

### Пример 1. Система восьмёрки

```python
шаг = 8

print("Допустимые отступы:", [шаг * k for k in range(1, 9)])
print()
for отступ in [8, 12, 16, 20, 24, 30]:
    подходит = "да" if отступ % шаг == 0 else "нет — выбивается из ритма"
    print(f"{отступ:>3} px → {подходит}")
```

### Пример 2. Высота экрана: помещается ли макет

```python
экран = 812          # высота экрана телефона в точках
шапка = 56
карточка = 88
отступ = 16

сколько = (экран - шапка) // (карточка + отступ)

print(f"Экран {экран}, шапка {шапка}")
print(f"Карточка {карточка} + отступ {отступ} = {карточка + отступ}")
print(f"Помещается карточек: {сколько}")
print("Остаток внизу:", экран - шапка - сколько * (карточка + отступ), "px")
```

Такой расчёт делают до рисования: он сразу показывает, сколько
элементов человек увидит без прокрутки.

### Пример 3. Контраст текста

Контраст считается по яркости двух цветов. Формула из стандарта
упрощённо выглядит так.

```python
def яркость(r, g, b):
    return 0.2126 * r + 0.7152 * g + 0.0722 * b


def контраст(цвет1, цвет2):
    я1 = яркость(*цвет1) / 255
    я2 = яркость(*цвет2) / 255
    светлее, темнее = max(я1, я2), min(я1, я2)
    return round((светлее + 0.05) / (темнее + 0.05), 2)


пары = [
    ("чёрный на белом", (0, 0, 0), (255, 255, 255)),
    ("тёмно-серый на белом", (51, 51, 51), (255, 255, 255)),
    ("светло-серый на белом", (170, 170, 170), (255, 255, 255)),
    ("белый на синем", (255, 255, 255), (0, 90, 180)),
]

for название, цвет, фон in пары:
    к = контраст(цвет, фон)
    вердикт = "проходит" if к >= 4.5 else "НЕ проходит"
    print(f"{название:<24} {к:>6} : 1  — {вердикт}")
```

Светло-серый текст на белом — самый частый способ испортить
интерфейс. Выглядит «стильно», не читается.

### Пример 4. Близость решает

```python
плохо = [
    "Имя",
    "",
    "[______]",
    "Фамилия",
    "",
    "[______]",
]

хорошо = [
    "Имя",
    "[______]",
    "",
    "Фамилия",
    "[______]",
]

print("Подпись одинаково далеко от обоих полей:")
for строка in плохо:
    print("   ", строка)

print()
print("Подпись ближе к своему полю:")
for строка in хорошо:
    print("   ", строка)
```

Один и тот же набор элементов. Разница только в том, где пустая
строка, — а читается совершенно по-разному.

### Пример 5. Сколько состояний рисовать

```python
состояния = ["обычное", "наведение", "нажатие", "отключено", "загрузка"]
кнопок_на_экране = 3

print("Состояний у одной кнопки:", len(состояния))
print("Всего вариантов на экране:", len(состояния) * кнопок_на_экране)
print()
print("Поэтому кнопку делают компонентом: правишь образец —")
print("меняются все копии во всех макетах сразу.")
```

### Пример 6. Длина строки

```python
текст = ("Дизайн интерфейса нужен не для красоты, а для того, "
         "чтобы человек справился с задачей и не ошибся")

for ширина in [30, 60, 100]:
    строк = -(-len(текст) // ширина)
    вердикт = "хорошо" if 45 <= ширина <= 75 else "плохо"
    print(f"{ширина:>4} символов в строке → {строк} строк, {вердикт}")
```
