## Разбираемся

### Дизайн — это не «покрасивее»

Слово «дизайн» в школе обычно означает «украсить». В работе оно
означает другое: **сделать так, чтобы человек справился с задачей
и не ошибся**.

Кнопка «Удалить» рядом с «Сохранить», форма, где непонятно, какое
поле обязательное, приложение, где не видно, что происходит после
нажатия, — это не «некрасиво». Это сломано.

### Сначала структура, потом красота

Порядок работы над интерфейсом всегда один:

1. **Задача.** Что человек хочет сделать на этом экране?
2. **Содержание.** Какие данные и действия нужны для этого.
3. **Структура.** Что главное, что второстепенное, что рядом с чем.
4. **Вёрстка.** Сетка, отступы, размеры.
5. **Оформление.** Цвет, шрифт, иконки.

Сразу начать с пятого пункта — типичная ошибка новичка. Получается
красивый экран, на котором непонятно, что делать.

### Сетка, близость, иерархия

* **Сетка.** Отступы и размеры кратны одному шагу, обычно 8:
  8, 16, 24, 32, 40 пикселей. Глаз замечает несовпадение в пару
  пикселей, даже если не понимает, что не так.
* **Близость.** Связанное по смыслу стоит ближе друг к другу, чем
  к остальному. Подпись поля ближе к своему полю, чем к соседнему.
* **Иерархия.** Главное на экране выделяют одним-двумя способами:
  размером, жирностью, цветом. Если выделено всё — не выделено ничто.

### Типографика

| Правило | Почему |
|---|---|
| Не больше двух шрифтов | иначе экран разваливается |
| Размеры из короткого набора (12, 14, 16, 20, 24, 32) | тот же ритм, что у отступов |
| Строка 45–75 символов | длиннее — глаз теряет строку |
| Контраст текста к фону не ниже 4,5 : 1 | иначе не прочтут при солнце |

Последнее — требование стандарта доступности WCAG. Светло-серый
текст на белом красив ровно до первого человека со слабым зрением.

### Состояния элемента

Кнопка — это не одна картинка, а несколько состояний: обычное,
наведение, нажатие, отключено, загрузка. Про последние два
забывают чаще всего — и получают интерфейс, где человек жмёт
кнопку пять раз, потому что ничего не происходит.

### Figma: четыре слова

**Figma** — редактор интерфейсов, который работает в браузере
и бесплатен для учебных задач. Интерфейс у неё английский, поэтому
ниже рядом с русским названием стоит английское — то, что вы
увидите на экране.

| Понятие | В Figma | Что это |
|---|---|---|
| фрейм | Frame | экран или блок: рамка со своими размерами |
| автолейаут | Auto layout | фрейм сам расставляет содержимое с заданным отступом |
| компонент | Component | образец элемента; копии обновляются вслед за ним |
| прототип | Prototype | связи между экранами: нажал здесь — попал туда |

Макет — это техническое задание на вёрстку. Если в нём указаны
размеры, отступы, цвета и состояния, вёрстка занимает час. Если
макета нет, программист придумывает всё сам — и почти всегда
получается хуже и дольше.
