## Разбираемся

### Зачем нужна операционная система

Представьте, что каждая программа сама умеет работать с диском,
клавиатурой, сетью и звуковой картой — с каждой моделью отдельно.
Именно так и было в 1950-е: программу писали под конкретную машину,
и на соседней она не запускалась.

**Операционная система** — это посредник. Она берёт на себя работу
с железом и распределяет его между программами, а программам даёт
одинаковые правила обращения, независимо от того, какой стоит диск.

### Четыре задачи ОС

| Задача | Что делает система |
|---|---|
| процессы | запускает программы и переключает между ними процессор |
| память | выдаёт каждой программе её область и не даёт залезть в чужую |
| файлы | хранит данные на диске в виде файлов и папок |
| устройства | общается с железом через драйверы |

Плюс к этому — пользователи и права доступа, сеть, безопасность.

### Как несколько программ работают одновременно

На одном ядре в каждый момент выполняется ровно **одна** программа.
Ощущение одновременности создаёт **вытесняющая многозадачность**:
система даёт программе несколько миллисекунд, потом прерывает её,
сохраняет состояние и передаёт процессор следующей. За секунду таких
переключений сотни.

Программа при этом ничего не замечает — её как будто поставили
на паузу и потом продолжили. Именно поэтому зависшую программу можно
закрыть: система в любой момент способна отобрать у неё процессор.

### Виртуальная память

Каждой программе система показывает её собственное адресное
пространство, начинающееся с нуля. Настоящие адреса в микросхемах
памяти другие — соответствие поддерживает процессор вместе с ОС.

Отсюда две важные вещи:

* программы изолированы: чужую память просто не видно;
* если оперативной памяти не хватает, редко используемые страницы
  уходят на диск (это называют файлом подкачки). Работает всё
  по-прежнему, только в тысячи раз медленнее — тот самый случай,
  когда компьютер «намертво задумался».

### Виды программного обеспечения

| Вид | Примеры |
|---|---|
| системное | операционная система, драйверы, утилиты обслуживания |
| прикладное | браузер, редактор, игры, мессенджеры |
| инструментальное | компиляторы, среды разработки, отладчики |

По условиям распространения ПО делят иначе: проприетарное (продают
лицензию), свободное (можно изучать, менять, распространять),
бесплатное с закрытым кодом.

### Unix и его потомки

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g10/unix.png" width="420" alt="Unix шестой редакции на PDP-11 в эмуляторе">

*Unix шестой редакции на PDP-11 в эмуляторе*

<sub>Huihermit · CC0 · Wikimedia Commons</sub>

В 1969 году Кен Томпсон и Деннис Ритчи написали в лаборатории Bell
Labs небольшую систему для машины PDP-11. Она называлась Unix
и держалась на нескольких идеях:

* всё есть файл — и документ, и принтер, и диск;
* маленькие программы, каждая делает одну вещь хорошо;
* программы соединяются в цепочки, выход одной идёт на вход другой;
* система написана на языке высокого уровня (C), а не в машинных
  кодах, поэтому её можно перенести на другую машину.

На скриншоте — та самая система в эмуляторе. Из неё выросли Linux,
macOS, Android и iOS, а команды `ls`, `cd` и `cat` работают
и сегодня — в том числе в Colab.

### Виртуализация и облако

**Виртуальная машина** — компьютер, которого нет: программа,
притворяющаяся железом. На одном физическом сервере их запускают
десятки, и каждая думает, что работает одна.

Colab устроен именно так: когда вы нажимаете «Подключиться», Google
выдаёт вам виртуальную машину с Linux на несколько часов. Отсюда все
её особенности:

* файлы в `/content` исчезают после отключения;
* можно устанавливать любые библиотеки — ломать нечего;
* машина может отключиться сама, если долго простаивает.

Это и есть облако: вы арендуете вычисления на время, а не покупаете
железо.

### Пример 3. Процессы и переключения

Посмотрим, как ОС переключается между задачами. Запустим счётчик
и посмотрим, сколько процессорного времени он получил.

````python
import time

начало_стены = time.time()
начало_процессора = time.process_time()

сумма = 0
for i in range(3_000_000):
    сумма += i

прошло_стены = time.time() - начало_стены
прошло_процессора = time.process_time() - начало_процессора

print(f"По часам на стене:  {прошло_стены:.3f} с")
print(f"Процессорное время: {прошло_процессора:.3f} с")
print(f"Доля процессора:    {прошло_процессора / прошло_стены * 100:.0f}%")
````

Процессорное время — то, сколько наша программа реально считала.
Разница со временем «по часам» ушла другим процессам: система
отбирала процессор и отдавала его соседям.

### Пример 4. Файлы системы вместо диспетчера задач

В Linux всё есть файл, включая сведения о процессоре и памяти.

````python
def прочитать_поле(путь, что_искать):
    try:
        with open(путь, encoding="utf-8") as файл:
            for строка in файл:
                if что_искать in строка:
                    return строка.strip()
    except FileNotFoundError:
        return f"{путь} — нет такого файла (не Linux)"
    return "не найдено"


print(прочитать_поле("/proc/cpuinfo", "model name"))
print(прочитать_поле("/proc/meminfo", "MemTotal"))
````

Никакого специального интерфейса не понадобилось: обычное чтение
файла — и мы знаем модель процессора. Это и есть тот самый принцип
Unix «всё есть файл», которому больше пятидесяти лет.
