### Умножение и деление на степень основания

Самое полезное свойство позиционной записи:

| Действие | Что происходит с записью |
|---|---|
| умножить на q | приписать ноль справа |
| умножить на $q^n$ | приписать n нулей |
| разделить нацело на q | отбросить последнюю цифру |
| остаток от деления на q | последняя цифра |
| остаток от деления на $q^n$ | последние n цифр |

Последние две строки особенно полезны. «Найдите последнюю цифру числа
$7^{100}$ в восьмеричной системе» — это вопрос об остатке от деления
на 8, и его можно посчитать, не вычисляя само число.

### Делимость и последние цифры

В десятичной системе число делится на 10, если оканчивается нулём.
Общее правило: **число делится на q тогда и только тогда, когда его
запись в системе с основанием q оканчивается нулём**.

Обобщение: число делится на $q^n$, если запись оканчивается
n нулями.

### Задачи с уравнением

> В системе счисления с основанием q выполняется равенство
> $23_q + 14_q = 40_q$. Найдите q.

Переводим каждое число в десятичную форму:

$$(2q + 3) + (1q + 4) = 4q + 0$$
$$3q + 7 = 4q$$
$$q = 7$$

Проверяем: в семеричной системе $23_7 = 17$, $14_7 = 11$,
$40_7 = 28$. И действительно $17 + 11 = 28$.

**Обязательно проверяйте ответ** и следите за допустимостью цифр:
если бы получилось q = 3, ответ был бы неверным, потому что цифра 4
в троичной системе не существует.

### Сравнение чисел из разных систем

Единственный надёжный способ — привести к общей системе.
Но есть и быстрые оценки:

* при одинаковой длине записи больше то число, у которого больше
  основание;
* при одинаковом основании больше то, у которого запись длиннее;
* $1000_2 = 8$, а $1000_8 = 512$ — разница огромна.

### Сумма цифр и остаток

Полезное свойство, обобщающее школьный признак делимости на 9:

> Число сравнимо со суммой своих цифр по модулю (q − 1).

В десятичной системе это даёт привычное: число делится на 9, если
сумма его цифр делится на 9. В двоичной (q−1 = 1) свойство вырождается,
а вот в шестнадцатеричной работает для делимости на 15.

## Смотрим, как это работает

### Пример 1. Уравнение с основанием

```python
# 23q + 14q = 40q — перебираем допустимые основания
print("Проверяем равенство 23 + 14 = 40 в разных системах:")
for q in range(5, 17):          # цифра 4 требует основания не меньше 5
    первое = 2 * q + 3
    второе = 1 * q + 4
    сумма = 4 * q + 0
    отметка = "  ← подходит" if первое + второе == сумма else ""
    print(f"  q = {q:>2}: {первое:>3} + {второе:>3} = {первое + второе:>3}, "
          f"а 40 означает {сумма:>3}{отметка}")
```

Перебор начат с 5, а не с 2 — потому что в записи есть цифра 4,
и меньшие основания недопустимы. Это сокращение перебора нужно делать
всегда: оно и ускоряет, и защищает от ложных ответов.

### Пример 2. Последние цифры без вычисления числа

> Найдите две последние цифры числа $3^{50}$ в девятеричной системе.

```python
основание = 9
число_бы = 3 ** 50          # его можно и не вычислять, но для проверки удобно

# Последние две цифры — это остаток от деления на 9^2
остаток = pow(3, 50, основание ** 2)
цифры = ""
временное = остаток
for _ in range(2):
    цифры = str(временное % основание) + цифры
    временное //= основание

print(f"3^50 = {число_бы}")
print(f"Остаток от деления на 81: {остаток}")
print(f"Две последние цифры в девятеричной: {цифры}")

# Проверка полным переводом
полная = ""
временное = число_бы
while временное:
    полная = str(временное % 9) + полная
    временное //= 9
print(f"Полная запись оканчивается на: {полная[-2:]}")
```

Функция `pow(a, b, m)` вычисляет $a^b \bmod m$, **не вычисляя**
само $a^b$ — она использует умный алгоритм с последовательным
возведением в квадрат. Для настоящих задач с огромными степенями
это единственный способ получить ответ.

### Пример 3. Признак делимости через сумму цифр

```python
def сумма_цифр(число, основание):
    сумма = 0
    while число > 0:
        сумма += число % основание
        число //= основание
    return сумма


print(f"{'число':>8} {'сумма цифр':>12} {'остаток числа':>15} {'остаток суммы':>15}")
for число in [123, 999, 1000, 12345]:
    s = сумма_цифр(число, 10)
    print(f"{число:>8} {s:>12} {число % 9:>15} {s % 9:>15}")
```

Столбцы остатков совпали для всех чисел — свойство работает.
Именно на нём основан школьный признак делимости на 9,
и он же даёт быструю проверку арифметики: если остатки не сошлись,
где-то ошибка в вычислениях.
