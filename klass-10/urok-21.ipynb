## Разбираемся

### Зачем функции на самом деле

Обычное объяснение — «чтобы не повторять код». Оно верное,
но неполное. Главная польза функции в другом: она **прячет
сложность за именем**.

Прочитав `нод(a, b)`, вы понимаете строку, ничего не зная
про алгоритм Евклида. Программу из пятидесяти функций можно читать
сверху вниз как текст, а программу из тысячи строк подряд —
нельзя.

### Аргументы: позиционные и именованные

```python
def площадь(ширина, высота):
    return ширина * высота

площадь(3, 4)                      # позиционно
площадь(высота=4, ширина=3)        # по именам, порядок не важен
```

### Значения по умолчанию

```python
def площадь(ширина, высота=1):
    return ширина * высота

площадь(5)        # 5 — высота взялась по умолчанию
площадь(5, 3)     # 15
```

Параметры со значением по умолчанию идут **после** обычных —
иначе Python не поймёт, что кому передали.

> **Ловушка.** Значением по умолчанию не делают список:
> `def f(данные=[])`. Такой список создаётся **один раз**
> и потом живёт между вызовами, накапливая мусор. Правильно —
> `def f(данные=None)` и внутри `if данные is None: данные = []`.

### Возврат нескольких значений

```python
def минмакс(числа):
    return min(числа), max(числа)

а, б = минмакс([3, 8, 1])
```

На самом деле возвращается один кортеж, который распаковывается
по переменным.

Функция без `return` возвращает `None` — «ничего». Это частая
причина загадочных ошибок: забыли `return`, а дальше пытаются
сложить результат с числом.

### Области видимости

> **Локальная переменная** — созданная внутри функции. Снаружи
> её не существует.

```python
счёт = 10                # глобальная

def испортить():
    счёт = 99            # создаётся новая, локальная
    return счёт

испортить()              # 99
print(счёт)              # 10 — глобальная не изменилась
```

Это защита, а не досадное ограничение: функция не может случайно
испортить чужие данные.

Читать глобальные переменные функция может, а вот менять —
только объявив `global`. Но пользоваться `global` в нормальном коде
не нужно: если функции нужно значение, его передают аргументом,
а результат возвращают через `return`.

### Правило поиска имени

Python ищет имя по цепочке: сначала **локально** внутри функции,
потом в **объемлющей** функции, потом **глобально**, потом среди
**встроенных**. Первое найденное и побеждает.

Отсюда практический вывод: не называйте переменные `list`, `sum`,
`max`, `str`. Такое имя перекроет встроенную функцию, и она
перестанет работать во всей программе.

### Изменяемые аргументы

```python
def добавить(список):
    список.append(0)      # меняет тот самый список снаружи

def заменить(список):
    список = [1, 2, 3]    # а это создаёт новый, снаружи ничего
```

Списки передаются по ссылке: функция получает не копию, а тот же
самый объект. Изменение содержимого видно снаружи, переприсваивание
имени — нет.

Это одна из самых частых причин «почему у меня список испортился».

## Смотрим, как это работает

### Пример 1. Аргументы и значения по умолчанию

````python
def приветствие(имя, обращение="Привет"):
    return f"{обращение}, {имя}!"


print(приветствие("Аня"))
print(приветствие("Аня", "Здравствуйте"))
print(приветствие(обращение="Салют", имя="Борис"))
````

### Пример 2. Возврат нескольких значений

````python
def статистика(числа):
    return min(числа), max(числа), sum(числа) / len(числа)


наименьшее, наибольшее, среднее = статистика([3, 8, 1, 9, 4])

print(f"минимум {наименьшее}, максимум {наибольшее}, среднее {среднее}")
print("На самом деле вернулся кортеж:", статистика([1, 2, 3]))
````

### Пример 3. Забыли return

````python
def сумма_без_return(a, b):
    a + b        # результат вычислен и выброшен


результат = сумма_без_return(2, 3)
print("Вернулось:", результат, "типа", type(результат).__name__)
````

`None` — самая частая причина ошибки
`unsupported operand type(s) for +: 'NoneType' and 'int'`.
Если увидели такую — ищите функцию без `return`.

### Пример 4. Локальное и глобальное

````python
счёт = 10


def попытка_изменить():
    счёт = 99
    return счёт


print("Внутри функции:", попытка_изменить())
print("Снаружи:", счёт)


def прочитать():
    return счёт * 2


print("Читать глобальную можно:", прочитать())
````

### Пример 5. Список приходит по ссылке

````python
данные = [1, 2, 3]


def изменить_содержимое(список):
    список.append(99)


def переприсвоить(список):
    список = [0, 0, 0]


изменить_содержимое(данные)
print("После append:", данные)

переприсвоить(данные)
print("После переприсваивания:", данные, "— снаружи ничего не изменилось")
````

### Пример 6. Ловушка изменяемого умолчания

````python
def плохо(значение, копилка=[]):
    копилка.append(значение)
    return копилка


print("Первый вызов: ", плохо(1))
print("Второй вызов: ", плохо(2), "← а ожидали [2]")


def хорошо(значение, копилка=None):
    if копилка is None:
        копилка = []
    копилка.append(значение)
    return копилка


print()
print("Первый вызов: ", хорошо(1))
print("Второй вызов: ", хорошо(2))
````

### Пример 7. Разбиение задачи на функции

Задача: найти все простые числа до n. Решим её тремя маленькими
функциями вместо одной большой.

````python
def делится(n, k):
    return n % k == 0


def простое(n):
    if n < 2:
        return False
    for k in range(2, int(n ** 0.5) + 1):
        if делится(n, k):
            return False
    return True


def простые_до(n):
    return [x for x in range(2, n + 1) if простое(x)]


print(простые_до(50))
````

Каждую функцию можно проверить отдельно, а читается это почти
как условие задачи.

### Любопытно

Идея, что программу надо резать на маленькие независимые куски,
кажется очевидной — но её пришлось изобретать. В 1968 году
на конференции в Гармише впервые заговорили о «кризисе
программного обеспечения»: проекты срывали сроки, а код становился
таким, что его не понимал даже автор.

Тогда же Эдсгер Дейкстра написал знаменитую заметку «О вреде
оператора goto»: он предлагал отказаться от прыжков по программе
в пользу функций и структурных конструкций. Спорили лет десять.

Сегодня `goto` нет почти нигде, а функции есть везде — и никто
уже не помнит, что это был спорный вопрос.
