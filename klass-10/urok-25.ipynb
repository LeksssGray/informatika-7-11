## Разбираемся

### Строка неизменяема

Главное свойство строки, из которого следует всё остальное:
**строку нельзя изменить**. `слово[0] = "к"` — ошибка. Любой метод
не меняет строку, а возвращает новую.

Отсюда практическое следствие: `строка += кусок` в цикле — дорогая
операция. Каждый раз создаётся новая строка целиком. Для тысячи
кусков это тысяча копирований, и правильный способ — собрать список
и склеить его один раз через `join`.

### Срезы

```python
s[2:5]      # символы 2, 3, 4
s[:3]       # первые три
s[-3:]      # последние три
s[::2]      # каждый второй
s[::-1]     # перевёрнутая строка
```

Срез никогда не выходит за границы: `s[:100]` для короткой строки
вернёт её целиком, а не ошибку. Это отличает срез от индекса —
`s[100]` обрушит программу.

### Поиск

| Запись | Что даёт | Если не нашлось |
|---|---|---|
| `x in s` | `True` / `False` | `False` |
| `s.find(x)` | индекс первого вхождения | `-1` |
| `s.index(x)` | индекс первого вхождения | **ошибка** |
| `s.rfind(x)` | индекс последнего вхождения | `-1` |
| `s.count(x)` | сколько раз встречается | `0` |

`find` и `index` отличаются только поведением при неудаче —
и поэтому `find` безопаснее.

### Проверки

```python
s.startswith("при")     # начинается ли
s.endswith(".txt")      # заканчивается ли
s.isdigit()             # состоит ли только из цифр
s.isalpha()             # только из букв
s.isupper()             # все буквы заглавные
```

`isdigit` — стандартный способ проверить, можно ли превратить
строку в число, прежде чем звать `int()`.

### Изменяющие методы

```python
s.upper()               # "ПРИВЕТ"
s.lower()               # "привет"
s.strip()               # убрать пробелы по краям
s.replace("a", "b")     # заменить все вхождения
s.replace("a", "b", 1)  # заменить только первое
```

`strip()` незаменим при обработке ввода: пользователь почти всегда
добавляет лишний пробел, и сравнение внезапно даёт `False`.

### Разбиение и сборка

```python
"a,b,c".split(",")          # ["a", "b", "c"]
"мама мыла раму".split()    # по любым пробелам
"-".join(["2026", "09"])    # "2026-09"
```

`split()` без аргументов умнее, чем `split(" ")`: он схлопывает
несколько пробелов подряд и игнорирует их по краям.

### Форматирование

```python
f"{x:>10}"      # прижать вправо, ширина 10
f"{x:<10}"      # влево
f"{x:^10}"      # по центру
f"{x:*^10}"     # по центру, заполнить звёздочками
f"{3.14159:.2f}"   # два знака
```

Такое выравнивание превращает печать таблицы в одну строку кода
вместо ручного подсчёта пробелов.

### Строки и символы

```python
ord("A")        # 65
chr(97)         # "a"
```

Сравнение строк идёт по кодам символов, посимвольно слева направо.
Поэтому `"Яблоко" < "яблоко"` истинно: заглавные буквы в Unicode
стоят раньше строчных. Именно из-за этого список слов, отсортированный
«по алфавиту», иногда выглядит странно — и лечится сортировкой
по `str.lower`.

## Смотрим, как это работает

### Пример 1. Срезы

````python
s = "информатика"

print("s        =", s)
print("s[2:5]   =", s[2:5])
print("s[:3]    =", s[:3])
print("s[-4:]   =", s[-4:])
print("s[::2]   =", s[::2])
print("s[::-1]  =", s[::-1])
print("s[:100]  =", s[:100], "— срез не выходит за границы")
````

### Пример 2. Поиск

````python
фраза = "мама мыла раму"

print("'мыла' in фраза:", "мыла" in фраза)
print("find('мыла'):   ", фраза.find("мыла"))
print("find('папа'):   ", фраза.find("папа"), "— не нашлось")
print("rfind('ма'):    ", фраза.rfind("ма"), "— последнее вхождение")
print("count('ма'):    ", фраза.count("ма"))
````

### Пример 3. Проверки перед преобразованием

````python
ввод = ["42", " 42 ", "4.2", "сорок два", ""]

for значение in ввод:
    чистое = значение.strip()
    если_число = чистое.isdigit()
    итог = int(чистое) if если_число else "не число"
    print(f"{значение!r:<12} → {итог}")
````

### Пример 4. Разбор строки

````python
строка = "  Иванов ;  9А ;  85  "

части = [часть.strip() for часть in строка.split(";")]
фамилия, класс, балл = части

print("Части:", части)
print(f"Фамилия: {фамилия!r}, класс: {класс!r}, балл: {int(балл)}")
````

Без `.strip()` в каждой части остались бы пробелы, и `int(" 85 ")`
ещё сработал бы, а вот сравнение `класс == "9А"` — уже нет.

### Пример 5. Склейка: медленно и быстро

````python
import time

кусков = 30_000

начало = time.perf_counter()
медленно = ""
for i in range(кусков):
    медленно += "x"
время_склейки = time.perf_counter() - начало

начало = time.perf_counter()
куски = []
for i in range(кусков):
    куски.append("x")
быстро = "".join(куски)
время_join = time.perf_counter() - начало

print(f"Склейка в цикле: {время_склейки * 1000:>7.1f} мс")
print(f"Через join:      {время_join * 1000:>7.1f} мс")
print("Результаты одинаковые:", медленно == быстро)
````

### Пример 6. Выравнивание

````python
данные = [("Иванов", 92), ("Петрова", 78), ("Кузнецова", 95)]

print(f"{'Фамилия':<12}|{'Балл':>6}")
print("-" * 19)
for фамилия, балл in данные:
    print(f"{фамилия:<12}|{балл:>6}")

print()
print(f"{' ИТОГО ':*^19}")
````

### Пример 7. Сортировка строк

````python
слова = ["яблоко", "Яблоко", "банан", "Арбуз"]

print("Как есть:       ", sorted(слова))
print("Без учёта регистра:", sorted(слова, key=str.lower))
print()
print('Почему так: ord("Я") =', ord("Я"), ', ord("я") =', ord("я"))
````

### Любопытно

Неизменяемость строк кажется неудобством, но она даёт важную вещь:
одинаковые строки можно хранить в памяти **один раз**. Python так
и делает с короткими строками и именами переменных — это называется
интернированием.

Побочный эффект: сравнение одинаковых строк часто оказывается
мгновенным, потому что сравниваются не символы, а адреса.
Если бы строки можно было менять, такой фокус был бы невозможен:
одно изменение испортило бы все копии сразу.
