## Повторяем

### 1. Типы и операции

`//` округляет **вниз**: `-9 // 4 == -3`, а остаток берёт знак делителя:
`-9 % 4 == 3`. Деление `/` всегда даёт `float`.

### 2. Ветвления и циклы

Количество чисел с условием — счётчик или формула: двузначных, кратных
6, — `99 // 6 - 9 // 6`. Цифры числа — `% 10` и `// 10` в `while`.

### 3. Функции и области видимости

Переменная, которой присвоили значение внутри функции, —
**локальная**: снаружи её не видно, и внешнюю с тем же именем она
не меняет. Параметр со значением по умолчанию: `def f(a, b=3)`.

### 4. Списки и их методы

`sort()` и `reverse()` меняют список на месте и возвращают `None`.
`b = a` не копирует: это второе имя того же списка; копия — `list(a)`
или `a[:]`. Генератор: `[x for x in range(20) if x % 3 == 0]`.

### 5. Двумерные списки

`m[i][j]` — строка i, столбец j. Главная диагональ — `m[i][i]`.
`len(m)` — строк, `len(m[0])` — столбцов.

### 6. Строки

Срезы с шагом: `s[::2]` — каждый второй, `s[::-1]` — задом наперёд.
`split()` делит по пробелам, `join` склеивает, f-строки форматируют.

### 7. Словари и множества

`d.get(k, 0)` — значение или 0. Множество не хранит повторов:
`len(set("мама"))` — 2. Пересечение `&`, объединение `|`.

### 8. Таблицы и интерфейсы

`=СУММЕСЛИ(где; условие; что)`, `=ОКРУГЛ(x; 0)`. В Figma экран
собирают во **фрейме**.
