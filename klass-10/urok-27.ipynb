### Словарь как таблица

Словарь хранит пары «ключ — значение». Обратиться можно по ключу, а не по номеру:
`баллы["Иванова"]`. Чтобы посчитать частоту значений, заводят пустой словарь и на каждом
шаге увеличивают счётчик по ключу. Ключа ещё нет — его нужно сначала завести.

```python
слова = ["кот", "пёс", "кот", "кот"]
частоты = {}
for слово in слова:
    частоты[слово] = частоты.get(слово, 0) + 1
print(частоты)
```
