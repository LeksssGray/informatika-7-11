## Разбираемся

### Звук — это волна

Звук в природе непрерывен: колебания давления воздуха плавно меняются
во времени. Компьютер же хранит только дискретные числа. Значит,
непрерывную волну нужно превратить в набор чисел — этот процесс
называется **оцифровкой**.

Она состоит из двух шагов, и каждый вносит свою потерю точности.

### Шаг 1. Дискретизация по времени

Мы измеряем громкость волны не непрерывно, а через равные промежутки —
делаем «снимки». Количество снимков в секунду называется **частотой
дискретизации** и измеряется в герцах.

```
  Волна:      ╱‾╲    ╱‾╲
             ╱   ╲__╱   ╲

  Замеры:    •  •  •  •  •  •  •  •
             ↑
       частота дискретизации = сколько таких точек в секунду
```

| Частота | Где применяется |
|---|---|
| 8 кГц | телефонная связь |
| 22,05 кГц | среднее качество |
| 44,1 кГц | аудио-CD, музыка |
| 48 кГц | видео, профессиональный звук |
| 96–192 кГц | студийная запись |

Почему именно 44,1 кГц? Человек слышит частоты примерно до 20 кГц,
а **теорема Котельникова** утверждает: чтобы восстановить сигнал
без потерь, частота дискретизации должна быть **вдвое выше** самой
высокой частоты сигнала. 20 × 2 = 40 кГц, плюс небольшой запас.

### Шаг 2. Квантование по уровню

Каждый замер — вещественное число, а хранить его нужно целым.
Диапазон громкости делят на уровни, и **глубина кодирования** —
это количество бит на один замер.

| Глубина | Уровней | Качество |
|---|---|---|
| 8 бит | 256 | телефон, заметный шум |
| 16 бит | 65 536 | CD, стандарт |
| 24 бита | 16,7 млн | студия |

Связь та же: $N = 2^i$.

### Главная формула

$$V = f \cdot i \cdot t \cdot k$$

где f — частота дискретизации (Гц), i — глубина кодирования (бит),
t — длительность (с), k — количество каналов.

Каналов обычно один (моно) или два (стерео).

**Пример.** Минута стереозвука в качестве CD:

$$V = 44100 \cdot 16 \cdot 60 \cdot 2 = 84\,672\,000 \text{ бит} = 10\,584\,000 \text{ байт} \approx 10 \text{ Мбайт}$$

Отсюда, кстати, вместимость компакт-диска: 700 Мбайт делённые
на 10 Мбайт в минуту дают около 70 минут музыки. Стандарт CD
был рассчитан ровно так.

### Почему MP3 в десять раз меньше

Несжатый звук (формат WAV) занимает те самые 10 Мбайт в минуту.
MP3 сжимает в 10–12 раз — за счёт **психоакустики**: алгоритм
отбрасывает то, что ухо всё равно не различит.

* Частоты выше 16–18 кГц взрослый человек не слышит.
* Тихий звук сразу после громкого маскируется и не воспринимается.
* Разница между каналами в низких частотах не различима.

Это сжатие **с потерями**: исходный сигнал восстановить невозможно.
Для архивного хранения используют форматы без потерь — FLAC, ALAC:
они сжимают вдвое, но сохраняют всё.

### Битрейт

Часто вместо объёма указывают **битрейт** — количество бит в секунду:

$$\text{битрейт} = f \cdot i \cdot k$$

Для CD это 1411 кбит/с. У MP3 обычно 128–320 кбит/с. Разделив битрейт
на 8, получим байты в секунду, а умножив на длительность — объём файла.

## Смотрим, как это работает

### Пример 1. Объём аудиофайла

````python
def объём_звука(частота, глубина, секунд, каналов):
    бит = частота * глубина * секунд * каналов
    return бит / 8 / 1024 / 1024      # в мегабайтах


варианты = [
    ("Телефон, моно, 1 мин", 8000, 8, 60, 1),
    ("Радио, моно, 1 мин", 22050, 16, 60, 1),
    ("CD, стерео, 1 мин", 44100, 16, 60, 2),
    ("Студия, стерео, 1 мин", 96000, 24, 60, 2),
]

print(f"{'вариант':<26} {'Мбайт':>10} {'битрейт, кбит/с':>18}")
for название, f, i, t, k in варианты:
    мбайт = объём_звука(f, i, t, k)
    битрейт = f * i * k / 1000
    print(f"{название:<26} {мбайт:>10.2f} {битрейт:>18.0f}")
````

Разброс огромен: студийная запись занимает почти в тридцать раз больше
телефонной. При этом на слух разница между CD и студией для большинства
людей неразличима — что и служит оправданием для сжатия.

### Пример 2. Обратная задача

> Файл занимает 5 Мбайт, записан в моно с частотой 22050 Гц
> и глубиной 16 бит. Какова длительность записи?

````python
объём_мбайт = 5
частота, глубина, каналов = 22050, 16, 1

бит = объём_мбайт * 1024 * 1024 * 8
секунд = бит / (частота * глубина * каналов)

print(f"Всего бит: {бит:,}")
print(f"Бит в секунду: {частота * глубина * каналов:,}")
print(f"Длительность: {секунд:.1f} с = {секунд / 60:.1f} мин")
````

Схема прежняя: перевести всё в биты, затем разделить на известное.
Обратите внимание, что мегабайты переводятся в байты умножением
на 1024 дважды, и только потом в биты.

### Пример 3. Подбор параметров под ограничение

> Нужно записать 10 минут звука в моно так, чтобы файл поместился
> на дискету объёмом 1,44 Мбайта. Какие параметры выбрать?

````python
предел_бит = 1.44 * 1024 * 1024 * 8
секунд = 600

print(f"Доступно бит: {предел_бит:,.0f}")
print(f"Бит в секунду: не более {предел_бит / секунд:,.0f}")
print()

for частота in [8000, 11025, 22050, 44100]:
    for глубина in [8, 16]:
        нужно = частота * глубина * секунд
        помещается = "да" if нужно <= предел_бит else "нет"
        print(f"  {частота:>6} Гц, {глубина:>2} бит → "
              f"{нужно / 8 / 1024 / 1024:>6.2f} Мбайт  {помещается}")
````

Результат неожиданный: **не подходит ни один вариант**. Даже самое
скромное телефонное качество даёт 4,6 Мбайта — втрое больше дискеты.

Вывод из этого важнее самой задачи: несжатый звук огромен, и без сжатия
хранить его практически невозможно. Чтобы уложиться в 1,44 Мбайта,
нужен либо MP3, либо запись вчетверо короче.

Перебор вариантов полезен именно этим — он честно показывает, когда
задача не имеет решения в заданных рамках.
