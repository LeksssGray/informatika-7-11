## Разбираемся

### Изображение как таблица чисел

Растровое изображение — прямоугольная сетка точек, **пикселей**.
У каждого пикселя есть цвет, а цвет — это число. Значит картинка
в памяти — просто большая таблица чисел.

Три параметра определяют всё:

* **разрешение** — сколько пикселей по горизонтали и вертикали;
* **глубина цвета** — сколько бит отведено на один пиксель;
* **палитра** — сколько различных цветов доступно.

### Глубина цвета и количество цветов

Связь та же самая, что между битами и вариантами:

$$N = 2^i$$

где i — глубина цвета в битах, N — количество различимых цветов.

| Глубина | Цветов | Где применяется |
|---|---|---|
| 1 бит | 2 | чёрно-белые факсы |
| 4 бита | 16 | старые интерфейсы EGA |
| 8 бит | 256 | GIF, индексированные палитры |
| 16 бит | 65 536 | High Color |
| 24 бита | 16 777 216 | True Color, фотографии |
| 32 бита | 16,7 млн + прозрачность | графика с альфа-каналом |

24 бита — это по 8 бит на каждый из трёх каналов RGB. Ровно поэтому
каждая составляющая цвета принимает значения от 0 до 255,
а код цвета в вебе состоит из трёх пар шестнадцатеричных цифр.

### Главная формула

$$V = W \cdot H \cdot i$$

Объём равен ширине, умноженной на высоту и на глубину цвета.
Результат в битах — не забывайте делить на 8 для байтов.

**Пример.** Фотография 1920 × 1080 в True Color:

$$V = 1920 \cdot 1080 \cdot 24 = 49\,766\,400 \text{ бит} = 6\,220\,800 \text{ байт} \approx 5{,}9 \text{ Мбайт}$$

Почему же реальный файл с фотографии весит 2 Мбайта, а не 6?
Потому что он сжат.

### Сжатие: с потерями и без

**Сжатие без потерь** сохраняет каждый пиксель в точности. Работает
на повторах: вместо «сто белых пикселей подряд» записывается
«белый × 100». Форматы: PNG, GIF, BMP с упаковкой.

**Сжатие с потерями** отбрасывает детали, которые человеческий глаз
почти не замечает, и достигает гораздо более сильного уменьшения.
Формат: JPEG.

| | Без потерь | С потерями |
|---|---|---|
| Качество | не меняется | ухудшается |
| Степень сжатия | в 2–3 раза | в 10–20 раз |
| Пересохранение | безопасно | накапливает искажения |
| Для чего | схемы, скриншоты, текст | фотографии |

> ⚠️ Скриншот с текстом в JPEG — плохая идея: вокруг букв появляются
> «грязные» ореолы. Для резких границ нужен PNG.

### Векторная графика

Совершенно другой подход: хранить не пиксели, а **описание фигур**.
Круг задаётся центром, радиусом и цветом — тремя числами вместо
тысяч пикселей.

| | Растр | Вектор |
|---|---|---|
| Что хранится | цвета пикселей | формулы фигур |
| Масштабирование | пикселизация | идеальное качество |
| Фотографии | да | нет |
| Логотипы, схемы | плохо | отлично |
| Размер файла | зависит от разрешения | зависит от сложности |
| Форматы | PNG, JPEG, BMP | SVG, EPS, AI |

Логотипы всегда делают в векторе: один файл годится и для визитки,
и для рекламного щита.

### Разрешение и физический размер

Разрешение печати измеряют в **dpi** — точках на дюйм.
Изображение 1200 × 1800 пикселей при 300 dpi даст отпечаток
размером 4 × 6 дюймов, то есть примерно 10 × 15 см.

Формула: физический размер = пиксели ÷ dpi.

## Смотрим, как это работает

### Пример 1. Объём изображения

```python
def объём_изображения(ширина, высота, глубина):
    бит = ширина * высота * глубина
    return бит, бит / 8, бит / 8 / 1024, бит / 8 / 1024 / 1024


варианты = [
    ("Иконка 32×32, 8 бит", 32, 32, 8),
    ("Аватар 256×256, 24 бита", 256, 256, 24),
    ("Full HD, 24 бита", 1920, 1080, 24),
    ("4K, 32 бита", 3840, 2160, 32),
]

print(f"{'изображение':<28} {'байт':>12} {'Кбайт':>10} {'Мбайт':>8}")
for название, ш, в, г in варианты:
    _, байт, кбайт, мбайт = объём_изображения(ш, в, г)
    print(f"{название:<28} {байт:>12,.0f} {кбайт:>10.1f} {мбайт:>8.2f}")
```

Обратите внимание на скачок между Full HD и 4K: пикселей вчетверо
больше, и объём вырос вчетверо. Именно поэтому видео в 4K требует
такого широкого канала связи.

Формат `:>12,.0f` вставляет разделители тысяч — читать большие числа
становится намного легче.

### Пример 2. Обратная задача

> Файл занимает 900 Кбайт, изображение имеет размер 640 × 480.
> Какова глубина цвета и сколько цветов в палитре?

```python
объём_кбайт = 900
ширина, высота = 640, 480

бит_всего = объём_кбайт * 1024 * 8
пикселей = ширина * высота
глубина = бит_всего / пикселей

print(f"Всего бит:      {бит_всего:,}")
print(f"Пикселей:       {пикселей:,}")
print(f"Глубина цвета:  {глубина:.0f} бит")
print(f"Цветов:         {2 ** int(глубина):,}")
```

Схема решения обратных задач всегда одна: перевести всё в биты,
затем делить на известное, чтобы найти неизвестное. Главное —
не запутаться в единицах: килобайты в байты, байты в биты.

### Пример 3. Задача с ограничением

> Какое наибольшее разрешение можно выбрать, чтобы изображение
> с глубиной 24 бита поместилось в 1 Мбайт, если стороны относятся
> как 4:3?

```python
предел_байт = 1024 * 1024
глубина = 24

# Пусть ширина = 4k, высота = 3k
k = 1
while (4 * k) * (3 * k) * глубина / 8 <= предел_байт:
    k += 1
k -= 1

print(f"Максимальный множитель k = {k}")
print(f"Разрешение: {4 * k} × {3 * k}")
print(f"Объём: {(4 * k) * (3 * k) * глубина / 8 / 1024:.1f} Кбайт")
```

Здесь применён приём подбора: увеличиваем параметр, пока не нарушится
условие, затем откатываемся на шаг. Так решаются многие задачи
с ограничениями — и это законный метод, если вариантов конечное число.
