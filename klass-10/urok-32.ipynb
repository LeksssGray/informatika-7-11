## Разбираемся

### Рекурсия

Рекурсивная функция вызывает саму себя, но с меньшими данными. Обязательны два элемента:
**базовый случай**, на котором функция останавливается, и **шаг**, который приводит к нему.

```python
def сумма_до(n):
    if n <= 0:
        return 0
    return n + сумма_до(n - 1)

print(сумма_до(5))
```

### Перебор и модуль itertools

Чтобы перебрать все перестановки, удобно использовать `itertools.permutations`. Он возвращает
все упорядоченные наборы элементов, и остаётся только собрать их в нужный вид.

### Мемоизация

Фибоначчи, посчитанные рекурсией напрямую, повторяют одни и те же вызовы много раз. Если запомнить
уже вычисленные значения в словаре, каждый вызов выполнится один раз.
