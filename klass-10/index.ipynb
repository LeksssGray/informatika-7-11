### К июню будешь уметь

- рассчитывать объём текста, графики и звука при любом способе кодирования
- решать логические уравнения и преобразовывать выражения
- свободно пользоваться списками, словарями, множествами и файлами
- реализовывать и оценивать алгоритмы поиска и сортировки

Уроки идут по порядку и опираются друг на друга — перепрыгивать через них не стоит.

### Как здесь работать

Подробно, с картинками, это разобрано в [первом уроке](https://colab.research.google.com/github/LeksssGray/informatika-7-11/blob/main/klass-10/urok-01.ipynb). Коротко — один и тот же порядок на каждом уроке:

1. Открыть урок по ссылке из таблицы ниже.
2. Нажать 🔑 на панели слева и включить переключатель у секрета `INFORMATIKA` (как его завести — шаг 5 первого урока). Секрета нет — вписать ФИО и класс в первую ячейку.
3. Запустить первую ячейку кнопкой ▶. На предупреждение «Google не имеет отношения к созданию этого блокнота» ответить **«Выполнить»**, на запрос доступа к аккаунту — **«Разрешить»**.
4. Идти по ячейкам сверху вниз, решать задачи и запускать проверки. В конце — «Отчёт по уроку».

Нужен Google-аккаунт: без него кнопки запуска работать не будут.

Результат каждой проверки уходит учителю сразу, так что сохранять урок не обязательно. Хочешь оставить свои решения себе — «Файл → Сохранить копию на Диске».
