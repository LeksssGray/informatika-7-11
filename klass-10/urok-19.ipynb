## Разбираемся

### Условия и циклы на уровне ЕГЭ

`if` и `for` вы пишете с седьмого класса. В десятом добавляется
то, что отличает аккуратный код от кода, который «вроде работает»:
досрочный выход, `else` у цикла, вложенные условия и перебор
с проверкой.

### Цепочки сравнений

```python
if 0 <= x <= 100:
if a == b == c:
```

Python проверяет всю цепочку целиком — это не то же самое,
что `a == b and b == c` только по виду, но ровно то же по смыслу.
А вот `5 < 7 < 6` — ложь, потому что второе сравнение не выполнено.

### Досрочный выход

| Слово | Что делает |
|---|---|
| `break` | немедленно выйти из цикла |
| `continue` | пропустить остаток тела и перейти к следующему шагу |
| `return` | выйти из функции (и заодно из цикла) |

Типичное применение — поиск: как только нашли, продолжать незачем.

```python
for i in range(len(данные)):
    if данные[i] == цель:
        найдено = i
        break
```

### `else` у цикла

Редкая, но очень полезная конструкция:

```python
for x in числа:
    if x < 0:
        print("нашлось отрицательное")
        break
else:
    print("отрицательных нет")
```

`else` у цикла срабатывает, **если цикл закончился сам**, без
`break`. Читается как «если не нашли». Без него пришлось бы заводить
переменную-флажок.

### Флажок

Если `else` у цикла кажется странным, ту же задачу решают флажком:

```python
нашли = False
for x in числа:
    if x < 0:
        нашли = True
        break
```

Оба способа правильные. Главное — не писать `if нашли == True`:
`if нашли` короче и понятнее.

### Перебор с проверкой

Классика ЕГЭ: найти все числа в диапазоне, удовлетворяющие условию.

```python
for n in range(1, 1000):
    if условие(n):
        print(n)
```

Обычно требуется наименьшее или наибольшее такое число — тогда
либо `break` на первом найденном, либо перебор с конца
(`range(999, 0, -1)`).

### Вложенность и как её избегать

```python
# три уровня — тяжело читать
for x in данные:
    if x > 0:
        if x % 2 == 0:
            ...

# один уровень — то же самое
for x in данные:
    if x > 0 and x % 2 == 0:
        ...

# и ещё вариант: ранний выход
for x in данные:
    if x <= 0:
        continue
    ...
```

Правило: если отступов стало больше трёх, почти всегда условие
можно объединить или перевернуть.

### Бесконечный цикл с выходом

```python
while True:
    ...
    if условие:
        break
```

Выглядит пугающе, но это законный и частый приём, когда условие
выхода удобнее проверять в середине тела. Главное — чтобы `break`
действительно достигался.

## Смотрим, как это работает

### Пример 1. Цепочки сравнений

````python
for x in [-5, 0, 50, 100, 101]:
    print(f"x = {x:>4}  0 <= x <= 100 → {0 <= x <= 100}")

print()
print("5 < 7 < 6 →", 5 < 7 < 6, "— второе сравнение ложно")
````

### Пример 2. Поиск с досрочным выходом

````python
данные = [4, 8, 15, 16, 23, 42]
цель = 16

шагов = 0
найдено = -1
for i in range(len(данные)):
    шагов += 1
    if данные[i] == цель:
        найдено = i
        break

print(f"Индекс: {найдено}, просмотрено элементов: {шагов} из {len(данные)}")
````

### Пример 3. else у цикла

````python
for набор in [[3, 5, -2, 8], [3, 5, 8]]:
    for x in набор:
        if x < 0:
            print(f"{набор} → есть отрицательное: {x}")
            break
    else:
        print(f"{набор} → отрицательных нет")
````

### Пример 4. continue вместо вложенности

````python
данные = [-3, 4, 7, 12, -8, 20]

# вложенный вариант
сумма_1 = 0
for x in данные:
    if x > 0:
        if x % 2 == 0:
            сумма_1 += x

# плоский вариант
сумма_2 = 0
for x in данные:
    if x <= 0 or x % 2 != 0:
        continue
    сумма_2 += x

print("Сумма положительных чётных:", сумма_1, "и", сумма_2)
````

### Пример 5. Наименьшее подходящее число

Найдём наименьшее число больше 1000, которое делится на 7 и на 11.

````python
for n in range(1001, 10000):
    if n % 7 == 0 and n % 11 == 0:
        print("Наименьшее:", n)
        break
````

А теперь наибольшее трёхзначное с тем же свойством — тот же перебор,
но с конца.

````python
for n in range(999, 99, -1):
    if n % 7 == 0 and n % 11 == 0:
        print("Наибольшее трёхзначное:", n)
        break
````

### Пример 6. while True с выходом

Делим число пополам, пока оно чётное, и считаем шаги.

````python
n = 1000
шагов = 0

while True:
    if n % 2 != 0:
        break
    n = n // 2
    шагов += 1

print(f"Делили {шагов} раз, осталось {n}")
````

### Пример 7. Сколько раз выполнится

````python
for a, b, шаг in [(2, 20, 3), (1, 10, 1), (10, 0, -2), (5, 5, 1)]:
    повторов = len(range(a, b, шаг))
    значения = list(range(a, b, шаг))
    print(f"range({a}, {b}, {шаг}) → {повторов} повторов: {значения}")
````

`len(range(...))` считает количество без перебора — удобно
для проверки себя в задачах ЕГЭ.
