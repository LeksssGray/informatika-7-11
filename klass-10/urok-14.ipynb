### Идея, которой восемьдесят лет

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g10/fon-neyman.jpg" width="220" alt="Джон фон Нейман">

*Джон фон Нейман*

<sub>wikispaces · CC BY-SA 3.0 · Wikimedia Commons</sub>

Первые вычислительные машины программировали проводами: чтобы решать
другую задачу, ENIAC перекоммутировали несколько дней. В 1945 году
Джон фон Нейман, участвовавший в проекте следующей машины, изложил
в отчёте принцип, который всё изменил: **программа хранится в той же
памяти, что и данные**.

Идея обсуждалась целой группой инженеров — Экертом, Мокли и другими, —
но в историю принцип вошёл под именем фон Неймана, потому что подписан
отчёт был только им. С тех пор так устроены почти все компьютеры,
включая ваш телефон.

### Принципы фон Неймана

| Принцип | Что означает |
|---|---|
| однородность памяти | программа и данные лежат в одной памяти и неразличимы по виду |
| адресность | память разбита на пронумерованные ячейки, к любой можно обратиться по номеру |
| программное управление | процессор выполняет команды по порядку, пока не встретит команду перехода |
| двоичное кодирование | и команды, и данные записаны нулями и единицами |

Из однородности следует любопытное: программа может менять сама
себя — просто записав в память новые байты. Так делали в 1950-е ради
экономии памяти, а сегодня операционные системы это, наоборот,
запрещают: слишком удобная лазейка для вирусов.

### Из чего состоит компьютер

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g10/materinka.jpg" width="420" alt="Материнская плата с процессором и слотами">

*Материнская плата с процессором и слотами*

<sub>Andrew Dunn · CC BY-SA 2.0 · Wikimedia Commons</sub>

* **процессор** — считает и управляет;
* **оперативная память** — хранит программы и данные, пока включено
  питание;
* **внешняя память** — диск, SSD, флешка: хранит и после выключения;
* **устройства ввода-вывода** — клавиатура, экран, сеть;
* **шина** — набор проводников, по которым всё это обменивается
  данными.

Шин на самом деле три: **шина данных** (что передаём), **шина
адреса** (куда) и **шина управления** (что именно делать — читать
или писать). Разрядность шины адреса определяет, сколько памяти
компьютер вообще способен видеть: 32 бита — это 2³² адресов, то есть
4 гигабайта. Именно поэтому 32-разрядные системы не работают
с большим объёмом памяти.

### Внутри процессора

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g10/kristall.jpg" width="380" alt="Кристалл процессора Intel 286 под микроскопом">

*Кристалл процессора Intel 286 под микроскопом*

<sub>rasimick · CC BY 2.0 · Wikimedia Commons</sub>

| Блок | Задача |
|---|---|
| АЛУ (арифметико-логическое устройство) | складывает, сравнивает, выполняет логические операции |
| устройство управления | читает команды и командует всем остальным |
| регистры | несколько десятков сверхбыстрых ячеек внутри процессора |
| кэш | небольшая быстрая память для копий данных из оперативной |

Процессор работает по циклу: **выбрать команду → расшифровать →
выполнить → перейти к следующей**. Миллиарды раз в секунду.

### Иерархия памяти

Быстрая память дорога, дешёвая — медленна. Выход один: сделать
несколько уровней и держать нужное поближе.

| Уровень | Объём | Время доступа | Во сколько раз медленнее регистров |
|---|---|---|---|
| регистры | сотни байт | доли наносекунды | 1 |
| кэш L1 | десятки КБ | ~1 нс | ~3 |
| кэш L3 | десятки МБ | ~10 нс | ~30 |
| оперативная память | гигабайты | ~60 нс | ~200 |
| SSD | сотни ГБ | ~50 мкс | ~150 000 |
| жёсткий диск | терабайты | ~5 мс | ~15 000 000 |

Чтобы почувствовать разницу, переведём в человеческие масштабы:
если обращение к регистру — одна секунда, то к оперативной памяти —
три минуты, к SSD — два дня, а к жёсткому диску — полгода.

Работает всё это благодаря **локальности**: программы почти всегда
обращаются к данным, которые лежат рядом с только что использованными.
Поэтому маленький кэш успевает обслужить большинство обращений.

### Что на самом деле влияет на скорость

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g10/ibm-pc.jpg" width="360" alt="IBM PC 5150, 1981 год">

*IBM PC 5150, 1981 год*

<sub>Ruben de Rijcke · CC BY-SA 3.0 · Wikimedia Commons</sub>

У первого IBM PC 1981 года было 16 килобайт памяти и процессор
на 4,77 МГц. Современный телефон быстрее в десятки тысяч раз — но
ощущения «мгновенно» это не даёт, потому что программы выросли ещё
сильнее.

На практике скорость определяют:

* **тип накопителя** — замена жёсткого диска на SSD меняет ощущения
  сильнее, чем любая другая деталь;
* **объём памяти** — когда её не хватает, система начинает
  использовать диск вместо памяти, и всё замирает;
* **число ядер** — помогает, только если программа умеет считать
  параллельно;
* **частота процессора** — важна для одиночных тяжёлых задач;
* **видеокарта** — для игр, видео и нейросетей.

Частота в гигагерцах давно перестала расти: около 2005 года упёрлись
в тепло. Дальше производители пошли вширь — стали добавлять ядра.

## Смотрим, как это работает

### Пример 1. Считаем, сколько памяти адресуется

```python
for разрядов in (16, 20, 32, 48, 64):
    адресов = 2 ** разрядов
    print(f"{разрядов:>2} бит адреса → {адресов:>20,} ячеек = "
          f"{адресов / 1024 ** 3:.3g} ГБ".replace(",", " "))
```

64-разрядная шина адреса даёт запас на десятилетия вперёд: реальные
процессоры используют не все 64 бита, обычно 48.

### Пример 2. Иерархия памяти в человеческих масштабах

```python
уровни = [
    ("регистр", 0.3),
    ("кэш L1", 1),
    ("кэш L3", 10),
    ("оперативная память", 60),
    ("SSD", 50_000),
    ("жёсткий диск", 5_000_000),
]

база = 0.3
print(f"{'Уровень':<20} {'нс':>12} {'если регистр = 1 с':>22}")
for название, нс in уровни:
    во_сколько = нс / база
    секунд = во_сколько
    if секунд < 60:
        по_человечески = f"{секунд:.0f} с"
    elif секунд < 3600:
        по_человечески = f"{секунд / 60:.0f} мин"
    elif секунд < 86400:
        по_человечески = f"{секунд / 3600:.1f} ч"
    else:
        по_человечески = f"{секунд / 86400:.0f} дней"
    print(f"{название:<20} {нс:>12,.0f} {по_человечески:>22}".replace(",", " "))
```

### Пример 3. Локальность видно на замерах

Пройдём по большому списку двумя способами: подряд и вразнобой.
Действий одинаково, а времени уйдёт разное — потому что подряд идущие
данные попадают в кэш.

```python
import random
import time

размер = 2_000_000
данные = list(range(размер))
порядок_подряд = list(range(размер))
порядок_вразнобой = порядок_подряд[:]
random.shuffle(порядок_вразнобой)

for название, порядок in (("подряд", порядок_подряд), ("вразнобой", порядок_вразнобой)):
    начало = time.time()
    сумма = 0
    for i in порядок:
        сумма += данные[i]
    print(f"{название:<11} {time.time() - начало:.2f} с")
```

Разница обычно в полтора-два раза, хотя сложений столько же.
В программах на C, где нет накладных расходов Python, разрыв доходит
до десяти раз. Отсюда практическое правило: обрабатывайте данные
в том порядке, в котором они лежат в памяти.

### Пример 4. Что за железо под Colab

Colab — это виртуальная машина в дата-центре Google. Посмотрим,
что нам досталось.

```python
import os

print("Ядер процессора:", os.cpu_count())

try:
    with open("/proc/meminfo", encoding="utf-8") as файл:
        для_памяти = файл.readline()
    килобайт = int(для_памяти.split()[1])
    print(f"Оперативной памяти: {килобайт / 1024 / 1024:.1f} ГБ")
except FileNotFoundError:
    print("Файл /proc/meminfo есть только в Linux — на своём компьютере "
          "посмотрите в диспетчере задач")
```

В Linux сведения о железе лежат в обычных файлах — их можно просто
прочитать. Если запускаете урок не в Colab, а на Windows, второй
части вывода не будет: там таких файлов нет.

### Любопытно

«Узкое место фон Неймана» — так называют главную проблему этой
архитектуры: процессор и память соединены одной шиной, и процессор
постоянно ждёт данные. За восемьдесят лет разрыв только вырос: память
стала быстрее в сотни раз, процессоры — в десятки тысяч. Поэтому
современный процессор больше чем наполовину состоит из кэша: это
не вычислитель с памятью, а скорее память с вычислителем.
