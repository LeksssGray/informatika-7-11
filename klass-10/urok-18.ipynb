### Типы и их проверка

| Тип | Что хранит | Пример |
|---|---|---|
| `int` | целое любой длины | `2 ** 1000` посчитается точно |
| `float` | вещественное, приближённо | `0.1 + 0.2` даст `0.30000000000000004` |
| `str` | строку | `"42"` — это не число |
| `bool` | истину или ложь | `True`, `False` |

Две особенности Python, которых нет в большинстве языков:

**Целые не переполняются.** `2 ** 1000` — это точное число
из 302 цифр. В C или Java такое давно бы переполнило переменную.

**`bool` — это подвид `int`.** `True == 1` истинно, а `sum([True,
True, False])` даёт 2. На этом основан приём подсчёта: сумма условий
равна количеству истинных.

### Три деления

```python
17 / 5      # 3.4   — всегда float
17 // 5     # 3     — целая часть вниз
17 % 5      # 2     — остаток
```

Тонкость, которую спрашивают: как ведут себя `//` и `%`
с отрицательными числами.

```python
-17 // 5    # -4, а не -3: округление идёт вниз, а не к нулю
-17 % 5     # 3, а не -2: остаток в Python неотрицательный
```

Зато всегда верно тождество: `a == (a // b) * b + a % b`.

### Приведение типов

```python
int("42")       # 42
int(3.9)        # 3 — дробная часть отбрасывается, не округляется
round(3.9)      # 4 — а вот это округление
float("3.14")   # 3.14
str(42)         # "42"
```

`int()` и `round()` путают постоянно: `int(-3.7)` даст −3,
а `round(-3.7)` даст −4.

И ещё: `round` в Python округляет «к чётному»: `round(2.5)` — это 2,
а `round(3.5)` — 4. Так придумано, чтобы ошибки при массовых
округлениях не копились в одну сторону.

### Ввод нескольких значений

```python
a, b = input().split()          # две строки
a, b = map(int, input().split())   # два числа
числа = list(map(int, input().split()))   # сколько угодно чисел
```

`map(int, ...)` применяет `int` к каждой части. Это стандартная
строчка на олимпиадах и в ЕГЭ: запомните её как есть.

### Множественное присваивание

```python
a, b = 1, 2
a, b = b, a          # обмен без третьей переменной
x = y = 0            # обоим ноль
```

В первой строке справа на самом деле кортеж `(1, 2)`, который
распаковывается по переменным. Отсюда и работает обмен: правая
часть вычисляется целиком **до** присваивания.

### Форматированный вывод

```python
x = 3.14159
print(f"{x:.2f}")        # 3.14   — два знака
print(f"{x:8.3f}")       # ‹  3.142› — ширина 8
print(f"{42:05d}")       # 00042  — с ведущими нулями
print(f"{1234567:,}")    # 1,234,567
print(f"{255:b}")        # 11111111 — двоичная запись
print(f"{255:x}")        # ff — шестнадцатеричная
```

Последние две строки — удобный способ проверить себя в задачах
на системы счисления, которые мы разбирали в начале года.

## Смотрим, как это работает

### Пример 1. Типы и их сюрпризы

````python
print("2 ** 100 =", 2 ** 100)
print("Тип 10 / 2:", type(10 / 2).__name__, "→", 10 / 2)
print("Тип 10 // 2:", type(10 // 2).__name__, "→", 10 // 2)
print()
print("0.1 + 0.2 =", 0.1 + 0.2)
print("Равно ли 0.3:", 0.1 + 0.2 == 0.3)
print("Сравнивать дробные надо так:", abs((0.1 + 0.2) - 0.3) < 1e-9)
````

### Пример 2. True — это единица

````python
оценки = [5, 4, 5, 3, 5]

пятёрок = sum(оценка == 5 for оценка in оценки)

print("Оценки:", оценки)
print("Пятёрок:", пятёрок)
print("Потому что True == 1:", True == 1, "и False == 0:", False == 0)
````

### Пример 3. Деление отрицательных

````python
for a in [17, -17]:
    for b in [5, -5]:
        print(f"{a:>4} // {b:>3} = {a // b:>4}    {a:>4} % {b:>3} = {a % b:>4}"
              f"    проверка: {(a // b) * b + a % b}")
````

Последний столбец всюду равен исходному числу — тождество
выполняется при любых знаках.

### Пример 4. int против round

````python
for x in [3.9, -3.9, 2.5, 3.5, -2.5]:
    print(f"x = {x:>5}   int(x) = {int(x):>3}   round(x) = {round(x):>3}")
````

Строки с 2.5 и 3.5 — то самое округление к чётному.
На ЕГЭ это не спрашивают, но в расчётах знать полезно.

### Пример 5. Разбор строки ввода

Настоящий `input()` здесь не запустить — автопроверка пропускает
такие ячейки. Поэтому подставим строку руками: разбор от этого
не меняется.

````python
строка = "12 45 7 90 3"

числа = list(map(int, строка.split()))

print("Строка:", repr(строка))
print("Числа:", числа)
print("Сумма:", sum(числа), " Максимум:", max(числа))

a, b = map(int, "10 20".split())
print(f"a = {a}, b = {b}")
````

### Пример 6. Обмен значений

````python
a, b = 1, 2
print("До обмена: ", a, b)

a, b = b, a
print("После:     ", a, b)
````

### Пример 7. Форматирование

````python
x = 3.14159
n = 255

print(f"{x:.2f}")
print(f"{x:10.4f}|")
print(f"{42:05d}")
print(f"{1234567:,}".replace(",", " "))
print(f"{n} = {n:b} (дв.) = {n:o} (восьм.) = {n:x} (шестн.)")
````
