## Разбираемся

### Какой график выбрать

* **Линия** — изменение во времени: температура, цена, рост.
* **Столбцы** — сравнение категорий: выручка магазинов.
* **Точечная диаграмма** — связь двух величин: рост и вес.
* **Гистограмма** — распределение значений по интервалам: оценки по диапазонам.

### Честность графика

График врёт, если ось начинается не с нуля там, где это меняет смысл, если масштаб скрыт или
если подписи не сообщают единиц. Всегда подписывай оси и указывай, что показано.

```python
# colab-only: рисует matplotlib
import matplotlib.pyplot as plt

месяцы = ["янв", "фев", "мар"]
выручка = [120, 150, 135]
plt.bar(месяцы, выручка)
plt.title("Выручка по месяцам")
plt.show()
```
