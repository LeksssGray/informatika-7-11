### Словарь в Python

**Словарь** хранит пары «ключ → значение». По ключу значение находится
мгновенно, сколько бы в словаре ни было записей.

````python
телефоны = {
    "Аня": "8-900-111-22-33",
    "Боря": "8-900-444-55-66",
    "Вика": "8-900-777-88-99",
}

print(телефоны["Боря"])
print("Записей:", len(телефоны))
````

Пишется в фигурных скобках, ключ от значения отделяется двоеточием.
Обращение — по ключу в квадратных скобках, как у списка, только вместо
номера ключ.

### Основные действия

| Действие | Запись |
|---|---|
| пустой словарь | `d = {}` |
| добавить или изменить | `d["Гриша"] = "8-900-000"` |
| получить значение | `d["Гриша"]` |
| получить безопасно | `d.get("Гриша", "нет номера")` |
| проверить наличие ключа | `"Гриша" in d` |
| удалить | `del d["Гриша"]` |
| количество записей | `len(d)` |
| перебрать ключи | `for ключ in d:` |
| перебрать пары | `for ключ, значение in d.items():` |

Главная ловушка: обращение к несуществующему ключу — ошибка
`KeyError`. Поэтому там, где ключа может не быть, пользуются `.get()`:
он вернёт значение по умолчанию вместо ошибки.

### Что может быть ключом

Ключ должен быть **неизменяемым**: строка, число, кортеж. Список
ключом быть не может. Значение — что угодно, хоть список, хоть другой
словарь.

Ключи в словаре уникальны. Если записать один ключ дважды, останется
последнее значение — никакой ошибки не будет.

### Словарь или список

| Задача | Что брать |
|---|---|
| хранить оценки по порядку | список |
| хранить оценки по предметам | словарь |
| найти элемент по номеру | список |
| найти данные по имени, коду, слову | словарь |
| посчитать, сколько раз что встречается | словарь |

Правило: если данные ищут **по имени**, а не по порядку — нужен словарь.

## Смотрим, как это работает

### Пример 1. Телефонная книга

````python
книга = {"Аня": "111-22-33", "Боря": "444-55-66"}

книга["Вика"] = "777-88-99"          # добавили
книга["Аня"] = "111-00-00"           # изменили
del книга["Боря"]                    # удалили

for имя, номер in книга.items():
    print(f"{имя:<6} {номер}")

print()
print("Есть Боря?", "Боря" in книга)
print("Номер Бори:", книга.get("Боря", "не найден"))
````

`.items()` на каждом шаге отдаёт сразу пару — ключ и значение.
Форматирование `{имя:<6}` выравнивает имена по левому краю в шесть
позиций, чтобы номера встали столбиком.

### Пример 2. Частотный словарь букв

Классическая задача: сколько раз каждая буква встречается в тексте.

````python
текст = "Толковый словарь живого великорусского языка"

частоты = {}
for буква in текст.lower():
    if буква == " ":
        continue
    частоты[буква] = частоты.get(буква, 0) + 1

print(частоты)
````

Вся работа — в одной строке: `частоты.get(буква, 0) + 1`. Если буква
встречалась, берём её счётчик, если нет — ноль, и в обоих случаях
прибавляем единицу. Без `.get()` пришлось бы писать `if` с проверкой.

### Пример 3. Сортировка по значению

Словарь не упорядочивает записи по значению — это делаем мы сами.

````python
пары = sorted(частоты.items(), key=lambda пара: пара[1], reverse=True)

print("Самые частые буквы:")
for буква, сколько in пары[:5]:
    print(f"  {буква} — {сколько}")
````

`sorted` получил список пар и указание, по чему сравнивать: `key`
берёт из пары второй элемент, то есть счётчик. `reverse=True`
переворачивает порядок — от большего к меньшему.

Запись `lambda пара: пара[1]` — это функция без имени, написанная
в одну строку. То же самое можно сделать обычной функцией:

```python
def по_счётчику(пара):
    return пара[1]

пары = sorted(частоты.items(), key=по_счётчику, reverse=True)
```

### Пример 4. Оценки по предметам

````python
оценки = {
    "информатика": [5, 5, 4],
    "математика": [4, 4, 5, 4],
    "история": [3, 4],
}

for предмет, список in оценки.items():
    средний = sum(список) / len(список)
    print(f"{предмет:<13} {средний:.2f}")

лучший = max(оценки, key=lambda предмет: sum(оценки[предмет]) / len(оценки[предмет]))
print("\nЛучший средний балл:", лучший)
````

Значение словаря — целый список. Так в одной переменной уместился
весь дневник.
