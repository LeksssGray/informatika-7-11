## Разбираемся

### Что спрашивают на экзамене

В ОГЭ есть задания, где дают текстовый файл с данными и просят
ответить по нему на вопрос: сколько строк удовлетворяют условию,
какое наибольшее число встречается, сколько раз попалось слово.
Файл большой — руками его не разобрать, нужна программа.

Формулировки меняются от года к году, а приёмы — нет. Их всего пять,
и все они уже были на прошлых уроках.

### Пять приёмов

| Что просят | Чем решается |
|---|---|
| посчитать подходящие строки | счётчик + `if` |
| найти максимум или минимум | переменная-рекордсмен |
| посчитать сумму или среднее | накопитель `сумма` |
| посчитать, сколько раз что встретилось | словарь частот |
| отобрать строки и записать в новый файл | список + запись |

### Шаблон решения

Держите его в голове целиком — он подходит почти к любой задаче
на файл:

```python
ответ = 0                                    # накопитель
with open("данные.txt", encoding="utf-8") as файл:
    for строка in файл:
        строка = строка.strip()
        if not строка:
            continue
        поля = строка.split(";")             # разделитель из условия!
        # ... разбор и проверка условия
print(ответ)
```

### Три ошибки, которые стоят баллов

1. **Забыли `int()`.** `"10" > "9"` — истина, потому что строки
   сравниваются по буквам. У чисел всё наоборот. Сравнивать нужно
   числа, а не строки.
2. **Забыли `strip()`.** `int("5\n")` сработает, а вот `"5\n" == "5"`
   — нет, и сравнение строк развалится.
3. **Не учли пустую строку в конце файла.** Проверка `if not строка:
   continue` стоит одну секунду и спасает от ошибки.

### Как проверять себя

На экзамене нет автопроверки. Приём: сделайте маленький файл
из трёх-четырёх строк, где ответ считается в уме, и запустите
программу на нём. Если на маленьком файле ответ верный — значит,
алгоритм правильный, и можно запускать на большом.

## Смотрим, как это работает

### Готовим данные

Сделаем два файла: один маленький для проверки, один побольше —
на нём и будем решать.

````python
маленький = """Аня;7А;информатика;5
Боря;7Б;математика;3
Вика;7А;информатика;4
"""

with open("мал.txt", "w", encoding="utf-8") as файл:
    файл.write(маленький)

большой = """Аня;7А;информатика;5
Боря;7Б;математика;3
Вика;7А;информатика;4
Гриша;7Б;информатика;5
Даша;7А;математика;4
Егор;7Б;физика;2
Жанна;7А;информатика;3
Зоя;7Б;математика;5
Игорь;7А;физика;4
Катя;7Б;информатика;5
Лёша;7А;математика;2
Марина;7Б;информатика;4
"""

with open("бол.txt", "w", encoding="utf-8") as файл:
    файл.write(большой)

print("Готово: мал.txt (3 записи) и бол.txt (12 записей)")
````

### Пример 1. Сколько строк удовлетворяют условию

> Сколько учеников получили по информатике оценку выше тройки?

Сначала на маленьком файле: там таких двое — Аня и Вика. Проверяем.

````python
def сильных_по_информатике(имя_файла):
    счётчик = 0
    with open(имя_файла, encoding="utf-8") as файл:
        for строка in файл:
            строка = строка.strip()
            if not строка:
                continue
            имя, класс, предмет, оценка = строка.split(";")
            if предмет == "информатика" and int(оценка) > 3:
                счётчик += 1
    return счётчик


print("мал.txt:", сильных_по_информатике("мал.txt"))
print("бол.txt:", сильных_по_информатике("бол.txt"))
````

На маленьком файле получилось 2 — как мы и посчитали в уме. Значит,
ответу на большом файле можно верить.

Обратите внимание: `encoding="utf-8"` написан, а режим `"r"` — нет.
Чтение — режим по умолчанию, его можно не указывать.

### Пример 2. Максимум и кто его набрал

````python
лучшая_оценка = 0
лучший = ""

with open("бол.txt", encoding="utf-8") as файл:
    for строка in файл:
        строка = строка.strip()
        if not строка:
            continue
        имя, класс, предмет, оценка = строка.split(";")
        if int(оценка) > лучшая_оценка:
            лучшая_оценка = int(оценка)
            лучший = имя

print(f"Первая максимальная оценка {лучшая_оценка} — у {лучший}")
````

Строгое `>` оставляет первого из рекордсменов, `>=` оставило бы
последнего. В задании всегда сказано, какой ответ нужен, — читайте
внимательно.

### Пример 3. Словарь частот: сколько записей по каждому предмету

````python
по_предметам = {}

with open("бол.txt", encoding="utf-8") as файл:
    for строка in файл:
        строка = строка.strip()
        if not строка:
            continue
        имя, класс, предмет, оценка = строка.split(";")
        по_предметам[предмет] = по_предметам.get(предмет, 0) + 1

for предмет in sorted(по_предметам):
    print(f"{предмет:<13} {по_предметам[предмет]}")

самый_частый = max(по_предметам, key=lambda п: по_предметам[п])
print("\nЧаще всего встречается:", самый_частый)
````

### Пример 4. Отбор строк в новый файл

> Выпишите в отдельный файл все записи класса 7А.

````python
отобранные = []

with open("бол.txt", encoding="utf-8") as файл:
    for строка in файл:
        строка = строка.strip()
        if строка and строка.split(";")[1] == "7А":
            отобранные.append(строка)

with open("только_7а.txt", "w", encoding="utf-8") as файл:
    файл.write("\n".join(отобранные) + "\n")

print(f"Отобрано строк: {len(отобранные)}")
with open("только_7а.txt", encoding="utf-8") as файл:
    print(файл.read())
````

`строка.split(";")[1]` — второе поле без распаковки всей строки.
Так короче, когда нужно ровно одно поле.
