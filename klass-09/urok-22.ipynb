## Разбираемся

### Зачем вообще диаграмма

Таблица на двадцать строк — это двадцать чисел, которые надо
сравнивать глазами. Диаграмма показывает то же самое так, что
вывод виден за секунду: этот столбик выше, эта линия растёт,
этот кусок — почти половина круга.

Но у диаграммы есть и обратная сторона. Она **убеждает** сильнее,
чем таблица, и потому ею удобно врать. Во второй половине урока
разберём, как именно, — чтобы уметь распознавать.

### Какой тип выбрать

Тип диаграммы выбирают не по красоте, а по вопросу, на который
она отвечает.

| Вопрос | Тип | Пример |
|---|---|---|
| Что больше, что меньше? | **столбчатая** | средний балл по классам |
| Как меняется со временем? | **линейная** (график) | температура по дням |
| Из чего состоит целое? | **круговая** | доли ответов в опросе |
| Связаны ли две величины? | **точечная** | рост и вес учеников |
| Как распределены значения? | **гистограмма** | сколько человек набрали 60–70 баллов |

Два правила, которые закрывают большинство ошибок:

1. **Время — всегда линия, а не круг.** Круговая диаграмма
   показывает состав в один момент, а не изменение.
2. **Круговая — только если части складываются в целое.**
   «Сколько человек занимается спортом, музыкой и рисованием» —
   не круг: кто-то занимается всем сразу, и сумма даст больше 100 %.

И ещё: **больше пяти-шести секторов — круг не читается**. Мелкие
части объединяют в «прочее».

### Оформление

Диаграмма без подписей бесполезна. Обязательный минимум:

* **заголовок** — о чём она;
* **подписи осей** с единицами измерения («баллы», «°C», «шт.»);
* **легенда** — если рядов больше одного;
* **подписи значений** — если чисел немного и точность важна.

Чего быть не должно: трёхмерных эффектов, теней, градиентов
и восьми разных цветов. Всё это мешает сравнивать — а диаграмма
нужна ровно для сравнения.

### Как диаграммой вводят в заблуждение

**Обрезанная ось.** Ось значений начинается не с нуля, а с 95.
Разница в один процент превращается в столбик вдвое выше соседнего.
Это приём номер один в рекламе и предвыборных листовках.

> Правило: у столбчатой диаграммы ось **обязана** начинаться
> с нуля. У линейной допускается обрезать — но тогда об этом
> честно пишут.

**Объёмные столбики.** В 3D ближний столбец выглядит больше
дальнего просто потому, что он ближе.

**Площадь вместо высоты.** Картинку увеличивают вдвое по обеим
сторонам — площадь растёт вчетверо, а данные выросли в два раза.

**Неравномерная шкала времени.** По оси идут 2010, 2015, 2016,
2017, 2025 через одинаковые промежутки — линия ведёт себя как ей
велели, а не как данные.

**Выбор удобного отрезка.** Взяли три года из двадцати — те,
где линия растёт. Формально правда, по сути обман.

Проверять чужую диаграмму стоит в таком порядке: **ось с нуля?
какой период? что по осям? сумма частей — сто процентов?**

## Смотрим, как это работает

Рисовать диаграммы будем прямо здесь, а строить их в таблице
вы умеете с седьмого класса — идея одна и та же.

### Пример 1. Данные

```python
классы = ["9А", "9Б", "9В", "9Г"]
средний_балл = [72, 68, 75, 70]

for класс, балл in zip(классы, средний_балл):
    print(f"{класс}: {балл}")
```

### Пример 2. Столбчатая диаграмма из символов

Самый честный вариант: ось начинается с нуля, длина столбика
пропорциональна значению.

```python
for класс, балл in zip(классы, средний_балл):
    столбик = "█" * (балл // 3)
    print(f"{класс} |{столбик} {балл}")

print()
print("Ось начинается с нуля — разница видна такой, какая она есть.")
```

### Пример 3. Та же диаграмма, но с обрезанной осью

Теперь начнём отсчёт не с нуля, а с 65 — и посмотрим,
что станет с теми же данными.

```python
начало = 65

for класс, балл in zip(классы, средний_балл):
    столбик = "█" * ((балл - начало) * 6)
    print(f"{класс} |{столбик} {балл}")

print()
print("Разница 9Б и 9В была 10 %, а выглядит как «в семь раз».")
```

Числа не изменились ни на единицу. Изменилась только точка отсчёта —
и картинка стала врать.

### Пример 4. Круговая: когда можно и когда нельзя

```python
ответы = [("да", 12), ("нет", 5), ("не знаю", 3)]
всего = sum(значение for _, значение in ответы)

print("Опрос: части складываются в целое")
for название, значение in ответы:
    доля = значение / всего * 100
    print(f"  {название:<9} {значение:>3}  {доля:5.1f} %  {'●' * round(доля / 5)}")
print(f"  {'итого':<9} {всего:>3}  {100.0:5.1f} %")

print()
кружки = [("спорт", 15), ("музыка", 9), ("рисование", 7)]
сумма = sum(значение for _, значение in кружки)
print("Кружки: сумма долей больше ста процентов —")
print(f"  человек в классе 25, а записей {сумма}: кто-то ходит на два кружка.")
print("  Круговая диаграмма здесь соврёт, нужна столбчатая.")
```

### Пример 5. Выбор удобного отрезка

```python
годы = [2019, 2020, 2021, 2022, 2023, 2024, 2025, 2026]
значения = [100, 96, 91, 88, 85, 88, 92, 95]

print("Весь период:")
for год, значение in zip(годы, значения):
    print(f"  {год} {'▇' * (значение // 6)} {значение}")

print()
print("А теперь возьмём только последние три года:")
for год, значение in zip(годы[-3:], значения[-3:]):
    print(f"  {год} {'▇' * (значение // 6)} {значение}")

print()
print("Во втором случае — уверенный рост. В первом — падение,")
print("которое только-только начало отыгрываться.")
```

### Пример 6. Распределение

Гистограмма отвечает на вопрос «сколько человек в каждом интервале».

```python
баллы = [45, 52, 67, 71, 73, 78, 82, 85, 88, 91, 62, 69, 74, 77, 95]

интервалы = [(40, 59), (60, 69), (70, 79), (80, 89), (90, 100)]

for низ, верх in интервалы:
    сколько = 0
    for балл in баллы:
        if низ <= балл <= верх:
            сколько += 1
    print(f"{низ}–{верх} |{'█' * сколько} {сколько}")
```
