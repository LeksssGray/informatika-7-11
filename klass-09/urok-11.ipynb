### Зачем разбивать программу на функции

Пока программа умещается в тридцать строк, разбивать нечего.
Но как только задач становится несколько, отдельные функции дают сразу
четыре выгоды:

* **понятность** — по именам видно, что программа делает, ещё до чтения
  кода;
* **проверяемость** — каждую функцию можно проверить отдельно;
* **повторное использование** — написали один раз, вызываете отовсюду;
* **изменяемость** — исправляя алгоритм внутри функции, не трогаете
  остальную программу.

Правило простое: **одна функция — одна задача**. Если в названии функции
просится союз «и» (`посчитать_и_вывести`), это две функции.

### Возврат нескольких значений

После `return` можно перечислить несколько значений через запятую.
Python соберёт их в **кортеж** — это почти список, только менять его
нельзя.

````python
def мин_макс(числа):
    return min(числа), max(числа)


результат = мин_макс([7, 2, 9, 4])
print(результат)

наименьшее, наибольшее = мин_макс([7, 2, 9, 4])
print("минимум:", наименьшее, "| максимум:", наибольшее)
````

Вторая запись называется **распаковкой**: слева столько же имён,
сколько значений вернула функция. Так читается гораздо лучше, чем
`результат[0]` и `результат[1]`.

### Список в функции: осторожно

Число внутрь функции попадает копией, а список — нет. Функция получает
тот же самый список, что и снаружи, и все изменения видны вызывающему.

````python
def испортить(числа):
    числа.append(0)


мои = [1, 2, 3]
испортить(мои)
print(мои)
````

Иногда это именно то, что нужно. Но чаще функция должна возвращать
**новый** список и не трогать исходный — тогда её проще проверять
и невозможно случайно испортить чужие данные.

```python
def без_нулей(числа):
    return [x for x in числа if x != 0]      # исходный список цел
```

### Значения по умолчанию и именованные аргументы

Параметру можно задать значение по умолчанию — тогда при вызове его
разрешается не указывать.

````python
def округлить(число, знаков=2):
    return round(число, знаков)


print(округлить(3.14159))
print(округлить(3.14159, 4))
print(округлить(3.14159, знаков=0))
````

Именованные аргументы (`знаков=0`) удобны, когда параметров много:
по вызову видно, что означает каждое число.

### Модули

**Модуль** — файл с функциями, который можно подключить к программе.
Часть модулей идёт в комплекте с Python:

| Модуль | Что внутри |
|---|---|
| `math` | корень, округление, тригонометрия, число π |
| `random` | случайные числа и выбор |
| `statistics` | среднее, медиана, мода |
| `time` | время и замеры длительности |

Подключают их двумя способами:

```python
import math              # math.sqrt(16)
from math import sqrt    # sqrt(16)
```

Первый способ длиннее, зато по коду видно, откуда взялась функция.

## Смотрим, как это работает

### Пример 1. Одна большая программа

Задача: по списку оценок напечатать отчёт — сколько оценок, средний
балл, лучшая и худшая оценка, и выставить итог за четверть.
Сначала как обычно пишут в первый раз — сплошным куском.

````python
оценки = [5, 4, 4, 5, 3, 4, 5, 4]

сумма = 0
for о in оценки:
    сумма += о
средний = сумма / len(оценки)
лучшая = оценки[0]
худшая = оценки[0]
for о in оценки:
    if о > лучшая:
        лучшая = о
    if о < худшая:
        худшая = о
if средний >= 4.5:
    итог = 5
elif средний >= 3.5:
    итог = 4
elif средний >= 2.5:
    итог = 3
else:
    итог = 2

print(f"Оценок: {len(оценки)}, средний балл: {средний:.2f}")
print(f"Лучшая: {лучшая}, худшая: {худшая}, итог за четверть: {итог}")
````

Работает. Но найдите здесь ответ на вопрос «как считается итог» —
придётся прочитать всё.

### Пример 2. То же самое по функциям

````python
def средний_балл(оценки):
    return sum(оценки) / len(оценки)


def крайние(оценки):
    return min(оценки), max(оценки)


def итог_четверти(средний):
    if средний >= 4.5:
        return 5
    if средний >= 3.5:
        return 4
    if средний >= 2.5:
        return 3
    return 2


def отчёт(оценки):
    средний = средний_балл(оценки)
    худшая, лучшая = крайние(оценки)
    строки = [
        f"Оценок: {len(оценки)}, средний балл: {средний:.2f}",
        f"Лучшая: {лучшая}, худшая: {худшая}, итог за четверть: {итог_четверти(средний)}",
    ]
    return "\n".join(строки)


print(отчёт([5, 4, 4, 5, 3, 4, 5, 4]))
````

Строк стало даже больше. Зато теперь видно устройство программы,
`итог_четверти` можно проверить отдельно на десятке значений,
а `отчёт` работает с любым списком оценок — хоть с другим классом.

Обратите внимание: `отчёт` не печатает, а возвращает текст. Функция,
которая считает, и функция, которая выводит, — разные вещи. Так текст
отчёта можно сохранить в файл или отправить куда-то ещё.

### Пример 3. Проверка функции без автопроверки

`assert` проверяет условие: если оно истинно, ничего не происходит,
если ложно — программа останавливается с ошибкой. Это способ проверить
свои функции самому.

````python
assert итог_четверти(5.0) == 5
assert итог_четверти(4.5) == 5
assert итог_четверти(4.49) == 4
assert итог_четверти(2.0) == 2
print("Все проверки прошли")
````

Граничные значения (ровно 4.5, чуть меньше) — самые важные: ошибки
прячутся именно там.

### Пример 4. Готовые модули

````python
import math
import statistics

оценки = [5, 4, 4, 5, 3, 4, 5, 4]

print("Среднее:", round(statistics.mean(оценки), 2))
print("Медиана:", statistics.median(оценки))
print("Корень из 144:", math.sqrt(144))
print("Округление вверх 4.1 →", math.ceil(4.1))
````

`statistics.mean` делает то же, что наша `средний_балл`. Прежде чем
писать функцию, стоит проверить: может быть, её уже написали.
