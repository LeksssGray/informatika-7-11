## Разбираемся

### Почему одна программа быстрая, а другая нет

Обе программы правильные, обе выдают верный ответ. Но одна считает
мгновенно, а вторая на тех же данных думает минуту.

Дело почти никогда не в «мощности компьютера» и не в языке
программирования. Дело в **количестве действий**, которое программа
выполняет. Компьютер быстрый, но не волшебный: миллиард операций —
это секунды, а не мгновение.

### Считаем операции, а не секунды

Секунды зависят от машины: на школьном компьютере одно, на сервере
другое. Поэтому алгоритмы сравнивают не временем, а **числом
операций в зависимости от размера данных**.

Пусть в списке $n$ элементов.

| Алгоритм | Операций | Как растёт |
|---|---|---|
| взять элемент по индексу | 1 | не растёт вообще |
| линейный поиск | до $n$ | вдвое больше данных — вдвое дольше |
| двоичный поиск | около $\log_2 n$ | вдвое больше данных — **на один шаг** дольше |
| сортировка пузырьком | около $n^2$ | вдвое больше данных — вчетверо дольше |
| встроенная сортировка | около $n \log_2 n$ | почти как линейно |
| перебор всех пар | $n^2/2$ | вчетверо |

Главное в этой таблице — правая колонка. Вопрос не «сколько
операций сейчас», а **что будет, когда данных станет вдесятеро
больше**.

### Как это выглядит в числах

| $n$ | $\log_2 n$ | $n$ | $n \log_2 n$ | $n^2$ |
|---|---|---|---|---|
| 10 | 3 | 10 | 33 | 100 |
| 1 000 | 10 | 1 000 | 10 000 | 1 000 000 |
| 1 000 000 | 20 | 1 000 000 | 20 000 000 | 1 000 000 000 000 |

Последняя строка объясняет всё. Миллион элементов: двоичный поиск —
двадцать шагов, линейный — миллион, пузырёк — триллион, то есть
часы работы.

### Правило больших чисел

На маленьких данных разницы нет. Десять элементов отсортируются
мгновенно любым способом, и выбирать алгоритм незачем.

Разница появляется на больших. Поэтому вопрос «какой алгоритм лучше»
всегда означает «какой лучше, когда данных **много**».

### Что делает программу медленной

Три самых частых школьных случая:

**Вложенный цикл там, где хватит одного.** Ищете пары с нужной
суммой перебором всех пар — это $n^2$. Часто задачу можно решить
за один проход, запомнив увиденное во множестве.

**Поиск в списке внутри цикла.** Проверка `if x in список`
сама по себе перебор. Внутри цикла получается $n^2$, хотя
множество `set` даёт ту же проверку мгновенно.

**Склейка строк в цикле.** Каждое `строка += кусок` создаёт новую
строку целиком. Тысяча кусков — тысяча копирований. Правильно:
собирать в список и склеить один раз через `"".join`.

### Как измерить

В Python есть модуль `time`:

```python
import time

начало = time.perf_counter()
# ... что-то считаем ...
прошло = time.perf_counter() - начало
```

`perf_counter` меряет время точнее обычных часов и специально
предназначен для замеров.

Важная оговорка: одно измерение ничего не значит. Компьютер
параллельно занят своими делами, и результат скачет. Поэтому
замеряют несколько раз и смотрят на порядок величины,
а не на третий знак.

## Смотрим, как это работает

### Пример 1. Считаем операции, а не секунды

````python
def шагов_линейного(n):
    return n


def шагов_двоичного(n):
    шагов = 0
    while n > 0:
        n = n // 2
        шагов += 1
    return шагов


print(f"{'n':>12} {'линейный':>12} {'двоичный':>10}")
for n in [10, 100, 1000, 10**6, 10**9]:
    print(f"{n:>12,} {шагов_линейного(n):>12,} {шагов_двоичного(n):>10}")
````

Миллиард элементов — тридцать шагов. Это не опечатка: каждый шаг
выбрасывает половину оставшегося.

### Пример 2. Замер: линейный против двоичного

````python
import time

данные = list(range(1_000_000))
искомое = 999_999          # худший случай: элемент в самом конце

начало = time.perf_counter()
найдено = -1
for i in range(len(данные)):
    if данные[i] == искомое:
        найдено = i
        break
линейно = time.perf_counter() - начало

начало = time.perf_counter()
лево, право = 0, len(данные) - 1
while лево <= право:
    середина = (лево + право) // 2
    if данные[середина] == искомое:
        break
    elif данные[середина] < искомое:
        лево = середина + 1
    else:
        право = середина - 1
двоично = time.perf_counter() - начало

print(f"Линейный поиск: {линейно * 1000:.2f} мс")
print(f"Двоичный поиск: {двоично * 1000:.4f} мс")
print(f"Быстрее примерно в {линейно / двоично:,.0f} раз")
````

### Пример 3. Сортировки на замерах

````python
import random

random.seed(26)


def пузырьком(числа):
    числа = list(числа)
    for i in range(len(числа)):
        for j in range(len(числа) - 1):
            if числа[j] > числа[j + 1]:
                числа[j], числа[j + 1] = числа[j + 1], числа[j]
    return числа


for размер in [200, 400, 800]:
    исходные = [random.randint(1, 1000) for _ in range(размер)]

    начало = time.perf_counter()
    пузырьком(исходные)
    медленно = time.perf_counter() - начало

    начало = time.perf_counter()
    sorted(исходные)
    быстро = time.perf_counter() - начало

    print(f"n = {размер:>4}: пузырёк {медленно * 1000:>8.2f} мс, "
          f"sorted {быстро * 1000:>6.3f} мс")
````

Посмотрите на столбец пузырька: размер вырос вдвое — время выросло
примерно вчетверо. Это и есть $n^2$, увиденное своими глазами.

А `sorted` почти не замечает роста: у него $n \log n$.

### Пример 4. Список против множества

````python
размер = 20_000
список = list(range(размер))
множество = set(список)
проверяемые = [random.randint(0, размер) for _ in range(2000)]

начало = time.perf_counter()
нашлось = 0
for x in проверяемые:
    if x in список:
        нашлось += 1
по_списку = time.perf_counter() - начало

начало = time.perf_counter()
нашлось = 0
for x in проверяемые:
    if x in множество:
        нашлось += 1
по_множеству = time.perf_counter() - начало

print(f"Проверка «есть ли» в списке:    {по_списку * 1000:>8.2f} мс")
print(f"Проверка «есть ли» в множестве: {по_множеству * 1000:>8.2f} мс")
print(f"Разница примерно в {по_списку / по_множеству:,.0f} раз")
````

Код отличается одним словом — `список` или `множество`, — а время
в сотни раз. Множество хранит элементы так, что проверка почти
не зависит от их числа.

### Пример 5. Склейка строк

````python
кусков = 20_000

начало = time.perf_counter()
строка = ""
for i in range(кусков):
    строка += "x"
склейкой = time.perf_counter() - начало

начало = time.perf_counter()
куски = []
for i in range(кусков):
    куски.append("x")
результат = "".join(куски)
через_join = time.perf_counter() - начало

print(f"Склейка в цикле: {склейкой * 1000:>7.2f} мс")
print(f"Через join:      {через_join * 1000:>7.2f} мс")
````

### Пример 6. Одна задача, два алгоритма

Есть ли в списке два числа с нужной суммой? Переберём все пары —
и сравним с одним проходом.

````python
числа = [random.randint(1, 100_000) for _ in range(3000)]
цель = 150_000

начало = time.perf_counter()
есть_парами = False
for i in range(len(числа)):
    for j in range(i + 1, len(числа)):
        if числа[i] + числа[j] == цель:
            есть_парами = True
            break
    if есть_парами:
        break
парами = time.perf_counter() - начало

начало = time.perf_counter()
видели = set()
есть_проходом = False
for x in числа:
    if цель - x in видели:
        есть_проходом = True
        break
    видели.add(x)
проходом = time.perf_counter() - начало

print("Ответы совпали:", есть_парами == есть_проходом)
print(f"Перебор пар:  {парами * 1000:>8.2f} мс")
print(f"Один проход:  {проходом * 1000:>8.2f} мс")
````

Идея второго способа: для каждого числа мы знаем, какое второе
нужно, — и просто спрашиваем, не встречалось ли оно раньше.
Один проход вместо миллионов пар.
