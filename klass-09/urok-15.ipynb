### Текстовый файл

Текстовый файл — просто длинная строка символов. Там, где мы видим
перенос строки, в файле стоит невидимый символ `\n`.

```
Аня 5
Боря 4
Вика 5
```

В памяти это `"Аня 5\nБоря 4\nВика 5\n"`.

### Открыть, поработать, закрыть

```python
with open("оценки.txt", "r", encoding="utf-8") as файл:
    текст = файл.read()
```

* `open` открывает файл и возвращает объект, через который с ним
  работают;
* **режим** говорит, что мы собираемся делать;
* `encoding="utf-8"` — в какой кодировке записаны буквы;
* `with` гарантирует, что файл закроется, даже если внутри случится
  ошибка. Без `with` пришлось бы вызывать `файл.close()` вручную —
  и когда-нибудь забыть.

### Режимы открытия

| Режим | Что делает | Если файла нет |
|:-:|---|---|
| `"r"` | чтение | ошибка `FileNotFoundError` |
| `"w"` | запись — **стирает** старое содержимое | создаст новый |
| `"a"` | дозапись в конец | создаст новый |

Самая дорогая ошибка новичка — открыть в режиме `"w"` файл, который
хотели прочитать. Содержимое исчезает мгновенно и навсегда.

### Три способа прочитать

| Способ | Что получаем | Когда удобно |
|---|---|---|
| `файл.read()` | весь файл одной строкой | файл маленький |
| `файл.readlines()` | список строк | нужен список целиком |
| `for строка in файл:` | строки по одной | файл большой |

Третий способ — основной: он не держит в памяти весь файл и читает
хоть гигабайтную таблицу.

Важно: при чтении в конце каждой строки остаётся `\n`. Его убирают
методом `.strip()` — иначе `"5\n"` не превратится в число.

### Запись

`файл.write(текст)` пишет ровно то, что дали, **без** перевода строки.
Символ `\n` в конце добавляют сами.

### Кодировка

Кодировка — договор о том, каким числом записана каждая буква.
Сейчас почти везде используют **utf-8**, в старых русских файлах
встречается windows-1251. Если открыть файл не в той кодировке,
вместо текста получится «крокозябра» вроде `ÐŸÑ€Ð¸Ð²ÐµÑ‚` — или
сразу ошибка. Поэтому `encoding="utf-8"` пишут всегда.

## Смотрим, как это работает

### Пример 1. Записали и прочитали

````python
with open("оценки.txt", "w", encoding="utf-8") as файл:
    файл.write("Аня 5\n")
    файл.write("Боря 4\n")
    файл.write("Вика 5\n")

with open("оценки.txt", "r", encoding="utf-8") as файл:
    содержимое = файл.read()

print(repr(содержимое))
print("---")
print(содержимое)
````

`repr` показывает строку так, как её видит Python, — с символами `\n`.
Обычный `print` эти символы выполняет, превращая в переносы.

### Пример 2. Читаем построчно и разбираем

````python
with open("оценки.txt", "r", encoding="utf-8") as файл:
    for строка in файл:
        имя, оценка = строка.strip().split()
        print(f"{имя:<6} оценка {оценка}, это {int(оценка)} баллов")
````

Схема, которая пригодится сотню раз: **строка → strip → split →
нужные части → int, если это число**.

### Пример 3. Дозапись

````python
with open("оценки.txt", "a", encoding="utf-8") as файл:
    файл.write("Гриша 3\n")

with open("оценки.txt", "r", encoding="utf-8") as файл:
    строк = len(файл.readlines())

print("Теперь строк в файле:", строк)
````

Режим `"a"` дописал строку в конец, ничего не стерев. Если бы мы
написали `"w"`, в файле осталась бы одна строка про Гришу.

### Пример 4. Прочитать, обработать, записать результат

Полный цикл обработки данных: читаем один файл, считаем, пишем другой.

````python
with open("числа.txt", "w", encoding="utf-8") as файл:
    файл.write("15\n7\n42\n3\n28\n")

числа = []
with open("числа.txt", "r", encoding="utf-8") as файл:
    for строка in файл:
        строка = строка.strip()
        if строка:                      # пустые строки пропускаем
            числа.append(int(строка))

отчёт = [
    f"Чисел: {len(числа)}",
    f"Сумма: {sum(числа)}",
    f"Максимум: {max(числа)}",
    f"Среднее: {sum(числа) / len(числа):.2f}",
]

with open("отчёт.txt", "w", encoding="utf-8") as файл:
    файл.write("\n".join(отчёт) + "\n")

with open("отчёт.txt", "r", encoding="utf-8") as файл:
    print(файл.read())
````

Проверка `if строка:` спасает от пустой строки в конце файла: после
последнего `\n` при чтении может появиться пустая строка, а `int("")`
— это ошибка.
