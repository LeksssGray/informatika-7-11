## Разбираемся

### Сколько бит нужно на символ

Если алфавит из `N` символов кодируют одинаковыми кодами, нужно `k` бит, где `2^k ≥ N`.
Для 16 символов хватит четырёх бит, для 17 — уже пяти.

### Перевод между системами

В двоичной системе только цифры 0 и 1, в шестнадцатеричной — 0–9 и A–F. Число переводят
делением на основание с остатками, либо по разрядам: `1101` = 1·8 + 1·4 + 0·2 + 1·1 = 13.

### Единицы измерения

1 Кб = 1024 байта, 1 байт = 8 бит. Объём файла в байтах делят на 1024, чтобы получить килобайты.

```python
число = 13
print("двоичная:", bin(число)[2:])
print("шестнадцатеричная:", hex(255)[2:].upper())
```
