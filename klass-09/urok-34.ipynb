### Типичные потери баллов

* перепутанные системы счисления и единицы измерения (Кб и Мб, бит и байт);
* неверная граница цикла: `range(a, b)` не включает `b`;
* невнимательность к формулировке: «не меньше» и «больше» дают разные ответы;
* ответ записан не в том формате: нужно число, а написано слово или наоборот.

Проверяй каждый ответ по простому примеру: подставь маленькое число и посмотри, совпадает ли результат.
