## Смотрим, как это работает

### Вопрос и данные

Вопрос: **«Отличаются ли оценки у тех, кто спит меньше 8 часов
и 8 и больше?»** Данные — опрос двенадцати человек: часы сна и средний
балл за четверть.

```python
опрос = [(6, 3.4), (7, 3.9), (8, 4.2), (9, 4.5), (6.5, 3.6), (8.5, 4.4),
         (7.5, 4.0), (9, 4.3), (5.5, 3.2), (8, 4.6), (7, 3.5), (9.5, 4.1)]

print("Ответов:", len(опрос))
```

### Обработка: группируем словарём

```python
группы = {"меньше 8 часов": [], "8 и больше": []}
for сон, балл in опрос:
    if сон < 8:
        группы["меньше 8 часов"].append(балл)
    else:
        группы["8 и больше"].append(балл)

средние = {}
for группа, баллы in группы.items():
    средние[группа] = round(sum(баллы) / len(баллы), 2)
    print(группа, "— человек:", len(баллы), "средний балл:", средние[группа])
```

### Визуализация без библиотек

Столбик из значков: длина пропорциональна среднему баллу.

```python
for группа, среднее in средние.items():
    print(f"{группа:>15} | {'█' * round(среднее * 10)} {среднее}")
```

### Вывод

«В нашем опросе у тех, кто спит 8 часов и больше, средний балл выше:
4,35 против 3,6. Опрошено всего 12 человек, поэтому вывод
предварительный. Причину по этим данным определить нельзя:
возможно, выспавшимся легче учиться, а возможно, и сон, и оценки
зависят от режима дня».

Вот так выглядит честный вывод: число, сравнение, оговорки.
