## Разбираемся

### IP-адрес и доменное имя

У каждого устройства в сети есть **IP-адрес**. Для IPv4 это четыре числа от 0 до 255,
записанные через точку: `192.168.0.1`. Каждое число — один байт, то есть восемь бит.

Людям неудобно запоминать числа, поэтому у сайтов есть **доменные имена**: `school.ru`.
Превращением имени в адрес занимается служба **DNS** — что-то вроде телефонной книги интернета.

```python
ip = "192.168.0.1"
части = ip.split(".")
print("частей:", len(части))
print("каждая от 0 до 255:", all(0 <= int(x) <= 255 for x in части))
```

### Структура адреса страницы

`https://www.school.ru/lessons/inf/index.html` делится так: протокол `https`, сервер
`www.school.ru`, путь к файлу `/lessons/inf/index.html`. Браузер сначала узнаёт IP-адрес
через DNS, потом отправляет запрос, а сервер отвечает страницей в виде HTML-кода.

### Поисковые запросы — это логика

Слова в запросе связаны логическими операциями. **И** — нужны оба слова: `кот И собака`.
**ИЛИ** — подходит хотя бы одно: `кот ИЛИ котёнок`. **НЕ** — исключает слово: `кот НЕ собака`.
Точные условия задаются кавычками и операциями вместе, поэтому полезно заранее понять,
что должно получиться.
