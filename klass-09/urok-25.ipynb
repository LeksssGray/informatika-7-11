## Разбираемся

### Сеть — это про договорённости

Два компьютера, соединённые проводом, сами по себе ничего друг другу
не скажут: сигнал придёт, а как его понимать — неизвестно. Сеть
начинается не с провода, а с **договорённостей**: где начало
сообщения, как подтвердить получение, что делать, если часть
данных потерялась.

> **Протокол** — набор правил, по которым устройства обмениваются
> данными.

Именно поэтому телефон Samsung спокойно открывает сайт, лежащий
на сервере с незнакомой операционной системой: они оба соблюдают
одни и те же протоколы.

### Локальные и глобальные

| | Локальная (LAN) | Глобальная (WAN) |
|---|---|---|
| Размер | кабинет, школа, здание | город, страна, мир |
| Кто владеет | один хозяин | много операторов |
| Скорость | выше | ниже |
| Пример | школьная сеть, Wi-Fi дома | интернет |

Интернет — не «одна большая сеть», а **сеть сетей**: миллионы
локальных сетей, соединённых между собой по общим правилам.

### Топологии

Топология — схема соединения узлов.

| Топология | Как устроена | Слабое место |
|---|---|---|
| **шина** | все на одном общем кабеле | обрыв кабеля убивает всю сеть |
| **кольцо** | каждый соединён с двумя соседями | обрыв разрывает кольцо |
| **звезда** | все через один центральный узел | отказ центра убивает всё |
| **ячеистая** | много связей между узлами | дорого, зато надёжно |

Сегодня почти везде звезда: центральным узлом работает коммутатор
или Wi-Fi-роутер. А интернет в целом — ячеистая: между двумя точками
есть много маршрутов, и потому обрыв одного кабеля его не обрушивает.

### Клиент и сервер

**Клиент** — тот, кто запрашивает. **Сервер** — тот, кто отвечает.
Это роли, а не железки: ваш ноутбук обычно клиент, но может раздать
файл и стать сервером.

Диалог всегда один и тот же: клиент послал запрос — сервер прислал
ответ. Открыли страницу — браузер (клиент) попросил у веб-сервера
файл, получил его и показал.

### Как данные едут по сети

Файл не отправляется целиком. Он режется на **пакеты** — небольшие
куски с адресом отправителя, адресом получателя и номером. Пакеты
едут по сети каждый своим маршрутом и собираются обратно у получателя
по номерам.

Зачем так сложно:

* потерялся пакет — переспрашивают один кусок, а не весь файл;
* один канал не занят одним файлом — в нём едут пакеты всех сразу;
* оборвался маршрут — следующие пакеты поедут в обход.

За это отвечают два главных протокола, которые обычно называют
вместе — **TCP/IP**:

* **IP** доставляет пакет по адресу — но не обещает, что доставит;
* **TCP** нумерует пакеты, проверяет доставку и переспрашивает
  потерянное.

### Скорость передачи

Здесь школьников подстерегает главная путаница урока.

> **Скорость** измеряется в **битах** в секунду, а объём файлов —
> в **байтах**. Разница в восемь раз.

100 Мбит/с — это не 100 мегабайт в секунду, а 12,5.

Формула простая:

$$t = \frac{I}{v}$$

где $I$ — объём данных, $v$ — скорость. Но перед делением обе
величины надо привести к одним единицам: обычно всё переводят в биты.

Порядок действий, который не даёт ошибиться:

1. объём — в биты (байты × 8, килобайты × 1024 × 8 и так далее);
2. скорость — в биты в секунду;
3. поделить;
4. перевести секунды в минуты, если просят.

## Смотрим, как это работает

### Пример 1. Биты и байты

```python
скорость_мбит = 100

байт_в_секунду = скорость_мбит * 1_000_000 / 8
мегабайт_в_секунду = байт_в_секунду / 1_048_576

print(f"Канал {скорость_мбит} Мбит/с — это:")
print(f"  {байт_в_секунду:,.0f} байт в секунду")
print(f"  {мегабайт_в_секунду:.2f} Мбайт в секунду")
```

Вот почему фильм «на сто мегабит» качается не за секунду:
провайдер считает в битах, а файловый менеджер — в байтах.

### Пример 2. Время передачи

```python
объём_кбайт = 1500
скорость_кбит = 256

объём_бит = объём_кбайт * 1024 * 8
скорость_бит = скорость_кбит * 1024

время = объём_бит / скорость_бит

print(f"Файл {объём_кбайт} Кбайт = {объём_бит:,} бит")
print(f"Канал {скорость_кбит} Кбит/с = {скорость_бит:,} бит/с")
print(f"Время: {время:.0f} с = {время / 60:.1f} мин")
```

### Пример 3. Типичная ошибка

```python
неверно = объём_кбайт * 1024 / скорость_бит

print(f"Правильно:            {время:.0f} с")
print(f"Забыли умножить на 8: {неверно:.0f} с")
print(f"Ответ занижен в {время / неверно:.0f} раз")
```

Ровно в восемь. Если ответ в задаче ОГЭ отличается от ожидаемого
в восемь раз — вы знаете, где искать.

### Пример 4. Сколько времени на разных каналах

```python
файл_мбайт = 700

for название, скорость_бит_с in [
    ("модем 56 Кбит/с", 56 * 1024),
    ("ADSL 8 Мбит/с", 8 * 1_000_000),
    ("Wi-Fi 100 Мбит/с", 100 * 1_000_000),
    ("оптика 1 Гбит/с", 1_000_000_000),
]:
    бит = файл_мбайт * 1024 * 1024 * 8
    секунд = бит / скорость_бит_с
    if секунд >= 3600:
        печать = f"{секунд / 3600:.1f} ч"
    elif секунд >= 60:
        печать = f"{секунд / 60:.1f} мин"
    else:
        печать = f"{секунд:.1f} с"
    print(f"{название:<20} {печать}")
```

### Пример 5. Пакеты

Посмотрим, на сколько пакетов разрежется файл и почему последний
пакет почти всегда неполный.

```python
файл_байт = 10_000
пакет_байт = 1460          # столько данных везёт обычный пакет

пакетов = -(-файл_байт // пакет_байт)      # деление с округлением вверх
последний = файл_байт - (пакетов - 1) * пакет_байт

print(f"Файл {файл_байт} байт режется на {пакетов} пакетов")
print(f"Полных пакетов: {пакетов - 1}, последний везёт {последний} байт")
```

Приём `-(-a // b)` — это деление с округлением вверх, знакомое
ещё по задаче про коробки из седьмого класса.

создавался в конце 1960-х по заказу военных, и главным требованием
