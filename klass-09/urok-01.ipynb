### Моделирование как метод познания

Начнём с идеи, которая объединяет всю информатику.

> **Модель** — упрощённое представление объекта, процесса или явления,
> сохраняющее существенные для решаемой задачи свойства.

Ключевые слова здесь — «упрощённое» и «существенные для задачи».
Модель всегда беднее оригинала, и это не недостаток, а смысл:
нам нужно отбросить всё несущественное, иначе задача не решится.

Глобус — модель Земли. Он не передаёт ни массу планеты, ни состав
атмосферы, ни движение материков. Но для задачи «показать взаимное
расположение материков» этого достаточно, а всё остальное только мешало бы.

### Классификация моделей

**По способу представления:**

* **материальные** — макет здания, глобус, манекен;
* **информационные** — описание, формула, чертёж, таблица, программа.

Информационные модели делятся дальше:

| Вид | Чем задаётся | Пример |
|---|---|---|
| словесная | текстом | описание погоды в прогнозе |
| математическая | формулами | $s = v t$ |
| табличная | таблицей | расписание уроков |
| графическая | схемой, чертежом | карта метро |
| компьютерная | программой | симулятор полёта |

**По учёту времени:**

* **статические** — описывают состояние в один момент (фотография класса);
* **динамические** — описывают изменение (рост населения по годам).

### Этапы моделирования

Порядок работы, который годится для любой задачи с компьютером:

```
  1. Постановка задачи      что именно требуется узнать
           ↓
  2. Формализация           отбросить лишнее, выбрать величины
           ↓
  3. Построение модели      формула, таблица, схема, программа
           ↓
  4. Компьютерный опыт      расчёты, прогоны, эксперименты
           ↓
  5. Анализ результатов     верим ли мы этому? если нет — назад к п. 2
```

Обратите внимание на стрелку назад. Моделирование почти никогда
не проходит с первого раза: первая модель оказывается слишком грубой,
её уточняют и считают заново.

### Адекватность модели

Модель **адекватна**, если её выводы совпадают с реальностью
в пределах нужной точности.

Классический пример: модель падения тела $h = gt^2/2$ не учитывает
сопротивление воздуха. Для стального шарика она адекватна.
Для бумажного листа — совершенно нет. Модель не «плохая» и не «хорошая»
сама по себе: она адекватна или неадекватна **для конкретной задачи**.

Отсюда важное правило: у любой модели есть **границы применимости**,
и их нужно знать. Прогноз погоды на завтра неплох, на две недели —
почти бесполезен, хотя модель одна и та же.

## Смотрим, как это работает

### Пример 1. Одна задача — три модели

Задача: спрогнозировать, сколько человек будет в школе через 5 лет,
если сейчас 800 и каждый год прибавляется 3 %.

Построим три модели разной сложности и сравним.

````python
начало = 800

# Модель 1: линейная — «каждый год плюс 24 человека»
линейная = начало + 24 * 5

# Модель 2: экспоненциальная — «каждый год плюс 3 % от текущего»
показательная = начало
for год in range(5):
    показательная *= 1.03

# Модель 3: с ограничением — здание вмещает не больше 900
с_потолком = начало
for год in range(5):
    с_потолком = min(с_потолком * 1.03, 900)

print(f"Линейная модель:        {линейная:.0f} человек")
print(f"Показательная модель:   {показательная:.0f} человек")
print(f"С учётом вместимости:   {с_потолком:.0f} человек")
````

Три модели дали три ответа. Какая правильная? Тот самый вопрос
об адекватности.

Линейная проще всего, но занижает результат: проценты считаются
от растущего числа, а не от исходного. Показательная точнее математически,
но игнорирует физическую реальность — в здание больше 900 человек
не поместится. Третья модель учитывает ограничение и потому адекватнее
для нашей задачи.

Мораль: усложнять модель нужно не ради красоты, а когда простая
даёт заведомо неверный ответ.

### Пример 2. Табличная модель

Расписание — типичная табличная модель. Посмотрим, как с ней работать
в коде.

````python
расписание = [
    ["Понедельник", "Алгебра", "Русский", "Информатика", "История"],
    ["Вторник", "Геометрия", "Физика", "Литература", "Химия"],
    ["Среда", "Информатика", "Алгебра", "Биология", "Физкультура"],
]

# Сколько раз за неделю встречается информатика
счётчик = 0
for день in расписание:
    счётчик += день.count("Информатика")

print(f"Информатика в расписании {счётчик} раза")
print()

for день in расписание:
    print(f"{день[0]:<14} {', '.join(день[1:])}")
````

Таблица здесь — список списков: каждая строка это день, элементы
внутри — уроки. Такая структура называется **двумерным массивом**,
и мы подробно разберём её на уроке 7.

### Пример 3. Проверка адекватности

Сравним модель с реальными данными и увидим, где она перестаёт работать.

````python
# Модель свободного падения без сопротивления воздуха: h = g·t²/2
def высота_по_модели(время):
    return 9.8 * время ** 2 / 2


# Реальные измерения падения бумажного листа (условные данные)
измерения = [(1, 3.0), (2, 8.5), (3, 14.0), (4, 19.5)]

print(f"{'время':>6} {'модель':>10} {'реальность':>12} {'расхождение':>13}")
for время, реальная_высота in измерения:
    модельная = высота_по_модели(время)
    расхождение = модельная - реальная_высота
    print(f"{время:>6} {модельная:>10.1f} {реальная_высота:>12.1f} {расхождение:>13.1f}")
````

Расхождение растёт с каждой секундой: для бумажного листа модель
свободного падения неадекватна. Причина понятна — не учтено сопротивление
воздуха, которое для лёгкого предмета с большой площадью решает всё.

Заметьте, что мы обнаружили это не рассуждением, а **сравнением
с данными**. Это и есть четвёртый и пятый этапы моделирования.
