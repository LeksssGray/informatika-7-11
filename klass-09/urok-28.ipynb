### 1. Формулы и ссылки

### 2. Условные функции

`=ЕСЛИ(условие; если_да; если_нет)`, `=СЧЁТЕСЛИ(диапазон; условие)`,
`=СУММЕСЛИ(где_проверять; условие; что_складывать)`. Сравнение
в условии — в кавычках: `">=50"`.

величин — гистограмма. Размах = МАКС − МИН.

### 6. Компьютерные сети

Локальная сеть — в здании; коммутатор соединяет компьютеры в ней,
маршрутизатор связывает сеть с интернетом. Сервер отвечает на запросы,
клиент их отправляет.

### 7. Адресация

IPv4 — 4 байта, четыре числа от 0 до 255. DNS переводит имя сайта
в IP-адрес. Маска `/n`: в сети `2 ** (32 − n)` адресов, из них два
служебных.
