### Формальный исполнитель

**Формальный исполнитель** понимает только команды из своей системы и выполняет их буквально.
Робот на клетчатом поле, черепашка, калькулятор — примеры. Задача исполнителя — проследить,
что получится после последовательности команд, и не додумывать лишнего.

Чтобы не ошибиться, веди таблицу: после каждой команды записывай текущее положение
или текущее значение переменной.

```python
s = 0
for i in range(1, 5):
    s += i * i
print("s =", s)
```

Такой цикл — типичная задача на анализ: выполни его на бумаге, шаг за шагом, и сравни
результат с программой. Совпало — значит, понял.
