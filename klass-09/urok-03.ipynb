## Справочник

### Ветвления

```python
if условие:
    ...
elif другое:
    ...
else:
    ...
```

Условия проверяются сверху вниз, срабатывает **первое подошедшее**.
Поэтому порядок важен: если начать с `if баллы >= 50`, то пятёрка
никогда не выпадет.

Сравнения: `==` равно, `!=` не равно, `<`, `>`, `<=`, `>=`.
Один знак `=` — это присваивание, не сравнение.

Логические связки: `and`, `or`, `not`.

### Циклы

```python
for i in range(5):          # 0, 1, 2, 3, 4
for i in range(1, 6):       # 1, 2, 3, 4, 5
for i in range(0, 10, 2):   # 0, 2, 4, 6, 8
for элемент in список:      # по значениям
```

Правая граница `range` **не входит**. Это источник примерно половины
ошибок «на единицу».

```python
while условие:              # пока условие верно
```

`for` — когда число повторений известно заранее, `while` — когда нет.

### Вложенные циклы

```python
for i in range(3):
    for j in range(3):
        ...                 # выполнится 9 раз
```

Число шагов перемножается. Два вложенных цикла по тысяче элементов —
это миллион операций.

### Списки

```python
числа = [5, 3, 8, 1]
числа[0]          # первый, нумерация с нуля
числа[-1]         # последний
len(числа)        # длина
числа.append(7)   # добавить в конец
числа[1:3]        # срез: элементы 1 и 2
sorted(числа)     # новый отсортированный список
```

### Пять приёмов, которыми решается почти всё

| Задача | До цикла | Внутри |
|---|---|---|
| сумма | `сумма = 0` | `сумма += x` |
| произведение | `пр = 1` | `пр *= x` |
| количество | `к = 0` | `if условие: к += 1` |
| максимум | `м = список[0]` | `if x > м: м = x` |
| отбор | `новый = []` | `if условие: новый.append(x)` |

Максимум начинают **с первого элемента списка**, а не с нуля.
Произведение — с единицы, а не с нуля.

### Работа с индексами

Когда нужен номер элемента, а не только значение:

```python
for i in range(len(список)):
    ...список[i]...
```

А когда сравниваем соседей — диапазон короче на единицу:

```python
for i in range(len(список) - 1):
    if список[i] > список[i + 1]:
        ...
```

Забыть `- 1` здесь — гарантированная ошибка выхода за границу.

### Поиск и сортировка

**Линейный поиск** — перебрать всё подряд. Работает всегда,
но медленно на больших данных.

**Двоичный поиск** — только в отсортированном списке: смотрим середину,
отбрасываем половину, повторяем. В миллионе элементов находит
за 20 шагов.

### Числа по цифрам

```python
while n > 0:
    цифра = n % 10      # последняя цифра
    n //= 10            # отбрасываем её
```

Этот шаблон вы напишете в этом году ещё раз двадцать.

## Смотрим, как это работает

### Пять приёмов за один проход

````python
числа = [4, -2, 7, 0, 9, -5, 12]

сумма = 0
положительных = 0
максимум = числа[0]
чётные = []

for x in числа:
    сумма += x
    if x > 0:
        положительных += 1
    if x > максимум:
        максимум = x
    if x % 2 == 0:
        чётные.append(x)

print("Сумма:", сумма)
print("Положительных:", положительных)
print("Максимум:", максимум)
print("Чётные:", чётные)
print("Среднее:", round(сумма / len(числа), 2))
````

Один цикл, пять результатов. Если бы мы вызвали `sum`, `max` и другие
готовые функции по очереди, список был бы просмотрен пять раз.
На семи элементах разницы нет, на миллионе — заметная.

### Соседние элементы

````python
числа = [1, 3, 2, 5, 4, 6]

возрастаний = 0
for i in range(len(числа) - 1):
    if числа[i] < числа[i + 1]:
        возрастаний += 1

print("Пар, где следующий больше предыдущего:", возрастаний)
````

Обратите внимание на `len(числа) - 1`. Без минуса единицы на последнем
шаге программа попытается взять элемент за границей списка и упадёт
с `IndexError`.

### Двоичный поиск против линейного

````python
данные = list(range(1, 1000001))       # миллион чисел по порядку
искомое = 999999

# линейный: считаем шаги
шагов_линейно = 0
for x in данные:
    шагов_линейно += 1
    if x == искомое:
        break

# двоичный
лево, право = 0, len(данные) - 1
шагов_двоично = 0
while лево <= право:
    шагов_двоично += 1
    середина = (лево + право) // 2
    if данные[середина] == искомое:
        break
    if данные[середина] < искомое:
        лево = середина + 1
    else:
        право = середина - 1

print("Линейный поиск, шагов:", шагов_линейно)
print("Двоичный поиск, шагов:", шагов_двоично)
````

Почти миллион против двадцати. Двоичный поиск на каждом шаге
выбрасывает половину оставшихся вариантов — отсюда такая разница.

Плата за это одна: список должен быть отсортирован.
