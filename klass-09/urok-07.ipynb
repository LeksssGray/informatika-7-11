## Разбираемся

### Таблица как массив массивов

Класс, расписание, шахматная доска, изображение, матрица смежности —
всё это таблицы. В программе таблицу хранят **двумерным массивом**:
списком, элементы которого сами списки.

```python
таблица = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
```

Обращение к элементу — **сначала строка, потом столбец**:

```python
таблица[0][2]     # первая строка, третий столбец → 3
таблица[2][0]     # третья строка, первый столбец → 7
```

Порядок индексов запомнить легко: как при чтении адреса — сначала дом,
потом квартира.

### Размеры

```python
строк = len(таблица)          # сколько списков внутри
столбцов = len(таблица[0])    # сколько элементов в первом списке
```

### Обход двойным циклом

Внешний цикл идёт по строкам, внутренний — по столбцам:

```python
for i in range(len(таблица)):
    for j in range(len(таблица[0])):
        print(таблица[i][j], end=" ")
    print()
```

Если индексы не нужны, а нужны только значения, читается ещё проще:

```python
for строка in таблица:
    for элемент in строка:
        ...
```

### Создание таблицы

> ⚠️ **Главная ловушка двумерных массивов.**

```python
плохо = [[0] * 3] * 3      # НЕ ДЕЛАЙТЕ ТАК
хорошо = [[0] * 3 for _ in range(3)]
```

В первом случае создаётся **одна** строка, а список содержит три ссылки
на неё. Изменив `плохо[0][0]`, вы измените первый элемент во всех трёх
строках сразу. Ошибка неочевидная и очень обидная — потратите час,
пока найдёте.

Во втором случае генератор выполняется трижды и создаёт три
независимые строки.

### Типовые обходы

**По строкам** — сумма каждой строки:

```python
for строка in таблица:
    print(sum(строка))
```

**По столбцам** — здесь индексы обязательны, причём внешний цикл
идёт по столбцам, а внутренний по строкам:

```python
for j in range(столбцов):
    сумма = 0
    for i in range(строк):
        сумма += таблица[i][j]
```

**Главная диагональ** — элементы, у которых номер строки равен номеру
столбца: `таблица[i][i]`.

**Побочная диагональ** — из правого верхнего в левый нижний угол:
`таблица[i][n - 1 - i]`.

```
  ⋱ · ·        · · ⋰
  · ⋱ ·        · ⋰ ·
  · · ⋱        ⋰ · ·
  главная      побочная
```

### Транспонирование

Поменять строки и столбцы местами — операция, которая называется
**транспонированием**:

```
  1 2 3        1 4
  4 5 6   →    2 5
               3 6
```

Новый элемент `[i][j]` берётся из старого `[j][i]`.

## Смотрим, как это работает

### Пример 1. Печать и суммы

````python
таблица = [
    [5, 3, 4],
    [4, 5, 5],
    [3, 3, 4],
    [5, 4, 5],
]

print("Таблица оценок (4 ученика × 3 предмета):")
for i, строка in enumerate(таблица):
    print(f"  ученик {i + 1}: {строка}  средний {sum(строка) / len(строка):.2f}")

print("\nСредние по предметам:")
предметы = ["Алгебра", "Физика", "Информатика"]
for j in range(len(таблица[0])):
    столбец = [таблица[i][j] for i in range(len(таблица))]
    print(f"  {предметы[j]:<14} {sum(столбец) / len(столбец):.2f}")
````

Обратите внимание на строку с генератором:
`[таблица[i][j] for i in range(len(таблица))]` — она собирает столбец
в отдельный список, после чего с ним можно работать как с обычным
одномерным массивом. Приём удобный и часто экономит нервы.

### Пример 2. Ловушка с созданием таблицы

Убедимся, что предупреждение из теории не преувеличено.

````python
плохо = [[0] * 3] * 3
плохо[0][0] = 9
print("Так делать нельзя:", плохо)

хорошо = [[0] * 3 for _ in range(3)]
хорошо[0][0] = 9
print("А так правильно:  ", хорошо)
````

В первом случае девятка появилась во **всех трёх** строках — потому
что строка на самом деле одна, а ссылок на неё три. Во втором случае
всё как ожидалось.

### Пример 3. Диагонали и транспонирование

````python
матрица = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
n = len(матрица)

главная = [матрица[i][i] for i in range(n)]
побочная = [матрица[i][n - 1 - i] for i in range(n)]

print("Матрица:")
for строка in матрица:
    print("  ", строка)

print(f"\nГлавная диагональ:  {главная}, сумма {sum(главная)}")
print(f"Побочная диагональ: {побочная}, сумма {sum(побочная)}")

транспонированная = [[матрица[j][i] for j in range(n)] for i in range(n)]
print("\nТранспонированная:")
for строка in транспонированная:
    print("  ", строка)
````

Двойной генератор `[[матрица[j][i] for j in ...] for i in ...]` читается
изнутри наружу: внутренний собирает одну строку результата, внешний —
все строки. Заметьте перестановку индексов `[j][i]` — в ней вся суть
транспонирования.
