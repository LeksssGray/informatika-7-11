## Разбираемся

### Что такое разметка

HTML — язык разметки: текст обёрнут в **теги**, которые говорят браузеру, что это такое.
Тег открывается `<имя>` и закрывается `</имя>`. Заголовок, абзац, ссылка — всё это теги.

Структура страницы всегда одна: `<html>` содержит `<body>`, а внутри лежат заголовки,
абзацы, ссылки и картинки.

````python
страница = """<html>
<body>
<h1>Обо мне</h1>
<p>Я учусь в девятом классе.</p>
<a href="https://school.ru">Школа</a>
<img src="foto.jpg">
</body>
</html>"""
print(страница)
````

Ссылка задаётся тегом `<a href="адрес">текст</a>`, картинка — тегом `<img src="файл">`.
Тег `<img>` закрывать не нужно: у него нет текста внутри.

Список — тег `<ul>`, каждый пункт в нём — `<li>`:

```html
<ul>
  <li>читать</li>
  <li>рисовать</li>
</ul>
```

Чтобы увидеть страницу в браузере, сохрани такой текст в файл с расширением `.html` и
открой его двойным щелчком. Браузер прочитает теги и покажет результат.
