## Разбираемся

### Угрозы и защита

Информация в сети подвергается угрозам: её крадут, подменяют или делают недоступной.
Защита строится в несколько слоёв. Самый доступный слой — пароль, но и он работает
только при условии, что его сложно угадать и его никому не отдают.

### Надёжный пароль

Оценим пароль по четырём признакам: длина не меньше восьми символов, есть цифра,
есть заглавная буква и есть знак, который не буква и не цифра. Каждый признак даёт один балл.

```python
пароль = "Abcdefg1!"
print("длина:", len(пароль))
print("есть цифра:", any(c.isdigit() for c in пароль))
print("есть заглавная:", any(c.isupper() for c in пароль))
print("есть знак:", any(not c.isalnum() for c in пароль))
```

### Двухфакторная аутентификация и социальная инженерия

**Двухфакторная аутентификация** — второй способ подтвердить вход, например код из SMS.
Даже украденный пароль без второго фактора не открывает аккаунт.

**Социальная инженерия** — обман через доверие: звонок «из банка», письмо «от учителя».
Злоумышленник не взламывает систему, а уговаривает человека отдать данные. Поэтому
никому не сообщай код из SMS и не вводи пароль по ссылке из письма.

### Персональные данные и цифровой след

К персональным данным относится всё, что указывает на конкретного человека: ФИО, дата
рождения, адрес, фотографии. **Цифровой след** — всё, что ты оставляешь в сети: посты,
лайки, геометки. Его трудно стереть, поэтому подумай заранее, что публикуешь.
