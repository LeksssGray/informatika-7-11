## Разбираемся

### Тысяча строк вместо десяти

Таблицы вы видели и в седьмом, и в восьмом классе. Но там были
учебные таблички на десять строк, где всё видно сразу. Настоящие
данные другие: выгрузка из журнала, список заказов, показания
датчика за месяц — это тысячи строк, которые не помещаются на экран.

С такими таблицами работают иначе, и сегодня — про эти приёмы.
Заодно это подготовка к заданию 14 ОГЭ: там как раз дают файл
на несколько тысяч строк.

### Навигация

Мышкой тысячу строк не пролистаешь. Клавиши:

| Сочетание | Куда попадём |
|---|---|
| `Ctrl` + `↓` | в последнюю заполненную ячейку столбца |
| `Ctrl` + `→` | в последнюю заполненную ячейку строки |
| `Ctrl` + `Home` | в ячейку A1 |
| `Ctrl` + `End` | в последнюю ячейку с данными |
| `Ctrl` + `Shift` + `↓` | выделить столбец до конца данных |

Последняя строка — самая полезная: так выделяют диапазон для формулы,
не прокручивая таблицу.

### Закрепление областей

Прокрутили на тысячную строку — и заголовки столбцов уехали вверх.
Теперь неясно, где оценка, а где номер варианта.

Лечится закреплением: **Вид → Закрепить области**. Первая строка
(или первый столбец) остаётся на месте при любой прокрутке.

Правило: закрепляют **перед** работой с большой таблицей, а не
после того, как запутались.

### Диапазоны и их имена

Диапазон `B2:B1001` — тысяча ячеек. Писать такое в каждой формуле
неудобно и легко ошибиться на единицу.

Диапазону можно дать **имя**: выделить, щёлкнуть поле слева
от строки формул, написать `оценки`. Дальше формула читается
по-человечески:

```
=СРЗНАЧ(оценки)
=СЧЁТЕСЛИ(оценки; 5)
```

И сравните это с `=СРЗНАЧ(B2:B1001)` — второе короче, зато первое
понятно через месяц.

### Копирование формулы на тысячу строк

Три способа, от худшего к лучшему:

1. тянуть мышкой за уголок — на тысяче строк это упражнение
   на терпение;
2. выделить диапазон и нажать `Ctrl` + `D` — формула размножится
   вниз по выделению;
3. написать формулу в первой строке и **дважды щёлкнуть** уголок —
   таблица сама протянет её до конца данных.

Третий способ — тот, которым пользуются. Но он требует, чтобы
в соседнем столбце не было пустых строк: на первой же пустой
заполнение остановится.

### Типичные ошибки в данных

Данные из реального мира почти всегда грязные. Вот что встречается
постоянно.

| Что не так | Как выглядит | Чем плохо |
|---|---|---|
| Число записано текстом | `«5 »`, `5,0`, `5 ` | не попадает в СУММ и СРЗНАЧ |
| Лишние пробелы | `«Иванов »` | не находится поиском, не совпадает при сравнении |
| Дубликаты | две строки на одного ученика | завышают все итоги |
| Пустые ячейки | ничего | СРЗНАЧ их пропускает, а СУММ считает нулём |
| Разный формат даты | `01.09.26` и `2026-09-01` | не сортируется |

Отсюда главное правило работы с чужой таблицей: **сначала проверь
данные, потом считай**. Иначе получится красивый и неверный ответ.

### Сообщения об ошибках

| Ошибка | Что случилось |
|---|---|
| `#ДЕЛ/0!` | деление на ноль или на пустую ячейку |
| `#ЗНАЧ!` | в формулу попал текст вместо числа |
| `#ССЫЛКА!` | формула ссылается на удалённую ячейку |
| `#Н/Д` | ВПР не нашла значение |
| `#####` | столбец узкий — это не ошибка, расширьте его |

### Проверка данных

Чтобы грязь не появлялась вообще, на столбец ставят **проверку
данных**: «Данные → Проверка данных → целое число от 2 до 5».
Тогда таблица просто не даст вписать «шесть» или «пять».

Это ровно та же мысль, что и в программировании: проверять ввод
на входе дешевле, чем искать причину странного результата потом.
