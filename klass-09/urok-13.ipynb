*Наборные литеры в кассе*

<sub>Willi Heidelbach · CC BY 2.5 · Wikimedia Commons</sub>

До изобретения наборного шрифта книгу переписывали от руки. Иоганн
Гутенберг в середине XV века предложил другое: отлить каждую букву
отдельной металлической литерой, а страницу собирать из них, как
конструктор. Буквы хранились в кассе по ячейкам, наборщик доставал
нужные и выстраивал в строку.

Компьютерная строка устроена ровно так же: это последовательность
отдельных символов, к каждому можно обратиться по номеру. Разница
в скорости — набор страницы занимал часы, программа обрабатывает
тысячу страниц за секунду.

### Что уже умеем

| Действие | Запись | Результат |
|---|---|---|
| длина | `len("привет")` | 6 |
| символ по номеру | `"привет"[0]` | `"п"` |
| последний символ | `"привет"[-1]` | `"т"` |
| срез | `"привет"[1:4]` | `"рив"` |
| срез с конца | `"привет"[-3:]` | `"вет"` |
| переворот | `"привет"[::-1]` | `"тевирп"` |
| склейка | `"при" + "вет"` | `"привет"` |
| повтор | `"ха" * 3` | `"хахаха"` |
| проверка вхождения | `"ив" in "привет"` | `True` |

### Методы строк

Метод — функция, которая пишется после точки и работает с этой строкой.
Главное помнить: **строку изменить нельзя**. Любой метод возвращает
новую строку, а старая остаётся прежней.

| Метод | Что делает | Пример |
|---|---|---|
| `.lower()` / `.upper()` | регистр | `"Аня".lower()` → `"аня"` |
| `.strip()` | убирает пробелы по краям | `"  да  ".strip()` → `"да"` |
| `.replace(a, b)` | заменяет все вхождения | `"ко-ко".replace("-", "")` → `"коко"` |
| `.split()` | режет строку на список | `"a b c".split()` → `["a","b","c"]` |
| `.join(список)` | собирает список в строку | `" ".join(["a","b"])` → `"a b"` |
| `.count(x)` | сколько раз встречается | `"ананас".count("на")` → 2 |
| `.find(x)` | номер первого вхождения или −1 | `"привет".find("в")` → 3 |
| `.startswith(x)` / `.endswith(x)` | начинается / заканчивается | `"файл.txt".endswith(".txt")` → `True` |
| `.isdigit()` / `.isalpha()` | из цифр / из букв | `"123".isdigit()` → `True` |

`split` и `join` — противоположности, и работают они чаще всего в паре:
разрезали текст на слова, обработали список, собрали обратно.

### Ловушка: перекрывающиеся вхождения

Сколько раз строка `"аа"` встречается в `"ааааа"`? Метод `.count()`
ответит 2: он ищет только непересекающиеся куски. Но вхождений
на самом деле четыре — на позициях 0, 1, 2 и 3.

```
  а а а а а
  └┬┘
    └┬┘
      └┬┘
        └┬┘
```

Если в задаче не сказано «непересекающиеся», считайте сами циклом
по всем позициям. Это одна из самых частых ловушек в задачах ОГЭ
на обработку текста.

## Смотрим, как это работает

### Пример 1. Разбор текста на слова

````python
текст = "  Мама   мыла раму.  Рама   была чистой.  "

слова = текст.split()
print("Слов:", len(слова))
print(слова)

нормальный = " ".join(текст.split()).lower()
print(нормальный)
````

`split()` без аргументов режет по любым пробелам и сам выбрасывает
лишние. Поэтому связка `" ".join(текст.split())` — самый короткий
способ привести текст в порядок.

Знаки препинания при этом остаются приклеенными к словам: в списке
есть `"раму."` с точкой. Если они мешают, их убирают заранее.

````python
def очистить(текст):
    for знак in ".,!?;:—()\"":
        текст = текст.replace(знак, " ")
    return " ".join(текст.split()).lower()


print(очистить("Мама мыла раму. Рама — была чистой!"))
````

### Пример 2. Считаем вхождения честно

````python
def вхождений(текст, что):
    счётчик = 0
    for i in range(len(текст) - len(что) + 1):
        if текст[i:i + len(что)] == что:
            счётчик += 1
    return счётчик


print("count():", "ааааа".count("аа"))
print("с перекрытиями:", вхождений("ааааа", "аа"))
print("count():", "абабаб".count("абаб"))
print("с перекрытиями:", вхождений("абабаб", "абаб"))
````

Цикл идёт по всем возможным началам и сравнивает срез нужной длины
с образцом. Верхняя граница `len(текст) - len(что) + 1` — чтобы срез
не выехал за конец строки.

### Пример 3. Самое длинное слово

````python
текст = "Наборный шрифт перевернул книгопечатание в пятнадцатом веке"

слова = очистить(текст).split()
длиннейшее = слова[0]
for слово in слова:
    if len(слово) > len(длиннейшее):
        длиннейшее = слово

print("Самое длинное слово:", длиннейшее, f"({len(длиннейшее)} букв)")
print("То же через max:", max(слова, key=len))
````

`max(слова, key=len)` сравнивает слова не по алфавиту, а по длине:
`key` задаёт, что именно сравнивать. При равной длине оба способа
берут первое подходящее слово.

### Пример 4. Проверка формата

Задача из жизни: проверить, похожа ли строка на дату вида `12.05.2026`.

````python
def похоже_на_дату(строка):
    части = строка.split(".")
    if len(части) != 3:
        return False
    день, месяц, год = части
    if not (день.isdigit() and месяц.isdigit() and год.isdigit()):
        return False
    return len(день) == 2 and len(месяц) == 2 and len(год) == 4


for проба in ("12.05.2026", "1.5.2026", "12/05/2026", "аб.вг.дежз"):
    print(проба, "→", похоже_на_дату(проба))
````

Функция проверяет форму, а не смысл: `99.99.2026` она пропустит.
Проверку значений добавляют отдельно — так код остаётся читаемым.
