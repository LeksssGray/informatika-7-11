## Разбираемся

### Модель связей

Многие задачи сводятся к одному и тому же: есть объекты, и между
некоторыми из них есть связи. Города и дороги. Люди и знакомства.
Страницы сайта и ссылки. Станции метро и перегоны.

Для всех этих задач придумана одна модель — **граф**.

> **Граф** — это множество **вершин** и соединяющих их **рёбер**.

Вершины рисуют кружками, рёбра — линиями. Больше в графе ничего нет:
ни расстояний на рисунке, ни углов, ни расположения. Важно только то,
**что с чем соединено**.

```
       А ───── Б
       │  \    │
       │    \  │
       В ───── Г ───── Д
```

Этот же граф можно нарисовать совершенно иначе — рёбра могут быть
кривыми, вершины расположены как угодно. Пока связи те же, граф тот же.

### Основные понятия

**Степень вершины** — количество рёбер, выходящих из неё. У вершины Г
на рисунке степень 4 (связи с Б, В, Д и А через диагональ... впрочем,
посчитайте сами по рисунку).

**Путь** — последовательность вершин, в которой каждые соседние соединены
ребром. Путь А → Б → Г → Д существует, а вот А → Д напрямую — нет.

**Цикл** — путь, который возвращается в начальную вершину.

**Связный граф** — тот, в котором из любой вершины можно добраться
до любой другой. Если граф распадается на несвязанные куски,
эти куски называют **компонентами связности**.

### Виды графов

**Неориентированный** — рёбра работают в обе стороны. Дружба взаимна:
если А дружит с Б, то и Б с А.

**Ориентированный** (орграф) — у рёбер есть направление, их называют
дугами и рисуют стрелками. Ссылки между сайтами односторонние:
страница А ссылается на Б, но не наоборот.

**Взвешенный** — каждому ребру приписано число: длина дороги,
стоимость перелёта, время в пути.

### Задача на подсчёт путей

Классическая задача ОГЭ звучит так:

> Из города А в город К ведут дороги (схема прилагается). Сколько
> существует различных путей из А в К?

Приём решения — **расстановка чисел по вершинам**. У каждой вершины
записываем, сколькими путями в неё можно попасть.

1. В начальной вершине пишем 1.
2. Для каждой следующей вершины складываем числа всех вершин,
   из которых в неё ведут дороги.
3. Двигаемся строго по направлению дорог, слева направо.

```
    A(1) ──→ Б(1) ──→ Г(3)
      │        │        │
      ↓        ↓        ↓
    В(1) ──→ Д(2) ──→ К(5)
```

Разберём: в Б только из А, значит 1. В В только из А, значит 1.
В Д ведут дороги из Б и В: 1 + 1 = 2. В Г из Б и Д: 1 + 2 = 3.
В К из Г и Д: 3 + 2 = 5.

**Ответ: 5 путей.**

Этот приём работает всегда, если в графе нет циклов — а в задачах
этого типа их не бывает. Главное правило: считать вершину можно только
после того, как посчитаны **все** вершины, ведущие в неё.

### Как записать граф в программе

Рисунок компьютеру не покажешь. Самый удобный способ — **список
смежности**: для каждой вершины перечислить, куда из неё можно пойти.

```python
граф = {
    "А": ["Б", "В"],
    "Б": ["Г", "Д"],
    "В": ["Д"],
    "Г": ["К"],
    "Д": ["Г", "К"],
    "К": [],
}
```

Такая запись читается почти как условие задачи и занимает мало места.

## Смотрим, как это работает

### Пример 1. Степени вершин

````python
граф = {
    "А": ["Б", "В", "Г"],
    "Б": ["А", "Г"],
    "В": ["А", "Г"],
    "Г": ["А", "Б", "В", "Д"],
    "Д": ["Г"],
}

for вершина, соседи in граф.items():
    print(f"  {вершина}: степень {len(соседи)}, соседи {', '.join(соседи)}")

сумма_степеней = sum(len(с) for с in граф.values())
print(f"\nСумма степеней: {сумма_степеней}, рёбер в графе: {сумма_степеней // 2}")
````

Обратите внимание на последнюю строку. Сумма степеней всех вершин
всегда **вдвое больше** числа рёбер — каждое ребро посчитано с двух
концов. Это утверждение называют леммой о рукопожатиях, и из него
следует любопытный вывод: вершин нечётной степени в любом графе
всегда чётное количество.

### Пример 2. Подсчёт путей

Реализуем приём с расстановкой чисел.

````python
# Ориентированный граф: куда ведут дороги из каждого города
дороги = {
    "А": ["Б", "В"],
    "Б": ["Г", "Д"],
    "В": ["Д"],
    "Г": ["К"],
    "Д": ["Г", "К"],
    "К": [],
}

# Порядок обработки: вершина считается после всех, кто в неё ведёт
порядок = ["А", "Б", "В", "Д", "Г", "К"]

пути = {вершина: 0 for вершина in дороги}
пути["А"] = 1

for вершина in порядок:
    for сосед in дороги[вершина]:
        пути[сосед] += пути[вершина]
    print(f"  обработали {вершина}: {пути}")

print(f"\nПутей из А в К: {пути['К']}")
````

Каждая вершина «раздаёт» своё число всем, куда ведут её дороги.
К моменту обработки вершины в ней уже накоплена сумма от всех
предшественников — при условии, что мы соблюдаем правильный порядок.

Запись `{вершина: 0 for вершина in дороги}` — это генератор словаря,
короткий способ создать словарь с нулями для всех вершин.

### Пример 3. Проверка существования пути

Проверим, можно ли добраться из одной вершины в другую. Будем идти
по графу, отмечая, где уже побывали.

````python
def достижима(граф, старт, финиш):
    посещённые = set()
    очередь = [старт]

    while очередь:
        текущая = очередь.pop()
        if текущая == финиш:
            return True
        if текущая in посещённые:
            continue
        посещённые.add(текущая)
        очередь.extend(граф[текущая])

    return False


print("Из А в К:", достижима(дороги, "А", "К"))
print("Из К в А:", достижима(дороги, "К", "А"))
print("Из В в Г:", достижима(дороги, "В", "Г"))
````

Множество `посещённые` защищает от зацикливания: без него программа
могла бы бесконечно ходить по кругу. Обратите внимание, что из К в А
пути нет — граф ориентированный, дороги односторонние.

Этот алгоритм называется **обходом графа**, и в 11 классе мы разберём
его подробно. Пока достаточно понимать идею: берём вершину из очереди,
смотрим соседей, добавляем их в очередь.
