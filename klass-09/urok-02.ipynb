## Справочник: системы счисления

### Три системы, которые нужны

| Система | Основание | Цифры |
|---|---|---|
| двоичная | 2 | 0, 1 |
| восьмеричная | 8 | 0 … 7 |
| шестнадцатеричная | 16 | 0 … 9, A … F |

Буквы означают числа: A = 10, B = 11, C = 12, D = 13, E = 14, F = 15.

В системе с основанием q цифры идут **от 0 до q−1**. Записи `189`
в восьмеричной не существует.

### Перевод в десятичную

Умножаем цифры на степени основания и складываем:

$$237_8 = 2 \cdot 8^2 + 3 \cdot 8^1 + 7 \cdot 8^0 = 128 + 24 + 7 = 159$$

Удобнее считать по схеме Горнера, слева направо: результат умножаем
на основание и прибавляем следующую цифру.

```
  0 · 8 + 2 = 2
  2 · 8 + 3 = 19
 19 · 8 + 7 = 159
```

### Перевод из десятичной

Делим на основание с остатком, **остатки читаем снизу вверх**.

### Быстрый перевод между 2, 8 и 16

Через десятичную ходить не нужно — работает группировка:

| Переход | Как |
|---|---|
| 2 → 8 | группы по **3** разряда справа |
| 2 → 16 | группы по **4** разряда справа |
| 8 → 2 | каждая цифра → 3 разряда |
| 16 → 2 | каждая цифра → 4 разряда |

Слева при нехватке дописываем нули. Группировать всегда **справа**.

### Полезные факты о двоичной записи

| Число | Запись | Единиц |
|---|---|---|
| $2^n$ | 1 и n нулей | 1 |
| $2^n - 1$ | n единиц подряд | n |
| $2^a - 2^b$ (a > b) | (a−b) единиц, потом b нулей | a − b |

И приведение оснований: $4^n = 2^{2n}$, $8^n = 2^{3n}$, $16^n = 2^{4n}$.

## Справочник: логика

### Три основные операции

| Операция | Обозначения | Когда истина |
|---|---|---|
| И (конъюнкция) | ∧, `and` | когда **оба** истинны |
| ИЛИ (дизъюнкция) | ∨, `or` | когда **хотя бы один** истинен |
| НЕ (инверсия) | ¬, `not` | когда исходное ложно |

### Таблицы истинности

| A | B | A ∧ B | A ∨ B | ¬A |
|---|---|---|---|---|
| 0 | 0 | 0 | 0 | 1 |
| 0 | 1 | 0 | 1 | 1 |
| 1 | 0 | 0 | 1 | 0 |
| 1 | 1 | 1 | 1 | 0 |

Строк в таблице всегда $2^n$, где n — число переменных.
Три переменные — восемь строк.

### Приоритет

Сначала НЕ, потом И, потом ИЛИ. Скобки главнее всего.

### Законы, которые пригодятся

**Правила де Моргана** — самое полезное:

```
  ¬(A ∧ B)  =  ¬A ∨ ¬B
  ¬(A ∨ B)  =  ¬A ∧ ¬B
```

Отрицание «раскрывается»: И меняется на ИЛИ, каждая часть отрицается.

## Смотрим, как это работает

### Переводы между системами

````python
число = 159

print(f"{число} в десятичной")
print(f"  в двоичной:          {число:b}")
print(f"  в восьмеричной:      {число:o}")
print(f"  в шестнадцатеричной: {число:X}")
print()
print("Обратно:", int("237", 8), int("9F", 16), int("10011111", 2))
````

Запись `f"{число:b}"` сразу даёт двоичную форму без приставки — короче,
чем `bin(число)[2:]`.

### Группировка вместо вычислений

````python
двоичное = "10011111"

# по четыре разряда справа → шестнадцатеричная
левая = двоичное[:4]
правая = двоичное[4:]

print(f"{двоичное} режем на {левая} и {правая}")
print(f"  {левая} = {int(левая, 2):X}")
print(f"  {правая} = {int(правая, 2):X}")
print(f"  итого: {int(левая, 2):X}{int(правая, 2):X}")
print(f"  проверка: {int(двоичное, 2):X}")
````

Никаких умножений на степени — просто разрезали и перевели каждый
кусочек. На экзамене это экономит минуты.

### Таблица истинности программой

````python
print(" A  B | A and B | A or B | not A")
print("-" * 38)
for A in (False, True):
    for B in (False, True):
        print(f" {int(A)}  {int(B)} |    {int(A and B)}    |   {int(A or B)}    |   {int(not A)}")
````

Перебор всех вариантов двумя вложенными циклами — надёжный способ
проверить любое логическое выражение. Если сомневаетесь в ответе,
постройте таблицу кодом.
