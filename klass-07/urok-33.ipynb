## Разбираемся

### Из чего состоит адрес сайта

Адрес `https://school.ru/lesson/07` делится на части:

* **протокол** — `https` — правила, по которым браузер и сайт обмениваются данными;
* **домен** — `school.ru` — имя сайта в сети;
* **путь** — `/lesson/07` — какая страница внутри сайта.

Если после адреса стоит `?`, дальше идут параметры. Их легко увидеть в адресной строке
поиска: там записаны слова, которые ты вводил.

```python
адрес = "https://school.ru/lesson/07"
протокол, остаток = адрес.split("://")
домен = остаток.split("/")[0]
print("протокол:", протокол)
print("домен:", домен)
```

### Поиск: три приёма

* **Кавычки** — `"кодирование ascii"` ищет точную фразу.
* **Минус** — `кодирование -unicode` исключает слово из результатов.
* **Источник** — `site:gov.ru` ищет только на одном сайте.

### Можно ли верить найденному

Проверяй по четырём признакам: есть ли автор и дата, ссылки на первоисточник,
совпадает ли факт в двух независимых источниках, не продаёт ли страница что-то
под видом знаний.

### Пароли и фишинг

Надёжный пароль длинный (не меньше восьми символов), содержит строчные и заглавные
буквы и цифры, не совпадает с твоим именем или датой рождения. **Фишинг** — поддельное
письмо или сайт, которые выманивают пароль. Главное правило: никому не называй пароль
и не вводи его по ссылке из письма, сначала зайди на сайт сам.
