| своё | копия примера | пример чуть изменён | придумано самим |
| защита | не рассказал | без показа | рассказал, показал, ответил |

9–10 баллов — «5», 7–8 — «4», 5–6 — «3».

### Пример: калькулятор оценок

```python
def средняя(оценки):
    return sum(оценки) / len(оценки)


def итог_четверти(оценки):
    с = средняя(оценки)
    if с >= 4.5:
        return 5
    if с >= 3.5:
        return 4
    if с >= 2.5:
        return 3
    return 2


журнал = [
    ["математика", [5, 4, 5, 5]],
    ["русский", [4, 3, 4, 4]],
    ["информатика", [5, 5, 4]],
]
for строка in журнал:
    предмет = строка[0]
    оценки = строка[1]
    print(предмет, "— средняя", round(средняя(оценки), 2),
          "— за четверть", итог_четверти(оценки))
```

Две функции, список, цикл, понятный вывод. Свой проект может быть
таким же небольшим — важно, чтобы он работал и ты мог его объяснить.
