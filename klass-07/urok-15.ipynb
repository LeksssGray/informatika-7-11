## Разбираемся

### Программы бывают разные

Без программ компьютер — просто железо, которое греется. Всё, что
он умеет, ему дают программы. Вместе их называют **программным
обеспечением** (ПО). Его делят на три группы.

| Вид ПО | Для чего | Примеры |
|---|---|---|
| **системное** | чтобы работал сам компьютер | операционная система, драйверы, антивирус |
| **прикладное** | чтобы решать задачи пользователя | браузер, текстовый редактор, игры, мессенджер |
| **инструментальное** | чтобы писать новые программы | Python, Colab, среды разработки |

**Драйвер** — маленькая программа, которая объясняет системе, как
работать с конкретным устройством: принтером, видеокартой, мышью.

### Операционная система

**Операционная система** (ОС) — главная программа компьютера. Она
запускается первой, как только нажал кнопку, и работает до выключения.

Что она делает:

1. **Запускает программы** и следит, чтобы они не мешали друг другу.
2. **Делит процессор и память**: у тебя открыты браузер, музыка
   и игра, и каждой достаётся свой кусочек.
3. **Управляет файлами**: где что лежит на диске, кто может открыть.
4. **Работает с устройствами** через драйверы.
5. **Показывает интерфейс**: рабочий стол, окна, значки, меню.

| ОС | Где работает |
|---|---|
| Windows | большинство домашних и школьных компьютеров |
| macOS | компьютеры Apple |
| Linux | серверы, суперкомпьютеры, часть домашних компьютеров |
| Android | большинство телефонов |
| iOS | iPhone |

В России делают свои системы на основе Linux, например Astra Linux
и РЕД ОС: на них работают в госучреждениях и многих школах.

### Лицензии: что можно делать с программой

Покупая программу, ты обычно покупаешь не саму программу, а **право
пользоваться** ею по правилам — **лицензию**.

| Вид | Что можно | Пример |
|---|---|---|
| **проприетарное** платное | пользоваться после оплаты, код закрыт | Microsoft Office, многие игры |
| **бесплатное** | пользоваться даром, код закрыт | браузер, мессенджер |
| **условно-бесплатное** | попробовать бесплатно, потом заплатить | пробные версии программ |
| **свободное** | пользоваться, смотреть код, менять и распространять | Linux, Firefox, Python |

Скачать платную программу с «взломанного» сайта — это **пиратство**:
нарушение закона. А ещё в таких программах часто спрятаны вирусы.

### Вредоносные программы

| Кто | Что делает |
|---|---|
| **вирус** | прячется внутри других файлов и заражает новые |
| **червь** | сам расползается по сети с компьютера на компьютер |
| **троян** | выдаёт себя за полезную программу, а сам крадёт данные |
| **шпион** | тайно следит, что ты вводишь и какие сайты открываешь |
| **вымогатель** | шифрует твои файлы и требует деньги за расшифровку |

Первым вирусом для IBM-совместимых компьютеров считают **Brain**
1986 года. Его написали два брата из Пакистана. Вирус записывался
в начало дискеты, и в нём даже были их имена, адрес и телефон — видно
на снимке справа. Вреда он почти не причинял, но за несколько месяцев
разошёлся по всему миру вместе с дискетами.

Через два года студент Роберт Моррис запустил в интернет червя —
в качестве эксперимента. Из-за ошибки в программе червь размножался
слишком быстро и за сутки вывел из строя около шести тысяч
компьютеров, примерно каждый десятый в тогдашнем интернете.

### Как защититься

1. **Обновления** — ставь их: они закрывают дыры, через которые
   пролезают вредоносные программы.
2. **Программы — только с официальных сайтов** и из магазинов
   приложений.
3. **Не запускай** файлы `.exe` и не открывай вложения от незнакомых.
4. **Антивирус** — встроенный в Windows обычно достаточен.
5. **Копии важных файлов** — на случай вымогателя.
