## Разбираемся

### Третья базовая конструкция

Мы уже умеем две вещи: выполнять команды **по порядку** и **выбирать**
с помощью `if`. Осталась третья — **повторять**. Из этих трёх
кирпичиков собирается любая программа на свете.

Напечатать «Привет» пять раз можно пятью строчками. А сто раз?
Тысячу? Для этого есть **цикл**.

```python
for i in range(5):
    print("Привет")
```

Читается так: «для каждого `i` из диапазона от 0 до 5 — напечатай».
Строки с отступом под `for` называются **телом цикла**. Они выполнятся
пять раз.

### Переменная цикла

`i` — обычная переменная, которая на каждом повторе принимает следующее
значение. Её можно использовать внутри:

```python
for i in range(5):
    print("Шаг", i)
```

напечатает `Шаг 0`, `Шаг 1`, … `Шаг 4`. Обрати внимание: **счёт идёт
с нуля**, а пятёрка в диапазон **не входит**.

### Три формы range

| Запись | Какие числа | Пример |
|---|---|---|
| `range(n)` | от 0 до n−1 | `range(4)` → 0, 1, 2, 3 |
| `range(a, b)` | от a до b−1 | `range(2, 6)` → 2, 3, 4, 5 |
| `range(a, b, шаг)` | от a до b−1 с шагом | `range(1, 10, 3)` → 1, 4, 7 |

Шаг бывает и отрицательным — тогда счёт идёт вниз:
`range(5, 0, -1)` → 5, 4, 3, 2, 1.

> Правую границу `range` никогда не включает. Хочешь числа
> от 1 до 10 — пиши `range(1, 11)`.

### Накопление

Самый полезный приём с циклом — **копить результат**. Сумма чисел
от 1 до 5:

```python
сумма = 0
for i in range(1, 6):
    сумма = сумма + i
print(сумма)
```

До цикла заводим «копилку» и кладём в неё ноль. На каждом шаге
добавляем очередное число. После цикла в копилке сумма: 15.

Запись `сумма = сумма + i` можно сократить до `сумма += i` — это
одно и то же.

## Смотрим, как это работает

### Повторяем

````python
for i in range(5):
    print("Привет, шаг", i)
````

### Три формы range

````python
print("range(4):")
for i in range(4):
    print(i)

print("range(2, 6):")
for i in range(2, 6):
    print(i)

print("range(10, 0, -3):")
for i in range(10, 0, -3):
    print(i)
````

В последнем цикле счёт идёт вниз с шагом 3: 10, 7, 4, 1. Ноль
не напечатался — правая граница не входит.

### Таблица умножения на 7

````python
for i in range(1, 11):
    print(f"7 × {i} = {7 * i}")
````

Десять строк программой в две строчки. Поменяй 7 на своё число.

### Накопление суммы

````python
сумма = 0
for i in range(1, 6):
    сумма += i
    print(f"прибавили {i}, в копилке {сумма}")

print("Итого:", сумма)
````

Видно, как копилка растёт шаг за шагом: 1, 3, 6, 10, 15.

### Цикл и условие вместе

Внутри цикла можно проверять условие. Посчитаем, сколько чисел
от 1 до 30 делятся на 3.

````python
счётчик = 0
for i in range(1, 31):
    if i % 3 == 0:
        счётчик += 1

print("Делятся на 3:", счётчик)
````
