## Смотрим, как это работает

### Компьютер печатает

Запусти ячейку ниже.

````python
print("Привет! Это моя первая программа.")
````

Разберём по кусочкам:

* `print` — команда «напечатай»;
* скобки `( )` — в них кладут то, что печатать;
* кавычки `" "` — говорят компьютеру: внутри обычный текст, а не команда.

Кавычки на экран не попадают, они только обозначают, где текст начинается
и где заканчивается.

### Компьютер считает

Если кавычки убрать и написать числа, он посчитает.

````python
print(2 + 2)
print(365 * 24)
print("365 * 24")
````

Посмотри на третью строку. `"365 * 24"` в кавычках — это **текст**,
и компьютер напечатал его как есть, ничего не считая. А без кавычек
это **пример**, и он его решил.

Разница между текстом и числом ещё не раз всплывёт. Запомни на будущее:
кавычки выключают вычисление.

### Несколько штук сразу

В одном `print` можно перечислить всё через запятую.

````python
print("В сутках", 24, "часа, а в году", 365, "дней")
````

````python
# калькулятор: меняй числа и запускай
print(365 * 24)
````

Пробелы между кусками компьютер расставил сам. Текст — в кавычках,
числа — без, и всё это спокойно живёт в одной команде.

### Ошибка — это нормально

Ломается у всех и постоянно. Важно не «никогда не ошибаться», а уметь
прочитать, что именно не понравилось.

Ячейка ниже сломана специально. Запусти её.

````python
# СПЕЦИАЛЬНО СЛОМАННАЯ ЯЧЕЙКА  demo-error
print("Здесь забыли закрыть кавычку)
````

Появился красный текст. Не пугайся: в нём важна **последняя строка**.

```
SyntaxError: unterminated string literal
```

`SyntaxError` — ошибка в записи, `unterminated string` — незакрытая строка.
Компьютер даже показывает стрелочкой, где споткнулся.

Почини ячейку выше: поставь закрывающую кавычку перед `)` и запусти снова.
«Запустил → сломалось → починил» — обычная работа программиста,
так бывает по многу раз на дню.
