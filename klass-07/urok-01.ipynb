Ты открыл не учебник, а рабочую тетрадь, в которой можно нажимать кнопки.

### Страница собрана из ячеек

Ячейки бывают двух видов.

**Текстовые** — вроде той, которую ты читаешь. В них объяснения, картинки,
таблицы.

**Кодовые** — в них программа. Слева у такой ячейки круглая кнопка ▶.
Нажал — программа выполнилась, результат появился внизу.

### Как запускать

Три способа, выбирай любой:

* кнопка ▶ слева от ячейки;
* **Shift + Enter** — выполнить и перейти к следующей;
* **Ctrl + Enter** — выполнить и остаться на месте.

Пока ячейка думает, слева крутится кружок. Когда закончит — появится
номер. Он показывает, в каком порядке ты запускал ячейки.

> **Одно правило, которое сэкономит тебе кучу нервов.** Ячейки выполняются
> в том порядке, в котором ты нажимаешь ▶, а не в том, в котором они идут
> на странице. Пропустишь ячейку — следующая может не заработать.
> Поэтому иди сверху вниз и ничего не перепрыгивай.

## Смотрим, как это работает

### Компьютер печатает

Запусти ячейку ниже.

````python
print("Привет! Это моя первая программа.")
````

Разберём по кусочкам:

* `print` — команда «напечатай»;
* скобки `( )` — в них кладут то, что печатать;
* кавычки `" "` — говорят компьютеру: внутри обычный текст, а не команда.

Кавычки на экран не попадают, они только обозначают, где текст начинается
и где заканчивается.

### Компьютер считает

Если кавычки убрать и написать числа, он посчитает.

````python
print(2 + 2)
print(365 * 24)
print("365 * 24")
````

Третья строка — самое интересное. `"365 * 24"` в кавычках это **текст**,
и компьютер честно напечатал его как есть, ничего не считая.
А без кавычек — это **пример**, и он его решил.

Разница между текстом и числом ещё не раз всплывёт. Запомни на будущее:
кавычки выключают вычисление.

### Несколько штук сразу

В одном `print` можно перечислить всё через запятую.

````python
print("В сутках", 24, "часа, а в году", 365, "дней")
````

````python
# калькулятор: меняй числа и запускай
print(365 * 24)
````

Пробелы между кусками компьютер расставил сам. Текст — в кавычках,
числа — без, и всё это спокойно живёт в одной команде.

### Ошибка — это нормально

Ломается у всех и постоянно. Важно не «никогда не ошибаться», а уметь
прочитать, что именно не понравилось.

Ячейка ниже сломана специально. Запусти её.

````python
# СПЕЦИАЛЬНО СЛОМАННАЯ ЯЧЕЙКА  demo-error
print("Здесь забыли закрыть кавычку)
````

Вылезло красное. Не пугайся: в нём важна **последняя строка**.

```
SyntaxError: unterminated string literal
```

`SyntaxError` — ошибка в записи, `unterminated string` — незакрытая строка.
Компьютер даже показывает стрелочкой, где споткнулся.

Почини ячейку выше: поставь закрывающую кавычку перед `)` и запусти снова.
Вот это «запустил → сломалось → починил» и есть обычная работа
программиста. Не исключение, а рутина.
