## Часть 1. Теория — 15 минут

### Что такое информатика

Информатика — наука о том, как **представлять**, **хранить**, **передавать**
и **обрабатывать** информацию. Ключевое слово здесь — «обрабатывать»:
компьютер не понимает смысл того, с чем работает, он умеет только выполнять
точные инструкции. Наша задача — научиться такие инструкции составлять.

Инструкция для компьютера называется **программой**, а язык, на котором она
записана, — **языком программирования**. Мы будем писать на языке **Python**:
на нём работают поисковые системы, нейросети и научные расчёты, и при этом
он достаточно простой, чтобы начать с первого урока.

### Где мы будем работать

Этот учебник — не бумажная книга, а набор **ноутбуков Google Colab**.
Ноутбук — это документ, в котором текст соседствует с живым кодом:
код можно запускать прямо со страницы и сразу видеть результат.
Ничего устанавливать не нужно, программа выполняется на серверах Google,
а вам достаточно браузера.

### Из чего состоит ноутбук

Ноутбук собран из **ячеек** двух видов.

| Вид ячейки | Что содержит | Что делает при запуске |
|---|---|---|
| Текстовая | объяснения, формулы, картинки | превращается в красиво оформленный текст |
| Кодовая | программу на Python | выполняет программу и показывает результат под собой |

Ячейка, которую вы читаете прямо сейчас, — текстовая. Ниже будет кодовая:
у неё слева есть круглая кнопка ▶.

### Как запустить ячейку

Три способа, выбирайте любой:

1. нажать кнопку ▶ слева от ячейки;
2. нажать **Shift + Enter** — ячейка выполнится, и курсор перейдёт к следующей;
3. нажать **Ctrl + Enter** — ячейка выполнится, курсор останется на месте.

Пока ячейка работает, слева крутится кружок. Когда она закончит,
появится номер — он показывает, в каком порядке вы запускали ячейки.

> ⚠️ **Важное правило.** Ячейки выполняются в том порядке, в котором вы
> нажимаете ▶, а не в том, в котором они расположены на странице.
> Если пропустить ячейку, следующая может не заработать. Поэтому на уроке
> запускайте всё сверху вниз, ничего не пропуская.

## Часть 2. Примеры с разбором — 10 минут

### Пример 1. Первая программа

Запустите ячейку ниже и посмотрите, что появится под ней.

````python
print("Привет! Это моя первая программа.")
````

Разберём по частям:

* `print` — команда «напечатай». По-английски это «печатать»;
* круглые скобки `( )` — в них указывают, **что именно** печатать;
* кавычки `" "` — говорят Python: внутри обычный текст, а не команда.

Текст в кавычках называется **строкой**. Кавычки на экран не выводятся —
они нужны только для того, чтобы обозначить границы текста.

### Пример 2. Компьютер считает

Если убрать кавычки и написать числа, Python посчитает выражение.

````python
print(2 + 2)
print(365 * 24)
print("2 + 2")
````

Обратите внимание на третью строку: `"2 + 2"` в кавычках — это **текст**,
и Python честно печатает его как есть, ничего не вычисляя.
А `2 + 2` без кавычек — это **вычисление**, и печатается результат `4`.

Разница между текстом и числом — одна из самых важных идей в программировании.
Мы вернёмся к ней ещё не раз.

### Пример 3. Несколько значений сразу

В одном `print` можно перечислить несколько значений через запятую.

````python
print("В сутках", 24, "часа, а в году", 365, "дней")
````

Python сам расставил пробелы между значениями. Заметьте: текст пишется
в кавычках, а числа — без них, и всё это спокойно уживается в одной команде.

### Пример 4. Ошибка — это нормально

Программисты ошибаются постоянно. Важно не «не ошибаться», а уметь читать,
что именно не понравилось компьютеру. Запустите ячейку — она сломается
специально.

````python
# СПЕЦИАЛЬНО СЛОМАННАЯ ЯЧЕЙКА  demo-error
print("Здесь забыли закрыть кавычку)
````

Появилось красное сообщение. В нём главное — **последняя строка**:

```
SyntaxError: unterminated string literal
```

`SyntaxError` значит «ошибка в записи», а `unterminated string` —
«незакрытая строка». Python даже показывает стрелочкой место, где споткнулся.

Исправьте ячейку выше: поставьте закрывающую кавычку перед `)` — и запустите снова.
Такой цикл «запустил → сломалось → починил» и есть обычная работа программиста.
