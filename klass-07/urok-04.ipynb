## Часть 1. Теория — 15 минут

### Почему компьютер считает двойками

Внутри компьютера нет цифр — там есть электрические сигналы.
Надёжно различить можно всего два состояния: **есть напряжение** и **нет
напряжения**. Попытка различать десять уровней сигнала провалилась бы
на первой же помехе: «пять вольт» легко спутать с «четырьмя с половиной»,
а вот «есть ток» и «нет тока» не спутаешь.

Поэтому вся техника работает с двумя цифрами — **0** и **1**. Система
счисления, в которой используются только они, называется **двоичной**.

### Как устроена привычная десятичная запись

Прежде чем перейти к двоичной, разберём знакомую. Запись `275` означает:

$$275 = 2 \cdot 100 + 7 \cdot 10 + 5 \cdot 1 = 2 \cdot 10^2 + 7 \cdot 10^1 + 5 \cdot 10^0$$

Каждая цифра умножается на степень десятки, а показатель степени равен
номеру позиции, если считать справа налево с нуля. Такая система называется
**позиционной**: значение цифры зависит от её места.

### Двоичная запись устроена так же

Отличие ровно одно — вместо десятки берём двойку:

$$1011_2 = 1 \cdot 2^3 + 0 \cdot 2^2 + 1 \cdot 2^1 + 1 \cdot 2^0 = 8 + 0 + 2 + 1 = 11$$

Маленькая двойка внизу — основание системы. Её пишут, чтобы `1011`
не приняли за десятичную тысячу одиннадцать.

Полезно выучить степени двойки — они нужны постоянно:

| Степень | 2⁰ | 2¹ | 2² | 2³ | 2⁴ | 2⁵ | 2⁶ | 2⁷ | 2⁸ |
|---|---|---|---|---|---|---|---|---|---|
| Значение | 1 | 2 | 4 | 8 | 16 | 32 | 64 | 128 | 256 |

### Перевод из двоичной в десятичную

Надпишите над разрядами степени двойки справа налево и сложите те,
где стоит единица:

```
  разряды:   16   8   4   2   1
  число:      1   1   0   1   0     →   16 + 8 + 2 = 26
```

### Перевод из десятичной в двоичную

Обратный ход — деление с остатком на 2. Делим, пока не получим ноль,
и **записываем остатки снизу вверх**:

```
  26 : 2 = 13,  остаток 0   ↑
  13 : 2 =  6,  остаток 1   │
   6 : 2 =  3,  остаток 0   │  читаем
   3 : 2 =  1,  остаток 1   │  в эту
   1 : 2 =  0,  остаток 1   │  сторону

  Ответ: 11010₂
```

Проверим обратным переводом: 16 + 8 + 2 = 26. Сходится.

### Сложение в двоичной системе

Правила почти как в десятичной, только перенос возникает раньше:

```
  0 + 0 = 0        1 + 1 = 10   (ноль пишем, единицу в уме)
  0 + 1 = 1        1 + 1 + 1 = 11
```

Пример:

```
    1 0 1 1        (11)
  +   1 1 0         (6)
  ─────────
  1 0 0 0 1        (17)
```

## Часть 2. Примеры с разбором — 10 минут

### Пример 1. Переводим вручную — и проверяем компьютером

В Python есть готовые инструменты для перевода. Но сначала всегда считайте
сами, а машиной — проверяйте: на контрольной калькулятора не будет.

````python
# int(строка, 2) — прочитать строку как двоичное число
print("1011 в двоичной это", int("1011", 2))
print("11010 в двоичной это", int("11010", 2))

# bin(число) — получить двоичную запись
print("26 в двоичной это", bin(26))
````

Обратите внимание: `bin` добавляет приставку `0b` — так Python помечает
двоичные числа. Чтобы получить чистую запись, приставку отрезают:
`bin(26)[2:]` даст `'11010'`.

### Пример 2. Перевод вручную, шаг за шагом

Напишем перевод из двоичной в десятичную сами, без готовых функций —
буквально повторяя формулу из теории.

````python
двоичное = "11010"
результат = 0
степень = len(двоичное) - 1   # старший разряд слева

for цифра in двоичное:
    вклад = int(цифра) * 2 ** степень
    print(f"  цифра {цифра} в разряде 2^{степень} даёт {вклад}")
    результат += вклад
    степень -= 1

print("Итого:", результат)
````

Цикл идёт по символам строки слева направо, а степень убывает от старшей
к младшей. Каждая единица добавляет свою степень двойки, каждый ноль —
ничего.

### Пример 3. Деление с остатком

Теперь обратный перевод. Понадобятся две операции:

* `//` — деление нацело: `26 // 2` даёт `13`;
* `%` — остаток от деления: `26 % 2` даёт `0`.

````python
число = 26
остатки = ""

while число > 0:
    остаток = число % 2
    остатки = str(остаток) + остатки   # приписываем слева!
    print(f"  {число} : 2 = {число // 2}, остаток {остаток}")
    число = число // 2

print("Двоичная запись:", остатки)
````

Ключевая хитрость — в строке `остатки = str(остаток) + остатки`.
Каждый новый остаток приписывается **слева**, поэтому в конце они
оказываются в нужном порядке, и переворачивать ничего не нужно.
