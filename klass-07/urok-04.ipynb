## Разбираемся

### Как измерить то, чего не видно

В прошлый раз договорились: информация уменьшает незнание. Значит
и мерить её можно тем, насколько сильно это незнание уменьшилось.

Самый простой случай. Я загадал: орёл или решка. Вариантов два, оба
одинаково вероятны. Ты спросил, я ответил — и незнания больше нет.

> Количество информации в сообщении, которое выбирает **один вариант
> из двух равновозможных**, приняли за единицу. Она называется **бит**.

Слово от английского *binary digit* — «двоичная цифра». Один бит это
ответ «да» или «нет». Ноль или единица. Есть сигнал или нет сигнала.

### Вариантов больше — информации больше

А если вариантов четыре? Скажем, я загадал масть карты. Одним вопросом
не обойтись, но двумя — легко:

```
Красная?  ──да──→  Черви?   ──да──→  ЧЕРВИ
                            ──нет─→  БУБНЫ
          ──нет─→  Пики?    ──да──→  ПИКИ
                            ──нет─→  ТРЕФЫ
```

Два вопроса — два бита. Видишь закономерность? 2 варианта — 1 бит,
4 варианта — 2 бита, 8 вариантов — 3 бита. Каждый новый бит **удваивает**
число вариантов, которые можно различить.

### Главная формула

$$N = 2^i$$

**N** — сколько равновозможных вариантов, **i** — сколько бит информации.

| N вариантов | 2 | 4 | 8 | 16 | 32 | 64 | 128 | 256 |
|---|---|---|---|---|---|---|---|---|
| i бит | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 |

Эту табличку стоит выучить. Она будет нужна почти в каждой задаче
до конца школы.

### Байт и всё, что больше

Одним битом много не скажешь, поэтому их объединили по восемь:

> **1 байт = 8 бит**

Восемь бит дают 256 вариантов — как раз хватает на все буквы, цифры
и знаки препинания. Поэтому один символ текста исторически весил
ровно байт.

Дальше единицы растут **не по тысяче, а по 1024**, потому что 1024 —
это два в десятой степени, круглое число в двоичном мире:

| Единица | Это сколько | В байтах |
|---|---|---|
| 1 килобайт | 1024 байта | 1 024 |
| 1 мегабайт | 1024 Кбайт | 1 048 576 |
| 1 гигабайт | 1024 Мбайт | 1 073 741 824 |

### Сколько весит текст

Порядок такой:

1. считаем **мощность алфавита** N — сколько разных символов может
   встретиться, включая пробелы и запятые;
2. по формуле находим i — сколько бит на **один** символ;
3. умножаем i на количество символов в тексте.

Например: алфавит из 32 символов, текст из 100 символов. Раз 32 это два
в пятой, символ весит 5 бит. Весь текст — 500 бит, то есть 62,5 байта.

Дробный ответ — это нормально: байт нужен для хранения, а символ вовсе
не обязан занимать целое их количество.

## Смотрим, как это работает

### Степени двойки

Запусти ячейку и сверься с таблицей из теории.

````python
print("1 бит  различает", 2 ** 1, "варианта")
print("2 бита различают", 2 ** 2, "варианта")
print("3 бита различают", 2 ** 3, "вариантов")
print("5 бит  различают", 2 ** 5, "вариантов")
print("8 бит  различают", 2 ** 8, "вариантов")
````

Каждая строчка вдвое больше предыдущей. Восемь бит — уже 256 вариантов,
и это ровно один байт.

### Переводим единицы

Три мегабайта в биты — по шагам, чтобы видеть каждое действие.

````python
print("3 Мбайт =", 3 * 1024, "Кбайт")
print("3 Мбайт =", 3 * 1024 * 1024, "байт")
print("3 Мбайт =", 3 * 1024 * 1024 * 8, "бит")
````

Правило простое: **вниз по таблице умножаем, вверх — делим**.
Из мегабайтов в байты — дважды умножить на 1024. Обратно — дважды
разделить.

### Считаем объём текста

Алфавит из 32 символов, в тексте 100 символов.

````python
print("Бит на один символ:", 5)
print("Всего бит:", 5 * 100)
print("Это байт:", 5 * 100 / 8)
````
