## Разбираемся

### Почему компьютер считает двойками

Внутри компьютера нет цифр. Там электрические сигналы, и надёжно
различить можно всего два состояния: **есть напряжение** и **нет
напряжения**.

Попробуй компьютер различать десять уровней — он бы путался на первой же
помехе: «пять вольт» легко спутать с «четырьмя с половиной». А вот «есть
ток» и «нет тока» не спутаешь никак.

Поэтому вся техника работает с двумя цифрами: **0** и **1**. Система
записи чисел, где кроме них ничего нет, называется **двоичной**.

### Сначала про привычную запись

Прежде чем лезть в двоичную, разберём знакомую. Что означает `275`?

```
  2 сотни  +  7 десятков  +  5 единиц
  2 · 100  +  7 · 10      +  5 · 1
```

Каждая цифра умножается на свой «вес», и веса растут справа налево:
1, 10, 100, 1000. Такая запись называется **позиционной**: значение
цифры зависит от её места. Единица в конце — это один, а в начале — сотня.

### Двоичная устроена точно так же

Отличие ровно одно: веса растут не десятками, а двойками.

```
  веса:     16   8   4   2   1
  число:     1   1   0   1   0
             ↓   ↓       ↓
            16 + 8  +    2  = 26
```

Складываешь веса там, где стоит единица. Всё.

Веса стоит помнить наизусть, они будут нужны постоянно:

| 1 | 2 | 4 | 8 | 16 | 32 | 64 | 128 | 256 |
|---|---|---|---|---|---|---|---|---|

Чтобы `1011` не приняли за тысячу одиннадцать, внизу приписывают
основание: `1011₂`.

### Из двоичной в обычную

Надпиши веса справа налево и сложи те, где единица.

```
  веса:     8   4   2   1
  число:    1   0   1   1     →   8 + 2 + 1 = 11
```

### Из обычной в двоичную

Обратный ход — деление на 2 с остатком. Делим, пока не дойдём до нуля,
и **читаем остатки снизу вверх**:

```
  26 : 2 = 13,  остаток 0   ↑
  13 : 2 =  6,  остаток 1   │
   6 : 2 =  3,  остаток 0   │  читаем
   3 : 2 =  1,  остаток 1   │  сюда
   1 : 2 =  0,  остаток 1   │

  Ответ: 11010₂
```

Проверяем обратно: 16 + 8 + 2 = 26. Сошлось.

> Читать остатки снизу вверх — то место, где ошибаются чаще всего.
> Прочтёшь сверху вниз — получишь число задом наперёд.

### Сложение

Правила почти как в столбик, только перенос случается раньше:

```
  0 + 0 = 0        1 + 1 = 10   (ноль пишем, единицу в уме)
  0 + 1 = 1        1 + 1 + 1 = 11
```

Пример:

```
    1 0 1 1        (11)
  +   1 1 0         (6)
  ─────────
  1 0 0 0 1        (17)
```

## Смотрим, как это работает

### Компьютер умеет переводить сам

Считать всё равно надо уметь руками — на контрольной калькулятора
не будет. Но проверять себя удобно машиной.

````python
# int(строка, 2) читает строку как двоичное число
print("1011 в двоичной это", int("1011", 2))
print("11010 в двоичной это", int("11010", 2))

# bin(число) выдаёт двоичную запись
print("26 в двоичной это", bin(26))
````

Заметил приставку `0b`? Так Python помечает, что число двоичное.
Сама запись — это то, что после неё.

### Складываем веса вручную

Разберём `11010` так же, как в теории, но пусть считает компьютер.

````python
print("Разряды числа 11010 справа налево:")
print("  1 стоит на весе 2   →", 1 * 2)
print("  0 стоит на весе 4   →", 0 * 4)
print("  1 стоит на весе 8   →", 1 * 8)
print("  1 стоит на весе 16  →", 1 * 16)
print()
print("Итого:", 2 + 8 + 16)
````

Нули можно вообще не считать — они ничего не добавляют. Смотри только
на единицы и складывай их веса.

### Деление с остатком

Для обратного перевода нужны две операции:

* `//` — деление нацело: `26 // 2` даёт `13`;
* `%` — остаток: `26 % 2` даёт `0`.

````python
print("26 // 2 =", 26 // 2, "  остаток", 26 % 2)
print("13 // 2 =", 13 // 2, "  остаток", 13 % 2)
print(" 6 // 2 =", 6 // 2, "  остаток", 6 % 2)
print(" 3 // 2 =", 3 // 2, "  остаток", 3 % 2)
print(" 1 // 2 =", 1 // 2, "  остаток", 1 % 2)
print()
print("Остатки снизу вверх: 1 1 0 1 0")
print("Проверяем:", bin(26))
````

Эта пара операций — `//` и `%` — будет попадаться тебе постоянно.
Деление нацело отвечает «сколько целых раз поместилось»,
остаток — «сколько не поместилось».
