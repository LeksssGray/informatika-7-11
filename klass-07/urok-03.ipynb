## Часть 1. Теория — 15 минут

### Как измерить то, чего не видно

На прошлом уроке мы договорились: информация уменьшает неопределённость.
Значит, измерять её можно тем, **насколько сильно** она эту неопределённость
уменьшила.

Возьмём самый простой случай. Я загадал: орёл или решка. Вариантов два,
они равновероятны. Вы спрашиваете — я отвечаю. После ответа
неопределённости не осталось.

> Количество информации в сообщении, которое выбирает **один вариант
> из двух равновозможных**, принято за единицу и называется **бит**.

Слово «бит» — от английского *binary digit*, «двоичная цифра».
Один бит — это ответ «да» или «нет», ноль или единица, есть сигнал
или нет сигнала.

### Больше вариантов — больше информации

А если вариантов не два, а четыре? Скажем, я загадал масть карты.
Одним вопросом «да/нет» тут не обойтись, но двумя — вполне:

```
Красная масть?  ──да──→  Черви?   ──да──→  ЧЕРВИ
                                  ──нет─→  БУБНЫ
                ──нет─→  Пики?    ──да──→  ПИКИ
                                  ──нет─→  ТРЕФЫ
```

Два вопроса — два бита. Заметьте закономерность: 2 варианта — 1 бит,
4 варианта — 2 бита, 8 вариантов — 3 бита. Каждый новый бит **удваивает**
число различимых вариантов.

### Главная формула

Связь числа вариантов и количества информации записывают так:

$$N = 2^i$$

где **N** — количество равновозможных вариантов, **i** — количество
информации в битах.

| N (вариантов) | 2 | 4 | 8 | 16 | 32 | 64 | 128 | 256 |
|---|---|---|---|---|---|---|---|---|
| i (бит) | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 |

Эту таблицу стоит выучить — она будет нужна почти в каждой задаче.

### Байт и его старшие братья

Одного бита мало для чего угодно полезного, поэтому биты объединили
в группы по восемь:

> **1 байт = 8 бит**

Восемь бит дают $2^8 = 256$ вариантов — как раз хватает, чтобы
закодировать все буквы, цифры и знаки препинания. Поэтому исторически
один символ текста занимал ровно один байт.

Дальше единицы растут не по тысяче, а **по 1024** — потому что 1024
это $2^{10}$, круглое число в двоичном мире:

| Единица | Сколько это | В байтах |
|---|---|---|
| 1 килобайт (Кбайт) | 1024 байта | 1 024 |
| 1 мегабайт (Мбайт) | 1024 Кбайт | 1 048 576 |
| 1 гигабайт (Гбайт) | 1024 Мбайт | 1 073 741 824 |
| 1 терабайт (Тбайт) | 1024 Гбайт | ≈ 10<sup>12</sup> |

### Объём текста: алфавитный подход

Чтобы посчитать, сколько «весит» текст, поступают так.

1. Считают **мощность алфавита** N — сколько различных символов может
   встретиться (включая пробелы и знаки препинания).
2. По формуле $N = 2^i$ находят i — сколько бит занимает **один символ**.
3. Умножают i на количество символов в тексте.

**Пример.** В алфавите 32 символа, в тексте 100 символов.
$32 = 2^5$, значит один символ весит 5 бит, а весь текст — 500 бит,
то есть 62,5 байта.

## Часть 2. Примеры с разбором — 10 минут

### Пример 1. Считаем биты по числу вариантов

Найдём i для разных N перебором степеней двойки.

````python
def бит_на_символ(вариантов):
    """Сколько бит нужно, чтобы различить заданное число вариантов."""
    бит = 0
    while 2 ** бит < вариантов:
        бит += 1
    return бит


for N in [2, 4, 8, 16, 32, 64, 256]:
    print(f"{N:>4} вариантов → {бит_на_символ(N)} бит")
````

Цикл увеличивает счётчик, пока `2 ** бит` не дорастёт до нужного числа.
Такой перебор — самый честный способ понять формулу: мы буквально
спрашиваем «а трёх бит хватит? а четырёх?».

Обратите внимание на строку `f"{N:>4} ..."` — это f-строка, в неё можно
подставлять значения прямо внутрь текста. Запись `:>4` выравнивает число
по правому краю на четыре знака, чтобы столбик получился ровным.

### Пример 2. Перевод единиц

Переведём 3 мегабайта в биты — шаг за шагом, чтобы видеть каждое действие.

````python
мегабайты = 3
килобайты = мегабайты * 1024
байты = килобайты * 1024
биты = байты * 8

print(мегабайты, "Мбайт =", килобайты, "Кбайт")
print(килобайты, "Кбайт =", байты, "байт")
print(байты, "байт =", биты, "бит")
````

Правило простое: **вверх по таблице делим, вниз — умножаем.**
Из мегабайтов в байты — умножаем дважды на 1024. Обратно — делим.

### Пример 3. Задача про алфавитный подход

> В алфавите некоторого языка 16 символов. Сообщение состоит из 200
> символов. Каков его объём в байтах?

Решаем по шагам из теории.

````python
N = 16
символов = 200

i = бит_на_символ(N)             # шаг 1-2: бит на один символ
объём_бит = i * символов          # шаг 3: весь текст в битах
объём_байт = объём_бит / 8        # переводим в байты

print(f"Один символ: {i} бит")
print(f"Всё сообщение: {объём_бит} бит = {объём_байт} байт")
````

Ответ: 100 байт. Заметьте, что $16 = 2^4$, поэтому символ занял 4 бита —
ровно половину байта. Такое бывает часто, и в этом нет ничего странного:
байт — единица хранения, а не обязательный размер символа.
