## Разбираемся

### Что считать информацией

Строгого определения у слова «информация» нет — как нет его у слов «время»
или «жизнь». Но для работы хватит такого:

> Информация — это то, что уменьшает незнание.

Вторая половина тут важнее первой. Сообщение несёт информацию, только если
ты чего-то не знал. Фраза «завтра будет либо дождь, либо нет» формально
верна, но не сообщает ровным счётом ничего: до неё ты знал столько же.

### Как информация до нас доходит

Через органы чувств, и распределение довольно неожиданное:

| Откуда | Чем воспринимаем | Сколько примерно |
|---|---|---|
| глазами | зрение | 80–90 % |
| ушами | слух | 8–10 % |
| носом, языком, кожей | остальное | около 2 % |

Компьютеру органы чувств не нужны. Ему важно другое — **в каком виде**
информация записана: числом, текстом, картинкой, звуком, видео.

Одно и то же можно подать по-разному. Температуру за окном — числом `+18`,
словом «тепло», графиком за неделю или цветом на карте погоды. Информация
та же, форма разная.

### Три вещи, которые с информацией происходят

Что бы ни случилось, всё сводится к трём процессам.

**Хранение.** Информацию записали куда-то, чтобы вернуться позже.
Носитель — лист бумаги, флешка, память компьютера, твоя собственная память.

**Передача.** Информация переехала от одного к другому. Тут всегда есть
трое участников и одна помеха:

```
  ИСТОЧНИК  ──→  КАНАЛ СВЯЗИ  ──→  ПРИЁМНИК
                      ↑
                     ШУМ
```

Шум — это всё, что мешает: треск в трубке, клякса в тетради, потерянный
кусочек данных в сети.

**Обработка.** Из того, что было, получили новое. Решил задачу — обработка.
Перевёл текст — тоже, только тут изменилась форма, а не содержание.

### Как отличить одно от другого

Правило, которое работает почти всегда:

* информация та же, лежит и ждёт → **хранение**;
* информация та же, но сменила место → **передача**;
* информация изменилась, появилась новая → **обработка**.

### Что делать с шумом

Спасает **избыточность**: важное повторяют по нескольку раз.

Допустим, сообщение «ВСТРЕЧА В ПЯТЬ» отправили трижды, и каждый раз
что-то потерялось:

```
  копия 1:   ВСТ?ЕЧА В ПЯТЬ
  копия 2:   ВСТРЕЧА В П?ТЬ
  копия 3:   ВСТРЕЧ? В ПЯТ?
  ───────────────────────────
  собрали:   ВСТРЕЧА В ПЯТЬ
```

Каждая копия покоцана, а вместе они дают целое сообщение: там, где в одной
дырка, в другой буква уцелела.

Ровно на этой идее держится защита данных на дисках и в сетях. Именно
поэтому скачанный файл иногда «проверяется» перед установкой.

## Смотрим, как это работает

### Одно действие — три процесса

> Ученик прочитал условие задачи, решил её и записал ответ в тетрадь.

Здесь всё сразу:

1. **передача** — условие переехало из учебника в голову;
2. **обработка** — из условия получилось решение;
3. **хранение** — ответ лёг в тетрадь.

Так почти всегда: бытовое действие раскладывается на несколько процессов.
Когда спрашивают про «главный» — имеют в виду тот, ради которого всё
затевалось. Здесь это обработка: задачу решали, а не переписывали.

### То же самое, но делает компьютер

Запусти ячейку. Три строки — три процесса.

````python
# хранение: число легло в память под именем «температура»
температура = 18

# обработка: из старого получили новое
в_фаренгейтах = температура * 9 / 5 + 32

# передача: результат уехал на экран, к человеку
print("Было", температура, "°C, стало", в_фаренгейтах, "°F")
````

Слово `температура` слева от знака равно — это **имя**, под которым
компьютер запомнил число. Такие имена называют переменными, и подробно
мы разберём их через несколько уроков.

Пока достаточно понимать: любая программа только и делает, что хранит,
обрабатывает и передаёт.
