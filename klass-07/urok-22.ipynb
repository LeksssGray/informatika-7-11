## Разбираемся

### Картинка из клеточек

Экран, на который ты сейчас смотришь, состоит из крошечных
светящихся точек. Одна такая точка называется **пиксель**
(от английского *picture element* — «элемент изображения»).

> **Растровое изображение** — прямоугольная сетка пикселей,
> у каждого из которых свой цвет.

Компьютеру для хранения такой картинки нужно запомнить цвет каждого
пикселя по порядку — как если бы ты диктовал по телефону: «первая
клетка красная, вторая красная, третья белая…».

Фотография с телефона — растровая. Скриншот — растровый. Почти всё,
что ты видел в интернете, — растровое.

### Разрешение

> **Разрешение** — размеры изображения в пикселях: ширина на высоту.

Запись `1920 × 1080` означает 1920 пикселей по горизонтали
и 1080 по вертикали. Общее число пикселей — это произведение:

$$K = ширина \cdot высота$$

1920 × 1080 = 2 073 600 пикселей, то есть больше двух миллионов
цветных точек на одном экране.

Чем больше разрешение, тем мельче детали — и тем больше файл.
Эти две вещи всегда идут вместе.

### Почему растр нельзя увеличить

Если растянуть фотографию вдвое, новым пикселям взять цвет неоткуда:
в файле записана ровно та сетка, что была. Программа честно
растягивает каждый пиксель в квадрат — и картинка рассыпается
на «лесенки» и квадратики.

Так в кино и появилась знаменитая неправда: «увеличьте вот этот
участок… ещё… а теперь улучшите». Улучшать нечего — информации
в файле не прибавилось.

### Вектор: картинка из описаний

Есть другой способ. Можно хранить не цвет каждой точки, а **описание
фигур**: «круг с центром здесь, радиусом такой-то, залит красным».

> **Векторное изображение** — набор геометрических фигур, заданных
> формулами и параметрами.

При увеличении фигуры пересчитываются заново, поэтому край круга
останется гладким при любом размере. Никаких квадратиков.

### Растр или вектор

| | Растр | Вектор |
|---|---|---|
| Что хранится | цвет каждого пикселя | описание фигур |
| Увеличение | появляются квадратики | качество не теряется |
| Фотография | подходит | не подходит |
| Логотип, схема, шрифт | плохо масштабируется | то, что нужно |
| Размер файла | зависит от разрешения | зависит от числа фигур |

Логотип школы, дорожный знак, буква любого шрифта — всё это вектор.
Фотография двора — растр. Ни один из способов не лучше другого,
у них разные задачи.

### Форматы и сжатие

Хранить цвет каждого пикселя «как есть» — расточительно. Поэтому
изображения **сжимают**, и делают это двумя разными способами.

**Сжатие без потерь.** Картинка восстанавливается точь-в-точь.
Работает на однотонных участках: вместо «сто белых пикселей подряд»
записывается «сто раз белый».

**Сжатие с потерями.** Часть мелких деталей выбрасывается насовсем —
зато файл в десятки раз меньше. Глаз подмены почти не замечает,
но исходное изображение уже не вернуть.

| Формат | Тип | Сжатие | Где применяют |
|---|---|---|---|
| BMP | растр | нет | почти не используется, файлы огромные |
| PNG | растр | без потерь | скриншоты, рисунки, прозрачность |
| JPEG | растр | с потерями | фотографии |
| GIF | растр | без потерь, 256 цветов | анимация |
| SVG | вектор | — | логотипы, иконки, схемы |

Правило на всю жизнь: **фотография — JPEG, скриншот и рисунок —
PNG, логотип — SVG**. Сохранишь скриншот в JPEG — получишь грязь
вокруг букв; сохранишь фотографию в PNG — файл вырастет в разы.

## Смотрим, как это работает

### Сколько точек на экране

```python
ширина = 1920
высота = 1080

пикселей = ширина * высота

print("Разрешение:", ширина, "×", высота)
print("Пикселей:", пикселей)
```

### Что даёт удвоение размера

Увеличим обе стороны вдвое и посмотрим, что станет с числом точек.

```python
ширина = 800
высота = 600

было = ширина * высота
стало = (ширина * 2) * (высота * 2)

print("Было пикселей:", было)
print("Стало пикселей:", стало)
print("Выросло в", стало // было, "раза")
```

Стороны выросли вдвое, а точек стало вчетверо больше — и файл
вырастет примерно так же. Это стоит помнить, когда отправляешь
фотографии: «чуть побольше» на деле означает «в четыре раза тяжелее».

### Экран телефона и экран кинотеатра

```python
print("HD:      ", 1280 * 720)
print("Full HD: ", 1920 * 1080)
print("4K:      ", 3840 * 2160)
```
