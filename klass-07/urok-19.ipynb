## Разбираемся

### Буква — это число

Компьютер не знает букв. В памяти лежат только числа, а точнее —
нули и единицы. Значит, каждой букве нужно **назначить номер**
и договориться, какой именно. Такая договорённость называется
**кодовой таблицей**.

> **Кодировка** — таблица, которая каждому символу сопоставляет число.

Договорённость должна быть общей. Если ты записал текст в одной
таблице, а открыл в другой, вместо слов получится знаменитая
«кракозябра»: `Ïðèâåò`. Это не поломка файла — это чтение чисел
по чужой таблице.

### ASCII: первая общая таблица

В 1963 году в США приняли таблицу **ASCII** — 128 символов:
латинские буквы, цифры, знаки препинания и служебные команды.

Сколько бит нужно на 128 вариантов? По знакомой формуле $N = 2^i$:
$128 = 2^7$, то есть **7 бит**. На практике выделили целый байт —
8 бит, и остались 128 свободных мест.

Несколько кодов, которые полезно узнавать:

| Символ | Код |
|---|---|
| пробел | 32 |
| `0` | 48 |
| `A` | 65 |
| `a` | 97 |

Из таблицы видно два удобных правила. Цифры идут подряд с 48,
буквы — подряд с 65 и с 97. Поэтому код буквы `C` — это 65 + 2,
а разница между большой и маленькой буквой всегда ровно **32**.

### Куда делась кириллица

В свободные 128 мест каждая страна поселила свой алфавит — и в этом
была беда. Русских кодировок набралось больше пяти: КОИ-8, CP1251,
CP866, ISO-8859-5. Один и тот же текст выглядел по-разному в разных
программах, а в одном файле нельзя было держать русский и греческий.

### Unicode: таблица на все языки сразу

В 1991 году придумали **Unicode** — единую таблицу для всех
письменностей мира: латиница, кириллица, иероглифы, арабская вязь,
ноты, математические знаки и эмодзи. Сейчас в ней больше 150 тысяч
символов, и место ещё осталось.

Кириллица в Unicode начинается с 1040: `А` — 1040, `Б` — 1041
и дальше подряд.

**UTF-8** — самый распространённый способ записать Unicode в байты.
Он экономный: латинская буква занимает 1 байт (и совпадает с ASCII),
кириллическая — 2 байта, иероглиф — 3 байта. Именно поэтому в наших
файлах написано `encoding="utf-8"`.

### Объём текста

Формула та же, что в четвёртом уроке:

$$I = K \cdot i$$

где $K$ — количество символов, $i$ — сколько занимает один символ.
Вся хитрость — не перепутать биты с байтами:

| Кодировка | Один символ |
|---|---|
| ASCII, однобайтовая | 1 байт = 8 бит |
| UTF-8, кириллица | 2 байта = 16 бит |
| UTF-16 | 2 байта = 16 бит |

### Две функции Python

* `ord("A")` — узнать код символа. Вернёт 65.
* `chr(65)` — узнать символ по коду. Вернёт `"A"`.

Название `ord` — от английского *ordinal*, «порядковый номер»,
а `chr` — от *character*, «символ». Они обратны друг другу:
`chr(ord("A"))` снова даст `"A"`.

## Смотрим, как это работает

### Код символа и символ по коду

```python
print("Код буквы A:", ord("A"))
print("Код буквы a:", ord("a"))
print("Код цифры 0:", ord("0"))
print("Код пробела:", ord(" "))

print("Символ с кодом 66:", chr(66))
print("Символ с кодом 100:", chr(100))
```

### Кириллица начинается с 1040

```python
print("Код буквы А:", ord("А"))
print("Код буквы Я:", ord("Я"))
print("Символ с кодом 1041:", chr(1041))
```

Внимание: русская `А` и латинская `A` выглядят одинаково, а коды
у них разные — 1040 и 65. Компьютер считает их **разными буквами**.
Отсюда половина загадочных ошибок в паролях и именах файлов.

### Печатаем алфавит циклом

```python
for i in range(26):
    print(chr(65 + i), end=" ")

print()
```

Первая буква — код 65, дальше просто прибавляем номер шага.
Параметр `end=" "` заставляет `print` ставить пробел вместо перевода
строки, поэтому алфавит печатается в одну строчку.

### Из большой буквы в маленькую

```python
буква = "G"
маленькая = chr(ord(буква) + 32)

print(буква, "→", маленькая)
```

### Шифр Цезаря на одном символе

Древнеримский шифр: каждую букву заменяют на стоящую в алфавите
через три позиции. Так Цезарь писал письма полководцам.

```python
буква = "d"
сдвиг = 3
зашифрованная = chr(ord(буква) + сдвиг)

print(буква, "→", зашифрованная)
```

### Сколько весит сообщение

```python
символов = 500
бит_на_символ = 8

объём_бит = символов * бит_на_символ
объём_байт = объём_бит // 8

print("Бит:", объём_бит)
print("Байт:", объём_байт)
print("Килобайт:", объём_байт / 1024)
```
