## Разбираемся

### print умеет больше, чем кажется

Мы пользовались этой командой на четверть возможностей. Разберём целиком.

**Одно значение.** Самый простой случай:

```python
print("Привет")
```

**Несколько через запятую.** Компьютер напечатает подряд, сам разделив
пробелами:

```python
print("Мне", 13, "лет")     →  Мне 13 лет
```

**Свой разделитель.** Пробел между кусками можно заменить:

```python
print("2024", "09", "01", sep="-")     →  2024-09-01
print("а", "б", "в", sep="")           →  абв
```

**Без перевода строки.** Обычно после печати курсор уходит на новую
строку. Это тоже меняется:

```python
print("Загрузка", end="...")
print("готово")                        →  Загрузка...готово
```

### Строки можно складывать и умножать

Звучит странно, но работает:

```python
"при" + "вет"      →  "привет"        склеивание
"-" * 20           →  "--------------------"   повтор
```

Плюс склеивает две строки в одну, звёздочка повторяет строку заданное
число раз. Из этих двух действий собирается почти любая картинка
из символов.

> Складывать можно только строку со строкой. `"возраст: " + 13` —
> ошибка. Число надо сначала превратить в строку: `str(13)`.

### Кавычки

Годятся и одинарные, и двойные — разницы нет:

```python
print("привет")
print('привет')
```

Разница появляется, когда кавычка нужна **внутри** текста. Тогда берут
кавычки другого вида:

```python
print("Он сказал: 'привет'")
print('Число "пи" равно 3.14')
```

### Спецсимволы

Некоторые вещи просто так не напечатаешь — их записывают через
обратный слеш:

| Запись | Что означает |
|---|---|
| `\n` | перевод строки |
| `\t` | широкий отступ |
| `\\` | сам обратный слеш |

Строка `"а\nб"` — это две строки: `а` и `б`.

### Комментарии

Всё после знака `#` компьютер пропускает. Это заметки для человека:

```python
# считаем площадь комнаты
площадь = 5 * 3      # длина на ширину
```

Комментарий пишут не про то, **что** делает строка (это и так видно),
а про то, **зачем**. `# складываем a и b` бесполезен. `# ищем среднюю
температуру за неделю` — полезен.

### Как читать ошибку

Смотри на **последнюю строку** красного текста:

| Что написано | Что случилось |
|---|---|
| `SyntaxError` | опечатка: скобка, кавычка, двоеточие |
| `NameError` | такого имени не существует |
| `TypeError` | смешал несовместимое, например текст и число |

## Смотрим, как это работает

### sep и end

Запусти и сравни, что напечатали разные команды.

````python
print("раз", "два", "три")
print("раз", "два", "три", sep=" | ")
print("раз", end=" → ")
print("два", end=" → ")
print("три")
````

Последние три команды напечатали всё в одну строку — потому что мы
заменили перевод строки на стрелочку. Только у последней `print`
осталось поведение по умолчанию, и она строку завершила.

### Рамка из символов

````python
верх = "+" + "-" * 20 + "+"
пусто = "|" + " " * 20 + "|"
текст = "|" + "  Привет, 7 класс!  " + "|"

print(верх)
print(пусто)
print(текст)
print(пусто)
print(верх)
````

Тут работают обе операции сразу: `"-" * 20` даёт двадцать дефисов,
а `+` склеивает куски. Текст внутри рамки — ровно двадцать символов,
иначе рамку перекосит. Попробуй изменить надпись и посмотри,
что произойдёт.

### Многострочный текст

````python
стих = "Программа — это текст,\nкоторый понимает\nи человек, и компьютер."

print(стих)
````

Один `\n` — один перевод строки. В переменной лежит одна длинная строка,
а на экране получилось три.
