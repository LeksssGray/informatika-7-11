## Часть 1. Теория — 15 минут

### Команда print во всех подробностях

С `print` мы уже знакомы, но использовали её на четверть возможностей.
Разберём команду целиком.

**Одно значение.** Простейший случай:

```python
print("Привет")
```

**Несколько значений через запятую.** Python напечатает их подряд,
автоматически разделив пробелом:

```python
print("Мне", 13, "лет")     →  Мне 13 лет
```

**Разделитель `sep`.** Пробел между значениями можно заменить на любой
другой текст:

```python
print("2024", "09", "01", sep="-")     →  2024-09-01
print("а", "б", "в", sep="")           →  абв
```

**Окончание `end`.** По умолчанию `print` переводит строку в конце.
Это поведение тоже меняется:

```python
print("Загрузка", end="...")
print("готово")                        →  Загрузка...готово
```

### Строки и кавычки

Текст можно заключать в одинарные или двойные кавычки — Python не делает
между ними разницы:

```python
print("привет")
print('привет')
```

Разница появляется, когда кавычка нужна **внутри** текста. Тогда берут
кавычки другого вида:

```python
print("Он сказал: 'привет'")
print('Число "пи" равно 3.14')
```

Если внутри нужны оба вида кавычек, используют тройные:

```python
print("""Он сказал: "привет" — и ушёл""")
```

Тройные кавычки умеют ещё кое-что: текст в них может занимать несколько
строк, и все переводы строк сохранятся.

### Специальные символы

Некоторые символы нельзя просто напечатать — их записывают обратным слешем:

| Запись | Что означает |
|---|---|
| `\n` | перевод строки |
| `\t` | табуляция (широкий отступ) |
| `\\` | сам обратный слеш |
| `\"` | кавычка внутри двойных кавычек |

### Комментарии

Всё, что идёт после знака `#`, Python игнорирует. Это заметки для человека:

```python
# Считаем площадь прямоугольника
площадь = 5 * 3      # длина умножить на ширину
```

Комментарии — не украшение. Через неделю вы не вспомните, почему написали
именно так, а комментарий вспомнит за вас. Правило хорошего тона: писать
в комментарии **зачем**, а не **что**. Строка `# складываем a и b` бесполезна
— это и так видно. Строка `# ищем среднюю температуру за неделю` полезна.

### Как читать ошибку

Ошибки будут всегда — важно уметь их читать. Смотрите на **последнюю строку**
красного сообщения, там суть:

| Сообщение | Что случилось |
|---|---|
| `SyntaxError` | опечатка в записи: скобка, кавычка, двоеточие |
| `NameError` | использовали имя, которого не существует |
| `TypeError` | смешали несовместимые типы, например текст и число |
| `IndentationError` | сбились отступы в начале строки |

## Часть 2. Примеры с разбором — 10 минут

### Пример 1. sep и end в работе

Запустите и сравните вывод трёх команд.

````python
print("раз", "два", "три")
print("раз", "два", "три", sep=" | ")
print("раз", end=" → ")
print("два", end=" → ")
print("три")
````

Третья, четвёртая и пятая команды напечатали всё в **одну строку**, потому
что мы заменили перевод строки на стрелочку. Только последний `print`
сработал по умолчанию и завершил строку.

### Пример 2. Рисуем рамку

Соберём небольшой рисунок, пользуясь `\n` и повторением строк.

````python
print("+" + "-" * 20 + "+")
print("|" + " " * 20 + "|")
print("|" + "  Привет, 7 класс!  " + "|")
print("|" + " " * 20 + "|")
print("+" + "-" * 20 + "+")
````

Здесь работает интересная возможность: строку можно **умножить** на число,
и она повторится. `"-" * 20` даёт двадцать дефисов, а `+` склеивает куски
в одну строку. Обратите внимание, что текст внутри рамки — ровно 20 символов,
иначе рамка перекосится.

### Пример 3. Многострочный текст

Тройные кавычки удобны для стихов, писем и любого текста, который должен
сохранить форму.

````python
print("""
    Программа — это текст,
    который понимает
    и человек, и компьютер.
""")

print("А это то же самое,\nно через \\n — короче, но хуже читается.")
````

Во второй команде обратите внимание на `\\n`: двойной слеш нужен, чтобы
напечатать сам символ `\n`, а не выполнить перевод строки.
