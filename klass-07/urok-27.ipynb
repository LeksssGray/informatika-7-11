## Повторяем

### 1. Цикл while

`while` повторяет тело, **пока условие истинно**. Число повторов
заранее неизвестно — цикл сам решает, когда остановиться.

```python
n = 100
while n > 1:
    n = n // 2
```

Проверь, что условие когда-нибудь станет ложным, иначе цикл не кончится.

### 2. Кодирование текста

Каждый символ хранится числом — **кодом**. `ord("D")` даёт код (68),
`chr(68)` — символ по коду. Буквы алфавита идут подряд: код `E` на один
больше, чем у `D`. В однобайтовой кодировке символ занимает 8 бит,
в Unicode UTF-16 — 16.

### 3. Строки: индексы и срезы

Индексы с нуля, `-1` — последний символ. Срез `слово[2:5]` берёт
индексы 2, 3, 4: правая граница не входит, как в `range`.
`слово[:3]` — первые три, `слово[-3:]` — последние три.

### 4. Строки: длина и методы

`len(s)` — длина, пробелы и знаки тоже считаются.
`s.count("а")` — сколько раз встречается, `s.replace("а", "о")` —
замена **всех** вхождений, `s.upper()` — заглавные.

### 5. Растровая и векторная графика

**Растр** — сетка пикселей (JPEG, PNG, BMP): подходит для фотографий,
при увеличении видны квадратики. **Вектор** — описание фигур (SVG):
линии остаются гладкими при любом масштабе.

### 6. Кодирование цвета

Глубина цвета `i` бит даёт `2 ** i` цветов. Объём картинки =
ширина × высота × глубина (в битах), делим на 8 — получаем байты.
RGB: три числа от 0 до 255 — красный, зелёный, синий.

### 7. Списки

`a[0]` — первый элемент, `a[-1]` — последний, `a.append(x)` — добавить
в конец, `len(a)` — сколько элементов.

### 8. Обработка списков

`sum`, `max`, `min` — готовые функции. Счётчик по условию — цикл
с `if` и переменная, которая растёт на 1.
