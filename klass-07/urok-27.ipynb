`слово[1:3]` даёт символы с индексами 1 и 2, а не 3. Правая граница не входит —
точно так же, как в `range`. Отрицательный индекс считает с конца.

```python
слово = "информатика"
print(слово[1:3])
print(слово[-1])
print(len(слово))
```
