## Разбираемся

### Зачем функции

Представь, что одну и ту же операцию нужно делать в десяти местах программы:
посчитать квадрат, поздороваться, проверить знак. Копировать код десять раз
неудобно: нашёл ошибку — исправляй все десять. **Функция** — это имя для куска кода.
Написали один раз, вызываем сколько нужно.

### Определение и вызов

`def` объявляет функцию. Имя `x` в скобках — **параметр**, место для значения.
Строки с отступом после `def` — тело функции. `квадрат(4)` — **вызов**: параметр
получает 4, и функция возвращает 16.

```python
def квадрат(x):
    return x * x

print(квадрат(4))
print(квадрат(7))
```

### return отдаёт результат

Функция возвращает значение, а печатает его тот, кто вызвал. Без `return` результата
нет: функция вернёт `None`, то есть «ничего». Поэтому `return` обязателен, если
функция должна что-то выдать.

### Локальные имена

Имя, которое появилось внутри функции, живёт только внутри неё. Переменная `x`
в функции и `x` в основной программе — разные переменные.

```python
x = 5

def поменять():
    x = 10
    print("внутри функции:", x)

поменять()
print("снаружи:", x)
```
