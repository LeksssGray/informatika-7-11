## Разбираемся

### Зачем нужны переменные

Пока наши программы считали и тут же забывали. Но обычно результат нужен
дальше: посчитал площадь — сравнил, сложил — поделил. Чтобы значение
не потерялось, его кладут в **переменную**.

Представь подписанную коробку. На коробке написано `возраст`, внутри
лежит 13. Нужно значение — заглядываешь в коробку по имени.

### Знак равно означает не то, что в математике

```python
возраст = 13
имя = "Аня"
рост = 1.62
```

Здесь `=` читается не «равно», а **«положи»**. Справа вычисляется
значение, слева стоит имя коробки, куда его положить.

Поэтому вот такая строка — не бессмыслица:

```python
счётчик = счётчик + 1
```

Она означает: возьми что лежит, прибавь единицу, положи обратно.
В математике такое равенство невозможно, в программе — обычное дело.

### Как называть

Правила короткие:

* буквы, цифры и подчёркивание;
* **не начинается с цифры**: `2сумма` — нельзя, `сумма2` — можно;
* без пробелов: пишут `средний_балл`, а не `средний балл`;
* `Сумма` и `сумма` — разные переменные;
* нельзя занимать служебные слова: `if`, `for`, `print`.

И главное правило, которого компьютер не требует, а здравый смысл
требует: **имя должно объяснять смысл**. Через неделю `x1` и `aaa`
не скажут тебе ничего, а `средний_балл` скажет.

### У значений есть тип

Тип определяет, что со значением можно делать.

| Тип | Как называется | Пример | Что это |
|---|---|---|---|
| `int` | целое | `13`, `-5`, `0` | числа без дробной части |
| `float` | вещественное | `1.62`, `-0.5` | числа с дробной частью |
| `str` | строка | `"Аня"`, `"7А"` | текст в кавычках |
| `bool` | логический | `True`, `False` | правда или неправда |

Узнать тип можно командой `type`.

> Дробная часть отделяется **точкой**, а не запятой. `1.62` — одно
> число, `1,62` — два разных значения через запятую.

### Тип решает, что произойдёт

Один и тот же плюс работает по-разному:

```python
2 + 3        →  5          сложили числа
"2" + "3"    →  "23"       склеили строки
2 + "3"      →  ошибка     так нельзя
```

Компьютер не догадывается, что ты имел в виду. Если нужно превратить
одно в другое, это делают явно:

```python
int("42")      →  42        строка стала числом
str(42)        →  "42"      число стало строкой
float("3.14")  →  3.14
```

### Как поменять значения местами

Классическая задачка. В `а` лежит 1, в `б` — 2, надо поменять.
Прямолинейно не выйдет:

```python
а = б      # теперь в обеих двойка, единица потеряна навсегда
б = а
```

Нужна третья коробка:

```python
временно = а
а = б
б = временно
```

В Python есть и короткая запись для того же самого, но она опирается
на вещи, которых мы ещё не проходили. Вернёмся к ней позже —
пока временная переменная работает ничем не хуже.

## Смотрим, как это работает

### Переменные и их типы

````python
имя = "Аня"
возраст = 13
рост = 1.62
отличница = True

print(имя, "— это", type(имя))
print(возраст, "— это", type(возраст))
print(рост, "— это", type(рост))
print(отличница, "— это", type(отличница))
````

Запись `<class 'int'>` читается просто как «тип int». Слово «class»
тут ничего особенного не значит, привыкать к нему не обязательно.

### Значение меняется

````python
счёт = 0
print("В начале:", счёт)

счёт = счёт + 10
print("Забили гол:", счёт)

счёт = счёт + 5
print("И ещё один:", счёт)

счёт = "много"
print("А теперь:", счёт, type(счёт))
````

Последняя строка показывает особенность Python: в переменную можно
положить что угодно, даже другого типа. Удобно, но и опасно —
легко запутаться. Без нужды тип лучше не менять.

### Обмен значениями

````python
а = "первое"
б = "второе"
print("До:   а =", а, ", б =", б)

временно = а
а = б
б = временно

print("После: а =", а, ", б =", б)
````

Третья коробка нужна, чтобы первое значение не потерялось: как только
мы напишем `а = б`, старое содержимое `а` исчезнет навсегда.
Поэтому его сначала откладывают в сторону.
