## Часть 1. Теория — 15 минут

### Программа, которая спрашивает

До сих пор все данные мы вписывали прямо в код. Такая программа умеет
ровно одно: посчитать площадь именно того прямоугольника, который мы
заложили. Чтобы она работала с любыми данными, их нужно **вводить**.

За это отвечает функция `input`:

```python
имя = input("Как тебя зовут? ")
print("Привет,", имя)
```

Программа останавливается, показывает вопрос, ждёт, пока человек напечатает
ответ и нажмёт Enter, и кладёт введённое в переменную.

### Главная ловушка ввода

> **`input` всегда возвращает строку.** Даже если человек ввёл число.

Это причина примерно половины ошибок у новичков:

```python
возраст = input("Возраст: ")     # человек ввёл 13
print(возраст + 1)               # TypeError: строка плюс число
```

Введённое `13` — это текст `"13"`, а не число. Чтобы получить число,
преобразование делают явно:

```python
возраст = int(input("Возраст: "))       # целое
рост = float(input("Рост в метрах: "))  # дробное
```

Обратите внимание на порядок: сначала `input` получает строку, потом `int`
превращает её в число. Читать такую запись удобно изнутри наружу.

### Арифметические операции

| Операция | Знак | Пример | Результат |
|---|---|---|---|
| сложение | `+` | `7 + 3` | `10` |
| вычитание | `-` | `7 - 3` | `4` |
| умножение | `*` | `7 * 3` | `21` |
| деление | `/` | `7 / 3` | `2.333...` |
| деление нацело | `//` | `7 // 3` | `2` |
| остаток | `%` | `7 % 3` | `1` |
| степень | `**` | `7 ** 3` | `343` |

Две операции заслуживают отдельного внимания.

**Деление нацело `//`** отбрасывает дробную часть: сколько целых раз одно
помещается в другом. 100 минут — это `100 // 60 = 1` полный час.

**Остаток `%`** — сколько осталось после того, как целые части забрали.
100 минут — это 1 час и `100 % 60 = 40` минут.

Вместе эта пара решает почти все задачи «переведи одно в другое»:
минуты в часы, копейки в рубли, дни в недели.

### Порядок действий

Такой же, как в математике: сначала степень, потом умножение и деление,
потом сложение и вычитание. Скобки меняют порядок:

```python
2 + 3 * 4        →  14
(2 + 3) * 4      →  20
```

Если сомневаетесь — ставьте скобки. Лишняя скобка ничего не сломает,
а недостающая испортит результат.

### f-строки: удобный вывод

Собирать текст из кусочков через `+` неудобно и легко ошибиться с типами.
Гораздо лучше **f-строка**: буква `f` перед кавычками разрешает вставлять
значения прямо в текст, в фигурных скобках.

```python
имя = "Аня"
лет = 13
print(f"{имя} учится в 7 классе, ей {лет} лет")
```

Внутри фигурных скобок можно писать и выражения:

```python
print(f"Через два года будет {лет + 2}")
```

А ещё — задавать формат вывода. Запись `:.2f` означает «вещественное
с двумя знаками после точки»:

```python
print(f"Средний балл: {4.6667:.2f}")     →  Средний балл: 4.67
```

## Часть 2. Примеры с разбором — 10 минут

### Пример 1. Ввод с преобразованием

В ноутбуке `input` показывает поле для ввода прямо под ячейкой.
Запустите и введите любое число.

Если ввести не число, а слово, программа упадёт с ошибкой
`ValueError: invalid literal for int()`. Это нормально: `int` честно
сообщает, что из слова «пять» число сделать не может.

### Пример 2. Деление нацело и остаток в связке

Переведём минуты в часы и минуты — классическая задача на эту пару операций.

````python
всего_минут = 250

часы = всего_минут // 60
минуты = всего_минут % 60

print(f"{всего_минут} минут — это {часы} ч {минуты} мин")
````

Проверим себя: 4 часа по 60 минут — это 240, плюс 10 минут, итого 250.
Сходится.

Тот же приём работает для любых единиц: секунды в минуты, копейки в рубли,
сантиметры в метры. Меняется только делитель.

### Пример 3. Расчёт сдачи

Задача: покупка стоит 347 рублей, покупатель дал 500. Какими купюрами
выдать сдачу, если есть номиналы 100, 50, 10 и 1?

````python
цена = 347
дали = 500
сдача = дали - цена

print(f"Сдача: {сдача} руб.")

сотни = сдача // 100
сдача = сдача % 100

полтинники = сдача // 50
сдача = сдача % 50

десятки = сдача // 10
рубли = сдача % 10

print(f"  по 100 руб. — {сотни} шт.")
print(f"  по 50 руб.  — {полтинники} шт.")
print(f"  по 10 руб.  — {десятки} шт.")
print(f"  по 1 руб.   — {рубли} шт.")
````

Схема одна и та же на каждом шаге: делением нацело узнаём, сколько купюр
этого номинала помещается, остатком — сколько денег ещё не выдано.
Дальше повторяем для следующего номинала.
