## Часть 1. Теория — 15 минут

### Зачем нужны переменные

Пока наши программы всё вычисляли на месте и тут же забывали.
Но обычно результат нужен дальше: посчитали площадь — потом сравнили,
сложили — потом разделили. Чтобы значение не потерялось, его **сохраняют
в переменную**.

> **Переменная** — это имя, за которым закреплено значение.

Представьте подписанную коробку: на коробке написано `возраст`,
внутри лежит число 13. Когда нужно значение — заглядываем в коробку по имени.

### Присваивание

Значение кладут в переменную знаком `=`:

```python
возраст = 13
имя = "Аня"
рост = 1.62
```

Знак `=` здесь читается не «равно», а **«присвоить»**. Это важно: справа
вычисляется значение, слева стоит имя, куда его положить. Поэтому запись

```python
счётчик = счётчик + 1
```

не бессмыслица, а совершенно нормальная строка: взять текущее значение,
прибавить единицу и положить обратно под тем же именем.

### Правила именования

Имя переменной можно придумать почти любое, но есть правила:

* состоит из букв, цифр и знака подчёркивания;
* **не начинается с цифры**: `2сумма` — ошибка, `сумма2` — нормально;
* нет пробелов: пишут `средний_балл`, а не `средний балл`;
* различается регистр: `Сумма` и `сумма` — две разные переменные;
* нельзя использовать служебные слова языка: `if`, `for`, `print` и другие.

И главное, чего не требует Python, но требует здравый смысл: **имя должно
объяснять смысл**. `x1`, `aaa` и `dannye` через месяц не скажут вам ничего,
а `средний_балл` скажет.

### Типы данных

У каждого значения есть **тип** — он определяет, что с этим значением
можно делать.

| Тип | Название | Пример | Что это |
|---|---|---|---|
| `int` | целое | `13`, `-5`, `0` | числа без дробной части |
| `float` | вещественное | `1.62`, `-0.5` | числа с дробной частью |
| `str` | строка | `"Аня"`, `"7А"` | текст в кавычках |
| `bool` | логический | `True`, `False` | истина или ложь |

Узнать тип значения можно функцией `type`.

> ⚠️ Дробная часть отделяется **точкой**, а не запятой: `1.62`, не `1,62`.
> Запятая разделяет разные значения, поэтому `print(1,62)` напечатает
> два числа, а не одно дробное.

### Тип определяет поведение

Один и тот же знак `+` работает по-разному в зависимости от типов:

```python
2 + 3        →  5          сложение чисел
"2" + "3"    →  "23"       склеивание строк
2 + "3"      →  TypeError  так нельзя
```

Python не догадывается, что вы имели в виду — он требует определённости.
Если нужно из строки получить число, преобразование делают явно:

```python
int("42")      →  42
float("3.14")  →  3.14
str(42)        →  "42"
```

### Обмен значений

Классическая задача: в переменной `а` лежит 1, в `б` — 2, надо поменять
местами. Прямолинейная попытка не работает:

```python
а = б      # теперь в обеих переменных 2 — первое значение потеряно!
б = а
```

Решение — временная переменная:

```python
врем = а
а = б
б = врем
```

В Python есть и короткий способ, который работает правильно:

```python
а, б = б, а
```

## Часть 2. Примеры с разбором — 10 минут

### Пример 1. Переменные и типы

````python
имя = "Аня"
возраст = 13
рост = 1.62
отличница = True

print(имя, "— тип", type(имя))
print(возраст, "— тип", type(возраст))
print(рост, "— тип", type(рост))
print(отличница, "— тип", type(отличница))
````

Запись `<class 'int'>` читается как «класс int», то есть целое число.
Слово «класс» здесь просто синоним слова «тип», привыкать к нему
не обязательно.

### Пример 2. Изменение значения

Проследим, как меняется содержимое переменной.

````python
счёт = 0
print("В начале:", счёт)

счёт = счёт + 10
print("После первого гола:", счёт)

счёт += 5          # краткая запись того же самого
print("После второго:", счёт)

счёт = "много"     # тип можно поменять на ходу
print("А теперь:", счёт, type(счёт))
````

Запись `счёт += 5` — сокращение для `счёт = счёт + 5`. Так же работают
`-=`, `*=`, `/=`.

Последняя строка показывает особенность Python: переменная не привязана
к типу навсегда, в неё можно положить что угодно. Это удобно, но и опасно —
легко запутаться, поэтому без нужды тип менять не стоит.

### Пример 3. Обмен значений

````python
а = "первое"
б = "второе"
print("До обмена:  а =", а, ", б =", б)

а, б = б, а

print("После обмена: а =", а, ", б =", б)
````

Python сначала вычисляет всё, что справа от `=`, и только потом
раскладывает результаты по именам слева. Поэтому оба старых значения
успевают сохраниться, и обмен проходит без потерь.
