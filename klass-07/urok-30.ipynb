## Разбираемся

### Одно и то же в таблице и в коде

У таблицы и у Python есть готовые функции для типовых расчётов. Соответствие простое:

| Что нужно | Python | Электронная таблица |
|---|---|---|
| сумма | `sum(список)` | `=СУММ(A1:A5)` |
| среднее | `sum(список) / len(список)` | `=СРЗНАЧ(A1:A5)` |
| наибольшее | `max(список)` | `=МАКС(A1:A5)` |
| наименьшее | `min(список)` | `=МИН(A1:A5)` |
| сколько чисел | `len(список)` | `=СЧЁТ(A1:A5)` |
| выбор по условию | `if … else` | `=ЕСЛИ(A1>3; "да"; "нет")` |

````python
оценки = [5, 4, 3, 5, 4]
print("Сумма:", sum(оценки))
print("Среднее:", sum(оценки) / len(оценки))
print("Максимум:", max(оценки))
print("Минимум:", min(оценки))
print("Количество:", len(оценки))
````

### Какую диаграмму выбрать

* **Круговая** — показывает долю каждой части в целом: расходы по статьям.
* **Столбчатая** — сравнивает величины друг с другом: выручка магазинов за месяц.
* **Линейная** — показывает изменение во времени: температура по дням.

Диаграмма не должна ничего скрывать. Если ось начинается не с нуля, это тоже нужно
заметить, иначе маленькая разница будет выглядеть огромной.
