### Магистраль

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g07/magistral.png" width="520" alt="Магистральная схема: процессор, память и устройства ввода-вывода соединены шинами">

*Магистральная схема: процессор, память и устройства ввода-вывода соединены шинами*

<sub>Д. Ильин · CC0 · Wikimedia Commons</sub>

Все части компьютера соединены общими линиями связи — **шинами**.
Вместе они называются **магистралью**. Это как улица, вдоль которой
стоят дома: по ней ездят все, а не проложены отдельные дорожки от каждого
дома к каждому.

| Шина | Что по ней передаётся |
|---|---|
| шина данных | сами данные: числа, буквы, кусочки картинок |
| шина адреса | куда их положить или откуда взять — номер ячейки памяти |
| шина управления | команды: «читай», «пиши», «жди» |

Благодаря магистрали новое устройство легко подключить: воткнул его
в разъём, и оно уже на общей «улице».

### Процессор

Процессор выполняет команды программы по одной, в очень быстром ритме.
Один удар этого ритма называется **тактом**.

* **Тактовая частота** — сколько тактов в секунду. Её меряют
  в гигагерцах: 1 ГГц — миллиард тактов в секунду. У обычных
  процессоров 2–5 ГГц.
* **Ядра** — сколько процессоров упаковано в один. У современных их
  от 4 до 16 и больше. Разные ядра выполняют разные задачи одновременно.

### Память

| Память | Скорость | После выключения | Где |
|---|---|---|---|
| **оперативная** (ОЗУ) | очень быстрая | стирается | планки на плате |
| **постоянная** (ПЗУ) | быстрая | сохраняется, но изменить трудно | чип на плате с программой запуска |
| **долговременная** | медленнее | сохраняется | жёсткий диск, SSD, флешка |

Когда ты запускаешь игру, она копируется с диска в оперативную память:
с ней процессор работает в сотни раз быстрее.

### Устройства ввода и вывода

| Ввода: информация идёт в компьютер | Вывода: информация идёт из компьютера |
|---|---|
| клавиатура, мышь, тачпад | монитор |
| микрофон | колонки, наушники |
| камера, сканер | принтер |
| джойстик, датчики | проектор |

Бывают и такие, что делают и то, и другое: сенсорный экран, гарнитура
с микрофоном, флешка.

### Как это было: первая мышь и первые окна

<img src="https://raw.githubusercontent.com/LeksssGray/informatika-7-11/main/img/g07/alto.jpg" width="320" alt="Xerox Alto, 1973 год: экран, клавиатура и мышь">

*Xerox Alto, 1973 год: экран, клавиатура и мышь*

<sub>Carlo Nardone · CC BY-SA 2.0 · Wikimedia Commons</sub>

Компьютерную мышь придумал американский инженер Дуглас Энгельбарт
в 1960-х: деревянная коробочка с двумя колёсиками. А в 1973 году
в лаборатории Xerox построили компьютер Alto: у него был экран с окнами,
значками и мышью — почти как у тебя сейчас. Продавать его не стали,
но идеи подсмотрели создатели Macintosh и Windows.

### Как читать характеристики

```
Процессор Intel Core i5, 6 ядер, 3,2 ГГц
ОЗУ 16 ГБ
SSD 512 ГБ
Видеокарта NVIDIA GeForce RTX 3050
```

| Строчка | Что значит | На что влияет |
|---|---|---|
| 6 ядер, 3,2 ГГц | процессор средней силы | общая скорость |
| ОЗУ 16 ГБ | рабочий стол на 16 гигабайт | сколько программ открыто сразу |
| SSD 512 ГБ | быстрый диск | скорость включения, место под файлы |
| GeForce RTX 3050 | отдельная видеокарта | игры, монтаж видео |

Под задачу подбирают не «самое мощное», а нужное: для учёбы
и браузера отдельная видеокарта не нужна, а для игр без неё никак.

## Смотрим, как это работает

### Сколько тактов в секунду

````python
частота_ггц = 3.2
тактов = частота_ггц * 1000000000

print(f"Процессор {частота_ггц} ГГц делает {тактов:.0f} тактов в секунду")
print(f"За одну тысячную секунды: {тактов / 1000:.0f} тактов")
````

Три миллиарда двести миллионов ударов в секунду. За время, пока ты
моргаешь, процессор успевает сделать около миллиарда тактов.

Запись `{тактов:.0f}` в f-строке печатает число без знаков после запятой.
