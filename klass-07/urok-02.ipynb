## Справочник

### Степени двойки

Главные числа информатики. Каждое следующее вдвое больше предыдущего.

| Степень | 2¹ | 2² | 2³ | 2⁴ | 2⁵ | 2⁶ | 2⁷ | 2⁸ | 2¹⁰ |
|---|---|---|---|---|---|---|---|---|---|
| Значение | 2 | 4 | 8 | 16 | 32 | 64 | 128 | 256 | 1024 |

Два числа отсюда встретятся особенно часто: **256** и **1024**.
Первое — сколько разных значений помещается в один байт, второе —
сколько байт в килобайте.

В коде степень пишется двумя звёздочками: `2 ** 10` это 1024.

### Проценты

Процент — это сотая доля. Чтобы найти процент от числа, умножают
на десятичную дробь:

| Проценты | Умножить на |
|---|---|
| 1 % | 0.01 |
| 10 % | 0.1 |
| 25 % | 0.25 |
| 50 % | 0.5 |
| 85 % | 0.85 |

25 % от 80 это `80 * 0.25 = 20`.

### Порядок действий

Как в математике: сначала степень, потом умножение и деление,
в конце сложение и вычитание. Скобки главнее всего.

```
  2 + 3 * 4       =  14      сначала умножили
  (2 + 3) * 4     =  20      скобки поменяли порядок
  2 + 3 * 2 ** 2  =  14      степень, потом умножение, потом плюс
```

### Деление нацело и остаток

Два действия, которых нет в обычной математике, но которые в информатике
используются постоянно.

| Запись | Что значит | Пример |
|---|---|---|
| `//` | сколько целых раз поместилось | `17 // 5` = 3 |
| `%` | сколько осталось | `17 % 5` = 2 |

17 конфет на пятерых: каждому по 3, две останутся.

### Как читать таблицу

В информатике много таблиц. Правило одно: сначала находишь **строку**,
потом **столбец**, на пересечении — ответ.

|  | утро | день | вечер |
|---|---|---|---|
| **понедельник** | 5 | 12 | 8 |
| **вторник** | 3 | 15 | 6 |

На пересечении «вторник» и «день» стоит 15.

### Округление

`round(число)` округляет до целого, `round(число, 2)` — до двух знаков
после точки.

```
  round(4.6)        →  5
  round(4.6667, 2)  →  4.67
```

## Смотрим, как это работает

### Степени двойки

````python
print("2 в 5 степени:", 2 ** 5)
print("2 в 8 степени:", 2 ** 8)
print("2 в 10 степени:", 2 ** 10)
````

Обрати внимание, как быстро растут числа: от пятой степени
до десятой — рост в тридцать два раза.

### Проценты

````python
всего = 80

print("25 % от 80:", всего * 0.25)
print("50 % от 80:", всего * 0.5)
print("10 % от 80:", всего * 0.1)
````

Заметил, что ответы дробные — `20.0` вместо `20`? Так работает
умножение на дробь. На правильность это не влияет.

### Порядок действий и скобки

````python
print("2 + 3 * 4     =", 2 + 3 * 4)
print("(2 + 3) * 4   =", (2 + 3) * 4)
print("10 - 2 * 3    =", 10 - 2 * 3)
print("(10 - 2) * 3  =", (10 - 2) * 3)
````

Одни и те же числа, разный порядок — разные ответы. Если сомневаешься,
ставь скобки: лишние не мешают, а забытые портят результат.

### Деление с остатком

````python
конфет = 17
детей = 5

print("Каждому достанется:", конфет // детей)
print("Останется лишних: ", конфет % детей)
print("Проверка:", детей * (конфет // детей) + конфет % детей)
````

Последняя строка — проверка: количество на каждого, умноженное
на число детей, плюс остаток должно дать исходное число.
Так можно проверять себя всегда.
