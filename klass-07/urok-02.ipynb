## Часть 1. Теория — 15 минут

### Что такое информация

Строгого определения информации нет — как нет его у слов «время»
или «пространство». Но для нашей работы достаточно рабочего описания:

> **Информация** — это сведения об окружающем мире, которые уменьшают
> неопределённость наших знаний.

Обратите внимание на вторую половину: сообщение несёт информацию, только
если вы чего-то не знали. Фраза «завтра будет либо дождь, либо нет»
формально верна, но информации не содержит — неопределённость осталась
ровно такой же.

### Виды информации по способу восприятия

Человек получает информацию органами чувств, и по каналу восприятия
её делят на пять видов.

| Вид | Орган чувств | Доля получаемой информации |
|---|---|---|
| Зрительная | глаза | около 80–90 % |
| Звуковая | уши | около 8–10 % |
| Обонятельная | нос | около 1 % |
| Вкусовая | язык | около 1 % |
| Осязательная | кожа | около 1 % |

### Виды информации по форме представления

Компьютеру органы чувств не нужны — ему важно, **в какой форме** записана
информация: числовая, текстовая, графическая, звуковая, видео.
Одну и ту же информацию можно представить по-разному: температуру воздуха —
числом `+18`, фразой «тепло», графиком за неделю или цветом на карте погоды.

### Три информационных процесса

Что бы ни происходило с информацией, это сводится к трём процессам.

**Хранение** — информацию записывают на носитель, чтобы вернуться к ней
позже. Носителем может быть лист бумаги, флешка, память компьютера
или человеческая память.

**Передача** — информация перемещается от источника к приёмнику.
В любой передаче есть три участника и одна помеха:

```
  ИСТОЧНИК  ──→  КАНАЛ СВЯЗИ  ──→  ПРИЁМНИК
                      ↑
                     ШУМ
```

**Шум** — это всё, что мешает передаче: треск в телефонной трубке, помарка
в тетради, потерянный пакет данных в сети. Чтобы шум не искажал сообщение,
используют **избыточность**: повторяют важное, добавляют проверочные коды.
Именно так работает контрольная сумма при скачивании файла.

**Обработка** — из имеющейся информации получают новую. Обработка бывает
двух типов: с получением нового содержания (решили задачу, сделали вывод)
и с изменением формы (перевели текст, построили график, отсортировали список).

### Как отличить один процесс от другого

Простое правило, которое почти всегда работает:

* информация **осталась той же**, но лежит и ждёт → хранение;
* информация **осталась той же**, но сменила место → передача;
* информация **изменилась** — появилась новая или другая форма → обработка.

## Часть 2. Примеры с разбором — 10 минут

### Пример 1. Разбираем ситуацию

> Ученик прочитал условие задачи, решил её и записал ответ в тетрадь.

Здесь целых три процесса подряд:

1. **передача** — условие перешло от учебника (источник) к ученику (приёмник);
2. **обработка** — из условия получилось решение, то есть новая информация;
3. **хранение** — ответ записан в тетрадь, то есть на носитель.

Одно бытовое действие почти всегда раскладывается на несколько процессов.
Когда в задаче спрашивают про «главный» процесс — имеют в виду тот,
ради которого всё делалось. Здесь это обработка.

### Пример 2. Компьютер как участник процессов

Запустите ячейку: она делает всё то же самое, но в исполнении компьютера.

````python
# Хранение: значение записано в память
температура = 18

# Обработка: из имеющегося получаем новое
в_фаренгейтах = температура * 9 / 5 + 32

# Передача: результат уходит на экран, к человеку
print("Было", температура, "°C, стало", в_фаренгейтах, "°F")
````

Три строки кода — три информационных процесса. Любая программа
в конечном счёте только этим и занимается.

### Пример 3. Информация и шум

Посмотрим, что делает шум с сообщением и как помогает избыточность.

````python
сообщение = "ВСТРЕЧА В ПЯТЬ"

# Шум испортил три символа
искажённое = "ВСТ?ЕЧА В П?Т?"
print("Пришло:", искажённое)

# Избыточность: то же самое отправили трижды
копии = ["ВСТ?ЕЧА В ПЯТЬ", "ВСТРЕЧА В П?ТЬ", "ВСТРЕЧ? В ПЯТ?"]

# Восстанавливаем: берём символ, который не потерялся хотя бы в одной копии
восстановлено = ""
for позиция in range(len(сообщение)):
    символы = [копия[позиция] for копия in копии if копия[позиция] != "?"]
    восстановлено += символы[0] if символы else "?"

print("Восстановлено:", восстановлено)
````

Каждая копия испорчена, но вместе они дают целое сообщение.
Ровно на этой идее построена защита данных на дисках и в сетях —
и мы к ней ещё вернёмся в старших классах.
