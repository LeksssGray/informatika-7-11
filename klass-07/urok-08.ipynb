## Часть 1. Разбор типовых ошибок — 15 минут

Четверть позади. Прежде чем писать контрольную, разберём ошибки, которые
встречаются у всех и стоят баллов на любой работе.

### Ошибка 1. Забыли, что input даёт строку

```python
n = input("Число: ")
print(n * 2)          # ввели 5, а напечаталось 55
```

Строка, умноженная на 2, повторяется дважды. Лечится преобразованием:
`n = int(input("Число: "))`.

### Ошибка 2. Путаница между `/` и `//`

`7 / 2` даёт `3.5`, а `7 // 2` даёт `3`. Если в ответе задачи должно быть
целое число купюр, страниц или коробок — нужно `//`. Если средний балл
или скорость — обычное `/`.

И наоборот: `10 / 2` даёт `5.0`, а не `5`. Если проверка ждёт целое —
добавьте `int()` или используйте `//`.

### Ошибка 3. Порядок в формуле N = 2ⁱ

В формуле $N = 2^i$ путают, что где стоит. Запомните по смыслу:
**вариантов всегда больше, чем бит**. 256 вариантов — 8 бит,
а не наоборот.

### Ошибка 4. Килобайт — это 1024, а не 1000

Все переводы между байтовыми единицами идут через 1024. Тысяча появляется
только в рекламе жёстких дисков.

### Ошибка 5. Потеря разряда при переводе в двоичную

При делении с остатком остатки читаются **снизу вверх**. Если прочитать
сверху вниз, число получится перевёрнутым: вместо `11010` выйдет `01011`.
Всегда проверяйте себя обратным переводом — это занимает десять секунд.

### Ошибка 6. Сравнение вместо присваивания

`=` присваивает, `==` сравнивает. Пока мы сравнения не использовали,
но со следующего урока начнём — и путаница станет очень частой.

## Часть 2. Разбор: находим ошибки в чужом коде — 10 минут

### Пример 1. Что здесь не так

Функция должна возвращать площадь квадрата. Найдите ошибку до запуска.

````python
def площадь_квадрата_неверно(сторона):
    площадь = сторона * 2
    print(площадь)


print("Результат вызова:", площадь_квадрата_неверно(5))
````

Ошибок сразу две:

1. **`сторона * 2` вместо `сторона ** 2`** — это периметр половины квадрата,
   а не площадь. Площадь — сторона в квадрате.
2. **`print` вместо `return`** — функция напечатала значение, но ничего
   не вернула, поэтому вызов дал `None`. Автопроверка такое не засчитает.

Правильный вариант:

````python
def площадь_квадрата(сторона):
    return сторона ** 2


print("Результат вызова:", площадь_квадрата(5))
````

### Пример 2. Крайние случаи

Функция считает, сколько коробок нужно, чтобы уложить `n` предметов
по `k` штук в коробку.

````python
def коробок_наивно(предметов, вместимость):
    return предметов // вместимость


print("10 предметов по 3:", коробок_наивно(10, 3))
````

Для десяти предметов по три ответ `3` — но в трёх коробках поместится
только девять! Оставшийся предмет требует четвёртую коробку.

Деление нацело отбрасывает остаток, а нам нужно **округление вверх**.
Классический приём:

````python
def коробок(предметов, вместимость):
    return (предметов + вместимость - 1) // вместимость


for n in [9, 10, 12, 0]:
    print(f"{n:>2} предметов по 3 → {коробок(n, 3)} коробок")
````

Хитрость в том, что мы добавляем «почти целую» порцию перед делением.
Если остаток был — он дотягивает до следующего целого; если не было —
ничего не меняется. Этот приём встретится вам ещё много раз.
