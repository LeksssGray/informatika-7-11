## Разбираемся

### Программа, которая спрашивает

До сих пор все числа мы вписывали прямо в код. Такая программа умеет
ровно одно: посчитать именно тот случай, который мы в неё заложили.
Чтобы она работала с любыми данными, их надо **вводить**.

За это отвечает `input`:

```python
имя = input("Как тебя зовут? ")
print("Привет,", имя)
```

Программа останавливается, показывает вопрос, ждёт, пока человек
напечатает ответ и нажмёт Enter, и кладёт введённое в переменную.

### Главная ловушка

> **`input` всегда возвращает строку.** Даже если ввели число.

Отсюда примерно половина ошибок у новичков:

```python
возраст = input("Возраст: ")     # ввели 13
print(возраст + 1)               # ошибка: строка плюс число
```

Введённое `13` — это текст `"13"`, а не число. Чтобы получить число,
надо сказать об этом прямо:

```python
возраст = int(input("Возраст: "))       # целое
рост = float(input("Рост в метрах: "))  # дробное
```

Читается такая запись изнутри наружу: сначала `input` берёт строку,
потом `int` превращает её в число.

### Все арифметические действия

| Что делает | Знак | Пример | Ответ |
|---|---|---|---|
| сложение | `+` | `7 + 3` | `10` |
| вычитание | `-` | `7 - 3` | `4` |
| умножение | `*` | `7 * 3` | `21` |
| деление | `/` | `7 / 3` | `2.333...` |
| деление нацело | `//` | `7 // 3` | `2` |
| остаток | `%` | `7 % 3` | `1` |
| степень | `**` | `7 ** 3` | `343` |

Две последние строки заслуживают внимания.

**Деление нацело `//`** отбрасывает дробную часть — отвечает на вопрос
«сколько целых раз одно поместилось в другом». 100 минут это
`100 // 60 = 1` полный час.

**Остаток `%`** — «сколько осталось после того, как целые забрали».
Те же 100 минут: `100 % 60 = 40` минут сверх часа.

Вместе эта пара решает почти все задачи вида «переведи одно в другое»:
минуты в часы, копейки в рубли, дни в недели.

### Порядок действий

Такой же, как в математике: сначала степень, потом умножение и деление,
потом сложение и вычитание. Скобки всё меняют:

```python
2 + 3 * 4        →  14
(2 + 3) * 4      →  20
```

Сомневаешься — ставь скобки. Лишняя ничего не испортит, недостающая
испортит ответ.

### Как красиво напечатать

Склеивать текст с числами через `+` и `str` неудобно. Есть способ
лучше — **f-строка**. Буква `f` перед кавычками разрешает вставлять
значения прямо в текст, в фигурных скобках:

```python
имя = "Аня"
лет = 13
print(f"{имя} учится в 7 классе, ей {лет} лет")
```

Внутри скобок можно писать и вычисления:

```python
print(f"Через два года будет {лет + 2}")
```

### Деление нацело и остаток в паре

Переведём минуты в часы и минуты.

````python
всего_минут = 250

часы = всего_минут // 60
минуты = всего_минут % 60

print(f"{всего_минут} минут — это {часы} ч {минуты} мин")
````

Проверим: 4 часа по 60 минут это 240, плюс 10 минут — как раз 250.
Сошлось.

Тот же приём работает для чего угодно: секунды в минуты, копейки
в рубли, сантиметры в метры. Меняется только число, на которое делим.

### Сдача купюрами

Покупка стоит 347, дали 500. Какими купюрами выдать сдачу,
если есть номиналы 100, 50, 10 и 1?

````python
сдача = 500 - 347
print(f"Сдача: {сдача} руб.")

сотни = сдача // 100
остаток = сдача % 100

полтинники = остаток // 50
остаток = остаток % 50

десятки = остаток // 10
рубли = остаток % 10

print(f"  по 100 — {сотни} шт.")
print(f"  по 50  — {полтинники} шт.")
print(f"  по 10  — {десятки} шт.")
print(f"  по 1   — {рубли} шт.")
````

Схема на каждом шаге одна и та же: делением нацело узнаём, сколько
купюр этого номинала влезет, остатком — сколько денег ещё не выдано.
Потом повторяем для следующего номинала.
