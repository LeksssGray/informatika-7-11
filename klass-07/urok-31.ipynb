## Разбираемся

### Модуль random

Случайность в программе берут из модуля `random`. Подключают его одной строчкой:

* `random.randint(a, b)` — случайное целое от `a` до `b`, **обе границы входят**;
* `random.choice(список)` — случайный элемент списка;
* `random.shuffle(список)` — перемешивает список на месте.

```python
import random

random.seed(1)  # одна и та же «случайность» при каждом запуске
print(random.randint(1, 6))
print(random.choice(["орёл", "решка"]))

колода = [1, 2, 3, 4, 5]
random.shuffle(колода)
print(колода)
```

Строка `random.seed(1)` задаёт начальное значение генератора. Числа выглядят
беспорядочно, но при одинаковом начальном значении идут одинаково. Так можно
повторить опыт и проверить его: у всех в классе выпадут одни и те же числа,
и ответы можно сверить.

### Моделируем: бросаем кубик много раз

Один бросок ничего не говорит. Тысяча бросков показывает, как часто на самом деле
выпадает каждая грань. Это и называется **моделированием**: вместо реального
кубика крутит случайность.

```python
import random

random.seed(2)
шестёрки = 0
for _ in range(1000):
    if random.randint(1, 6) == 6:
        шестёрки += 1

print("Шестёрок:", шестёрки, "из 1000")
print("Доля:", шестёрки / 1000)
```

Доля будет близка к одной шестой — примерно 0,17. Чем больше опытов, тем точнее.
