## Где обычно ошибаются

Четверть позади. Разберём ошибки, которые встречаются у всех
и стоят баллов на любой работе.

### Забыли, что input даёт строку

```python
n = input("Число: ")
print(n * 2)          # ввели 5, а вышло 55
```

Строка, умноженная на 2, повторилась дважды. Лечится так:
`n = int(input("Число: "))`.

### Перепутали `/` и `//`

`7 / 2` даёт `3.5`, а `7 // 2` даёт `3`. Если в ответе должно быть целое
число купюр, коробок или страниц — нужно `//`. Если средний балл
или скорость — обычное деление.

И наоборот: `10 / 2` даёт `5.0`, а не `5`.

### Перепутали местами N и i

В формуле $N = 2^i$ путают, что где. Запомни по смыслу:
**вариантов всегда больше, чем бит**. 256 вариантов — 8 бит,
никак не наоборот.

### Килобайт посчитали как тысячу

Все переводы между байтовыми единицами идут через 1024. Тысяча
появляется только в рекламе жёстких дисков.

### Потеряли разряд при переводе в двоичную

Остатки читаются **снизу вверх**. Прочтёшь сверху вниз — получишь
число задом наперёд: вместо `11010` выйдет `01011`.

Всегда проверяй себя обратным переводом, это двадцать секунд.

### Что проверить перед сдачей

1. Результат того типа, который ждут: число, а не строка?
2. Единицы переведены правильно — везде 1024, а не 1000?
3. Проверил обратным действием?

### Что здесь не так

Программа должна посчитать площадь квадрата со стороной 5.
Найди ошибку до запуска.

````python
сторона = 5
площадь = сторона * 2

print("Площадь:", площадь)
````

`сторона * 2` — это не площадь, а сумма двух сторон. Площадь квадрата —
сторона, умноженная **сама на себя**, то есть `сторона ** 2`
или `сторона * сторона`.

Такую ошибку трудно заметить: программа не падает, а печатает число —
просто не то.

````python
сторона = 5
площадь = сторона ** 2

print("Площадь:", площадь)
````

### Коробки, которых не хватило

Считаем, сколько нужно коробок, чтобы уложить 10 предметов по 3 штуки.

````python
предметов = 10
вместимость = 3

коробок = предметов // вместимость
print("Коробок:", коробок)
````

Получилось 3 — но в трёх коробках поместится только девять предметов!
Десятому нужна четвёртая.

Деление нацело отбрасывает остаток, а нам надо **округлить вверх**.
Есть классический приём: добавить «почти целую» порцию перед делением.

````python
предметов = 10
вместимость = 3

коробок = (предметов + вместимость - 1) // вместимость
print("Коробок:", коробок)
````

Если остаток был — добавка дотягивает до следующего целого.
Если не было — ничего не меняется. Приём встретится тебе ещё много раз,
запомни его.
