## Разбираемся

### Документ — это не просто буквы

Текст в тетради и текст в компьютере устроены по-разному. В тетради
заголовок — это то, что ты написал крупнее и подчеркнул. В документе
заголовок — **особый вид абзаца**, о котором программа знает: она
сама соберёт из заголовков оглавление, сама перенесёт их на новую
страницу, сама покрасит все сразу, если ты передумаешь.

Отсюда главное правило работы с документами: **сначала структура,
потом красота**.

### Из чего состоит документ

| Элемент | Зачем |
|---|---|
| Заголовки | делят текст на части, собираются в оглавление |
| Абзацы | основной текст, один абзац — одна мысль |
| Списки | перечисления: маркированные и нумерованные |
| Таблицы | данные, которые сравнивают между собой |
| Изображения | иллюстрации с подписями |

### Правила набора текста

Их немного, и они одинаковы во всех программах мира.

1. **Пробел ставится после знака препинания, а не до.**
   Правильно: `Привет, мир!` Неправильно: `Привет ,мир !`
2. **Один пробел, а не несколько.** Ряд пробелов для отступа —
   верный признак неумелой работы. Отступ задаётся настройкой абзаца.
3. **Enter — только в конце абзаца.** Внутри абзаца текст переносится
   сам. Нажмёшь Enter в каждой строке — и при изменении шрифта
   текст рассыплется.
4. **Дефис и тире — разные знаки.** Дефис короткий и без пробелов
   (`кто-то`), тире длинное и с пробелами (`знание — сила`).
5. **Кавычки-ёлочки** в русском тексте: «так», а не "так".

### Стили вместо ручного форматирования

Можно сделать заголовок так: выделить строку, поставить шрифт 16,
нажать «жирный», выбрать цвет. А можно применить **стиль
«Заголовок 1»**.

Выглядит одинаково, но разница огромна. Заголовков в документе
двадцать. Нужно сделать их все синими:

* с ручным форматированием — двадцать раз повторить всю цепочку;
* со стилями — один раз изменить стиль, и все двадцать заголовков
  изменятся сами.

> **Стиль** — именованный набор оформления, который применяется
> ко многим местам сразу. Меняешь стиль — меняется всё, где он
> применён.

Это та же идея, что переменная в программе: задал значение в одном
месте — оно подставилось везде.

### Markdown: разметка для тех, кто печатает

Все текстовые ячейки этого учебника написаны на **Markdown** —
способе размечать текст прямо символами, не отрывая рук
от клавиатуры.

| Что написать | Что получится |
|---|---|
| `# Заголовок` | заголовок первого уровня |
| `## Заголовок` | заголовок второго уровня |
| `**текст**` | **жирный** |
| `*текст*` | *курсив* |
| `* пункт` | маркированный список |
| `1. пункт` | нумерованный список |
| `` `код` `` | `моноширинный код` |

Заголовков бывает шесть уровней — от одной решётки до шести.
Больше не нужно: документ с семью уровнями вложенности не прочтёт
никто.

Markdown понимают Colab, GitHub, Telegram, заметки и половина
современных программ. Выучив семь строк таблицы выше, ты сможешь
оформлять текст почти везде.

## Смотрим, как это работает

### Считаем символы

Компьютер видит текст так же, как мы его разметили: пробелы
и знаки препинания — полноправные символы.

```python
фраза = "Привет, мир!"

print("Символов всего:", len(фраза))
print("Пробелов:", фраза.count(" "))
print("Запятых:", фраза.count(","))
```

### Лишние пробелы видно программе

Человек не отличит один пробел от двух, а `len` отличит.

```python
хорошо = "Привет, мир!"
плохо = "Привет ,  мир !"

print("Правильный набор:", len(хорошо), "символов")
print("Неряшливый набор:", len(плохо), "символов")
```

Три лишних символа на одну короткую фразу. В документе на двадцать
страниц такие мелочи превращаются в кашу при любом изменении шрифта.

### Чистим текст методом

```python
текст = "Мама  мыла   раму"

пока_чистим = текст.replace("   ", " ")
чисто = пока_чистим.replace("  ", " ")

print("Было:", текст)
print("Стало:", чисто)
```

### Любопытно

Markdown придумал в 2004 году Джон Грубер с простой целью: писать
текст так, чтобы он оставался читаемым **до** всякого превращения
в красивый вид. Поэтому разметка похожа на то, как люди и так
выделяли текст в письмах: звёздочками вокруг важного слова,
чёрточками для списка.

Расчёт оказался верным. Сегодня на Markdown написаны справки
миллионов программ, вся документация GitHub — и этот учебник тоже.
