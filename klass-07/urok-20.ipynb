## Разбираемся

### Строка — это цепочка символов

На прошлом уроке мы выяснили, что каждый символ — число. Строка же
из них составлена: `"привет"` — это шесть символов, стоящих
в определённом порядке.

Порядок и есть главное свойство строки. У каждого символа своё место,
и по этому месту его можно достать.

### Длина

Сколько символов в строке, считает функция `len`:

```python
len("привет")      # 6
```

Пробелы и знаки препинания — такие же символы, они тоже считаются.
В строке `"да, да"` шесть символов, а не четыре.

### Индексы: нумерация с нуля

Место символа называется **индексом**, и считают их **с нуля**:

```
 п  р  и  в  е  т
 0  1  2  3  4  5
```

```python
слово = "привет"
слово[0]      # "п" — первый символ
слово[3]      # "в"
```

Последний символ здесь — `слово[5]`, а не `слово[6]`: индексов
ровно `len(слово)`, но начинаются они с нуля и кончаются на единицу
меньше длины.

Отсюда любимая ошибка всех начинающих — выход за границу:
`слово[6]` вызовет `IndexError`.

Чтобы не считать в уме, есть отрицательные индексы — они идут
с конца:

```python
слово[-1]     # "т" — последний
слово[-2]     # "е" — предпоследний
```

### Срезы: кусок строки

Срез берёт не один символ, а целый кусок: `строка[начало:конец]`.

```python
слово = "информатика"
слово[0:5]      # "инфор"
```

Правило то же, что у `range`: **левая граница входит, правая — нет**.
Поэтому `[0:5]` — это символы с нулевого по четвёртый, пять штук.
Длина среза всегда равна `конец - начало`.

Границу можно не писать — тогда срез идёт от начала или до конца:

```python
слово[:3]       # "инф" — от начала
слово[7:]       # "тика" — до конца
```

### Склейка и повтор

Это мы уже видели в шестом уроке, но теперь понятно, почему так:

```python
"мама" + " " + "мыла"      # "мама мыла"
"ха" * 3                   # "хахаха"
```

Складывать строку с числом нельзя: `"дом" + 5` — ошибка.
Нужно сначала превратить число в строку: `"дом" + str(5)`.

### Методы: команды самой строки

У строки есть свои встроенные команды — **методы**. Пишутся они
через точку после строки:

| Метод | Что делает | Пример |
|---|---|---|
| `.upper()` | все буквы большие | `"дом".upper()` → `"ДОМ"` |
| `.lower()` | все буквы маленькие | `"ДОМ".lower()` → `"дом"` |
| `.count(x)` | сколько раз встречается | `"мама".count("м")` → `2` |
| `.replace(a, b)` | заменить одно на другое | `"кот".replace("о", "и")` → `"кит"` |

Скобки обязательны даже там, где внутри пусто: `.upper()`, а не
`.upper`.

> **Важно:** методы **не меняют** строку, а создают новую.
> После `слово.upper()` сама переменная `слово` остаётся прежней.
> Хочешь сохранить результат — присвой его: `слово = слово.upper()`.

## Смотрим, как это работает

### Длина и индексы

````python
слово = "информатика"

print("Строка:", слово)
print("Длина:", len(слово))
print("Первый символ:", слово[0])
print("Четвёртый символ:", слово[3])
print("Последний символ:", слово[-1])
````

### Перебираем символы циклом

Раз у каждого символа есть индекс, по ним можно пройти циклом.

````python
слово = "привет"

for i in range(len(слово)):
    print(i, "→", слово[i])
````

`range(len(слово))` даёт числа от 0 до 5 — ровно все допустимые
индексы. Такая связка встречается постоянно.

### Что будет, если выйти за границу

````python
# demo-error
слово = "привет"
print(слово[6])
````

`IndexError: string index out of range` — «индекс за пределами
строки». В слове из шести букв последний индекс — пятый.
Это самая частая ошибка при работе со строками, и теперь ты знаешь,
как она выглядит.

### Срезы

````python
слово = "информатика"

print(слово[0:5])
print(слово[:3])
print(слово[7:])
print(слово[3:6])
````

### Склейка и повтор

````python
имя = "Аня"
приветствие = "Привет, " + имя + "!"

print(приветствие)
print("-" * 20)
print("ля" * 4)
````

### Методы

````python
фраза = "Мама мыла раму"

print(фраза.upper())
print(фраза.lower())
print("Букв «а»:", фраза.count("а"))
print(фраза.replace("раму", "окно"))

print()
print("А сама фраза не изменилась:", фраза)
````

Последняя строка — то самое важное правило: метод вернул новую
строку, а `фраза` осталась прежней.

### Считаем гласные

Соберём вместе цикл, индексы и условие.

````python
слово = "информатика"
гласных = 0

for i in range(len(слово)):
    if слово[i] == "а" or слово[i] == "и" or слово[i] == "о":
        гласных += 1

print("Гласных а, и, о:", гласных)
````

### Любопытно

Нумерация с нуля кажется странной, но она старше самого Python.
В языке C индекс означал не «номер символа», а «на сколько шагов
отступить от начала». Первый символ стоит в самом начале —
отступ нулевой.

Большинство языков переняли эту привычку, и теперь программисты
всего мира считают с нуля. Говорят, поэтому они и в лифте
на первый этаж нажимают ноль.
