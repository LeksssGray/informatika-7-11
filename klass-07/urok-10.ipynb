## Разбираемся

### Программа, которая выбирает

Пока наши программы шли по прямой: посчитали, напечатали, конец.
Но в жизни почти всё зависит от условий. Дождь — берём зонт.
Оценка ниже тройки — пересдача. Возраст меньше шестнадцати — нельзя.

Компьютер тоже умеет выбирать. Для этого есть слово `if` — по-английски
«если».

```python
if погода == "дождь":
    print("Возьми зонт")
```

Читается почти как по-русски: **если** погода это дождь, **то**
напечатай напоминание.

### Как сравнивают

Внутри условия стоит сравнение. Знаков шесть:

| Знак | Что значит | Пример |
|---|---|---|
| `==` | равно | `возраст == 13` |
| `!=` | не равно | `имя != "Аня"` |
| `>` | больше | `баллы > 50` |
| `<` | меньше | `рост < 150` |
| `>=` | больше или равно | `возраст >= 16` |
| `<=` | меньше или равно | `цена <= 500` |

> **Два знака равно, а не один.** Один знак `=` означает «положи»,
> два `==` — «сравни». Перепутать легко, а ошибка обидная.

Результат сравнения — не число и не текст, а «правда» или «неправда».
По-английски `True` и `False`.

### Отступы решают всё

После условия ставится двоеточие, а всё, что относится к нему,
сдвигается вправо на четыре пробела:

```python
if баллы > 50:
    print("Зачёт")          ← это внутри условия
    print("Молодец")        ← и это тоже
print("Конец проверки")     ← а это уже снаружи, выполнится всегда
```

В большинстве языков для этого пишут скобки, а в Python роль скобок
играет отступ. Сдвинул строку — она внутри условия. Вернул к краю —
снаружи.

Colab обычно ставит отступ сам, когда видит двоеточие.

### А если условие не выполнилось

Тогда пригодится `else` — «иначе»:

```python
if возраст >= 16:
    доступ = "можно"
else:
    доступ = "нельзя"
```

Ровно одна из двух веток обязательно сработает. Третьего не дано.

### Чётность

Классический приём, который встретится сто раз. Число чётное, если
остаток от деления на 2 равен нулю:

```python
if число % 2 == 0:
    ответ = "чётное"
else:
    ответ = "нечётное"
```

Тот же значок `%`, что мы использовали для конфет. Здесь он отвечает
на вопрос «делится ли нацело».

## Смотрим, как это работает

### Первое ветвление

````python
температура = 3

if температура < 0:
    print("Мороз, одевайся теплее")
else:
    print("Плюс, можно налегке")

print("Хорошего дня!")
````

Последняя строка стоит **без отступа**, поэтому она печатается всегда —
и в мороз, и в тепло.

Поменяй температуру на минус пять и запусти снова.

### Чётное или нечётное

````python
число = 7

if число % 2 == 0:
    вид = "чётное"
else:
    вид = "нечётное"

print(число, "—", вид)
````

Обрати внимание: внутри веток мы не печатаем, а **кладём ответ
в переменную**. Так удобнее: результат можно использовать дальше.

### Сравнение и его результат

Само сравнение можно напечатать и посмотреть, что получится.

````python
возраст = 13

print("возраст >= 16 →", возраст >= 16)
print("возраст < 16  →", возраст < 16)
print("возраст == 13 →", возраст == 13)
print("возраст != 13 →", возраст != 13)
````

`True` — правда, `False` — неправда. Именно это значение и проверяет `if`:
если правда, ветка выполняется, если нет — пропускается.

### Частая ошибка

Ячейка ниже сломана специально: в условии стоит один знак равно
вместо двух. Запусти и посмотри на сообщение.

````python
# СПЕЦИАЛЬНО СЛОМАННАЯ ЯЧЕЙКА  demo-error
возраст = 13
if возраст = 13:
    print("Тринадцать")
````

`SyntaxError: invalid syntax` — компьютер не понял запись. Почини:
поставь `==` и запусти снова.

Это самая частая ошибка в ветвлениях. Если видишь такое сообщение
рядом с `if` — первым делом проверь количество знаков равно.
