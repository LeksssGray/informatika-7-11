## Разбираемся

### Пять алгоритмов на все случаи

Список у нас уже есть, перебирать его циклом мы умеем. Оказывается,
почти всё, что с ним делают, сводится к пяти типовым алгоритмам:

1. **сумма** — сложить все элементы;
2. **произведение** — перемножить все;
3. **максимум и минимум** — найти самый большой и самый маленький;
4. **счётчик** — сколько элементов подходят под условие;
5. **поиск индекса** — на каком месте стоит нужный элемент.

Это не «школьные упражнения»: из этих пяти кирпичиков собраны
и рейтинг в игре, и средний чек в магазине, и прогноз погоды.
В ОГЭ они встречаются каждый год.

### Сумма и произведение

Схема знакомая — копилка до цикла, накопление внутри:

```python
сумма = 0
for x in числа:
    сумма += x
```

Для произведения копилка начинается **с единицы**, иначе всё
обнулится:

```python
произведение = 1
for x in числа:
    произведение *= x
```

### Максимум: сравниваем с лучшим

Идея простая, как в соревновании. Объявляем чемпионом первый
элемент, потом идём по списку: кто сильнее — тот и новый чемпион.

```python
максимум = числа[0]
for x in числа:
    if x > максимум:
        максимум = x
```

> Начинать надо **с первого элемента списка**, а не с нуля.
> Иначе для списка из одних отрицательных чисел максимумом
> окажется ноль, которого в списке нет.

Минимум — то же самое со знаком `<`.

### Счётчик с условием

```python
чётных = 0
for x in числа:
    if x % 2 == 0:
        чётных += 1
```

Разница между счётчиком и копилкой — та самая, из урока 17:
`+= 1` отвечает на вопрос «сколько», `+= x` — на вопрос «на сколько
в сумме».

### Поиск индекса

Когда нужно не само значение, а его место, перебирают номера:

```python
индекс = 0
for i in range(len(числа)):
    if числа[i] > числа[индекс]:
        индекс = i
```

Здесь в `индекс` всё время лежит номер лучшего из просмотренных.

### Готовые функции

Три из пяти алгоритмов в Python уже написаны:

| Функция | Что делает |
|---|---|
| `sum(список)` | сумма всех элементов |
| `max(список)` | самый большой |
| `min(список)` | самый маленький |

```python
числа = [3, 8, 1, 9, 4]

sum(числа)      # 25
max(числа)      # 9
min(числа)      # 1
```

Тогда зачем мы писали циклы? По двум причинам.

Во-первых, на экзамене часто просят не просто максимум, а «максимум
среди чётных» или «сумму тех, что больше среднего» — готовая функция
такого не умеет, а цикл умеет всё.

Во-вторых, понимать, что внутри, важнее, чем знать имя. `max` —
это ровно тот цикл, который мы написали выше, просто уже набранный
за нас.

### Среднее арифметическое

Отдельной функции нет — среднее собирают из двух:

```python
среднее = sum(числа) / len(числа)
```

Деление обычное, не целочисленное: средний балл 4,25 — нормальный
ответ, а 4 — уже неправда.

## Смотрим, как это работает

### Сумма двумя способами

````python
числа = [3, 8, 1, 9, 4]

сумма = 0
for x in числа:
    сумма += x

print("Циклом:", сумма)
print("Функцией:", sum(числа))
````

### Максимум циклом

````python
числа = [3, 8, 1, 9, 4]

максимум = числа[0]
for x in числа:
    if x > максимум:
        максимум = x
        print("новый чемпион:", максимум)

print("Максимум:", максимум)
print("Проверка функцией:", max(числа))
````

Видно, как чемпион меняется по ходу: сначала 3, потом 8, потом 9.
Четвёрка в конце уже ничего не меняет.

### Минимум и среднее

````python
оценки = [5, 4, 5, 3, 4, 5]

print("Худшая оценка:", min(оценки))
print("Лучшая оценка:", max(оценки))
print("Средний балл:", sum(оценки) / len(оценки))
print("Округлённо:", round(sum(оценки) / len(оценки), 2))
````

### Счётчик с условием

````python
числа = [12, 7, 30, 5, 18, 3, 25]

больших = 0
for x in числа:
    if x > 10:
        больших += 1

print("Чисел больше десяти:", больших)
````

### Максимум среди чётных

Вот та задача, которую готовой функцией не решить.

````python
числа = [3, 8, 1, 12, 9, 6]

максимум_чётных = 0
for x in числа:
    if x % 2 == 0 and x > максимум_чётных:
        максимум_чётных = x

print("Самое большое чётное:", максимум_чётных)
````

### Где стоит максимум

````python
числа = [3, 8, 1, 9, 4]

индекс = 0
for i in range(len(числа)):
    if числа[i] > числа[индекс]:
        индекс = i

print("Максимум", числа[индекс], "стоит на месте с индексом", индекс)
````

### Любопытно

Поиск максимума за один проход кажется очевидным, но доказать,
что быстрее нельзя, — отдельная задача. Каждый элемент нужно хотя бы
раз посмотреть: пропустишь один — и именно он может оказаться самым
большим. Значит, меньше чем за $n$ сравнений не обойтись,
и наш цикл — лучшее, что бывает.

А вот найти одновременно максимум **и** минимум можно хитрее:
если брать элементы парами и сначала сравнивать их между собой,
выходит примерно на четверть меньше сравнений. Такими экономиями
занимается целая наука — теория сложности алгоритмов, и в одиннадцатом
классе вы к ней вернётесь.
