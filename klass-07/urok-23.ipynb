## Разбираемся

### Три лампочки в каждой точке

На прошлом уроке мы выяснили, что растровая картинка — сетка
пикселей, у каждого свой цвет. Осталось понять, как записать
сам цвет числом.

Если рассмотреть экран в лупу, видно, что каждый пиксель
на самом деле состоит из трёх крошечных светящихся полосок:
красной, зелёной и синей. Меняя их яркость, можно получить
любой цвет.

> **RGB** — цветовая модель, в которой цвет задаётся яркостью
> трёх составляющих: **R**ed (красной), **G**reen (зелёной),
> **B**lue (синей).

Яркость каждой составляющей записывают числом **от 0 до 255**.
Ноль — полоска не светится, 255 — светится на полную.

| Цвет | R | G | B |
|---|---|---|---|
| чёрный | 0 | 0 | 0 |
| белый | 255 | 255 | 255 |
| красный | 255 | 0 | 0 |
| зелёный | 0 | 255 | 0 |
| синий | 0 | 0 | 255 |
| жёлтый | 255 | 255 | 0 |
| серый | 128 | 128 | 128 |

Жёлтый из красного и зелёного — неожиданно, если помнить уроки
рисования. Но краски и свет смешиваются по-разному: краски
поглощают свет, а лампочки складывают его. Сложи красный луч
с зелёным — получишь жёлтый, сложи все три — белый.

### Почему именно 255

Потому что на одну составляющую отводится **один байт** —
8 бит, а это $2^8 = 256$ значений: от 0 до 255.

Три составляющие по байту — **3 байта, или 24 бита** на пиксель.
Сколько всего цветов получается?

$$256 \cdot 256 \cdot 256 = 16\,777\,216$$

Почти семнадцать миллионов оттенков. Это больше, чем человеческий
глаз способен различить, поэтому 24-битный цвет называют
**истинным цветом** (True Color).

### Глубина цвета

> **Глубина цвета** $i$ — сколько бит отводится на один пиксель.

Она напрямую задаёт размер палитры по знакомой формуле:

$$N = 2^i$$

| Глубина | Цветов | Где встречается |
|---|---|---|
| 1 бит | 2 | чёрно-белое изображение |
| 4 бита | 16 | старые игры |
| 8 бит | 256 | GIF, иконки |
| 16 бит | 65 536 | High Color |
| 24 бита | 16 777 216 | True Color, фотографии |

Формула та же, что в четвёртом уроке про алфавит: вариантов всегда
больше, чем бит. Перепутать нельзя — 8 бит дают 256 цветов,
а не наоборот.

### Объём изображения

Теперь соберём всё вместе. Объём файла — это количество пикселей,
умноженное на глубину цвета:

$$I = K \cdot i = ширина \cdot высота \cdot i$$

Результат получается **в битах**. Чтобы перевести в байты — делим
на 8, в килобайты — ещё на 1024.

Пример. Картинка 100 × 200 пикселей, глубина цвета 24 бита:

* пикселей: 100 · 200 = 20 000
* бит: 20 000 · 24 = 480 000
* байт: 480 000 : 8 = 60 000
* килобайт: 60 000 : 1024 ≈ 58,6

Это **несжатый** объём. Реальный JPEG будет в десятки раз меньше —
но в задачах ОГЭ сжатие обычно не учитывают, и об этом честно
пишут в условии.

### Палитра

Когда цветов в картинке немного, хранить по три байта на пиксель
расточительно. Тогда заводят **палитру** — пронумерованный список
используемых цветов, а в пикселе хранят только номер.

Палитра из 256 цветов означает 8 бит на пиксель вместо 24 —
файл втрое легче. Так устроены GIF и старые игры: узнаваемый
вид пиксельной графики родился именно из экономии памяти.

## Смотрим, как это работает

### Сколько цветов даёт глубина

```python
for глубина in range(1, 9):
    print(f"{глубина} бит → {2 ** глубина} цветов")
```

### Истинный цвет

```python
оттенков_на_канал = 256

всего = оттенков_на_канал * оттенков_на_канал * оттенков_на_канал

print("Оттенков на канал:", оттенков_на_канал)
print("Всего цветов:", всего)
print("То же через степень:", 2 ** 24)
```

### Объём картинки

```python
ширина = 100
высота = 200
глубина = 24

пикселей = ширина * высота
бит = пикселей * глубина
байт = бит // 8

print("Пикселей:", пикселей)
print("Бит:", бит)
print("Байт:", байт)
print("Килобайт:", round(байт / 1024, 1))
```

### Что даёт палитра

```python
пикселей = 640 * 480

истинный_цвет = пикселей * 24 // 8
палитра_256 = пикселей * 8 // 8

print("True Color, байт:", истинный_цвет)
print("Палитра 256, байт:", палитра_256)
print("Экономия в", истинный_цвет // палитра_256, "раза")
```

### Собираем цвет

Так выглядит запись цвета в виде трёх чисел. Поменяй значения
и посмотри, что получится: 255, 128, 0 — оранжевый;
128, 0, 128 — фиолетовый.

```python
r = 255
g = 165
b = 0

print(f"rgb({r}, {g}, {b})")
print("Сумма яркостей:", r + g + b)
```

### Любопытно

Человеческий глаз различает примерно десять миллионов оттенков —
меньше, чем даёт True Color. Казалось бы, запас есть.

Но глаз устроен неравномерно: тёмные тона он различает гораздо
лучше светлых. Поэтому в тёмных участках фотографий 256 градаций
на канал иногда не хватает, и на плавном переходе появляются
заметные полосы. Эту беду называют **бандингом**, и борются с ней,
добавляя едва заметный шум: глаз обманывается и видит гладкий
переход там, где его нет.
