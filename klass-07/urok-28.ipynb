## Разбираемся

### Таблица — это сетка

Столбцы обозначаются буквами A, B, C…, строки — цифрами 1, 2, 3… Место на пересечении
столбца и строки — **ячейка**. Её адрес — буква столбца и номер строки: `B3`.

**Диапазон** — прямоугольный кусок таблицы, записывается через двоеточие:
`A1:A5` — пять ячеек одного столбца, `A1:C1` — три ячейки одной строки.

### Формула начинается со знака равно

Если в ячейку вписать `=A1*2`, таблица не запишет текст, а посчитает: возьмёт число
из `A1`, умножит на 2 и покажет результат. Формула живёт в ячейке и пересчитывается
сама, когда меняется `A1`.

### Копируем формулу — адреса меняются

Пусть в `B1` записано `=A1*2`. Копируем формулу вниз, в `B2`. Ссылка сдвигается
вместе с ячейкой: в `B2` окажется `=A2*2`. Такая ссылка называется **относительной**.

Иногда нужно, чтобы ссылка не двигалась. Для этого ставят знаки доллара: `$A$1`.
Копируем `=$A$1*2` вниз — везде останется `=$A$1*2`. Это **абсолютная** ссылка.

Простое правило: относительная ссылка «едет» вместе с ячейкой, абсолютная стоит на месте.
