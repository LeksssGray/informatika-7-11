## Разбираемся

### Файл

**Файл** — информация на носителе, у которой есть имя: фотография,
документ, песня, программа. У каждого файла есть:

* **имя** — чтобы его отличать;
* **расширение** — часть имени после последней точки, по нему
  компьютер понимает тип файла;
* **размер** — сколько места он занимает;
* дата создания и изменения.

```
  доклад_вулканы.docx
  └──── имя ────┘└расширение┘
```

| Расширение | Тип файла |
|---|---|
| `.txt`, `.docx`, `.pdf` | текстовые документы |
| `.jpg`, `.png`, `.gif` | изображения |
| `.mp3`, `.wav` | звук |
| `.mp4`, `.avi` | видео |
| `.csv`, `.xlsx` | таблицы |
| `.py` | программа на Python |
| `.exe` | программа для Windows |
| `.zip`, `.rar` | архивы |

### Каталоги и дерево

Файлы хранят в **каталогах** (другое название — папки). Каталоги
вкладываются друг в друга, и получается **дерево**. Самый верхний
каталог диска называют **корневым**. В Windows он обозначается
буквой диска: `C:\`.

```
  C:\
  ├── Документы
  │   └── Школа
  │       ├── Математика
  │       │   └── доклад.docx
  │       └── История
  └── Фото
      └── Лето
          └── море.jpg
```

### Полный путь

**Полный путь** — адрес файла от корня: все каталоги по порядку
и в конце имя файла. В Windows каталоги разделяет обратная косая
черта `\`:

```
C:\Документы\Школа\Математика\доклад.docx
```

В Linux, в телефонах на Android и в Colab — прямая `/`:

```
/content/sample_data/README.md
```

Задачи ОГЭ часто описывают переходы: «пользователь поднялся на уровень
вверх и спустился в каталог…». **Подняться на уровень вверх** — перейти
в каталог, внутри которого лежит текущий. **Спуститься** — зайти
во вложенный.

### Маски имён

Чтобы найти или выбрать сразу много файлов, используют **маски** —
шаблоны с особыми знаками:

| Знак | Что означает |
|---|---|
| `*` | любое количество любых символов, в том числе ни одного |
| `?` | ровно один любой символ |

| Маска | Подходит | Не подходит |
|---|---|---|
| `*.jpg` | `кот.jpg`, `a.jpg` | `кот.png` |
| `кот*.*` | `кот.jpg`, `котёнок.png`, `кот2.gif` | `мой_кот.jpg` |
| `?от.txt` | `кот.txt`, `рот.txt` | `от.txt`, `крот.txt` |
| `*.do?` | `план.doc` | `план.docx` |

Проверяй маску посимвольно: сначала часть до точки, потом расширение.
