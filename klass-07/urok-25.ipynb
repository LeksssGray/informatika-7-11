### Что такое список

> **Список** — пронумерованный набор значений под одним именем.
> Записывается в квадратных скобках через запятую.

```python
оценки = [5, 4, 5, 3, 4]
```

Одна переменная вместо пяти — и все значения остаются на месте,
каждое на своём номере.

Внутри списка может лежать что угодно: числа, строки, даже другие
списки (как в картинке на прошлом уроке).

```python
имена = ["Аня", "Борис", "Вера"]
пустой = []
```

### Длина и индексы

Всё как у строки — и это не совпадение: строка и список
устроены одинаково, оба хранят элементы по порядку.

```python
оценки = [5, 4, 5, 3, 4]

len(оценки)        # 5 — сколько элементов
оценки[0]          # 5 — первый
оценки[2]          # 5 — третий
оценки[-1]         # 4 — последний
```

Нумерация снова **с нуля**, и последний индекс на единицу меньше
длины. Выход за границу даёт `IndexError` — старый знакомый.

### Чем список отличается от строки

Отличие ровно одно, зато важное: **список можно менять**.

```python
оценки = [5, 4, 5, 3, 4]
оценки[3] = 5          # исправили тройку на пятёрку
print(оценки)          # [5, 4, 5, 5, 4]
```

Со строкой так нельзя: `слово[0] = "к"` вызовет ошибку. Строку
можно только создать заново, а список — править на месте.

### Добавить элемент: append

Список умеет расти. Метод `.append(x)` дописывает значение в конец:

```python
оценки = [5, 4]
оценки.append(3)
print(оценки)          # [5, 4, 3]
```

Обрати внимание: `append` **меняет сам список** и ничего не
возвращает. Писать `оценки = оценки.append(3)` — частая ошибка,
после неё в переменной окажется пустота.

### Перебор циклом

Главная причина, ради которой списки и нужны: по ним можно пройти.

```python
for оценка in оценки:
    print(оценка)
```

Цикл сам возьмёт первый элемент, потом второй и так до конца —
сколько бы их ни было, пять или пять тысяч.

Если нужен ещё и номер элемента, берут привычный `range(len(...))`:

```python
for i in range(len(оценки)):
    print(i, "→", оценки[i])
```

| Что нужно | Как писать |
|---|---|
| только значения | `for x in список:` |
| значения и номера | `for i in range(len(список)):` |

Первый способ короче, поэтому его и выбирают, пока номер не нужен.

## Смотрим, как это работает

### Список оценок

````python
оценки = [5, 4, 5, 3, 4]

print("Список:", оценки)
print("Длина:", len(оценки))
print("Первая оценка:", оценки[0])
print("Последняя оценка:", оценки[-1])
````

### Исправляем оценку

````python
оценки = [5, 4, 5, 3, 4]
print("Было:", оценки)

оценки[3] = 5
print("Стало:", оценки)
````

### Список растёт

````python
дела = []
print("Сначала:", дела, "— длина", len(дела))

дела.append("математика")
дела.append("информатика")
дела.append("прогулка")

print("Потом:", дела, "— длина", len(дела))
````

### Перебираем значения

````python
имена = ["Аня", "Борис", "Вера"]

for имя in имена:
    print("Привет,", имя)
````

### Перебираем с номерами

````python
оценки = [5, 4, 5, 3, 4]

for i in range(len(оценки)):
    print(f"работа {i + 1}: оценка {оценки[i]}")
````

Номер работы на единицу больше индекса — людям привычнее считать
с единицы, а Python считает с нуля. Такая поправка встречается
постоянно.

### Копилка по списку

Знакомый приём, только вместо `range` теперь список.

````python
оценки = [5, 4, 5, 3, 4]
сумма = 0

for оценка in оценки:
    сумма += оценка

print("Сумма:", сумма)
print("Средний балл:", сумма / len(оценки))
````

### Собираем список циклом

Списки не только перебирают, но и наполняют.

````python
квадраты = []

for i in range(1, 8):
    квадраты.append(i * i)

print(квадраты)
````
