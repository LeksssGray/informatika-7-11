## Разбираемся

### Когда вариантов больше двух

`if` и `else` дают выбор из двух. Но оценка бывает не «зачёт или незачёт»,
а пятёрка, четвёрка, тройка или двойка. Четыре ветки.

Можно писать вложенные условия, но получится лесенка, в которой
невозможно разобраться. Поэтому придумали `elif` — сокращение
от «else if», то есть «иначе если».

```python
if баллы >= 85:
    оценка = 5
elif баллы >= 70:
    оценка = 4
elif баллы >= 50:
    оценка = 3
else:
    оценка = 2
```

Веток может быть сколько угодно. `else` в конце ловит все остальные
случаи — его можно и не писать, если он не нужен.

### Порядок веток решает всё

Условия проверяются **сверху вниз**, и срабатывает **первое подошедшее**.
Остальные даже не смотрятся.

Поэтому если начать с самого слабого условия, всё сломается:

```python
if баллы >= 50:      # ← 90 баллов подходят сюда
    оценка = 3       #   и получают тройку
elif баллы >= 85:
    оценка = 5       #   а сюда уже никто не дойдёт
```

Отличник с 90 баллами получит тройку. Правило простое: **сначала
самые строгие условия, потом более мягкие**.

### Два условия сразу

Иногда нужно проверить не одно, а сразу два. Для этого есть три слова.

**`and`** — «и». Правда, когда **оба** условия верны:

```python
if возраст >= 14 and рост >= 145:
    доступ = "можно"
```

На аттракцион пускают, если и возраст подходит, и рост.

**`or`** — «или». Правда, когда верно **хотя бы одно**:

```python
if день == "суббота" or день == "воскресенье":
    режим = "выходной"
```

**`not`** — «не». Переворачивает правду в неправду:

```python
if not идёт_дождь:
    print("Зонт не нужен")
```

### Как это запомнить

| Связка | Когда правда | Житейски |
|---|---|---|
| `and` | оба верны | «и то, и другое» |
| `or` | хотя бы одно | «хоть что-нибудь» |
| `not` | наоборот | «а вот и нет» |

С `and` условие становится **строже**: подходит меньше случаев.
С `or` — **мягче**: подходит больше.

### Диапазон

Частая задача: проверить, попадает ли число в промежуток.

```python
if 10 <= число and число <= 20:
    ответ = "внутри"
```

В Python это можно записать короче, прямо как в математике:

```python
if 10 <= число <= 20:
    ответ = "внутри"
```

Так умеют немногие языки, и это очень удобно.

## Смотрим, как это работает

### Оценка по баллам

````python
баллы = 78

if баллы >= 85:
    оценка = 5
elif баллы >= 70:
    оценка = 4
elif баллы >= 50:
    оценка = 3
else:
    оценка = 2

print("Баллов:", баллы, "→ оценка", оценка)
````

Проверка шла сверху вниз: 78 не дотянуло до 85, но прошло по 70 —
значит четвёрка, и остальные ветки уже не смотрелись.

Поменяй количество баллов и запусти снова.

### Что бывает при неверном порядке

````python
баллы = 90

# специально неправильный порядок веток
if баллы >= 50:
    оценка = 3
elif баллы >= 85:
    оценка = 5
else:
    оценка = 2

print("Отличник с 90 баллами получил:", оценка)
````

Девяносто баллов подошли под самое первое условие и получили тройку.
Ветка с пятёркой недостижима в принципе — до неё просто не доходит
очередь.

Ошибка коварна тем, что программа не падает. Она честно работает
и выдаёт неправильный ответ.

### and, or, not

````python
возраст = 15
рост = 150

print("возраст >= 14 and рост >= 145 →", возраст >= 14 and рост >= 145)
print("возраст >= 16 and рост >= 145 →", возраст >= 16 and рост >= 145)
print("возраст >= 16 or  рост >= 145 →", возраст >= 16 or рост >= 145)
print("not (возраст >= 16)           →", not (возраст >= 16))
````

Вторая строка: возраст не подошёл, и всё условие стало неправдой —
`and` требует, чтобы верны были обе части.

Третья строка: возраст не подошёл, но рост подошёл, и `or` этого хватило.

### Диапазон

````python
число = 15

if 10 <= число <= 20:
    где = "внутри"
else:
    где = "снаружи"

print(число, "—", где)
````
