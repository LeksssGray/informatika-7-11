## Разбираемся

### Что мы умеем к концу четверти

За вторую четверть в руках оказались три инструмента, которых хватает
для огромного числа задач:

* **ветвление** — `if`, `elif`, `else`, сложные условия с `and`, `or`, `not`;
* **цикл** — `for` и `range` в трёх формах;
* **счётчик и копилка** — переменная, которая накапливает результат.

Сегодня мы соединим их вместе, разберём типичные ошибки, а потом
будет контрольная работа.

### Цикл и условие вместе

Внутри цикла можно проверять условие — и это самая рабочая связка
во всём программировании. Схема всегда одна:

```python
счётчик = 0
for i in range(1, 101):
    if i % 7 == 0:
        счётчик += 1
print(счётчик)
```

Читается по строчкам: «заведи счётчик; пройди по числам от 1 до 100;
если число делится на 7 — прибавь единицу; напечатай итог».

Заметь, что `print` стоит **после** цикла и без отступа. Будь он внутри,
число печаталось бы сто раз.

### Счётчик или копилка

Эти две вещи путают чаще всего, хотя разница простая.

| | Счётчик | Копилка |
|---|---|---|
| Что считает | сколько раз | на сколько в сумме |
| Как растёт | `счётчик += 1` | `сумма += i` |
| Вопрос задачи | «сколько чисел…» | «какова сумма…» |

Для умножения копилку начинают с единицы, а не с нуля: `произведение = 1`.
Ноль обнулил бы всё произведение.

## Где обычно ошибаются

### Копилку завели внутри цикла

```python
for i in range(1, 6):
    сумма = 0          # ← каждый раз заново!
    сумма += i
```

На каждом шаге копилка обнуляется, и в конце в ней останется
только последнее число. **Копилку заводят до цикла.**

### Потеряли правую границу

`range(1, 10)` — это числа до девяти. Если в задаче сказано
«от 1 до 10 включительно», нужен `range(1, 11)`.

Проверка на пальцах: сколько чисел в `range(a, b)`? Ровно `b - a`.

### Печатают внутри цикла вместо итога

Если вопрос задачи — одно число, то `print` должен стоять после цикла.
Десяток строк вместо ответа — верный признак лишнего отступа.

### Перепутали = и ==

Одно равно — **присваивание**, два равно — **сравнение**.
В условии всегда два: `if i % 2 == 0:`.

### Забыли двоеточие или отступ

После `for ...` и `if ...` всегда двоеточие, а тело — с отступом
в четыре пробела. Python об этом честно скажет, но на бумаге
такую ошибку никто не подскажет.

### Что проверить перед сдачей

1. Копилка заведена **до** цикла?
2. Границы `range` включают то, что нужно?
3. `print` стоит после цикла?
4. В условии два знака равенства?

## Ищем чужие ошибки

### Сумма, которой не было

Программа должна сложить числа от 1 до 10. Найди ошибку до запуска.

````python
for i in range(1, 11):
    сумма = 0
    сумма += i

print("Сумма:", сумма)
````

Напечаталось 10 — последнее число, а не сумма. Копилка обнуляется
на каждом шаге, потому что строка `сумма = 0` попала внутрь цикла.
Вынесем её наверх:

````python
сумма = 0
for i in range(1, 11):
    сумма += i

print("Сумма:", сумма)
````

Теперь 55 — то, что нужно.

### Чётных оказалось слишком много

Программа должна посчитать чётные числа от 1 до 20.

````python
чётных = 0
for i in range(1, 21):
    if i % 2 == 0:
        чётных += i

print("Чётных чисел:", чётных)
````

Получилось 110 — столько чётных чисел быть не может, их всего двадцать
штук на весь диапазон. Здесь перепутаны счётчик и копилка: `+= i`
складывает сами числа, а вопрос был «сколько».

````python
чётных = 0
for i in range(1, 21):
    if i % 2 == 0:
        чётных += 1

print("Чётных чисел:", чётных)
````

Десять — вот теперь правда.

### Разминка перед работой

Посчитаем сразу две вещи за один проход: сколько чисел от 1 до 50
делятся на 6 и какова их сумма.

````python
сколько = 0
сумма = 0

for i in range(1, 51):
    if i % 6 == 0:
        сколько += 1
        сумма += i

print("Чисел:", сколько)
print("Их сумма:", сумма)
````

Один цикл, два накопителя — так делают почти всегда, когда вопросов
в задаче несколько.
