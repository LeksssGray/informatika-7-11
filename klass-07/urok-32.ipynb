## Разбираемся

### Таблица в виде списка списков

Строка таблицы в Python — это список. Вся таблица — список, в котором лежат строки,
то есть список списков. Первый индекс выбирает строку, второй — столбец.

```python
таблица = [
    ["Иван", 5, 4, 5],
    ["Мария", 3, 4, 4],
    ["Пётр", 5, 5, 5],
]

print(таблица[1][0])   # имя второго ученика
print(таблица[2])      # вся третья строка
print(len(таблица))    # сколько строк
```

### Когда что удобнее

Таблица удобна, когда данные нужно быстро увидеть, отсортировать глазами и посчитать
несколько простых величин. Программа выигрывает, когда одну и ту же обработку нужно
повторять с новыми данными или с множеством условий. Одну задачу полезно решить обоими
способами: так видно, где какой инструмент проще.
