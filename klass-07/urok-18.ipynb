## Разбираемся

### Когда `for` не подходит

Цикл `for` хорош, когда заранее известно, **сколько раз** повторять:
десять чисел, сто шагов, двадцать строк таблицы.

Но так бывает не всегда. «Дели число пополам, пока оно не станет
меньше единицы». «Складывай числа, пока сумма не перевалит за сто».
«Спрашивай пароль, пока не введут верный». Сколько будет повторов —
неизвестно до запуска. Зато известно **условие**, при котором
повторять нужно.

Для этого есть второй цикл — `while`.

```python
число = 100
while число > 1:
    число = число // 2
    print(число)
```

Читается: «пока число больше единицы — дели его пополам». Python
проверяет условие, выполняет тело, снова проверяет условие, снова
выполняет — и так, пока условие не станет ложным.

### Три обязательные части

У любого правильного `while` есть три части, и пропуск любой из них
ломает программу:

1. **Начальное значение** — до цикла. `число = 100`.
2. **Условие продолжения** — в заголовке. `while число > 1:`.
3. **Изменение** — в теле. `число = число // 2`.

Если забыть третью часть, условие никогда не станет ложным, и цикл
будет крутиться вечно. Это называется **бесконечный цикл**:

```python
число = 100
while число > 1:
    print(число)       # число не меняется — цикл никогда не кончится
```

Такую ячейку мы запускать не будем. Если это всё-таки случится,
в Colab есть кнопка **«Стоп»** — квадратик на месте кнопки запуска.
Ничего страшного не произойдёт: остановил и поправил.

### `for` или `while`

| | `for` | `while` |
|---|---|---|
| Когда брать | известно число повторов | известно условие |
| Типичная задача | «пройди по числам от 1 до 100» | «дели, пока не станет нулём» |
| Счётчик | заводит сам | заводишь и меняешь сам |
| Риск зациклиться | нет | есть |

Любой `for` можно переписать через `while`, но зачем: там, где число
повторов известно, `for` короче и безопаснее.

### Счётчик своими руками

В `while` переменная-счётчик не появляется сама — её заводят до цикла
и увеличивают в теле:

```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

Пропустишь `i += 1` — получишь бесконечную единицу на экране.

### Разбор числа на цифры

Самый частый приём с `while` во всей школьной информатике —
«раздеть» число по цифрам. Работают два действия:

* `число % 10` — последняя цифра;
* `число // 10` — число без последней цифры.

Повторяем, пока число не обнулится:

```python
число = 4821
while число > 0:
    цифра = число % 10      # 1, потом 2, потом 8, потом 4
    число = число // 10     # 482, потом 48, потом 4, потом 0
```

Этот приём встретится и в ОГЭ, и в ЕГЭ, и в любой олимпиаде.
Запомни его сейчас — сэкономишь годы.

## Смотрим, как это работает

### Обратный отсчёт

````python
секунд = 5
while секунд > 0:
    print(секунд)
    секунд -= 1

print("Пуск!")
````

### Пока не перевалит за сотню

Складываем числа 1, 2, 3, … и останавливаемся, как только сумма
станет больше 100. Заранее неизвестно, на каком числе это произойдёт —
ровно случай для `while`.

````python
сумма = 0
число = 0

while сумма <= 100:
    число += 1
    сумма += число
    print(f"прибавили {число}, сумма {сумма}")

print("Остановились на числе", число)
````

Обрати внимание: условие проверяется **до** очередного повтора,
поэтому последнее прибавление всё-таки перевалило за сотню.

### Делим пополам

````python
число = 1000
шагов = 0

while число > 1:
    число = число // 2
    шагов += 1
    print(f"шаг {шагов}: осталось {число}")
````

Из тысячи до единицы — всего девять шагов. Деление пополам
«съедает» большие числа поразительно быстро; на этом стоит
двоичный поиск, о котором речь пойдёт в старших классах.

### Считаем цифры числа

````python
число = 987654
цифр = 0

while число > 0:
    число = число // 10
    цифр += 1

print("Цифр в числе:", цифр)
````

### Сумма цифр

````python
число = 4821
сумма_цифр = 0

while число > 0:
    сумма_цифр += число % 10
    число = число // 10

print("Сумма цифр:", сумма_цифр)
````

Сначала берём последнюю цифру через `% 10`, потом отбрасываем её
через `// 10`. Порядок важен: отбросишь раньше — потеряешь цифру.

### Любопытно

В 1937 году японец Лотар Коллатц придумал простое правило: если число
чётное — дели пополам, если нечётное — умножай на три и прибавляй
единицу. Повторяй, пока не получится единица.

```python
число = 27
while число != 1:
    if число % 2 == 0:
        число = число // 2
    else:
        число = число * 3 + 1
```

Для 27 цикл сделает 111 шагов и поднимется до 9232, прежде чем
скатиться к единице. Похоже, что так заканчивается любое число —
компьютеры проверили это до чисел с двадцатью нулями. Но **доказать**
этого не смог никто за девяносто лет. Твой цикл из пяти строк
считает самую известную нерешённую задачу математики.
