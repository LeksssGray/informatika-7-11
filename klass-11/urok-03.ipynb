## Справочник

### Структуры данных: что когда брать

| Структура | Когда нужна | Проверка «есть ли элемент» |
|---|---|---|
| список | порядок важен, нужны индексы | $O(n)$ — медленно |
| множество | нужна только принадлежность | $O(1)$ — мгновенно |
| словарь | пары «ключ → значение», счётчики | $O(1)$ — мгновенно |

**Главная оптимизация ЕГЭ:** если внутри цикла проверяете `x in что-то`,
это «что-то» должно быть множеством или словарём, а не списком.
Иначе $O(n^2)$ вместо $O(n)$.

### Списки

```python
a[0], a[-1]              # первый, последний
a[1:4]                   # срез: элементы 1, 2, 3
a[::-1]                  # наоборот
len(a), sum(a), max(a)   # длина, сумма, максимум
a.append(x)              # добавить в конец
sorted(a)                # новый отсортированный
sorted(a, key=...)       # сортировка по своему признаку
[x * 2 for x in a]       # генератор списка
[x for x in a if x > 0]  # генератор с отбором
```

### Словари и множества

```python
счётчики = {}
счётчики[k] = счётчики.get(k, 0) + 1     # частотный счётчик

набор = set(данные)                       # множество из списка
set(a) & set(b)                           # пересечение
set(a) | set(b)                           # объединение
```

### Строки

```python
s.split()          # на слова
s.split(",")       # по разделителю
"".join(куски)     # склеить список в строку
s.count("а")       # сколько вхождений
s.find("аб")       # индекс первого вхождения или -1
s.replace("а", "б")
s.strip()          # убрать пробелы по краям
ord("A"), chr(65)  # символ ↔ код
```

Строка **неизменяема**: `s[0] = "х"` — ошибка. Меняют через срезы
и склейку.

### Матрицы

```python
m[i][j]                            # строка i, столбец j
[[0] * n for _ in range(n)]        # создание: только так
```

`[[0] * n] * n` — n ссылок на одну строку, классическая ловушка.

Диагонали: главная `m[i][i]`, побочная `m[i][n - 1 - i]`.

### Файлы

```python
with open("f.txt", encoding="utf-8") as файл:
    числа = [int(строка) for строка in файл]
```

### Числа

```python
n % 10, n // 10         # последняя цифра и её отбрасывание
n % 2 == 0              # чётность
d * d <= n              # перебор делителей до корня
pow(a, b, m)            # a^b mod m без вычисления степени
```

**Перебор делителей и проверка простоты — только до корня.**
Для чисел порядка $10^{12}$ это разница между секундой и часами.

### Сложность

| Сложность | 1000 элементов | 1 000 000 |
|---|---|---|
| $O(n)$ | мгновенно | мгновенно |
| $O(n \log n)$ | мгновенно | доли секунды |
| $O(n^2)$ | заметно | **не дождётесь** |

Правило для экзамена: если в файле 100 000 строк, вложенных циклов
по всем данным быть не должно.

### Динамическое программирование

Заводим таблицу ответов для меньших случаев и заполняем от простого
к сложному:

```python
таблица = [0] * (n + 1)
таблица[0] = 1
таблица[1] = 1
for i in range(2, n + 1):
    таблица[i] = таблица[i - 1] + таблица[i - 2]
```

Признаки задачи: спрашивают количество способов или оптимум, ответ
для n выражается через меньшие, наивный перебор не успевает.

## Смотрим, как это работает

### Множество против списка

````python
import time

данные = list(range(20000))
искомые = list(range(0, 20000, 7))

начало = time.perf_counter()
найдено = sum(1 for x in искомые if x in данные)          # список — медленно
время_список = time.perf_counter() - начало

начало = time.perf_counter()
набор = set(данные)
найдено2 = sum(1 for x in искомые if x in набор)          # множество — быстро
время_множество = time.perf_counter() - начало

print(f"Через список:    {время_список * 1000:8.1f} мс")
print(f"Через множество: {время_множество * 1000:8.1f} мс")
print(f"Ускорение в {время_список / время_множество:.0f} раз")
````

Одна строка `set(данные)` — и разница в десятки раз. Логика та же,
результат тот же, изменилась только структура данных.

### Делители за корень

````python
def делителей(n):
    количество = 0
    d = 1
    while d * d <= n:
        if n % d == 0:
            количество += 1
            if d != n // d:          # у полного квадрата не удваиваем
                количество += 1
        d += 1
    return количество


for n in [12, 36, 97, 1000000]:
    print(f"У числа {n:>8} делителей: {делителей(n)}")
````

Проверка `d != n // d` нужна для квадратов: у 36 делитель 6 нашёлся бы
дважды. Забыть её — типичная ошибка.

### Динамика: лестница

````python
n = 30
таблица = [0] * (n + 1)
таблица[0] = таблица[1] = 1
for i in range(2, n + 1):
    таблица[i] = таблица[i - 1] + таблица[i - 2]

print(f"Способов подняться на {n} ступенек: {таблица[n]}")
print("Первые десять значений:", таблица[:10])
````

Тридцать шагов вместо экспоненциального перебора. Рекурсия без таблицы
на этом же n думала бы заметно дольше, а на n = 50 не закончила бы
и за час.

### Обработка файла

````python
# готовим файл, как в задании 17
with open("числа.txt", "w", encoding="utf-8") as файл:
    for x in [17, 42, 8, 91, 23, 56, 4, 78]:
        файл.write(f"{x}\n")

with open("числа.txt", encoding="utf-8") as файл:
    числа = [int(строка) for строка in файл]

среднее = sum(числа) / len(числа)

print("Прочитано:", len(числа))
print("Среднее:", среднее)
print("Больше среднего:", sum(1 for x in числа if x > среднее))
print("Максимум:", max(числа))
````
