### Серии в строке

Серия — подряд идущие одинаковые символы. Чтобы найти самую длинную, идут по строке и держат
два счётчика: текущую серию и лучшую за всё время. Встретился другой символ — текущая серия
начинается заново.

```python
строка = "aaabbc"
лучшая = 0
текущая = 0
for i in range(len(строка)):
    if i > 0 and строка[i] == строка[i - 1]:
        текущая += 1
    else:
        текущая = 1
    лучшая = max(лучшая, текущая)
print(лучшая)
```
