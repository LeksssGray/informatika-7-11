|---|---|
| кодирование | N символов — `⌈log₂ N⌉` бит; пароль — целое число байт, округляем вверх |
| системы счисления | `int(s, b)`; цифры в системе b — деление на b с остатком; `bin(n).count("1")` |

| теория игр | выигрышная позиция — есть ход в проигрышную; проигрышная — все ходы в выигрышные |
| программирование | файлы, пары соседей, максимум среди подходящих, проверка на маленьких данных |
