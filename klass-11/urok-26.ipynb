## Разбираемся

### Последний шаг обработки данных

Схема была такая: загрузить → очистить → посчитать → **показать**.
Дошли до последнего шага, и он важнее, чем кажется: расчёт, который
никто не понял, не имеет ценности.

Задача визуализации — не украсить отчёт, а **сделать вывод
очевидным**. Если график нужно объяснять словами, он не работает.

### График под вопрос

| Вопрос | График |
|---|---|
| как менялось со временем | линейный |
| что больше, что меньше | столбчатая |
| из чего состоит целое | круговая (до 5–6 частей) |
| как распределены значения | гистограмма |
| связаны ли две величины | точечная |
| как отличается разброс групп | «ящик с усами» |

Последний вид вы, возможно, не встречали: он показывает медиану,
границы половины данных и выбросы — то есть ровно то, о чём шёл
прошлый урок.

### Обязательные подписи

* заголовок — какой вывод здесь виден;
* подписи осей с единицами;
* легенда, если рядов больше одного;
* источник и период данных.

Заголовок вроде «График 1» не годится. Хороший заголовок сам
формулирует вывод: «Медианный балл в 9А выше на 16 пунктов».

### Ошибки визуализации

| Ошибка | Что делает |
|---|---|
| обрезанная ось у столбцов | превращает 3 % разницы в «вдвое» |
| круговая на 12 секторов | ничего не читается |
| круговая, где сумма ≠ 100 % | обманывает по построению |
| двойная ось Y | позволяет «доказать» любую связь |
| 3D-эффекты | искажают сравнение |
| выбор удобного отрезка времени | честная ложь |
| восемь цветов без легенды | нечитаемо |

Правило: **у столбчатой диаграммы ось начинается с нуля**.
У линейной допустимо обрезать, но об этом пишут прямо.

### Сколько данных показывать

Показывают не всё, а то, что отвечает на вопрос. Двадцать
категорий на столбчатой диаграмме — это список, а не график;
оставляют топ-5 и «прочее».

### Отчёт как целое

Готовый отчёт — не набор картинок, а связный текст:

1. **вопрос** — что выясняли;
2. **данные** — откуда, сколько, что выбросили при очистке;
3. **метод** — как считали;
4. **результат** — таблица и один-два графика;
5. **вывод** — предложения с числами;
6. **оговорки** — чего эти данные не доказывают.

Пункт 6 отличает анализ от рекламы. Его почти всегда пропускают.

## Смотрим, как это работает

### Пример 1. Данные и агрегаты

````python
классы = ["9А", "9Б", "9В", "9Г"]
медианы = [81, 65, 74, 70]
человек = [26, 12, 24, 25]

print(f"{'класс':<6}{'медиана':>9}{'человек':>9}")
for i in range(len(классы)):
    print(f"{классы[i]:<6}{медианы[i]:>9}{человек[i]:>9}")
````

### Пример 2. Честная столбчатая: ось с нуля

````python
масштаб = 3
for i in range(len(классы)):
    полоса = "█" * (медианы[i] // масштаб)
    print(f"{классы[i]} |{полоса} {медианы[i]}")
print("   +" + "-" * 34)
print("    0" + " " * 28 + "100")
````

### Пример 3. Та же диаграмма с обрезанной осью

````python
низ = 60
for i in range(len(классы)):
    полоса = "█" * ((медианы[i] - низ) * 2)
    print(f"{классы[i]} |{полоса} {медианы[i]}")

print()
print(f"Разница 9А и 9Б: {медианы[0] - медианы[1]} пунктов "
      f"({(медианы[0] / медианы[1] - 1) * 100:.0f} %)")
print(f"А выглядит как разница в {(медианы[0] - низ) / (медианы[1] - низ):.0f} раз.")
````

### Пример 4. Когда круговая уместна

````python
ответы = [("да", 12), ("нет", 5), ("не знаю", 3)]
сумма = sum(з for _, з in ответы)

print("Опрос — части складываются в целое, круговая уместна:")
for название, значение in ответы:
    доля = значение / сумма * 100
    print(f"  {название:<9} {доля:>5.1f} % {'●' * round(доля / 4)}")

print()
кружки = [("спорт", 15), ("музыка", 9), ("рисование", 7)]
всего_детей = 25
сумма_кружков = sum(з for _, з in кружки)
print(f"Кружки: записей {сумма_кружков}, детей {всего_детей} —")
print("  сумма долей превысит 100 %, круговая соврёт. Нужна столбчатая.")
````

### Пример 5. Гистограмма распределения

````python
баллы = [45, 52, 67, 71, 73, 78, 82, 85, 88, 91, 62, 69, 74, 77, 95, 58, 81, 66]

интервалы = [(40, 59), (60, 69), (70, 79), (80, 89), (90, 100)]

print("Распределение баллов:")
for низ_и, верх_и in интервалы:
    сколько = sum(1 for б in баллы if низ_и <= б <= верх_и)
    print(f"  {низ_и}–{верх_и} |{'█' * сколько} {сколько}")
````

Гистограмма отвечает на вопрос, которого не видно в среднем:
где сгущаются результаты.

### Пример 6. Топ-5 и «прочее»

````python
категории = [("хлеб", 120), ("молоко", 95), ("сыр", 80), ("кофе", 60),
             ("чай", 45), ("сок", 20), ("вода", 15), ("конфеты", 12),
             ("печенье", 8), ("соль", 5)]

категории.sort(key=lambda пара: пара[1], reverse=True)
топ = категории[:5]
прочее = sum(з for _, з in категории[5:])

print("Что показываем на графике:")
for название, значение in топ:
    print(f"  {название:<10} {значение}")
print(f"  {'прочее':<10} {прочее}  ← {len(категории) - 5} мелких категорий вместе")
````

### Пример 7. График в matplotlib

Эта ячейка работает в Colab: там matplotlib установлен.

````python
# colab-only: рисует matplotlib
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 3.5))
plt.bar(классы, медианы, color="#4a7ebb")
plt.title("Медианный балл по классам")
plt.ylabel("баллы")
plt.ylim(0, 100)                 # ось с нуля — обязательно
for i, значение in enumerate(медианы):
    plt.text(i, значение + 2, str(значение), ha="center")
plt.show()
````

Строка `plt.ylim(0, 100)` — не украшение, а честность: без неё
matplotlib сам выберет удобный масштаб и обрежет ось.

### Пример 8. Отчёт целиком

````python
print("ВОПРОС: различается ли успеваемость по классам?")
print()
print(f"ДАННЫЕ: {sum(человек)} учеников, 4 класса, итоги четверти.")
print("МЕТОД: медианный балл по каждому классу (устойчив к выбросам).")
print()
print("РЕЗУЛЬТАТ:")
for i in range(len(классы)):
    print(f"  {классы[i]}: медиана {медианы[i]}, {человек[i]} чел.")
print()
разрыв = max(медианы) - min(медианы)
худший = классы[медианы.index(min(медианы))]
print(f"ВЫВОД: разрыв между крайними классами {разрыв} пунктов;")
print(f"  ниже всех {худший}.")
print()
print(f"ОГОВОРКИ: в {худший} только {человек[классы.index(худший)]} человек —")
print("  вдвое меньше остальных, разница может быть случайной.")
print("  Данные за одну четверть: тенденцию по ним судить нельзя.")
````
