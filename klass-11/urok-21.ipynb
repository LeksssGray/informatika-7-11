## Разбираемся

### Макет — это техническое задание

В выпускном классе дизайн интересует нас с конкретной стороны:
через неделю вы будете верстать HTML и CSS, а ещё через месяц —
делать итоговый проект. И то и другое начинается с макета.

Макет — не картинка «как будет красиво». Это **техническое задание**,
из которого вёрстка читается почти механически: вот размеры,
вот отступы, вот цвета, вот что происходит при наведении.

Если в макете этого нет, решения всё равно придётся принять —
но уже второпях и в коде.

### Порядок работы

1. **Задача пользователя.** Что человек пришёл сделать.
2. **Содержание.** Какие данные и действия нужны.
3. **Каркас** (вайрфрейм). Серые прямоугольники: что где лежит.
4. **Сетка и отступы.** Ритм, выравнивание.
5. **Оформление.** Цвет, шрифт, иконки.
6. **Состояния и прототип.** Что при наведении, что при ошибке,
   куда ведёт кнопка.

Пункт 3 не пропускают: на сером каркасе видно, работает ли
структура. Покрасить можно и потом, а перестроить готовый
цветной экран — дорого.

### Сетка и ритм

Все размеры и отступы кратны базовому шагу — обычно 8 пикселям.
Колоночная сетка для сайта: 12 колонок с промежутками, контент
выравнивается по ним.

Зачем: единый ритм убирает случайность. Проверить макет легко —
любой отступ должен делиться на 8.

### Типографика и контраст

| Параметр | Норма |
|---|---|
| Шрифтов на макете | один-два |
| Размеры | из набора 12/14/16/20/24/32 |
| Длина строки | 45–75 символов |
| Межстрочный | ≈ 1,4 от размера |
| Контраст к фону | не ниже 4,5 : 1 |

Последняя строка — требование стандарта доступности WCAG,
и его проверяют инструментами, а не глазами.

### Состояния

У каждого интерактивного элемента их минимум пять: обычное,
наведение, нажатие, отключено, загрузка. Для полей ввода
добавляются «в фокусе» и «ошибка».

Состояние «загрузка» забывают чаще всего — и получают интерфейс,
где пользователь жмёт кнопку пять раз подряд.

### Компоненты и автолейаут

**Компонент** — образец элемента. Исправили образец — изменились
все копии. Это функция в мире дизайна.

**Автолейаут** — фрейм, который сам расставляет содержимое
с заданным отступом и растягивается под контент. Это ровно то,
что в вёрстке называется flexbox: направление, отступ между
элементами, внутренние поля, выравнивание.

Совпадение не случайно: макет с автолейаутом переносится в CSS
почти построчно.

| Figma | CSS |
|---|---|
| автолейаут по горизонтали | `display: flex` |
| расстояние между элементами | `gap` |
| внутренние поля | `padding` |
| выравнивание | `align-items`, `justify-content` |
| скругление | `border-radius` |

### Что отдают верстальщику

Готовый макет — это не один экран, а набор:

* экраны во всех нужных размерах (телефон и десктоп);
* состояния элементов;
* пустые состояния: список без записей, поиск без результатов;
* состояния ошибок;
* подписи к размерам и цветам (в Figma они считываются сами).

Пустые состояния — признак взрослого макета. Именно они чаще всего
всплывают в реальном приложении в первый же день.

## Смотрим, как это работает

Figma живёт в браузере. Здесь посчитаем то, что в ней задают
руками, — и заодно получим числа для будущей вёрстки.

### Пример 1. Проверка макета по сетке

```python
шаг = 8
отступы = [8, 12, 16, 24, 30, 32, 40]

for отступ in отступы:
    вердикт = "ок" if отступ % шаг == 0 else f"выбивается (ближайшие {отступ // шаг * шаг} и {(отступ // шаг + 1) * шаг})"
    print(f"{отступ:>3} px — {вердикт}")
```

### Пример 2. Колоночная сетка

```python
ширина = 1200
колонок = 12
промежуток = 24
поля = 40

доступно = ширина - 2 * поля - промежуток * (колонок - 1)
колонка = доступно / колонок

print(f"Ширина {ширина}, поля {поля}, промежуток {промежуток}")
print(f"Ширина колонки: {колонка:.1f} px")
print(f"Блок на 4 колонки: {колонка * 4 + промежуток * 3:.1f} px")
print(f"Блок на 8 колонок: {колонка * 8 + промежуток * 7:.1f} px")
```

### Пример 3. Контраст по стандарту

```python
def яркость(цвет):
    r, g, b = цвет
    return (0.2126 * r + 0.7152 * g + 0.0722 * b) / 255


def контраст(текст, фон):
    a, b = яркость(текст), яркость(фон)
    светлее, темнее = max(a, b), min(a, b)
    return round((светлее + 0.05) / (темнее + 0.05), 2)


пары = [
    ("#333 на белом", (51, 51, 51), (255, 255, 255)),
    ("#999 на белом", (153, 153, 153), (255, 255, 255)),
    ("белый на синем", (255, 255, 255), (0, 90, 180)),
    ("белый на жёлтом", (255, 255, 255), (255, 214, 0)),
]

for название, текст, фон in пары:
    к = контраст(текст, фон)
    print(f"{название:<18} {к:>6} : 1 — {'проходит' if к >= 4.5 else 'НЕ проходит'}")
```

Последняя строка — классическая ошибка: белый текст на жёлтой
кнопке выглядит ярко и не читается.

### Пример 4. Автолейаут в числах

```python
элементов = 5
высота_элемента = 48
отступ_между = 8
поля_внутри = 16

высота = поля_внутри * 2 + элементов * высота_элемента + отступ_между * (элементов - 1)

print(f"Список из {элементов} элементов:")
print(f"  высота фрейма {высота} px")
print()
print("То же самое в CSS:")
print("  display: flex; flex-direction: column;")
print(f"  gap: {отступ_между}px; padding: {поля_внутри}px;")
```

### Пример 5. Сколько экранов рисовать

```python
экранов = 3
состояний_кнопки = 5
пустых_состояний = 2
размеров = 2       # телефон и десктоп

всего = экранов * размеров + состояний_кнопки + пустых_состояний

print(f"Экранов × размеров: {экранов * размеров}")
print(f"Состояний кнопки:   {состояний_кнопки}")
print(f"Пустых состояний:   {пустых_состояний}")
print(f"Итого артбордов:    {всего}")
print()
print("Кнопка при этом рисуется один раз — компонентом.")
```

### Пример 6. Проверяем длину строки

```python
текст = ("Макет — это техническое задание, из которого вёрстка "
         "читается почти механически")

for ширина_колонки, символов in [(320, 36), (600, 68), (1000, 110)]:
    вердикт = "хорошо" if 45 <= символов <= 75 else "плохо"
    строк = -(-len(текст) // символов)
    print(f"колонка {ширина_колонки:>5} px ≈ {символов:>3} симв. → {строк} строки, {вердикт}")
```

### Любопытно

Figma начиналась как спор с очевидностью: в 2012 году все
графические редакторы работали как настольные программы
с файлами, которые пересылают друг другу. Двое основателей
решили, что редактор должен жить в браузере, а над макетом можно
работать вдвоём одновременно, как в Google Документах.

Им говорили, что браузер не потянет графику. Потребовалось четыре
года и технология WebGL, но в 2016-м редактор вышел — и за пять
лет вытеснил настольных конкурентов почти отовсюду.

В 2022 году Adobe объявила о покупке Figma за 20 миллиардов
долларов; сделку в итоге запретили антимонопольные регуляторы.
Неплохо для идеи «а давайте сделаем это в браузере».
