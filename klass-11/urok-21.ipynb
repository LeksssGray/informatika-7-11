## Разбираемся

### Макет — это техническое задание

Через неделю вы будете верстать HTML и CSS, а ещё через месяц —
делать итоговый проект. И то и другое начинается с макета.

Макет — не картинка «как будет красиво». Это **техническое задание**,
из которого вёрстка читается почти механически: вот размеры,
вот отступы, вот цвета, вот что происходит при наведении.
Если в макете этого нет, решения всё равно придётся принять —
но уже второпях и в коде.

### Порядок работы

1. **Задача пользователя.** Что человек пришёл сделать.
2. **Содержание.** Какие данные и действия нужны.
3. **Каркас** (вайрфрейм). Серые прямоугольники: что где лежит.
4. **Сетка и отступы.** Ритм, выравнивание, шаг 8 пикселей.
5. **Оформление.** Цвет, шрифт, иконки.
6. **Состояния и прототип.** Что при наведении, что при ошибке,
   куда ведёт кнопка.

Пункт 3 не пропускают: на сером каркасе видно, работает ли
структура. Покрасить можно и потом, а перестроить готовый
цветной экран — дорого.

### Типографика и контраст

| Параметр | Норма |
|---|---|
| Шрифтов на макете | один-два |
| Размеры | из набора 12/14/16/20/24/32 |
| Длина строки | 45–75 символов |
| Межстрочный | ≈ 1,4 от размера |
| Контраст к фону | не ниже 4,5 : 1 |

### Состояния

У каждого интерактивного элемента их минимум пять: обычное,
наведение, нажатие, отключено, загрузка. Для полей ввода
добавляются «в фокусе» и «ошибка». В Figma состояния рисуют
**вариантами** (**Variants**) одного компонента.

### Компоненты и автолейаут

**Компонент** — образец элемента. Исправили образец — изменились
все копии. Это функция в мире дизайна.

**Автолейаут** — фрейм, который сам расставляет содержимое
с заданным отступом и растягивается под контент. Это ровно то,
что в вёрстке называется flexbox. Макет с автолейаутом переносится
в CSS почти построчно:

| Figma | CSS |
|---|---|
| автолейаут по горизонтали | `display: flex` |
| автолейаут по вертикали | `display: flex; flex-direction: column` |
| расстояние между элементами | `gap` |
| внутренние поля | `padding` |
| выравнивание | `align-items`, `justify-content` |
| скругление | `border-radius` |

### Что отдают верстальщику

* экраны во всех нужных размерах (телефон и десктоп);
* состояния элементов;
* пустые состояния: список без записей, поиск без результатов;
* состояния ошибок;
* карту сайта: какие страницы есть и как они связаны.

### Шаг 1. Кнопка-компонент с состояниями

Три кнопки медиаплеера будут на каждом слайде. Рисовать их на
каждом слайде заново — как копировать один и тот же код в десять
мест. Поэтому сначала делаем **компонент**.

1. Нарисуйте кнопку «вперёд»: квадрат **56 × 56**, скругление 8,
   заливка — акцентный цвет палитры; внутри треугольник-стрелка
   (**Polygon**, 3 угла, 24 × 24, повернуть на 90°).
2. Выделите квадрат и стрелку → `Shift` + `A`. Это **автолейаут**:
   стрелка встанет по центру. В правой панели в разделе
   **Auto layout** поставьте поля (**padding**) 16 со всех сторон.
3. Выделите получившийся фрейм → правая кнопка мыши →
   **Create component** (`Ctrl` + `Alt` + `K`). Фиолетовый ромб
   в списке слоёв — признак компонента. Назовите его «Кнопка».
4. Справа в разделе компонента нажмите **Add variant**
   (ромб с плюсом). Рядом появится копия — это **вариант**.
   Назовите варианты: `State=Default` и `State=Hover`.
5. Варианту **Hover** сделайте заливку на 20 % светлее и добавьте
   тень: **Effects → + → Drop shadow**.
6. Сделайте ещё два компонента по тому же образцу: «назад»
   (стрелка повёрнута на −90°) и «в начало» (квадрат 20 × 20 внутри
   вместо стрелки). Или — лучше — один компонент с вариантами
   `Icon=Next`, `Icon=Prev`, `Icon=Home`.

**Наведение.** Перейдите на вкладку **Prototype**, выделите вариант
`Default`, потяните от него стрелку к варианту `Hover` и поставьте:

| Поле | Значение |
|---|---|
| Trigger (когда) | **While hovering** |
| Action (что) | **Change to** → `Hover` |
| Animation | **Smart animate**, Ease out, 200 ms |

Теперь любая копия кнопки сама подсвечивается при наведении.

2. Все кнопки и карточки подсвечиваются при наведении —

   лабораторной.
3. Хотя бы **одна анимация Smart animate между фреймами**: например,
   слайдер на главной (три фрейма, на каждом другая картинка,
   стрелка ведёт на следующий) или раскрывающийся блок «Подробнее».

**Как работает Smart animate.** Figma сравнивает два фрейма
и ищет слои **с одинаковыми именами**. Если у слоя поменялись
место, размер или цвет — он плавно «перетекает». Поэтому:

* сделайте копию фрейма (`Ctrl` + `D`) и меняйте на ней только то,
  что должно анимироваться, — **одно изменение на пару фреймов**;
* не переименовывайте слои в копии: если в одном фрейме группа
  называется «Слайдер», а в другом «Group 12», анимации не будет —
  будет резкая смена;
* в связи поставьте **Animation → Smart animate**, Ease in and out,
  300–400 ms.

Если нужно и подсветить кнопку при наведении, и перейти по щелчку —
переход (**On click → Navigate to**) вешайте на вариант `Hover`,
ведь в момент щелчка курсор уже над кнопкой и на экране именно он.
