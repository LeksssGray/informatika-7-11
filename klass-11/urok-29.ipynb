## Разбираемся

### Функция подстановки

`ВПР` ищет значение в первом столбце диапазона и возвращает значение из другого столбца той же строки.
Так подтягивают цену по коду товара из соседней таблицы. Главное правило — искомое значение должно
стоять в первом столбце диапазона.

### Сводная таблица

Сводная таблица группирует строки по выбранному полю и считает итоги: сумму, количество, среднее.
Исходные данные при этом не меняются, а итоги пересчитываются, когда данные меняются.

### Подбор параметра

Если известен результат, а неизвестно исходное значение, таблица умеет подобрать его сама. Это
обратная задача: например, какую процентную ставку нужно, чтобы вклад вырос до заданной суммы.
