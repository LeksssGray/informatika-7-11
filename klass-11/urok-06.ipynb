## Часть 1. Теория — 15 минут

### Системы счисления на экзамене

В ЕГЭ эта тема встречается в нескольких заданиях, и почти все они
сводятся к четырём типам. Разберём каждый и подберём надёжный приём.

### Тип 1. Значение выражения в другой системе

> Найдите значение выражения $10_8 + 10_{16}$ в двоичной системе.

Приём: перевести всё в десятичную, вычислить, перевести обратно.
В Python это делается функцией `int(запись, основание)`, но на бумаге
нужно уметь и вручную.

### Тип 2. Поиск основания

> В системе с основанием q запись числа 15 имеет вид 21.
> Найдите q.

Два способа: уравнение и перебор. Перебор надёжнее, потому что
автоматически отсеивает недопустимые цифры.

> ⚠️ Всегда проверяйте: **все цифры записи должны быть меньше
> основания**. Ответ q = 3 для записи `14` неверен.

### Тип 3. Количество единиц или нулей в записи

Задачи вида «сколько единиц в двоичной записи $4^{20} + 2^{15} - 8$».
Решаются приведением к степеням двойки и рассуждением о структуре.

Сводка приёмов, которую стоит помнить наизусть:

| Выражение | Двоичная запись | Единиц |
|---|---|---|
| $2^n$ | 1 и n нулей | 1 |
| $2^n - 1$ | n единиц | n |
| $2^a + 2^b$, $a \ne b$ | две единицы в разных разрядах | 2 |
| $2^a - 2^b$, $a > b$ | $(a-b)$ единиц и b нулей | $a - b$ |

Приведение оснований: $4^n = 2^{2n}$, $8^n = 2^{3n}$, $16^n = 2^{4n}$.

### Тип 4. Последняя цифра или остаток

> Какое наименьшее основание q, при котором запись числа X
> оканчивается на цифру d?

Запись оканчивается на d ⟺ остаток от деления на q равен d.
Отсюда сразу следует условие: d < q и (X − d) делится на q.

### Универсальный приём: перебор

Заметьте, что почти все типы решаются перебором оснований от 2 до 36.
Это законный и быстрый метод — вариантов мало, а компьютер на экзамене
есть.

Схема, которая закрывает большинство задач:

```python
for q in range(2, 37):
    if условие_задачи(q):
        print(q)
```

Главное — правильно записать условие. На это и стоит тратить время,
а не на ручные выкладки.

### Полезные функции Python

| Функция | Что делает |
|---|---|
| `int("FF", 16)` | читает запись в заданной системе |
| `bin(n)`, `oct(n)`, `hex(n)` | переводит в 2, 8, 16 с приставкой |
| `f"{n:b}"`, `f"{n:o}"`, `f"{n:X}"` | то же без приставки |
| `n % q`, `n // q` | последняя цифра и «всё кроме неё» |
| `pow(a, b, m)` | $a^b \bmod m$ без вычисления степени |

Для оснований больше 16 готовой функции нет — переводить придётся
самому, циклом деления с остатком.

## Часть 2. Примеры с разбором — 10 минут

### Пример 1. Значение выражения

````python
выражения = [
    ("10₈ + 10₁₆", int("10", 8) + int("10", 16)),
    ("111₂ · 11₃", int("111", 2) * int("11", 3)),
    ("FF₁₆ − 377₈", int("FF", 16) - int("377", 8)),
    ("2024₁₀ в 16-ричной", 2024),
]

for описание, значение in выражения:
    print(f"  {описание:<22} = {значение:>6} = "
          f"{значение:b}₂ = {значение:o}₈ = {значение:X}₁₆")
````

Обратите внимание на третью строку: FF₁₆ и 377₈ — это одно и то же
число 255, поэтому разность нулевая. Такие «ловушки» в заданиях
встречаются: одинаковые значения в разных записях.

### Пример 2. Единицы в записи большого выражения

> Сколько единиц в двоичной записи числа $4^{2020} + 2^{2000} - 8$?

````python
# Рассуждение: 4^2020 = 2^4040, 8 = 2^3
# 2^4040 + 2^2000 - 2^3 = 2^4040 + (2^2000 - 2^3)
# Разность даёт 2000 - 3 = 1997 единиц, слагаемое 2^4040 добавляет одну

print("Рассуждение даёт:", 1 + (2000 - 3))

# Проверка на числе поменьше, с той же структурой
малое = 4 ** 20 + 2 ** 15 - 8
запись = bin(малое)[2:]
print(f"\nПроверка на 4^20 + 2^15 - 8:")
print(f"  рассуждение: {1 + (15 - 3)}")
print(f"  фактически:  {запись.count('1')}")
````

Настоящее число $4^{2020}$ имеет больше четырёх тысяч двоичных
разрядов — вычислять его бессмысленно. Но проверить рассуждение
на уменьшенной копии с той же структурой можно и нужно.

Это важный приём вообще: **если не уверены в рассуждении, проверьте
его на маленьком примере**.

### Пример 3. Перебор оснований

````python
# Задача: найти все основания, в которых запись числа 100
# оканчивается ровно двумя нулями

число = 100
print(f"Запись числа {число} в разных системах:")

for q in range(2, 20):
    запись = ""
    временное = число
    while временное > 0:
        цифра = временное % q
        запись = "0123456789ABCDEFGHIJ"[цифра] + запись
        временное //= q
    нулей = len(запись) - len(запись.rstrip("0"))
    отметка = "  ← ровно два нуля" if нулей == 2 else ""
    print(f"  q = {q:>2}: {запись:>10}, нулей на конце: {нулей}{отметка}")
````

Перебор сразу даёт полную картину, а не единственный ответ.
На экзамене это ценно: видно и то, что спрашивали, и то, что рядом, —
а значит, меньше шансов ошибиться в трактовке условия.
