## Повторяем

### 1. Сети и протоколы

IP доставляет пакет до компьютера, **порт** — до программы на нём.
**TCP** гарантирует доставку и порядок (страницы, файлы), **UDP** быстрее,
но без гарантий (видеозвонки, игры). **DNS** переводит имя в IP-адрес.

### 2. IP-адресация и маски

Маска `/n` — n единиц слева. Адрес сети = IP **AND** маска побайтно.
В сети `2 ** (32 − n)` адресов, узлов на два меньше: адрес сети
и широковещательный.

### 3. HTML и CSS

HTML — что на странице (`<a href>`, `<p>`, `<h1>`), CSS — как выглядит
(`color`, `font-size`). Селектор `.menu` — класс, `#main` — id, `p` — тег.

### 4. Защита информации и шифрование

Цезарь — сдвиг по алфавиту, ключей столько, сколько букв минус один.
**Симметричное** шифрование — один ключ, **асимметричное** — пара:
открытым шифруют, закрытым расшифровывают.

### 5. Табличные данные в Python

`csv.DictReader` даёт записи-словари, заголовок становится именами
полей. Типов в CSV нет: `"15" + "15"` — это `"1515"`, сначала `int`.

### 6. Группировка и агрегация

Сгруппировать → посчитать → сравнить: словарь «ключ → список» или
«ключ → сумма», потом средние. Групп столько, сколько разных ключей.

### 7. Визуализация

Доли целого — круговая, распределение — гистограмма, связь двух
величин — диаграмма рассеяния, изменение во времени — график.

### 8. Обработка файлов в задачах ЕГЭ

Типовые шаблоны: счётчик по условию, пары соседей `a[i], a[i + 1]`,
максимум среди подходящих, последняя цифра — `x % 10`.
