## Разбираемся

### Что делает браузер

Получив страницу, браузер проходит четыре шага:

1. **разбирает HTML** → дерево элементов (DOM);
2. **разбирает CSS** → правила оформления;
3. **считает раскладку** → где что лежит и какого размера;
4. **рисует** результат на экране.

Отсюда разделение ролей: **HTML — что это за содержимое**,
**CSS — как оно выглядит**. Смешивать их (делать заголовок жирным
текстом вместо `<h1>`) — значит ломать смысл документа.

### Структура документа

```html
<!DOCTYPE html>
<html lang="ru">
<head>
  <meta charset="utf-8">
  <title>Заголовок вкладки</title>
  <link rel="stylesheet" href="style.css">
</head>
<body>
  <h1>Заголовок страницы</h1>
  <p>Абзац текста.</p>
</body>
</html>
```

`head` — сведения о документе, `body` — то, что видно.
`charset="utf-8"` обязателен: без него кириллица превращается
в «кракозябры», как мы разбирали в теме кодировок.

### Семантические теги

| Тег | Смысл |
|---|---|
| `<header>` | шапка |
| `<nav>` | навигация |
| `<main>` | основное содержимое (один на страницу) |
| `<article>` | самостоятельная запись |
| `<section>` | раздел |
| `<aside>` | боковой блок |
| `<footer>` | подвал |

Можно было бы обойтись одними `<div>`. Но семантические теги
понимают поисковики и **программы экранного доступа** для незрячих:
по ним человек перемещается, как зрячий — глазами по странице.

### Частые теги

```html
<h1>…</h1> … <h6>…</h6>   заголовки шести уровней
<p>абзац</p>
<a href="адрес">ссылка</a>
<img src="файл.jpg" alt="описание">
<ul><li>пункт</li></ul>     маркированный список
<ol><li>пункт</li></ol>     нумерованный
<table><tr><td>ячейка</td></tr></table>
```

Атрибут `alt` у картинки — не мелочь: его читает экранный диктор
и показывает браузер, если картинка не загрузилась.

### Формы

```html
<form action="/search" method="get">
  <label for="q">Запрос</label>
  <input id="q" name="q" type="text" required>
  <button type="submit">Найти</button>
</form>
```

`label for` связывает подпись с полем: щелчок по подписи ставит
курсор в поле. `name` — под этим именем значение уйдёт на сервер.

### CSS: селекторы

```css
p            { color: #333; }        /* все абзацы */
.карточка    { padding: 16px; }      /* по классу */
#главное     { width: 600px; }       /* по идентификатору */
nav a        { color: blue; }        /* ссылки внутри nav */
a:hover      { text-decoration: underline; }  /* при наведении */
```

Классы (`.имя`) — основной инструмент. Идентификаторы (`#имя`)
применяют редко: они должны быть уникальны.

### Блочная модель

Каждый элемент — прямоугольник из четырёх слоёв:

```
content → padding → border → margin
```

`padding` — внутренний отступ, `margin` — внешний. Путаница между
ними — ошибка номер один у начинающих.

Строка, которую сегодня пишут почти всегда:

```css
* { box-sizing: border-box; }
```

Она заставляет `width` включать padding и border — то есть вести
себя так, как все и ожидают.

### Flexbox

```css
.список {
  display: flex;
  flex-direction: column;
  gap: 8px;
  padding: 16px;
  align-items: center;
}
```

Это ровно автолейаут из прошлого урока. Макет переносится
в CSS почти построчно — поэтому мы и рисовали его первым.

## Смотрим, как это работает

Ноутбук умеет показывать HTML прямо в выводе ячейки — так что
вёрстку можно проверять здесь же.

### Пример 1. Показать HTML в ноутбуке

````python
# colab-only: HTML рисует сам ноутбук
from IPython.display import HTML

HTML("""
<div style="font-family: sans-serif; border: 1px solid #ccc;
            border-radius: 8px; padding: 16px; max-width: 420px">
  <h3 style="margin: 0 0 8px">Карточка</h3>
  <p style="margin: 0; color: #555">Это настоящий HTML, отрисованный браузером.</p>
</div>
""")
````

### Пример 2. Структура страницы

````python
страница = """
<!DOCTYPE html>
<html lang="ru">
<head>
  <meta charset="utf-8">
  <title>Моя страница</title>
</head>
<body>
  <header><h1>Заголовок</h1></header>
  <main>
    <article><p>Текст статьи.</p></article>
  </main>
  <footer><p>2026</p></footer>
</body>
</html>
"""

for номер, строка in enumerate(страница.strip().split("\n"), 1):
    print(f"{номер:>2} | {строка}")
````

### Пример 3. Семантика против div

````python
плохо = "<div class='header'><div class='title'>Заголовок</div></div>"
хорошо = "<header><h1>Заголовок</h1></header>"

print("Для браузера выглядит одинаково.")
print("Для поисковика и экранного диктора:")
print("  плохо:  ", плохо, "→ просто два блока")
print("  хорошо: ", хорошо, "→ шапка и заголовок первого уровня")
````

### Пример 4. Блочная модель в числах

````python
def полная_ширина(content, padding, border, margin, border_box=False):
    if border_box:
        видимая = content              # width уже включает padding и border
    else:
        видимая = content + 2 * padding + 2 * border
    return видимая, видимая + 2 * margin


for режим in [False, True]:
    видимая, занимает = полная_ширина(300, 16, 1, 8, режим)
    название = "border-box" if режим else "content-box"
    print(f"{название:<12} width:300 → видимая ширина {видимая}, занимает {занимает} px")
````

Вот зачем `box-sizing: border-box`: без него элемент шириной 300
на деле занимает 334 пикселя, и вёрстка разъезжается.

### Пример 5. Flexbox из автолейаута

````python
# colab-only
HTML("""
<div style="display:flex; gap:8px; padding:16px; background:#f5f5f5;
            border-radius:8px; font-family:sans-serif; max-width:420px">
  <div style="flex:1; background:#fff; padding:12px; border-radius:6px">Раз</div>
  <div style="flex:1; background:#fff; padding:12px; border-radius:6px">Два</div>
  <div style="flex:1; background:#fff; padding:12px; border-radius:6px">Три</div>
</div>
""")
````

### Пример 6. Генерируем таблицу кодом

Типичная задача: есть данные, нужна страница. Собираем HTML
программой — так делают все генераторы отчётов.

````python
данные = [
    ("Иванов", "9А", 92),
    ("Петрова", "9Б", 78),
    ("Кузнецова", "9А", 95),
]

строки = []
for фамилия, класс, балл in данные:
    строки.append(f"<tr><td>{фамилия}</td><td>{класс}</td><td>{балл}</td></tr>")

таблица = (
    "<table style='border-collapse:collapse; font-family:sans-serif'>"
    "<tr><th style='border:1px solid #ccc;padding:6px'>Фамилия</th>"
    "<th style='border:1px solid #ccc;padding:6px'>Класс</th>"
    "<th style='border:1px solid #ccc;padding:6px'>Балл</th></tr>"
    + "".join(строки).replace("<td>", "<td style='border:1px solid #ccc;padding:6px'>")
    + "</table>"
)

# colab-only
HTML(таблица)
````
