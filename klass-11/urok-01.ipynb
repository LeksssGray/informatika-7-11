### Стратегия, которая работает

**Не решайте подряд.** Первым проходом берите то, что делаете уверенно
и быстро. Вторым — задачи посложнее. Задание 27 оставляйте напоследок:
оно самое дорогое по времени.

**Проверяйте кодом всё, что можно.** Даже задачу на логику, решённую
рассуждением, стоит подтвердить перебором за десять строк. Экзамен
на компьютере — пользуйтесь этим.

**Пишите шаблоны заранее.** Чтение файла, перебор пар, разбор числа
на цифры — эти куски кода вы напишете десятки раз. К маю они должны
писаться на автомате.

### Личный план подготовки

Составить его стоит сегодня, а не в апреле. Схема простая:

1. Прорешать диагностический вариант и записать, какие задания
   даются, а какие нет.
2. Слабые темы разбить на список и закрывать по одной в неделю.
3. Раз в месяц — полный вариант на время.

Главное — честность в первом пункте. «Вроде понимаю» и «решаю
за три минуты без ошибок» — разные вещи, и на экзамене засчитывается
только второе.

## Освежаем инструменты

### Пример 1. Шаблоны, которые понадобятся весь год

Соберём в одном месте приёмы, которые будут встречаться постоянно.

```python
число = 6072

# Разбор числа на цифры
цифры = []
временное = число
while временное > 0:
    цифры.append(временное % 10)
    временное //= 10
цифры.reverse()

print(f"Число {число}")
print(f"  цифры: {цифры}")
print(f"  сумма цифр: {sum(цифры)}")
print(f"  количество цифр: {len(цифры)}")
print(f"  наибольшая цифра: {max(цифры)}")
print(f"  число наоборот: {int(''.join(map(str, reversed(цифры))))}")
```

Разбор числа на цифры делением с остатком — самый частый приём
в заданиях 24–27. Есть и короткий путь через строку — `str(число)`, —
но он медленнее, а в задачах на большие объёмы данных это заметно.

### Пример 2. Перебор с условием

```python
# Найдём все числа от 1 до 200, у которых сумма цифр делится на 7
def сумма_цифр(n):
    сумма = 0
    while n > 0:
        сумма += n % 10
        n //= 10
    return сумма


подходящие = [n for n in range(1, 201) if сумма_цифр(n) % 7 == 0]

print(f"Найдено {len(подходящие)} чисел")
print(f"Первые десять: {подходящие[:10]}")
print(f"Наибольшее: {max(подходящие)}")
```

Полный перебор — совершенно законный метод на ЕГЭ, если диапазон
конечен. Компьютер переберёт миллион вариантов за секунду,
и попытка «решить красиво» здесь только отнимет время.

### Пример 3. Работа с файлом данных

Создадим файл и обработаем его — так устроены задания 17, 22, 26.

```python
# Готовим файл с данными
данные = [17, 42, 8, 91, 23, 56, 4, 78, 35, 62]
with open("данные.txt", "w", encoding="utf-8") as файл:
    for значение in данные:
        файл.write(f"{значение}\n")

# Читаем и обрабатываем
with open("данные.txt", encoding="utf-8") as файл:
    числа = [int(строка) for строка in файл]

print(f"Прочитано чисел: {len(числа)}")
print(f"Сумма: {sum(числа)}, среднее: {sum(числа) / len(числа):.1f}")
print(f"Больше среднего: {sum(1 for x in числа if x > sum(числа) / len(числа))}")
print(f"Максимум: {max(числа)}, минимум: {min(числа)}")
```

Конструкция `with open(...)` гарантирует, что файл закроется даже
при ошибке. Генератор `[int(строка) for строка in файл]` читает файл
построчно и сразу превращает строки в числа — компактно и быстро.

В настоящих заданиях файл содержит десятки тысяч строк, но код
от этого не меняется.
