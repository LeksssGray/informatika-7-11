## Где обычно ошибаются

Первая четверть выпускного года охватила алгоритмы, обработку чисел
и первый практикум ЕГЭ. Разберём ошибки, из-за которых на экзамене
теряют баллы при верном по сути решении.

### Ошибка 1. Перебор до n вместо перебора до корня

Проверка простоты и поиск делителей должны идти до $\sqrt{n}$.
Разница между $O(n)$ и $O(\sqrt{n})$ для чисел порядка $10^{12}$ —
это разница между «не дождётесь» и «доля секунды».

### Ошибка 2. Забыт случай полного квадрата

При поиске делителей парами: если $d \cdot d = n$, делитель считается
**один раз**. Забыв проверку `if d != n // d`, вы получите лишний
делитель у каждого квадрата.

### Ошибка 3. Единица объявлена простой

У единицы один делитель, а у простого числа их ровно два.
Проверка должна начинаться с `if n < 2: return False`.

### Ошибка 4. Наивная рекурсия вместо таблицы

Рекурсивное вычисление Фибоначчи или количества маршрутов работает
за $O(2^n)$. Для n = 50 это неразрешимо. Заводите таблицу.

### Ошибка 5. Неверная база динамики

Самая частая ошибка в задачах на исполнителей. Проверяйте таблицу
на первых трёх-четырёх значениях вручную: если они верны, дальше
формула обычно работает.

### Ошибка 6. Проигнорированы границы диапазона

«От 1 до 20» включает обе границы. В задачах на исполнителей
нужна проверка, что предшественник не меньше стартового числа.

### Ошибка 7. Поиск в списке вместо множества

`x in список` работает за $O(n)$. Внутри цикла это даёт $O(n^2)$
и не укладывается во время. Одна строка `набор = set(данные)` решает
проблему.

## Разбираем сложное

### Пример 1. Составная задача на числа

> Найдите все числа в диапазоне, которые являются произведением
> ровно двух различных простых чисел, и для каждого укажите
> наименьший из этих множителей.

```python
def разложить(n):
    множители = []
    d = 2
    while d * d <= n:
        while n % d == 0:
            множители.append(d)
            n //= d
        d += 1
    if n > 1:
        множители.append(n)
    return множители


результат = []
for n in range(2, 60):
    множители = разложить(n)
    if len(множители) == 2 and множители[0] != множители[1]:
        результат.append((n, множители[0]))

print("Числа вида p·q, где p и q — различные простые:")
for число, наименьший in результат:
    print(f"  {число:>3} = {' · '.join(map(str, разложить(число)))}, "
          f"наименьший множитель {наименьший}")
```

Условие «ровно два **различных**» проверяется двумя условиями сразу:
длина списка множителей равна двум и они не равны между собой.
Число 4 = 2·2 сюда не попадает, хотя множителей у него тоже два.

Такие тонкости формулировки — источник половины ошибок в заданиях
на числа. Читайте условие дважды.

### Пример 2. Динамика с двумя условиями

> Исполнитель умеет прибавлять 1 и умножать на 2. Сколько программ
> переводят 1 в 30, проходя через 10, но не через 20?

```python
def достижимость(старт, финиш, запрещённые=()):
    способов = {старт: 1}
    for n in range(старт + 1, финиш + 1):
        if n in запрещённые:
            способов[n] = 0
            continue
        итог = 0
        if n - 1 >= старт:
            итог += способов.get(n - 1, 0)
        if n % 2 == 0 and n // 2 >= старт:
            итог += способов.get(n // 2, 0)
        способов[n] = итог
    return способов[финиш]


до_десяти = достижимость(1, 10, {20})
после_десяти = достижимость(10, 30, {20})

print(f"Программ 1 → 10 (без 20):  {до_десяти}")
print(f"Программ 10 → 30 (без 20): {после_десяти}")
print(f"Всего через 10, но не через 20: {до_десяти * после_десяти}")
```

Два условия обработаны разными способами: обязательная точка —
разбиением пути на две части с перемножением, запрет — обнулением
ячейки.

Комбинировать эти приёмы приходится постоянно, и важно не перепутать:
«через» даёт **произведение**, «не через» — обнуление.
