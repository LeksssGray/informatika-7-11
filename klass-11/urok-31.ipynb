### Позиции и ходы

В игре с двумя игроками и полной информацией каждая позиция выигрышная или проигрышная для того,
кто ходит. Позиция **выигрышная**, если есть ход, после которого соперник оказывается в проигрышной
позиции. Позиция **проигрышная**, если любой ход ведёт в выигрышную для соперника.

Классическая игра: на столе `n` камней, за ход берут один или два. Кто взял последний — выиграл.
Пустой стол — проигрыш для того, кто должен ходить, потому что ходить уже некуда.

```python
for n in range(0, 7):
    print(n, "выигрышная" if n % 3 != 0 else "проигрышная")
```

Проверь эту закономерность на бумаге: из позиции с остатком не ноль можно сделать ход, который
оставит соперника на кратном трём. Так устроены многие задачи этого типа.
