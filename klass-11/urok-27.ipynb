### Шаблон «один проход»

Большой файл читают один раз и держат в памяти только то, что нужно для ответа: текущий
минимум, максимум, счётчик. Почти все задания этого типа решаются шаблоном:

1. завести переменные для ответа до цикла;
2. пройти по данным, обновляя их по условию;
3. вывести результат после цикла.

```python
цены = [7, 1, 5, 3, 6, 4]
минимум = цены[0]
лучшая = 0
for цена in цены:
    if цена < минимум:
        минимум = цена
    elif цена - минимум > лучшая:
        лучшая = цена - минимум
print("максимальная выгода:", лучшая)
```
