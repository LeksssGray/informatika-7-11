### Перебор наборов

Если переменных мало, все наборы значений перебирают прямо в коде. Для трёх переменных получается
восемь наборов, для четырёх — шестнадцать. Выражение вычисляется на каждом наборе, и считаются
те, на которых оно истинно.

```python
for x in (0, 1):
    for y in (0, 1):
        for z in (0, 1):
            print(x, y, z, (x or y) and not z)
```

Логические операции в Python: `and` — конъюнкция, `or` — дизъюнкция, `not` — отрицание.
Скобки ставь явно: так меньше шансов перепутать порядок.
